```python
# ============================================================
# OIP v1.0.36 — TANZANIA NPS WAVE 4
# CELL 02 — ARTIFACT INVENTORY & SHA-256
# ============================================================

from pathlib import Path
from datetime import datetime, timezone
import hashlib
import json

# ------------------------------------------------------------
# 1. PATHS
# ------------------------------------------------------------

KAGGLE_INPUT = Path("/kaggle/input")

RUN_ROOT = Path(
    "/kaggle/working/oip_v1_0_36_tanzania"
)

C02_DIR = RUN_ROOT / "c02_artifact_inventory"

C02_DIR.mkdir(
    parents=True,
    exist_ok=True
)

# ------------------------------------------------------------
# 2. FIXED IDENTITY
# ------------------------------------------------------------

PROTOCOL = "OIP v1.0.36"
STEP = "Step 3"
CELL = "C02"

DATASET = (
    "Tanzania National Panel Survey 2014-2015, Wave 4"
)

REFERENCE_ID = "TZA_2014_NPS-R4_v03_M"

timestamp_utc = (
    datetime.now(timezone.utc)
    .replace(microsecond=0)
    .isoformat()
)

# ------------------------------------------------------------
# 3. SHA-256 FUNCTION
# ------------------------------------------------------------

def sha256_file(path, chunk_size=1024 * 1024):

    h = hashlib.sha256()

    with open(path, "rb") as f:

        while True:

            chunk = f.read(chunk_size)

            if not chunk:
                break

            h.update(chunk)

    return h.hexdigest()

# ------------------------------------------------------------
# 4. RECURSIVE ARTIFACT INVENTORY
# ------------------------------------------------------------

artifacts = []

errors = []

if not KAGGLE_INPUT.exists():

    raise RuntimeError(
        "/kaggle/input is not accessible."
    )

for path in sorted(
    KAGGLE_INPUT.rglob("*"),
    key=lambda x: str(x).lower()
):

    if not path.is_file():
        continue

    try:

        relative_path = path.relative_to(
            KAGGLE_INPUT
        )

        size_bytes = path.stat().st_size

        sha256 = sha256_file(path)

        artifacts.append({

            "relative_path":
                str(relative_path),

            "filename":
                path.name,

            "extension":
                path.suffix.lower(),

            "size_bytes":
                size_bytes,

            "sha256":
                sha256,

        })

    except Exception as exc:

        errors.append({

            "path":
                str(path),

            "error":
                repr(exc),

        })

# ------------------------------------------------------------
# 5. BASIC INVENTORY SUMMARY
# ------------------------------------------------------------

extension_counts = {}

for item in artifacts:

    ext = item["extension"] or "[NO_EXTENSION]"

    extension_counts[ext] = (
        extension_counts.get(ext, 0) + 1
    )

# ------------------------------------------------------------
# 6. MACHINE-READABLE RECORD
# ------------------------------------------------------------

record = {

    "protocol":
        PROTOCOL,

    "step":
        STEP,

    "cell":
        CELL,

    "cell_purpose":
        "ARTIFACT_INVENTORY_AND_SHA256",

    "timestamp_utc":
        timestamp_utc,

    "dataset": {

        "name":
            DATASET,

        "reference_id":
            REFERENCE_ID,

    },

    "inventory": {

        "root":
            str(KAGGLE_INPUT),

        "artifact_count":
            len(artifacts),

        "extension_counts":
            extension_counts,

        "artifacts":
            artifacts,

        "errors":
            errors,

    },

    "claim_boundary": {

        "artifact_inventory_performed":
            True,

        "sha256_recorded":
            len(errors) == 0,

        "document_content_read":
            False,

        "variable_analysis":
            False,

        "construct_analysis":
            False,

        "construct_authorization":
            False,

        "identifier_authorization":
            False,

        "cohort_construction":
            False,

        "score_calculation":
            False,

        "outcome_construction":
            False,

        "empirical_evaluation":
            False,

        "empirical_validity_claim":
            False,

    },

    "next_cell":
        "C03 — DOCUMENTARY SOURCE INVENTORY",

}

# ------------------------------------------------------------
# 7. SAVE JSON
# ------------------------------------------------------------

output_json = (
    C02_DIR /
    "c02_artifact_inventory_sha256.json"
)

with open(
    output_json,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        record,
        f,
        indent=2,
        ensure_ascii=False
    )

# ------------------------------------------------------------
# 8. SAVE HUMAN-READABLE MANIFEST
# ------------------------------------------------------------

manifest_txt = (
    C02_DIR /
    "c02_artifact_manifest_sha256.txt"
)

with open(
    manifest_txt,
    "w",
    encoding="utf-8"
) as f:

    f.write(
        "OIP v1.0.36 — Tanzania NPS Wave 4\n"
    )

    f.write(
        "C02 — ARTIFACT INVENTORY & SHA-256\n"
    )

    f.write("=" * 72 + "\n\n")

    f.write(
        f"Timestamp UTC : {timestamp_utc}\n"
    )

    f.write(
        f"Artifact count: {len(artifacts)}\n\n"
    )

    for i, item in enumerate(
        artifacts,
        start=1
    ):

        f.write(
            f"{i}. {item['relative_path']}\n"
        )

        f.write(
            f"   Size   : {item['size_bytes']} bytes\n"
        )

        f.write(
            f"   SHA256 : {item['sha256']}\n\n"
        )

    if errors:

        f.write(
            "ERRORS\n"
        )

        for error in errors:

            f.write(
                f"{error}\n"
            )

# ------------------------------------------------------------
# 9. CONSOLE OUTPUT
# ------------------------------------------------------------

print("=" * 64)

print(
    "OIP v1.0.36 — TANZANIA NPS WAVE 4"
)

print(
    "C02 — ARTIFACT INVENTORY & SHA-256"
)

print("=" * 64)

print(
    f"Protocol : {PROTOCOL}"
)

print(
    f"Step     : {STEP}"
)

print(
    f"Cell     : {CELL}"
)

print(
    f"Artifacts inventoried : {len(artifacts)}"
)

print(
    f"SHA errors            : {len(errors)}"
)

print(
    "\nEXTENSIONS"
)

for ext, count in sorted(
    extension_counts.items()
):

    print(
        f"{ext} : {count}"
    )

print(
    "\nCLAIM BOUNDARY"
)

print(
    "Document content read = NO"
)

print(
    "Variable analysis     = NO"
)

print(
    "Construct analysis    = NO"
)

print(
    "Construct approval    = NO"
)

print(
    "Score calculation     = NO"
)

print(
    "Outcome construction  = NO"
)

print(
    "Empirical evaluation  = NO"
)

print(
    "\nOUTPUT"
)

print(
    output_json
)

print(
    manifest_txt
)

print(
    "\nNEXT"
)

print(
    "C03 — DOCUMENTARY SOURCE INVENTORY"
)

print("=" * 64)
```

```python
# ============================================================
# OIP v1.0.36 — TANZANIA NPS WAVE 4
# CELL 03 — ARTIFACT CLASSIFICATION
# ============================================================

from pathlib import Path
from datetime import datetime, timezone
from collections import Counter
import json

# ------------------------------------------------------------
# 1. PATHS
# ------------------------------------------------------------

RUN_ROOT = Path(
    "/kaggle/working/oip_v1_0_36_tanzania"
)

C02_JSON = (
    RUN_ROOT
    / "c02_artifact_inventory"
    / "c02_artifact_inventory_sha256.json"
)

C03_DIR = (
    RUN_ROOT
    / "c03_artifact_classification"
)

C03_DIR.mkdir(
    parents=True,
    exist_ok=True
)

# ------------------------------------------------------------
# 2. LOAD C02 — NO NEW FILE SCAN
# ------------------------------------------------------------

if not C02_JSON.exists():

    raise FileNotFoundError(
        f"C02 record not found: {C02_JSON}"
    )

with open(
    C02_JSON,
    "r",
    encoding="utf-8"
) as f:

    c02 = json.load(f)

artifacts = (
    c02["inventory"]["artifacts"]
)

# ------------------------------------------------------------
# 3. SEPARATE ARTIFACT TYPES
# ------------------------------------------------------------

pdfs = [
    a for a in artifacts
    if a["extension"] == ".pdf"
]

dtas = [
    a for a in artifacts
    if a["extension"] == ".dta"
]

others = [
    a for a in artifacts
    if a["extension"] not in (".pdf", ".dta")
]

# ------------------------------------------------------------
# 4. PDF CLASSIFICATION
# ------------------------------------------------------------
#
# Filename-based classification ONLY.
# This is NOT documentary semantic verification.
# ------------------------------------------------------------

def classify_pdf(filename):

    n = filename.lower()

    if (
        "ddi" in n
        or "documentation" in n
    ):
        return "DDI_DOCUMENTATION"

    if (
        "agriculture" in n
        or "_ag_" in n
        or "ag_" in n
    ):
        return "AGRICULTURAL_QUESTIONNAIRE"

    if (
        "community" in n
        or "_com_" in n
    ):
        return "COMMUNITY_QUESTIONNAIRE"

    if (
        "household" in n
        or "_hh_" in n
    ):
        return "HOUSEHOLD_QUESTIONNAIRE"

    if (
        "livestock" in n
        or "fishery" in n
        or "fisher" in n
    ):
        return "LIVESTOCK_FISHERIES_QUESTIONNAIRE"

    if (
        "interviewer" in n
        or "enumerator" in n
        or "manual" in n
    ):
        return "ENUMERATOR_INTERVIEWER_MANUAL"

    if (
        "basic_information" in n
        or "basic information" in n
        or "bid" in n
    ):
        return "BASIC_INFORMATION_DOCUMENT"

    if "report" in n:
        return "FINAL_REPORT"

    return "UNCLASSIFIED_PDF"

# ------------------------------------------------------------
# 5. DTA CLASSIFICATION
# ------------------------------------------------------------
#
# Prefix classification ONLY.
# No variable names or data values are interpreted.
# ------------------------------------------------------------

def classify_dta(filename):

    n = filename.lower()

    prefixes = {
        "ag_": "AGRICULTURE",
        "hh_": "HOUSEHOLD",
        "com_": "COMMUNITY",
        "ind_": "INDIVIDUAL",
        "fs_": "FOOD_SECURITY",
        "liv_": "LIVESTOCK",
        "fish_": "FISHERIES",
    }

    for prefix, label in prefixes.items():

        if n.startswith(prefix):

            return label

    return "UNCLASSIFIED_DTA"

# ------------------------------------------------------------
# 6. BUILD PDF RECORDS
# ------------------------------------------------------------

pdf_records = []

for a in pdfs:

    pdf_records.append({

        "filename":
            a["filename"],

        "relative_path":
            a["relative_path"],

        "size_bytes":
            a["size_bytes"],

        "sha256":
            a["sha256"],

        "filename_classification":
            classify_pdf(a["filename"]),

    })

# ------------------------------------------------------------
# 7. BUILD DTA RECORDS
# ------------------------------------------------------------

dta_records = []

for a in dtas:

    dta_records.append({

        "filename":
            a["filename"],

        "relative_path":
            a["relative_path"],

        "size_bytes":
            a["size_bytes"],

        "sha256":
            a["sha256"],

        "filename_module_classification":
            classify_dta(a["filename"]),

    })

# ------------------------------------------------------------
# 8. OTHER ARTIFACTS
# ------------------------------------------------------------

other_records = [

    {
        "filename":
            a["filename"],

        "relative_path":
            a["relative_path"],

        "extension":
            a["extension"],

        "size_bytes":
            a["size_bytes"],

        "sha256":
            a["sha256"],
    }

    for a in others
]

# ------------------------------------------------------------
# 9. COUNTS
# ------------------------------------------------------------

pdf_classes = Counter(
    x["filename_classification"]
    for x in pdf_records
)

dta_classes = Counter(
    x["filename_module_classification"]
    for x in dta_records
)

# ------------------------------------------------------------
# 10. MACHINE-READABLE RECORD
# ------------------------------------------------------------

record = {

    "protocol":
        "OIP v1.0.36",

    "step":
        "Step 3",

    "cell":
        "C03",

    "cell_purpose":
        "ARTIFACT_CLASSIFICATION",

    "timestamp_utc":
        datetime.now(timezone.utc)
        .replace(microsecond=0)
        .isoformat(),

    "input_record":
        str(C02_JSON),

    "counts": {

        "total_artifacts":
            len(artifacts),

        "pdf":
            len(pdfs),

        "dta":
            len(dtas),

        "other":
            len(others),
    },

    "pdf_class_counts":
        dict(sorted(pdf_classes.items())),

    "dta_module_class_counts":
        dict(sorted(dta_classes.items())),

    "pdf_artifacts":
        pdf_records,

    "dta_artifacts":
        dta_records,

    "other_artifacts":
        other_records,

    "claim_boundary": {

        "content_read":
            False,

        "pdf_content_semantic_analysis":
            False,

        "variable_analysis":
            False,

        "construct_analysis":
            False,

        "construct_authorization":
            False,

        "identifier_authorization":
            False,

        "cohort_construction":
            False,

        "score_calculation":
            False,

        "outcome_construction":
            False,

        "empirical_evaluation":
            False,

        "empirical_validity_claim":
            False,
    },

    "interpretation_boundary":
        (
            "Filename-based classifications are "
            "discovery labels only. They do not "
            "constitute documentary evidence, "
            "semantic verification, construct "
            "authorization, or variable authorization."
        ),

    "next_cell":
        "C04 — STATA SCHEMA INVENTORY",

}

# ------------------------------------------------------------
# 11. SAVE JSON
# ------------------------------------------------------------

output_json = (
    C03_DIR
    / "c03_artifact_classification.json"
)

with open(
    output_json,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        record,
        f,
        indent=2,
        ensure_ascii=False
    )

# ------------------------------------------------------------
# 12. HUMAN-READABLE SUMMARY
# ------------------------------------------------------------

summary_txt = (
    C03_DIR
    / "c03_artifact_classification.txt"
)

with open(
    summary_txt,
    "w",
    encoding="utf-8"
) as f:

    f.write(
        "OIP v1.0.36 — Tanzania NPS Wave 4\n"
    )

    f.write(
        "C03 — ARTIFACT CLASSIFICATION\n"
    )

    f.write("=" * 72 + "\n\n")

    f.write(
        f"Total artifacts : {len(artifacts)}\n"
    )

    f.write(
        f"PDF files       : {len(pdfs)}\n"
    )

    f.write(
        f"DTA files       : {len(dtas)}\n"
    )

    f.write(
        f"Other files     : {len(others)}\n\n"
    )

    f.write(
        "PDF CLASSIFICATIONS\n"
    )

    f.write("-" * 72 + "\n")

    for k, v in sorted(
        pdf_classes.items()
    ):

        f.write(
            f"{k}: {v}\n"
        )

    f.write(
        "\nDTA PREFIX CLASSIFICATIONS\n"
    )

    f.write("-" * 72 + "\n")

    for k, v in sorted(
        dta_classes.items()
    ):

        f.write(
            f"{k}: {v}\n"
        )

    f.write(
        "\nOTHER ARTIFACTS\n"
    )

    f.write("-" * 72 + "\n")

    if other_records:

        for item in other_records:

            f.write(
                f"{item['relative_path']}\n"
            )

    else:

        f.write(
            "NONE\n"
        )

    f.write(
        "\nCLAIM BOUNDARY\n"
    )

    f.write("-" * 72 + "\n")

    f.write(
        "Content read                 : NO\n"
    )

    f.write(
        "Semantic verification        : NO\n"
    )

    f.write(
        "Variable analysis            : NO\n"
    )

    f.write(
        "Construct authorization      : NO\n"
    )

    f.write(
        "Identifier authorization     : NO\n"
    )

    f.write(
        "Score calculation            : NO\n"
    )

    f.write(
        "Outcome construction         : NO\n"
    )

    f.write(
        "Empirical evaluation         : NO\n"
    )

# ------------------------------------------------------------
# 13. CONSOLE OUTPUT
# ------------------------------------------------------------

print("=" * 64)

print(
    "OIP v1.0.36 — TANZANIA NPS WAVE 4"
)

print(
    "C03 — ARTIFACT CLASSIFICATION"
)

print("=" * 64)

print(
    f"Total artifacts : {len(artifacts)}"
)

print(
    f"PDF files       : {len(pdfs)}"
)

print(
    f"DTA files       : {len(dtas)}"
)

print(
    f"Other files     : {len(others)}"
)

print(
    "\nPDF CLASSIFICATIONS"
)

for k, v in sorted(
    pdf_classes.items()
):

    print(
        f"  {k} : {v}"
    )

print(
    "\nDTA MODULE PREFIXES"
)

for k, v in sorted(
    dta_classes.items()
):

    print(
        f"  {k} : {v}"
    )

if others:

    print(
        "\nOTHER FILES — REVIEW REQUIRED"
    )

    for item in other_records:

        print(
            f"  {item['relative_path']}"
        )

print(
    "\nCLAIM BOUNDARY"
)

print(
    "Content read              = NO"
)

print(
    "Semantic verification     = NO"
)

print(
    "Variable analysis         = NO"
)

print(
    "Construct authorization   = NO"
)

print(
    "Identifier authorization  = NO"
)

print(
    "Score calculation         = NO"
)

print(
    "Outcome construction      = NO"
)

print(
    "Empirical evaluation      = NO"
)

print(
    "\nOUTPUT"
)

print(
    output_json
)

print(
    summary_txt
)

print(
    "\nNEXT"
)

print(
    "C04 — STATA SCHEMA INVENTORY"
)

print("=" * 64)
```

```python
# ============================================================
# OIP v1.0.36 — TANZANIA NPS WAVE 4
# C04-R4 — FINAL STATA SCHEMA INVENTORY
# ============================================================

from pathlib import Path
from datetime import datetime, timezone
import json
import pandas as pd

# ------------------------------------------------------------
# 1. PATHS
# ------------------------------------------------------------

KAGGLE_INPUT = Path("/kaggle/input")

RUN_ROOT = Path(
    "/kaggle/working/oip_v1_0_36_tanzania"
)

C04_DIR = RUN_ROOT / "c04_stata_schema"
C04_DIR.mkdir(parents=True, exist_ok=True)

C02_JSON = (
    RUN_ROOT
    / "c02_artifact_inventory"
    / "c02_artifact_inventory_sha256.json"
)

if not C02_JSON.exists():
    raise RuntimeError(
        f"C02 inventory not found: {C02_JSON}"
    )

# ------------------------------------------------------------
# 2. IDENTITY
# ------------------------------------------------------------

PROTOCOL = "OIP v1.0.36"
STEP = "Step 3"
CELL = "C04"

timestamp_utc = (
    datetime.now(timezone.utc)
    .replace(microsecond=0)
    .isoformat()
)

# ------------------------------------------------------------
# 3. LOAD C02
# ------------------------------------------------------------

with open(C02_JSON, "r", encoding="utf-8") as f:
    c02 = json.load(f)

all_artifacts = c02["inventory"]["artifacts"]

dta_artifacts = [
    a for a in all_artifacts
    if a["extension"].lower() == ".dta"
]

print(
    f"Total .dta artifacts from C02: "
    f"{len(dta_artifacts)}"
)

# ------------------------------------------------------------
# 4. SCHEMA INVENTORY
# ------------------------------------------------------------

schemas = []
errors = []

for idx, art in enumerate(
    dta_artifacts,
    start=1
):

    rel_path = art["relative_path"]
    full_path = KAGGLE_INPUT / rel_path

    record = {
        "relative_path": rel_path,
        "filename": art["filename"],
        "sha256": art["sha256"],
        "size_bytes": art["size_bytes"],

        "read_status": "NOT_ATTEMPTED",

        "rows": None,
        "columns": None,

        "data_label": None,

        "variable_names": [],
        "variable_labels": {},

        "value_label_names": {},

        "error": None,
    }

    # --------------------------------------------------------
    # FILE CHECK
    # --------------------------------------------------------

    if not full_path.exists():

        record["read_status"] = "FILE_NOT_FOUND"
        record["error"] = "FILE_NOT_FOUND"

        errors.append({
            "path": rel_path,
            "error": "FILE_NOT_FOUND"
        })

        schemas.append(record)
        continue

    # --------------------------------------------------------
    # STATA METADATA READER
    # --------------------------------------------------------

    try:

        # Context manager is the supported method.
        with pd.io.stata.StataReader(
            str(full_path)
        ) as reader:

            # ------------------------------------------------
            # INTERNAL STRUCTURAL METADATA
            # ------------------------------------------------

            nobs = getattr(
                reader,
                "_nobs",
                None
            )

            varlist = getattr(
                reader,
                "_varlist",
                None
            )

            nvar = getattr(
                reader,
                "_nvar",
                None
            )

            # ------------------------------------------------
            # VARIABLE NAMES
            # ------------------------------------------------

            if varlist is not None:

                record["variable_names"] = [
                    str(v)
                    for v in varlist
                ]

            else:

                # Public fallback:
                # variable_labels() returns variable-name keys.
                try:

                    labels = (
                        reader.variable_labels()
                    )

                    if isinstance(
                        labels,
                        dict
                    ):

                        record["variable_names"] = [
                            str(v)
                            for v in labels.keys()
                        ]

                except Exception:
                    pass

            # ------------------------------------------------
            # COLUMN COUNT
            # ------------------------------------------------

            if record["variable_names"]:

                record["columns"] = len(
                    record["variable_names"]
                )

            elif nvar is not None:

                try:

                    record["columns"] = int(nvar)

                except Exception:

                    record["columns"] = None

            # ------------------------------------------------
            # ROW COUNT
            # ------------------------------------------------

            if nobs is not None:

                try:

                    record["rows"] = int(nobs)

                except Exception:

                    record["rows"] = None

            # ------------------------------------------------
            # FILE-LEVEL DATA LABEL
            # ------------------------------------------------

            try:

                data_label = (
                    reader.data_label
                )

                record["data_label"] = (
                    str(data_label)
                    if data_label
                    else None
                )

            except Exception:
                pass

            # ------------------------------------------------
            # VARIABLE LABELS
            # ------------------------------------------------

            try:

                labels = (
                    reader.variable_labels()
                )

                if isinstance(
                    labels,
                    dict
                ):

                    record["variable_labels"] = {
                        str(k): str(v)
                        for k, v in labels.items()
                    }

            except Exception:
                pass

            # ------------------------------------------------
            # VALUE LABEL METADATA
            # ------------------------------------------------

            try:

                value_labels = (
                    reader.value_labels()
                )

                if isinstance(
                    value_labels,
                    dict
                ):

                    record["value_label_names"] = {
                        str(k): (
                            len(v)
                            if isinstance(
                                v,
                                dict
                            )
                            else 0
                        )
                        for k, v
                        in value_labels.items()
                    }

            except Exception:
                pass

            # ------------------------------------------------
            # SUCCESS CONDITION
            # ------------------------------------------------

            if (
                record["columns"] is not None
                and record["columns"] >= 0
                and len(
                    record["variable_names"]
                ) == record["columns"]
            ):

                record["read_status"] = "SUCCESS"

            else:

                record["read_status"] = (
                    "PARTIAL_METADATA"
                )

                record["error"] = (
                    "Structural metadata incomplete"
                )

    except Exception as exc:

        record["read_status"] = "ERROR"

        record["error"] = repr(exc)

        errors.append({
            "path": rel_path,
            "error": repr(exc)
        })

    schemas.append(record)

    if (
        idx % 15 == 0
        or idx == len(dta_artifacts)
    ):

        print(
            f"  schema progress: "
            f"{idx}/{len(dta_artifacts)}"
        )

# ------------------------------------------------------------
# 5. SUMMARY
# ------------------------------------------------------------

successful = [
    s for s in schemas
    if s["read_status"] == "SUCCESS"
]

partial = [
    s for s in schemas
    if s["read_status"] == "PARTIAL_METADATA"
]

failed = [
    s for s in schemas
    if s["read_status"] == "ERROR"
]

missing = [
    s for s in schemas
    if s["read_status"] == "FILE_NOT_FOUND"
]

total_rows = sum(
    s["rows"] or 0
    for s in successful
)

total_variable_slots = sum(
    s["columns"] or 0
    for s in successful
)

unique_var_names = set()

for s in successful:

    unique_var_names.update(
        s["variable_names"]
    )

# ------------------------------------------------------------
# 6. NAME-ONLY IDENTIFIER CANDIDATE DISCOVERY
# ------------------------------------------------------------

ID_CANDIDATE_TOKENS = [
    "hhid",
    "hh_id",
    "case_id",
    "caseid",
    "pid",
    "person_id",
    "personid",
    "individual_id",
    "individualid",
    "cluster",
    "ea",
    "ea_id",
    "strata",
    "stratum",
    "weight",
    "wgt",
    "round",
    "wave",
    "panel",
]

id_hits = {}

for s in successful:

    for variable in s["variable_names"]:

        variable_lower = (
            str(variable).lower()
        )

        for token in ID_CANDIDATE_TOKENS:

            if (
                variable_lower == token
                or variable_lower.startswith(
                    token + "_"
                )
            ):

                id_hits[token] = (
                    id_hits.get(token, 0) + 1
                )

# ------------------------------------------------------------
# 7. CLAIM BOUNDARY
# ------------------------------------------------------------

claim_boundary = {

    "schema_inventory_performed": True,

    "metadata_read": True,

    "substantive_data_values_read": False,

    "variable_interpretation": False,

    "construct_analysis": False,

    "construct_authorization": False,

    "identifier_candidate_discovery": True,

    "identifier_authorization": False,

    "outcome_construction": False,

    "outcome_authorization": False,

    "score_calculation": False,

    "empirical_evaluation": False,

    "empirical_validity_claim": False,
}

# ------------------------------------------------------------
# 8. FINAL RECORD
# ------------------------------------------------------------

record = {

    "protocol": PROTOCOL,

    "step": STEP,

    "cell": CELL,

    "cell_revision": "C04-R4",

    "cell_purpose":
        "FINAL_STATA_SCHEMA_INVENTORY",

    "timestamp_utc":
        timestamp_utc,

    "source_inventory":
        str(C02_JSON),

    "summary": {

        "dta_total":
            len(dta_artifacts),

        "schema_success":
            len(successful),

        "schema_partial":
            len(partial),

        "schema_error":
            len(failed),

        "schema_missing":
            len(missing),

        "total_rows_across_files":
            total_rows,

        "total_variable_slots_across_files":
            total_variable_slots,

        "unique_variable_names":
            len(unique_var_names),

        "identifier_token_hit_counts":
            id_hits,
    },

    "schemas":
        schemas,

    "errors":
        errors,

    "claim_boundary":
        claim_boundary,

    "next_cell":
        "C05 — DOCUMENTARY SOURCE INVENTORY",
}

# ------------------------------------------------------------
# 9. SAVE JSON
# ------------------------------------------------------------

out_json = (
    C04_DIR
    / "c04_stata_schema.json"
)

with open(
    out_json,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        record,
        f,
        indent=2,
        ensure_ascii=False
    )

# ------------------------------------------------------------
# 10. SAVE SUMMARY
# ------------------------------------------------------------

out_txt = (
    C04_DIR
    / "c04_stata_schema_summary.txt"
)

with open(
    out_txt,
    "w",
    encoding="utf-8"
) as f:

    f.write(
        "OIP v1.0.36 — TANZANIA NPS WAVE 4\n"
    )

    f.write(
        "C04-R4 — FINAL STATA SCHEMA INVENTORY\n"
    )

    f.write("=" * 64 + "\n")

    f.write(
        f".dta files              : "
        f"{len(dta_artifacts)}\n"
    )

    f.write(
        f"Schema success          : "
        f"{len(successful)}\n"
    )

    f.write(
        f"Schema partial          : "
        f"{len(partial)}\n"
    )

    f.write(
        f"Schema error            : "
        f"{len(failed)}\n"
    )

    f.write(
        f"Schema missing          : "
        f"{len(missing)}\n"
    )

    f.write(
        f"Total rows across files : "
        f"{total_rows}\n"
    )

    f.write(
        f"Variable slots          : "
        f"{total_variable_slots}\n"
    )

    f.write(
        f"Unique variable names   : "
        f"{len(unique_var_names)}\n"
    )

    f.write("\nCLAIM BOUNDARY\n")

    f.write(
        "Metadata read             = YES\n"
    )

    f.write(
        "Substantive values read   = NO\n"
    )

    f.write(
        "Variable interpretation   = NO\n"
    )

    f.write(
        "Construct analysis        = NO\n"
    )

    f.write(
        "Construct authorization   = NO\n"
    )

    f.write(
        "Identifier authorization  = NO\n"
    )

    f.write(
        "Outcome construction      = NO\n"
    )

    f.write(
        "Score calculation         = NO\n"
    )

    f.write(
        "Empirical evaluation      = NO\n"
    )

# ------------------------------------------------------------
# 11. CONSOLE OUTPUT
# ------------------------------------------------------------

print("=" * 64)
print(
    "OIP v1.0.36 — TANZANIA NPS WAVE 4"
)
print(
    "C04-R4 — FINAL STATA SCHEMA INVENTORY"
)
print("=" * 64)

print(f"Protocol : {PROTOCOL}")
print(f"Step     : {STEP}")
print(f"Cell     : {CELL}")

print()

print(
    f".dta files processed     : "
    f"{len(dta_artifacts)}"
)

print(
    f"Schema success           : "
    f"{len(successful)}"
)

print(
    f"Schema partial           : "
    f"{len(partial)}"
)

print(
    f"Schema error             : "
    f"{len(failed)}"
)

print(
    f"Schema missing           : "
    f"{len(missing)}"
)

print(
    f"Total rows across files  : "
    f"{total_rows}"
)

print(
    f"Unique variable names    : "
    f"{len(unique_var_names)}"
)

print()

print(
    "Identifier candidate hits "
    "(NAME ONLY):"
)

for key, value in sorted(
    id_hits.items()
):

    print(
        f"  {key}: {value}"
    )

print()
print("CLAIM BOUNDARY")

print(
    "Metadata read             = YES"
)

print(
    "Substantive values read   = NO"
)

print(
    "Variable interpretation   = NO"
)

print(
    "Construct analysis        = NO"
)

print(
    "Construct authorization   = NO"
)

print(
    "Identifier authorization  = NO"
)

print(
    "Outcome construction      = NO"
)

print(
    "Score calculation         = NO"
)

print(
    "Empirical evaluation      = NO"
)

print()
print("OUTPUT")
print(out_json)
print(out_txt)

print()
print(
    "NEXT: C05 — DOCUMENTARY SOURCE INVENTORY"
)

print("=" * 64)
```

```python
# ============================================================
# OIP v1.0.36 — TANZANIA NPS WAVE 4
# C05 — DOCUMENTARY SOURCE INVENTORY
# ============================================================

from pathlib import Path
from datetime import datetime, timezone
import json
import hashlib

# ------------------------------------------------------------
# 1. PATHS
# ------------------------------------------------------------

KAGGLE_INPUT = Path("/kaggle/input")

RUN_ROOT = Path(
    "/kaggle/working/oip_v1_0_36_tanzania"
)

C05_DIR = RUN_ROOT / "c05_documentary_inventory"
C05_DIR.mkdir(parents=True, exist_ok=True)

C02_JSON = (
    RUN_ROOT
    / "c02_artifact_inventory"
    / "c02_artifact_inventory_sha256.json"
)

if not C02_JSON.exists():
    raise RuntimeError(
        f"C02 inventory not found: {C02_JSON}"
    )

# ------------------------------------------------------------
# 2. IDENTITY
# ------------------------------------------------------------

PROTOCOL = "OIP v1.0.36"
STEP = "Step 3"
CELL = "C05"

timestamp_utc = (
    datetime.now(timezone.utc)
    .replace(microsecond=0)
    .isoformat()
)

# ------------------------------------------------------------
# 3. LOAD C02
# ------------------------------------------------------------

with open(C02_JSON, "r", encoding="utf-8") as f:
    c02 = json.load(f)

all_artifacts = c02["inventory"]["artifacts"]

pdf_artifacts = [
    a for a in all_artifacts
    if a["extension"].lower() == ".pdf"
]

print(
    f"Total PDF artifacts from C02: "
    f"{len(pdf_artifacts)}"
)

# ------------------------------------------------------------
# 4. PDF ENGINE
# ------------------------------------------------------------

try:
    import pypdf
    PDF_ENGINE = "pypdf"
except ImportError:
    try:
        import PyPDF2 as pypdf
        PDF_ENGINE = "PyPDF2"
    except ImportError:
        raise RuntimeError(
            "No supported PDF engine available "
            "(pypdf / PyPDF2)."
        )

# ------------------------------------------------------------
# 5. FILENAME-ONLY DOCUMENT CLASSIFICATION
# ------------------------------------------------------------

def classify_pdf(filename):

    name = filename.lower()

    if "agriculture" in name:
        return "AGRICULTURAL_QUESTIONNAIRE"

    if "community" in name:
        return "COMMUNITY_QUESTIONNAIRE"

    if "household" in name:
        return "HOUSEHOLD_QUESTIONNAIRE"

    if (
        "livestock" in name
        or "fishery" in name
    ):
        return "LIVESTOCK_FISHERIES_QUESTIONNAIRE"

    if (
        "interviewer" in name
        or "enumerator" in name
    ):
        return "ENUMERATOR_INTERVIEWER_MANUAL"

    if (
        "basic" in name
        or "bid" in name
    ):
        return "BASIC_INFORMATION_DOCUMENT"

    if "ddi" in name:
        return "DDI_DOCUMENTATION"

    if "report" in name:
        return "FINAL_REPORT"

    return "UNCLASSIFIED_PDF"

# ------------------------------------------------------------
# 6. SHA-256
# ------------------------------------------------------------

def sha256_file(path):

    h = hashlib.sha256()

    with open(path, "rb") as f:

        while True:

            chunk = f.read(1024 * 1024)

            if not chunk:
                break

            h.update(chunk)

    return h.hexdigest()

# ------------------------------------------------------------
# 7. DOCUMENT INVENTORY
# ------------------------------------------------------------

documents = []
errors = []

for idx, art in enumerate(
    pdf_artifacts,
    start=1
):

    rel_path = art["relative_path"]
    full_path = KAGGLE_INPUT / rel_path

    record = {

        "relative_path":
            rel_path,

        "filename":
            art["filename"],

        "document_category_filename_only":
            classify_pdf(art["filename"]),

        "sha256_c02":
            art["sha256"],

        "sha256_recomputed":
            None,

        "sha256_matches_c02":
            None,

        "size_bytes_c02":
            art["size_bytes"],

        "file_exists":
            False,

        "pdf_read_status":
            "NOT_ATTEMPTED",

        "page_count":
            None,

        "text_extraction_performed":
            False,

        "documentary_semantic_interpretation":
            False,

        "error":
            None,
    }

    # --------------------------------------------------------
    # FILE CHECK
    # --------------------------------------------------------

    if not full_path.exists():

        record["pdf_read_status"] = (
            "FILE_NOT_FOUND"
        )

        record["error"] = (
            "FILE_NOT_FOUND"
        )

        errors.append({
            "path": rel_path,
            "error": "FILE_NOT_FOUND"
        })

        documents.append(record)
        continue

    record["file_exists"] = True

    # --------------------------------------------------------
    # SHA RECOMPUTATION
    # --------------------------------------------------------

    try:

        recomputed = sha256_file(
            full_path
        )

        record["sha256_recomputed"] = (
            recomputed
        )

        record["sha256_matches_c02"] = (
            recomputed == art["sha256"]
        )

    except Exception as exc:

        record["error"] = repr(exc)

        errors.append({
            "path": rel_path,
            "error": repr(exc)
        })

    # --------------------------------------------------------
    # PDF STRUCTURAL READ ONLY
    # --------------------------------------------------------

    try:

        reader = pypdf.PdfReader(
            str(full_path)
        )

        record["page_count"] = len(
            reader.pages
        )

        record["pdf_read_status"] = (
            "SUCCESS"
        )

        # NO page text extraction.

        record["text_extraction_performed"] = (
            False
        )

    except Exception as exc:

        record["pdf_read_status"] = (
            "ERROR"
        )

        record["error"] = repr(exc)

        errors.append({
            "path": rel_path,
            "error": repr(exc)
        })

    documents.append(record)

    if (
        idx % 4 == 0
        or idx == len(pdf_artifacts)
    ):

        print(
            f"  documentary progress: "
            f"{idx}/{len(pdf_artifacts)}"
        )

# ------------------------------------------------------------
# 8. SUMMARY
# ------------------------------------------------------------

successful = [
    d for d in documents
    if d["pdf_read_status"] == "SUCCESS"
]

failed = [
    d for d in documents
    if d["pdf_read_status"] == "ERROR"
]

missing = [
    d for d in documents
    if d["pdf_read_status"] == "FILE_NOT_FOUND"
]

sha_mismatch = [
    d for d in documents
    if d["sha256_matches_c02"] is False
]

category_counts = {}

for d in documents:

    category = (
        d["document_category_filename_only"]
    )

    category_counts[category] = (
        category_counts.get(category, 0) + 1
    )

total_pages = sum(
    d["page_count"] or 0
    for d in successful
)

# ------------------------------------------------------------
# 9. CLAIM BOUNDARY
# ------------------------------------------------------------

claim_boundary = {

    "documentary_source_inventory":
        True,

    "pdf_structural_read":
        True,

    "filename_classification_only":
        True,

    "pdf_text_extraction":
        False,

    "documentary_semantic_interpretation":
        False,

    "variable_interpretation":
        False,

    "construct_analysis":
        False,

    "construct_authorization":
        False,

    "identifier_authorization":
        False,

    "outcome_construction":
        False,

    "outcome_authorization":
        False,

    "score_calculation":
        False,

    "empirical_evaluation":
        False,

    "empirical_validity_claim":
        False,
}

# ------------------------------------------------------------
# 10. FINAL RECORD
# ------------------------------------------------------------

record = {

    "protocol":
        PROTOCOL,

    "step":
        STEP,

    "cell":
        CELL,

    "cell_purpose":
        "DOCUMENTARY_SOURCE_INVENTORY",

    "timestamp_utc":
        timestamp_utc,

    "source_inventory":
        str(C02_JSON),

    "pdf_engine":
        PDF_ENGINE,

    "summary": {

        "pdf_total":
            len(pdf_artifacts),

        "pdf_read_success":
            len(successful),

        "pdf_read_error":
            len(failed),

        "pdf_missing":
            len(missing),

        "sha_mismatch":
            len(sha_mismatch),

        "total_pages":
            total_pages,

        "category_counts":
            category_counts,
    },

    "documents":
        documents,

    "errors":
        errors,

    "claim_boundary":
        claim_boundary,

    "next_cell":
        "C06 — DOCUMENTARY EVIDENCE EXTRACTION & SOURCE MAP",
}

# ------------------------------------------------------------
# 11. SAVE JSON
# ------------------------------------------------------------

out_json = (
    C05_DIR
    / "c05_documentary_inventory.json"
)

with open(
    out_json,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        record,
        f,
        indent=2,
        ensure_ascii=False
    )

# ------------------------------------------------------------
# 12. SAVE SUMMARY
# ------------------------------------------------------------

out_txt = (
    C05_DIR
    / "c05_documentary_inventory_summary.txt"
)

with open(
    out_txt,
    "w",
    encoding="utf-8"
) as f:

    f.write(
        "OIP v1.0.36 — TANZANIA NPS WAVE 4\n"
    )

    f.write(
        "C05 — DOCUMENTARY SOURCE INVENTORY\n"
    )

    f.write("=" * 64 + "\n")

    f.write(
        f"PDF files          : {len(pdf_artifacts)}\n"
    )

    f.write(
        f"Read success       : {len(successful)}\n"
    )

    f.write(
        f"Read error         : {len(failed)}\n"
    )

    f.write(
        f"Missing            : {len(missing)}\n"
    )

    f.write(
        f"SHA mismatches     : {len(sha_mismatch)}\n"
    )

    f.write(
        f"Total pages        : {total_pages}\n"
    )

    f.write("\nCATEGORY COUNTS\n")

    for key, value in sorted(
        category_counts.items()
    ):

        f.write(
            f"{key}: {value}\n"
        )

    f.write("\nCLAIM BOUNDARY\n")

    f.write(
        "PDF structural read          = YES\n"
    )

    f.write(
        "Filename classification      = YES\n"
    )

    f.write(
        "PDF text extraction          = NO\n"
    )

    f.write(
        "Semantic interpretation      = NO\n"
    )

    f.write(
        "Variable interpretation      = NO\n"
    )

    f.write(
        "Construct authorization      = NO\n"
    )

    f.write(
        "Identifier authorization     = NO\n"
    )

    f.write(
        "Outcome authorization        = NO\n"
    )

    f.write(
        "Score calculation            = NO\n"
    )

    f.write(
        "Empirical evaluation         = NO\n"
    )

# ------------------------------------------------------------
# 13. CONSOLE OUTPUT
# ------------------------------------------------------------

print("=" * 64)
print(
    "OIP v1.0.36 — TANZANIA NPS WAVE 4"
)
print(
    "C05 — DOCUMENTARY SOURCE INVENTORY"
)
print("=" * 64)

print(f"Protocol : {PROTOCOL}")
print(f"Step     : {STEP}")
print(f"Cell     : {CELL}")

print()

print(
    f"PDF files              : "
    f"{len(pdf_artifacts)}"
)

print(
    f"PDF read success       : "
    f"{len(successful)}"
)

print(
    f"PDF read error         : "
    f"{len(failed)}"
)

print(
    f"PDF missing            : "
    f"{len(missing)}"
)

print(
    f"SHA mismatches         : "
    f"{len(sha_mismatch)}"
)

print(
    f"Total pages            : "
    f"{total_pages}"
)

print()
print("DOCUMENT CATEGORIES")

for key, value in sorted(
    category_counts.items()
):

    print(
        f"  {key}: {value}"
    )

print()
print("CLAIM BOUNDARY")

print(
    "PDF structural read         = YES"
)

print(
    "Filename classification     = YES"
)

print(
    "PDF text extraction         = NO"
)

print(
    "Semantic interpretation     = NO"
)

print(
    "Variable interpretation     = NO"
)

print(
    "Construct authorization     = NO"
)

print(
    "Identifier authorization    = NO"
)

print(
    "Outcome authorization       = NO"
)

print(
    "Score calculation           = NO"
)

print(
    "Empirical evaluation        = NO"
)

print()
print("OUTPUT")
print(out_json)
print(out_txt)

print()
print(
    "NEXT: C06 — DOCUMENTARY EVIDENCE EXTRACTION & SOURCE MAP"
)

print("=" * 64)
```

```python
# ============================================================
# OIP v1.0.36 — TANZANIA NPS WAVE 4
# C06 — DOCUMENTARY EVIDENCE EXTRACTION & SOURCE MAP
# ============================================================

from pathlib import Path
from datetime import datetime, timezone
import json
import re

# ------------------------------------------------------------
# 1. PATHS
# ------------------------------------------------------------

KAGGLE_INPUT = Path("/kaggle/input")

RUN_ROOT = Path(
    "/kaggle/working/oip_v1_0_36_tanzania"
)

C05_DIR = (
    RUN_ROOT
    / "c05_documentary_inventory"
)

C05_JSON = (
    C05_DIR
    / "c05_documentary_inventory.json"
)

C06_DIR = (
    RUN_ROOT
    / "c06_documentary_evidence"
)

C06_DIR.mkdir(
    parents=True,
    exist_ok=True
)

if not C05_JSON.exists():
    raise RuntimeError(
        f"C05 inventory not found: {C05_JSON}"
    )

# ------------------------------------------------------------
# 2. IDENTITY
# ------------------------------------------------------------

PROTOCOL = "OIP v1.0.36"
STEP = "Step 3"
CELL = "C06"

timestamp_utc = (
    datetime.now(timezone.utc)
    .replace(microsecond=0)
    .isoformat()
)

# ------------------------------------------------------------
# 3. LOAD C05
# ------------------------------------------------------------

with open(
    C05_JSON,
    "r",
    encoding="utf-8"
) as f:

    c05 = json.load(f)

documents = c05["documents"]

if not documents:
    raise RuntimeError(
        "C05 contains no documentary records."
    )

# ------------------------------------------------------------
# 4. PDF ENGINE
# ------------------------------------------------------------

try:
    import pypdf
    PDF_ENGINE = "pypdf"

except ImportError:

    try:
        import PyPDF2 as pypdf
        PDF_ENGINE = "PyPDF2"

    except ImportError:

        raise RuntimeError(
            "No supported PDF engine available."
        )

# ------------------------------------------------------------
# 5. DOCUMENTARY EXTRACTION
# ------------------------------------------------------------

document_records = []

page_records = []

errors = []

total_pages_seen = 0
total_pages_with_text = 0
total_characters = 0

for doc_idx, doc in enumerate(
    documents,
    start=1
):

    rel_path = doc["relative_path"]

    full_path = (
        KAGGLE_INPUT
        / rel_path
    )

    document_record = {

        "document_index":
            doc_idx,

        "relative_path":
            rel_path,

        "filename":
            doc["filename"],

        "document_category":
            doc[
                "document_category_filename_only"
            ],

        "sha256":
            doc["sha256_c02"],

        "page_count_c05":
            doc["page_count"],

        "extraction_status":
            "NOT_ATTEMPTED",

        "pages_processed":
            0,

        "pages_with_text":
            0,

        "total_characters":
            0,

        "error":
            None,
    }

    if not full_path.exists():

        document_record[
            "extraction_status"
        ] = "FILE_NOT_FOUND"

        document_record[
            "error"
        ] = "FILE_NOT_FOUND"

        errors.append({
            "filename": doc["filename"],
            "error": "FILE_NOT_FOUND"
        })

        document_records.append(
            document_record
        )

        continue

    try:

        reader = pypdf.PdfReader(
            str(full_path)
        )

        actual_page_count = len(
            reader.pages
        )

        # ----------------------------------------------------
        # PAGE-BY-PAGE EXTRACTION
        # ----------------------------------------------------

        for page_number, page in enumerate(
            reader.pages,
            start=1
        ):

            try:

                text = page.extract_text()

                if text is None:
                    text = ""

                text = str(text)

            except Exception as exc:

                text = ""

                errors.append({
                    "filename":
                        doc["filename"],
                    "page":
                        page_number,
                    "error":
                        repr(exc)
                })

            character_count = len(text)

            has_text = (
                character_count > 0
            )

            if has_text:

                total_pages_with_text += 1
                document_record[
                    "pages_with_text"
                ] += 1

            document_record[
                "pages_processed"
            ] += 1

            document_record[
                "total_characters"
            ] += character_count

            total_characters += (
                character_count
            )

            total_pages_seen += 1

            # ------------------------------------------------
            # PAGE RECORD
            # ------------------------------------------------

            page_records.append({

                "document_index":
                    doc_idx,

                "filename":
                    doc["filename"],

                "relative_path":
                    rel_path,

                "document_category":
                    doc[
                        "document_category_filename_only"
                    ],

                "sha256":
                    doc["sha256_c02"],

                "page_number":
                    page_number,

                "page_character_count":
                    character_count,

                "has_text":
                    has_text,

                "text":
                    text,
            })

        # ----------------------------------------------------
        # STATUS
        # ----------------------------------------------------

        if (
            document_record["pages_processed"]
            == actual_page_count
        ):

            document_record[
                "extraction_status"
            ] = "SUCCESS"

        else:

            document_record[
                "extraction_status"
            ] = "PARTIAL"

        document_record[
            "page_count_extracted"
        ] = actual_page_count

    except Exception as exc:

        document_record[
            "extraction_status"
        ] = "ERROR"

        document_record[
            "error"
        ] = repr(exc)

        errors.append({
            "filename":
                doc["filename"],
            "error":
                repr(exc)
        })

    document_records.append(
        document_record
    )

    print(
        f"  documentary extraction: "
        f"{doc_idx}/{len(documents)} "
        f"— {doc['filename']}"
    )

# ------------------------------------------------------------
# 6. DOCUMENTARY TEXT INDEX
# ------------------------------------------------------------
#
# This is a retrieval aid only.
# It does NOT authorize any construct.
# ------------------------------------------------------------

SEARCH_TERMS = [

    "infrastructure",
    "dependency",
    "feedback",
    "adapt",
    "adaptive",
    "resilience",
    "shock",
    "livelihood",
    "income",
    "asset",
    "household",
    "individual",
    "community",
    "employment",
    "education",
    "health",
    "agriculture",
    "mobility",
    "migration",
    "coping",
    "decision",
    "risk",
    "outcome",
    "panel",
    "wave",
    "identifier",
    "household id",
    "person id",
    "questionnaire",
]

term_page_hits = {}

for term in SEARCH_TERMS:

    term_lower = term.lower()

    hits = []

    for page in page_records:

        text_lower = (
            page["text"].lower()
        )

        if term_lower in text_lower:

            hits.append({

                "filename":
                    page["filename"],

                "relative_path":
                    page["relative_path"],

                "page_number":
                    page["page_number"],

                "character_count":
                    page[
                        "page_character_count"
                    ],
            })

    term_page_hits[term] = hits

# ------------------------------------------------------------
# 7. QUESTION / ITEM CANDIDATE INDEX
# ------------------------------------------------------------
#
# Pattern-based retrieval only.
# These are NOT verified questions and NOT
# authorized measurements.
# ------------------------------------------------------------

question_candidates = []

question_pattern = re.compile(
    r"(^|\n)\s*"
    r"(?:Q(?:uestion)?\.?\s*)?"
    r"(?:[0-9]+(?:\.[0-9]+)*"
    r"|[A-Za-z]\d+(?:\.\d+)*)"
    r"[\)\.: -]+\s*"
    r".{5,250}",
    re.MULTILINE
)

for page in page_records:

    text = page["text"]

    if not text.strip():
        continue

    matches = question_pattern.findall(
        text
    )

    if matches:

        # Store only count here.
        # Full page text remains the source.
        question_candidates.append({

            "filename":
                page["filename"],

            "relative_path":
                page["relative_path"],

            "page_number":
                page["page_number"],

            "candidate_count":
                len(matches),

            "candidate_source":
                "PATTERN_ONLY",

        })

# ------------------------------------------------------------
# 8. EXTRACTION SUMMARY
# ------------------------------------------------------------

successful_documents = [
    d for d in document_records
    if d["extraction_status"]
    == "SUCCESS"
]

partial_documents = [
    d for d in document_records
    if d["extraction_status"]
    == "PARTIAL"
]

failed_documents = [
    d for d in document_records
    if d["extraction_status"]
    == "ERROR"
]

missing_documents = [
    d for d in document_records
    if d["extraction_status"]
    == "FILE_NOT_FOUND"
]

# ------------------------------------------------------------
# 9. CLAIM BOUNDARY
# ------------------------------------------------------------

claim_boundary = {

    "documentary_text_extraction":
        True,

    "page_level_provenance":
        True,

    "source_map_created":
        True,

    "keyword_retrieval":
        True,

    "question_pattern_discovery":
        True,

    "semantic_construct_verification":
        False,

    "variable_semantic_verification":
        False,

    "construct_authorization":
        False,

    "identifier_authorization":
        False,

    "outcome_authorization":
        False,

    "score_calculation":
        False,

    "empirical_evaluation":
        False,

    "empirical_validity_claim":
        False,

    "proxy_authorization":
        False,

    "silent_imputation":
        False,
}

# ------------------------------------------------------------
# 10. FINAL RECORD
# ------------------------------------------------------------

record = {

    "protocol":
        PROTOCOL,

    "step":
        STEP,

    "cell":
        CELL,

    "cell_purpose":
        "DOCUMENTARY_EVIDENCE_EXTRACTION_AND_SOURCE_MAP",

    "timestamp_utc":
        timestamp_utc,

    "source_inventory":
        str(C05_JSON),

    "pdf_engine":
        PDF_ENGINE,

    "summary": {

        "documents_total":
            len(documents),

        "documents_success":
            len(successful_documents),

        "documents_partial":
            len(partial_documents),

        "documents_error":
            len(failed_documents),

        "documents_missing":
            len(missing_documents),

        "pages_extracted":
            total_pages_seen,

        "pages_with_text":
            total_pages_with_text,

        "pages_without_text":
            (
                total_pages_seen
                - total_pages_with_text
            ),

        "total_characters":
            total_characters,

        "question_pattern_candidate_pages":
            len(question_candidates),

    },

    "documents":
        document_records,

    "term_page_hits":
        term_page_hits,

    "question_candidates":
        question_candidates,

    "errors":
        errors,

    "claim_boundary":
        claim_boundary,

    "next_cell":
        "C07 — VARIABLE/DOCUMENTARY CROSS-MAP",
}

# ------------------------------------------------------------
# 11. SAVE MASTER JSON
# ------------------------------------------------------------

out_json = (
    C06_DIR
    / "c06_documentary_evidence.json"
)

with open(
    out_json,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        record,
        f,
        indent=2,
        ensure_ascii=False
    )

# ------------------------------------------------------------
# 12. SAVE PAGE TEXT SEPARATELY
# ------------------------------------------------------------
#
# Keeps the master JSON manageable.
# This file preserves page-level source text.
# ------------------------------------------------------------

out_text = (
    C06_DIR
    / "c06_page_text.jsonl"
)

with open(
    out_text,
    "w",
    encoding="utf-8"
) as f:

    for page in page_records:

        f.write(
            json.dumps(
                page,
                ensure_ascii=False
            )
            + "\n"
        )

# ------------------------------------------------------------
# 13. SAVE SOURCE MAP
# ------------------------------------------------------------

source_map = {

    "protocol":
        PROTOCOL,

    "step":
        STEP,

    "cell":
        CELL,

    "source_map_type":
        "PAGE_LEVEL_DOCUMENTARY_MAP",

    "pages":
        [
            {
                "filename":
                    p["filename"],

                "relative_path":
                    p["relative_path"],

                "document_category":
                    p["document_category"],

                "sha256":
                    p["sha256"],

                "page_number":
                    p["page_number"],

                "character_count":
                    p["page_character_count"],

                "has_text":
                    p["has_text"],
            }

            for p in page_records
        ],
}

out_map = (
    C06_DIR
    / "c06_source_map.json"
)

with open(
    out_map,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        source_map,
        f,
        indent=2,
        ensure_ascii=False
    )

# ------------------------------------------------------------
# 14. CONSOLE OUTPUT
# ------------------------------------------------------------

print("=" * 64)

print(
    "OIP v1.0.36 — TANZANIA NPS WAVE 4"
)

print(
    "C06 — DOCUMENTARY EVIDENCE EXTRACTION "
    "& SOURCE MAP"
)

print("=" * 64)

print(
    f"Protocol : {PROTOCOL}"
)

print(
    f"Step     : {STEP}"
)

print(
    f"Cell     : {CELL}"
)

print()

print(
    f"Documents processed       : "
    f"{len(documents)}"
)

print(
    f"Extraction success        : "
    f"{len(successful_documents)}"
)

print(
    f"Extraction partial        : "
    f"{len(partial_documents)}"
)

print(
    f"Extraction errors         : "
    f"{len(failed_documents)}"
)

print(
    f"Documents missing         : "
    f"{len(missing_documents)}"
)

print(
    f"Pages extracted           : "
    f"{total_pages_seen}"
)

print(
    f"Pages with text           : "
    f"{total_pages_with_text}"
)

print(
    f"Pages without text        : "
    f"{total_pages_seen - total_pages_with_text}"
)

print(
    f"Total extracted chars     : "
    f"{total_characters}"
)

print(
    f"Question-pattern pages    : "
    f"{len(question_candidates)}"
)

print()

print(
    "SEARCH TERM PAGE HITS"
)

for term, hits in term_page_hits.items():

    print(
        f"  {term}: {len(hits)}"
    )

print()

print("CLAIM BOUNDARY")

print(
    "Documentary extraction       = YES"
)

print(
    "Page-level provenance        = YES"
)

print(
    "Source map                   = YES"
)

print(
    "Keyword retrieval            = YES"
)

print(
    "Semantic verification        = NO"
)

print(
    "Variable verification        = NO"
)

print(
    "Construct authorization      = NO"
)

print(
    "Identifier authorization     = NO"
)

print(
    "Outcome authorization       = NO"
)

print(
    "Score calculation            = NO"
)

print(
    "Empirical evaluation         = NO"
)

print(
    "Proxy authorization          = NO"
)

print()

print("OUTPUT")

print(out_json)
print(out_text)
print(out_map)

print()

print(
    "NEXT: C07 — VARIABLE/DOCUMENTARY CROSS-MAP"
)

print("=" * 64)
```

```python
# ================================================================
# OIP v1.0.36 — TANZANIA NPS WAVE 4
# C07 — VARIABLE / DOCUMENTARY CROSS-MAP
# FINAL CORRECTED VERSION
# ================================================================

from pathlib import Path
import json
import re

BASE = Path("/kaggle/working/oip_v1_0_36_tanzania")

C04_JSON = BASE / "c04_stata_schema" / "c04_stata_schema.json"
C06_JSON = BASE / "c06_documentary_evidence" / "c06_documentary_evidence.json"
C06_PAGE_JSONL = BASE / "c06_documentary_evidence" / "c06_page_text.jsonl"

OUT_DIR = BASE / "c07_variable_documentary_map"
OUT_DIR.mkdir(parents=True, exist_ok=True)

OUT_JSON = OUT_DIR / "c07_variable_documentary_cross_map.json"
OUT_SUMMARY = OUT_DIR / "c07_variable_documentary_cross_map_summary.txt"


# ------------------------------------------------
# LOAD C04
# ------------------------------------------------

with open(C04_JSON, "r", encoding="utf-8") as f:
    c04 = json.load(f)

schemas = c04["schemas"]

# ------------------------------------------------
# BUILD VARIABLE REGISTRY FROM ACTUAL C04 SCHEMA
# ------------------------------------------------

variables = {}

for schema in schemas:

    filename = schema.get("filename", "")
    variable_names = schema.get("variable_names", []) or []
    variable_labels = schema.get("variable_labels", {}) or {}

    for var_name in variable_names:

        if not isinstance(var_name, str) or not var_name.strip():
            continue

        var_name = var_name.strip()
        key = var_name.lower()

        if key not in variables:
            variables[key] = {
                "variable_name": var_name,
                "labels": set(),
                "files": set(),
            }

        variables[key]["files"].add(filename)

        label = variable_labels.get(var_name)

        if isinstance(label, str) and label.strip():
            variables[key]["labels"].add(label.strip())


variable_items = list(variables.values())


# ------------------------------------------------
# LOAD C06 PAGE TEXT
# ------------------------------------------------

pages = []

with open(C06_PAGE_JSONL, "r", encoding="utf-8") as f:

    for line in f:

        line = line.strip()

        if not line:
            continue

        rec = json.loads(line)

        text = ""

        for key in (
            "text",
            "page_text",
            "content",
            "extracted_text"
        ):
            if isinstance(rec.get(key), str):
                text = rec[key]
                break

        if not text:
            continue

        document = ""

        for key in (
            "document",
            "document_name",
            "file",
            "file_name",
            "filename",
            "source_file"
        ):
            if isinstance(rec.get(key), str):
                document = rec[key]
                break

        page_no = None

        for key in (
            "page",
            "page_number",
            "page_no"
        ):
            if key in rec:
                page_no = rec[key]
                break

        pages.append({
            "document": document,
            "page": page_no,
            "text": text
        })


# ------------------------------------------------
# LITERAL MATCH HELPER
# ------------------------------------------------

def literal_pattern(term):

    return re.compile(
        r"(?<![A-Za-z0-9_])"
        + re.escape(term)
        + r"(?![A-Za-z0-9_])",
        re.IGNORECASE
    )


# ------------------------------------------------
# VARIABLE / DOCUMENTARY CROSS-MAP
# ------------------------------------------------

variable_cross_map = []

print("=" * 64)
print("C07 — VARIABLE / DOCUMENTARY CROSS-MAP")
print("=" * 64)

print("Variables discovered :", len(variable_items))
print("Documentary pages    :", len(pages))
print()


for idx, item in enumerate(variable_items, start=1):

    var_name = item["variable_name"]
    labels = sorted(item["labels"])
    files = sorted(item["files"])

    name_hits = []

    name_regex = literal_pattern(var_name)

    for page in pages:

        if name_regex.search(page["text"]):

            name_hits.append({
                "document": page["document"],
                "page": page["page"],
            })


    label_hits = []

    for label in labels:

        label_regex = literal_pattern(label)

        for page in pages:

            if label_regex.search(page["text"]):

                label_hits.append({
                    "label": label,
                    "document": page["document"],
                    "page": page["page"],
                })


    variable_cross_map.append({

        "variable_name": var_name,

        "labels": labels,

        "files": files,

        "literal_variable_name_hits": name_hits,

        "literal_variable_label_hits": label_hits,

        "documentary_literal_hit":
            bool(name_hits or label_hits),

        # FAIL-CLOSED
        "semantic_verification": False,

        "construct_authorized": False,

        "identifier_authorized": False,

        "outcome_authorized": False,

        "next_required_step":
            "SEMANTIC_VERIFICATION_AND_LINKAGE_PROOF",
    })


    if idx % 200 == 0 or idx == len(variable_items):

        print(
            f"variable progress: "
            f"{idx}/{len(variable_items)}"
        )


# ------------------------------------------------
# IDENTIFIER CANDIDATE DISCOVERY
# NAME ONLY
# ------------------------------------------------

identifier_patterns = {

    "HOUSEHOLD_ID": re.compile(
        r"^(hhid|hh_id|household[_ ]?id|householdid)$",
        re.IGNORECASE
    ),

    "PERSON_ID": re.compile(
        r"^(pid|person[_ ]?id|personid|individual[_ ]?id|individualid)$",
        re.IGNORECASE
    ),

    "CLUSTER": re.compile(
        r"^(cluster|clusterid|cluster_id)$",
        re.IGNORECASE
    ),

    "EA": re.compile(
        r"^(ea|eaid|ea_id|enumeration[_ ]?area|enumeration[_ ]?area[_ ]?id)$",
        re.IGNORECASE
    ),

    "STRATA": re.compile(
        r"^(strata|stratum|strataid|strata_id|stratumid|stratum_id)$",
        re.IGNORECASE
    ),

    "WEIGHT": re.compile(
        r"^(weight|weights|hhweight|hh_weight|pweight|p_weight)$",
        re.IGNORECASE
    ),

    "WAVE_ROUND": re.compile(
        r"^(wave|round|waveid|wave_id|roundid|round_id)$",
        re.IGNORECASE
    ),
}


identifier_candidates = []

for item in variable_items:

    var_name = item["variable_name"]

    categories = []

    for category, pattern in identifier_patterns.items():

        if pattern.match(var_name):
            categories.append(category)

    if categories:

        identifier_candidates.append({

            "variable_name": var_name,

            "candidate_categories": categories,

            "files": sorted(item["files"]),

            "labels": sorted(item["labels"]),

            "identifier_authorized": False,

            "semantic_verification": False,

            "next_required_step":
                "SEMANTIC_VERIFICATION_AND_LINKAGE_PROOF",
        })


# ------------------------------------------------
# EXPLICIT ZERO-AUTHORIZATION STATE
# ------------------------------------------------

authorized_identifiers = []
authorized_constructs = []
authorized_outcomes = []


# ------------------------------------------------
# COUNTS
# ------------------------------------------------

literal_hit_count = sum(
    1
    for item in variable_cross_map
    if item["documentary_literal_hit"]
)


# ------------------------------------------------
# FINAL OUTPUT
# ------------------------------------------------

output = {

    "protocol": "OIP v1.0.36",

    "dataset": "Tanzania NPS Wave 4",

    "step": "Step 3",

    "cell": "C07",

    "cell_title":
        "Variable / Documentary Cross-Map",

    "status": "PASS",

    "counts": {

        "stata_schema_records":
            len(schemas),

        "variables_discovered":
            len(variable_items),

        "documentary_pages_loaded":
            len(pages),

        "identifier_candidates":
            len(identifier_candidates),

        "variables_with_literal_documentary_hits":
            literal_hit_count,
    },

    "variable_cross_map":
        variable_cross_map,

    "identifier_candidates":
        identifier_candidates,

    # Explicit zero authorization
    "authorized_identifiers":
        authorized_identifiers,

    "authorized_constructs":
        authorized_constructs,

    "authorized_outcomes":
        authorized_outcomes,

    "documentary_hits_are_evidence":
        False,

    "claim_boundary": {

        "documentary_literal_mapping":
            True,

        "semantic_verification":
            False,

        "variable_verification":
            False,

        "construct_authorization":
            False,

        "identifier_authorization":
            False,

        "outcome_authorization":
            False,

        "linkage_proof":
            False,

        "score_calculation":
            False,

        "empirical_evaluation":
            False,

        "proxy_authorization":
            False,
    },

    "next_cell":
        "C08 — DOCUMENTARY SEMANTIC VERIFICATION GATE",
}


# ------------------------------------------------
# WRITE JSON
# ------------------------------------------------

with open(OUT_JSON, "w", encoding="utf-8") as f:

    json.dump(
        output,
        f,
        indent=2,
        ensure_ascii=False
    )


# ------------------------------------------------
# WRITE SUMMARY
# ------------------------------------------------

summary = f"""
OIP v1.0.36 — TANZANIA NPS WAVE 4
C07 — VARIABLE / DOCUMENTARY CROSS-MAP
================================================

STATUS : PASS

Stata schema records       : {len(schemas)}
Variables discovered       : {len(variable_items)}
Documentary pages loaded   : {len(pages)}
Identifier candidates      : {len(identifier_candidates)}
Variables with literal hits: {literal_hit_count}

AUTHORIZATION STATE

Authorized identifiers : 0
Authorized constructs  : 0
Authorized outcomes    : 0

IMPORTANT

Documentary literal hits are NOT semantic evidence.
Candidate variables are NOT authorized variables.
Identifier candidates are NOT authorized identifiers.

CLAIM BOUNDARY

Documentary cross-map      = YES
Semantic verification      = NO
Variable verification      = NO
Construct authorization    = NO
Identifier authorization   = NO
Outcome authorization      = NO
Linkage proof              = NO
Score calculation          = NO
Empirical evaluation      = NO
Proxy authorization       = NO

NEXT

C08 — DOCUMENTARY SEMANTIC VERIFICATION GATE
"""

with open(OUT_SUMMARY, "w", encoding="utf-8") as f:
    f.write(summary)


# ------------------------------------------------
# CONSOLE
# ------------------------------------------------

print()
print("=" * 64)
print("C07 — FINAL CORRECTED RESULT")
print("=" * 64)

print("Stata schema records       :", len(schemas))
print("Variables discovered       :", len(variable_items))
print("Documentary pages loaded   :", len(pages))
print("Identifier candidates      :", len(identifier_candidates))
print("Variables with literal hits:", literal_hit_count)

print()
print("AUTHORIZATION STATE")
print("Authorized identifiers : 0")
print("Authorized constructs  : 0")
print("Authorized outcomes    : 0")

print()
print("CLAIM BOUNDARY")
print("Documentary cross-map      = YES")
print("Semantic verification      = NO")
print("Variable verification      = NO")
print("Construct authorization    = NO")
print("Identifier authorization   = NO")
print("Outcome authorization     = NO")
print("Linkage proof              = NO")
print("Score calculation          = NO")
print("Empirical evaluation       = NO")
print("Proxy authorization        = NO")

print()
print("OUTPUT")
print(OUT_JSON)
print(OUT_SUMMARY)

print()
print("NEXT")
print("C08 — DOCUMENTARY SEMANTIC VERIFICATION GATE")
```

```python
# ================================================================
# OIP v1.0.36 — TANZANIA NPS WAVE 4
# C15 — FINAL ARCHIVAL VERIFICATION & FREEZE
# ================================================================

from pathlib import Path
from datetime import datetime, timezone
import hashlib
import json

# ------------------------------------------------
# 1. PATHS
# ------------------------------------------------

BASE = Path("/kaggle/working/oip_v1_0_36_tanzania")

if not BASE.exists():
    raise RuntimeError(
        "OIP v1.0.36 working directory was not found: "
        + str(BASE)
    )

C15_DIR = BASE / "c15_archival_freeze"
C15_DIR.mkdir(parents=True, exist_ok=True)

MANIFEST_JSON = C15_DIR / "c15_archival_manifest_sha256.json"
MANIFEST_TXT = C15_DIR / "c15_archival_manifest_sha256.txt"
CORRECTION_JSON = C15_DIR / "c15_state_label_correction_note.json"
SUMMARY_TXT = C15_DIR / "c15_archival_freeze_summary.txt"

# ------------------------------------------------
# 2. SHA-256
# ------------------------------------------------

def sha256_file(path, chunk_size=1024 * 1024):

    h = hashlib.sha256()

    with open(path, "rb") as f:

        while True:

            chunk = f.read(chunk_size)

            if not chunk:
                break

            h.update(chunk)

    return h.hexdigest()


# ------------------------------------------------
# 3. DISCOVER ACTUAL C01-C14 JSON ARTIFACTS
# ------------------------------------------------
# Do NOT assume exact filenames.
# Search the actual working directory.

all_files = []

for p in BASE.rglob("*"):

    if not p.is_file():
        continue

    # Never include C15-generated files
    if C15_DIR in p.parents:
        continue

    all_files.append(p)


# ------------------------------------------------
# 4. IDENTIFY CELL ARTIFACTS
# ------------------------------------------------

cell_artifacts = {}

for p in all_files:

    name = p.name.lower()

    for n in range(1, 15):

        prefix = f"c{n:02d}_"

        if name.startswith(prefix) and name.endswith(".json"):

            cell_artifacts[n] = p

            break


# ------------------------------------------------
# 5. REQUIRED CELL CHECK
# ------------------------------------------------

required_cells = list(range(1, 15))

missing_cells = [
    n for n in required_cells
    if n not in cell_artifacts
]

found_cells = [
    n for n in required_cells
    if n in cell_artifacts
]


# ------------------------------------------------
# 6. VERIFY DISCOVERED CELL JSON FILES
# ------------------------------------------------

cell_records = []

for n in found_cells:

    p = cell_artifacts[n]

    record = {
        "cell": f"C{n:02d}",
        "relative_path": str(
            p.relative_to(BASE)
        ),
        "size_bytes": p.stat().st_size,
        "sha256": sha256_file(p),
        "readable_json": False
    }

    try:

        with open(
            p,
            "r",
            encoding="utf-8"
        ) as f:

            json.load(f)

        record["readable_json"] = True

    except Exception as exc:

        record["readable_json"] = False
        record["json_error"] = repr(exc)

    cell_records.append(record)


json_read_errors = [
    x for x in cell_records
    if not x["readable_json"]
]


# ------------------------------------------------
# 7. C14 DISCOVERY
# ------------------------------------------------

c14_record = None

for item in cell_records:

    if item["cell"] == "C14":
        c14_record = item
        break


c14_data = None

if c14_record is not None:

    c14_path = BASE / c14_record["relative_path"]

    try:

        with open(
            c14_path,
            "r",
            encoding="utf-8"
        ) as f:

            c14_data = json.load(f)

    except Exception:

        c14_data = None


# ------------------------------------------------
# 8. STATE LABEL CORRECTION NOTE
# ------------------------------------------------

correction_note = {

    "protocol": "OIP v1.0.36",

    "cell": "C15",

    "status": "ARCHIVAL_NOTE_ONLY",

    "purpose": (
        "Clarification of construct-state terminology "
        "for archival consistency."
    ),

    "audit_level_label": "UNRESOLVED",

    "frozen_archival_state": "PENDING_EVIDENCE",

    "clarification": (
        "The label UNRESOLVED appearing in the C14 "
        "audit output is preserved as part of the "
        "original C14 artifact. For archival comparison "
        "with the frozen OIP measurement-state vocabulary, "
        "the corresponding fail-closed state is "
        "PENDING_EVIDENCE."
    ),

    "boundary": (
        "This note does not modify, overwrite, or rerun "
        "C10 or C14 and does not introduce a new "
        "measurement state or qualification criterion."
    ),

    "construct_authorization": "NO",

    "score_authorization": "NO",

    "outcome_authorization": "NO",

    "empirical_evaluation": "NO"
}


with open(
    CORRECTION_JSON,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        correction_note,
        f,
        indent=2,
        ensure_ascii=False
    )


# ------------------------------------------------
# 9. HASH ALL PRE-C15 ARTIFACTS
# ------------------------------------------------

all_records = []

for p in sorted(
    all_files,
    key=lambda x: str(x)
):

    all_records.append(
        {
            "relative_path": str(
                p.relative_to(BASE)
            ),
            "size_bytes": p.stat().st_size,
            "sha256": sha256_file(p)
        }
    )


# ------------------------------------------------
# 10. ARCHIVAL STATUS
# ------------------------------------------------

integrity_pass = (
    len(missing_cells) == 0
    and len(json_read_errors) == 0
    and c14_record is not None
)

if integrity_pass:

    archival_status = "ARCHIVAL_FREEZE_PASS"

else:

    archival_status = "ARCHIVAL_FREEZE_BLOCKED"


# ------------------------------------------------
# 11. FINAL MANIFEST
# ------------------------------------------------

manifest = {

    "protocol": "OIP v1.0.36",

    "dataset":
        "Tanzania National Panel Survey 2014-2015, Wave 4",

    "reference_id":
        "TZA_2014_NPS-R4_v03_M",

    "step": "Step 3",

    "cell": "C15",

    "status": archival_status,

    "timestamp_utc":
        datetime.now(timezone.utc)
        .replace(microsecond=0)
        .isoformat(),

    "purpose":
        "Final archival verification and freeze.",

    "upstream_cells": {

        "expected": 14,

        "found": len(found_cells),

        "missing": missing_cells,

        "json_read_errors":
            json_read_errors,

        "records":
            cell_records
    },

    "c14_presence":
        c14_record is not None,

    "artifact_count_hashed":
        len(all_records),

    "artifact_manifest":
        all_records,

    "construct_archival_state": {

        "GFL": "PENDING_EVIDENCE",

        "IDS": "PENDING_EVIDENCE",

        "AML": "PENDING_EVIDENCE"
    },

    "authorization_boundary": {

        "constructs": "NO",

        "identifiers": "NO",

        "outcomes": "NO",

        "analytical_cohort": "NO",

        "score": "NO",

        "empirical_evaluation": "NO"
    },

    "qualification_boundary":
        "FAIL_CLOSED_QUALIFICATION_NOT_ESTABLISHED",

    "release_boundary":
        "AUDIT_ONLY / FAIL_CLOSED",

    "claim_boundary": {

        "empirical_validity": "NO",

        "predictive_validity": "NO",

        "causal_validity": "NO",

        "universal_validity": "NO",

        "superiority": "NO"
    },

    "prohibited_substitutions": {

        "unsupported_proxy": "NOT_USED",

        "silent_imputation": "NOT_USED",

        "synthetic_outcome": "NOT_USED",

        "random_fallback": "NOT_USED",

        "post_hoc_tuning": "NOT_USED"
    },

    "scientific_analysis_added":
        "NO",

    "construct_added":
        "NO",

    "outcome_added":
        "NO",

    "score_added":
        "NO"
}


with open(
    MANIFEST_JSON,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        manifest,
        f,
        indent=2,
        ensure_ascii=False
    )


# ------------------------------------------------
# 12. MANIFEST HASH
# ------------------------------------------------

manifest_sha256 = sha256_file(
    MANIFEST_JSON
)


with open(
    MANIFEST_TXT,
    "w",
    encoding="utf-8"
) as f:

    f.write(
        "OIP v1.0.36 — C15 ARCHIVAL MANIFEST\n"
    )

    f.write(
        "SHA-256: "
        + manifest_sha256
        + "\n"
    )

    f.write(
        "Verification: SHA-256 computed from the "
        "exact C15 manifest binary present at "
        "verification time.\n"
    )


# ------------------------------------------------
# 13. SUMMARY
# ------------------------------------------------

summary = []

summary.append("=" * 68)
summary.append(
    "OIP v1.0.36 — C15 FINAL ARCHIVAL VERIFICATION & FREEZE"
)
summary.append("=" * 68)

summary.append("")

summary.append("UPSTREAM CELL INTEGRITY")

summary.append(
    f"Expected C01-C14 : 14"
)

summary.append(
    f"Found C01-C14    : {len(found_cells)}"
)

summary.append(
    f"Missing cells    : {len(missing_cells)}"
)

if missing_cells:

    summary.append(
        "Missing : "
        + ", ".join(
            f"C{x:02d}"
            for x in missing_cells
        )
    )

summary.append(
    f"JSON read errors : {len(json_read_errors)}"
)

summary.append(
    f"Artifacts hashed : {len(all_records)}"
)

summary.append("")

summary.append("C14")

summary.append(
    "C14 present : "
    + ("YES" if c14_record else "NO")
)

summary.append(
    "Qualification : "
    "FAIL_CLOSED_QUALIFICATION_NOT_ESTABLISHED"
)

summary.append(
    "Release boundary : "
    "AUDIT_ONLY / FAIL_CLOSED"
)

summary.append("")

summary.append("ARCHIVAL STATE VOCABULARY")

summary.append(
    "GFL = PENDING_EVIDENCE"
)

summary.append(
    "IDS = PENDING_EVIDENCE"
)

summary.append(
    "AML = PENDING_EVIDENCE"
)

summary.append(
    "C14 UNRESOLVED label preserved in original artifact."
)

summary.append(
    "C15 records PENDING_EVIDENCE for frozen-vocabulary "
    "archival comparison."
)

summary.append("")

summary.append("AUTHORIZATION")

summary.append("Constructs          = NO")
summary.append("Identifiers         = NO")
summary.append("Outcomes            = NO")
summary.append("Analytical cohort   = NO")
summary.append("Score               = NO")
summary.append("Empirical evaluation= NO")

summary.append("")

summary.append("CLAIM BOUNDARY")

summary.append("Empirical validity  = NO")
summary.append("Predictive validity = NO")
summary.append("Causal validity     = NO")
summary.append("Universal validity  = NO")
summary.append("Superiority         = NO")

summary.append("")

summary.append(
    "C15 MANIFEST SHA-256"
)

summary.append(
    manifest_sha256
)

summary.append("")

summary.append(
    "C15 STATUS : "
    + archival_status
)

summary.append("")

summary.append("OUTPUTS")

summary.append(str(MANIFEST_JSON))
summary.append(str(MANIFEST_TXT))
summary.append(str(CORRECTION_JSON))
summary.append(str(SUMMARY_TXT))

summary.append("")
summary.append("=" * 68)

summary_text = "\n".join(summary)

with open(
    SUMMARY_TXT,
    "w",
    encoding="utf-8"
) as f:

    f.write(summary_text)


# ------------------------------------------------
# 14. FINAL CONSOLE OUTPUT
# ------------------------------------------------

print("=" * 68)

print(
    "OIP v1.0.36 — C15 FINAL ARCHIVAL VERIFICATION"
)

print("=" * 68)

print()

print(
    "EXPECTED C01-C14 : 14"
)

print(
    "FOUND C01-C14    :",
    len(found_cells)
)

print(
    "MISSING CELLS    :",
    len(missing_cells)
)

print(
    "JSON READ ERRORS  :",
    len(json_read_errors)
)

print(
    "ARTIFACTS HASHED  :",
    len(all_records)
)

print()

if missing_cells:

    print(
        "MISSING :",
        ", ".join(
            f"C{x:02d}"
            for x in missing_cells
        )
    )

print(
    "UPSTREAM INTEGRITY CHECK :",
    "PASS" if integrity_pass else "BLOCKED"
)

print()

print(
    "C14 PRESENT :",
    "YES" if c14_record else "NO"
)

print(
    "QUALIFICATION :",
    "FAIL_CLOSED_QUALIFICATION_NOT_ESTABLISHED"
)

print(
    "RELEASE :",
    "AUDIT_ONLY / FAIL_CLOSED"
)

print()

print("ARCHIVAL STATE")

print("GFL = PENDING_EVIDENCE")
print("IDS = PENDING_EVIDENCE")
print("AML = PENDING_EVIDENCE")

print()

print(
    "C14 UNRESOLVED : PRESERVED"
)

print(
    "C15 ARCHIVAL MAPPING : PENDING_EVIDENCE"
)

print()

print(
    "C15 MANIFEST SHA-256 :"
)

print(manifest_sha256)

print()

print(
    "C15 STATUS :",
    archival_status
)

print()

print("OUTPUT")

print(MANIFEST_JSON)
print(MANIFEST_TXT)
print(CORRECTION_JSON)
print(SUMMARY_TXT)

print()

print("=" * 68)

if integrity_pass:

    print(
        "C15 — ARCHIVAL FREEZE PASS"
    )

else:

    print(
        "C15 — ARCHIVAL FREEZE BLOCKED"
    )

print("=" * 68)
```
