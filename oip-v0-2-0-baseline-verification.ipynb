## 1. Research Objective

This notebook evaluates whether the original OIP v0.2.0 decision engine behaves according to its specified decision rules under controlled synthetic inputs.

The objective is to make the verification independently repeatable. Another researcher should be able to inspect the test definitions, reproduce the execution, compare expected and observed results, and verify the integrity of the exported artifacts.

This notebook evaluates software behavior. It does not, by itself, validate the scientific theory or every version of the OIP protocol.

## 2. Verification Scope

The verification covers:

- Core decision behavior
- Edge cases and invalid inputs
- Output structure and field requirements
- Configuration and field immutability
- Decision precedence
- A defined exhaustive categorical test matrix
- Reproducible result exports and integrity checks

The original decision engine must remain unchanged during baseline testing.

Any proposed improvements or modified implementations must be evaluated separately against the frozen baseline.

## 3. Source Identity and Integrity

## 4. Controlled Synthetic Scenarios

Synthetic inputs will represent fictional situations such as:

- **Medical review:** A required report is missing or a review remains incomplete.
- **Financial verification:** Supporting documents contain conflicting information.
- **Infrastructure safety:** A required inspection or review has failed.

These scenarios provide understandable labels for software tests. They are not real patient assessments, actual credit decisions, or engineering safety evaluations.

Expected outcomes must be defined from the applicable, documented decision rules before the corresponding tests run. They must not be invented from the scenario description or generated by simply copying the engine's own output.

```python
# Cell 2 — Verify and load the exact v0.2.0 baseline source

import hashlib

EXPECTED_SOURCE_SHA256 = (
    "1bc5ce14aa25d8788b28f1ee08a78ee2db343d0ea02b0e67bb08da5e7a0e4dc5"
)

SOURCE_TEXT = r'''# Cell 4 — OIP v0.2.0 Synthetic Decision Engine

FIELDS = {
    "case_id",
    "evidence_status",
    "review_result",
    "evidence_conflict",
    "prior_state",
}

EVIDENCE_VALUES = {"PRESENT", "MISSING", "UNKNOWN"}
REVIEW_VALUES = {"PASSED", "FAILED", "NOT_REVIEWED"}
STATE_VALUES = {
    "NOT_ASSESSED",
    "PENDING_EVIDENCE",
    "APPROVED",
    "BLOCKED",
}


def decide(case: dict) -> dict:
    """Apply the frozen v0.2.0 demo decision rules."""

    def invalid_result(reason):
        cid = case.get("case_id") if isinstance(case, dict) else None
        prior = case.get("prior_state") if isinstance(case, dict) else None

        if not isinstance(cid, str) or not cid.strip():
            cid = None

        if not isinstance(prior, str) or prior not in STATE_VALUES:
            prior = None

        return {
            "case_id": cid,
            "state": "INVALID_INPUT",
            "reason": reason,
            "prior_state": prior,
        }

    # 1. Input must be a dictionary.
    if not isinstance(case, dict):
        return invalid_result("Input must be a dictionary.")

    # 2. Required and unexpected fields.
    missing = FIELDS - set(case.keys())
    extra = set(case.keys()) - FIELDS

    if missing:
        return invalid_result(
            f"Missing required fields: {sorted(missing)}"
        )

    if extra:
        return invalid_result(
            f"Unexpected fields: {sorted(map(str, extra))}"
        )

    # 3. Validate each field before making a decision.
    cid = case["case_id"]
    evidence = case["evidence_status"]
    review = case["review_result"]
    conflict = case["evidence_conflict"]
    prior = case["prior_state"]

    if not isinstance(cid, str) or not cid.strip():
        return invalid_result("case_id must be a non-empty string.")

    if not isinstance(evidence, str) or evidence not in EVIDENCE_VALUES:
        return invalid_result("Invalid evidence_status.")

    if not isinstance(review, str) or review not in REVIEW_VALUES:
        return invalid_result("Invalid review_result.")

    if type(conflict) is not bool:
        return invalid_result("evidence_conflict must be True or False.")

    if not isinstance(prior, str) or prior not in STATE_VALUES:
        return invalid_result("Invalid prior_state.")

    # 4. Frozen decision precedence.
    if conflict:
        state = "BLOCKED"
        reason = "Contradictory evidence declared."

    elif prior == "BLOCKED":
        state = "BLOCKED"
        reason = "Previous BLOCKED state retained by demo policy."

    elif review == "FAILED":
        state = "BLOCKED"
        reason = "Failed review: v0.2.0 demo policy."

    elif evidence in {"MISSING", "UNKNOWN"}:
        state = "PENDING_EVIDENCE"
        reason = "Evidence is missing or unresolved."

    elif review == "NOT_REVIEWED":
        state = "PENDING_EVIDENCE"
        reason = "Review has not been completed."

    elif evidence == "PRESENT" and review == "PASSED":
        state = "APPROVED"
        reason = "All demo approval conditions are satisfied."

    else:
        state = "PENDING_EVIDENCE"
        reason = "No approval rule was satisfied."

    # 5. Return an explicit, auditable decision record.
    return {
        "case_id": cid,
        "state": state,
        "reason": reason,
        "prior_state": prior,
    }


print("Decision engine defined.")
print("Full rule tests are scheduled for Cell 6.")
'''

source_bytes = SOURCE_TEXT.encode("utf-8")
actual_source_sha256 = hashlib.sha256(source_bytes).hexdigest()

print("Expected SHA-256:", EXPECTED_SOURCE_SHA256)
print("Actual SHA-256:  ", actual_source_sha256)

if actual_source_sha256 != EXPECTED_SOURCE_SHA256:
    raise RuntimeError(
        "SOURCE IDENTITY CHECK FAILED. Stop: do not execute tests."
    )

print("SOURCE IDENTITY CHECK: PASS")

exec(
    compile(SOURCE_TEXT, "verified_oip_v0_2_0_source.py", "exec"),
    globals()
)

print("BASELINE SOURCE LOADED: PASS")
```

```python
# CELL 17 — Exhaustive Matrix Test (72 Combinations)

import itertools
import pandas as pd

statuses = ["PRESENT", "MISSING", "UNKNOWN"]
reviews = ["PASSED", "FAILED", "NOT_REVIEWED"]
conflicts = [False, True]
priors = [
    "NOT_ASSESSED",
    "PENDING_EVIDENCE",
    "APPROVED",
    "BLOCKED"
]

matrix_results = []

for idx, (status, review, conflict, prior) in enumerate(
    itertools.product(statuses, reviews, conflicts, priors),
    start=1
):
    case = {
        "case_id": f"MATRIX_{idx:03d}",
        "evidence_status": status,
        "review_result": review,
        "evidence_conflict": conflict,
        "prior_state": prior
    }

    # Reference expectations based on the frozen v0.2.0 rules
    if conflict:
        expected = "BLOCKED"
    elif prior == "BLOCKED":
        expected = "BLOCKED"
    elif review == "FAILED":
        expected = "BLOCKED"
    elif status in ["MISSING", "UNKNOWN"]:
        expected = "PENDING_EVIDENCE"
    elif review == "NOT_REVIEWED":
        expected = "PENDING_EVIDENCE"
    elif status == "PRESENT" and review == "PASSED":
        expected = "APPROVED"
    else:
        expected = "PENDING_EVIDENCE"

    try:
        result = decide(case.copy())
        actual = (
            result.get("state")
            if isinstance(result, dict)
            else None
        )
        passed = actual == expected
        error = None

    except Exception as exc:
        actual = None
        passed = False
        error = f"{type(exc).__name__}: {exc}"

    matrix_results.append({
        "case_id": case["case_id"],
        "evidence": status,
        "review": review,
        "conflict": conflict,
        "prior_state": prior,
        "expected": expected,
        "actual": actual,
        "status": "PASS" if passed else "FAIL",
        "error": error
    })

df_matrix = pd.DataFrame(matrix_results)

print("FIRST 10 CASES")
display(df_matrix.head(10))

print("LAST 10 CASES")
display(df_matrix.tail(10))

print("\nEXHAUSTIVE MATRIX SUMMARY")
print(df_matrix["status"].value_counts().to_dict())
print("Cases generated:", len(df_matrix))
print("Failures:", int((df_matrix["status"] == "FAIL").sum()))
print(
    "All 72 cases passed:",
    len(df_matrix) == 72
    and (df_matrix["status"] == "PASS").all()
)

if (df_matrix["status"] == "FAIL").any():
    print("\nFAILED CASES — INVESTIGATION REQUIRED")
    display(df_matrix[df_matrix["status"] == "FAIL"])
```
