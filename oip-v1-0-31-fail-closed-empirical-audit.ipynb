```python
# ============================================================
# OIP v1.0.31 — CELL 01
# PROVENANCE + VERSION LOCK
# ============================================================

import os
import json
import hashlib
from pathlib import Path
from datetime import datetime, timezone

print("=" * 80)
print("OIP v1.0.31 — CELL 01: PROVENANCE + VERSION LOCK")
print("=" * 80)

# ------------------------------------------------------------
# 1. FIXED EVALUATION IDENTITY
# ------------------------------------------------------------

OIP_VERSION = "v1.0.31"
GOVERNING_PROTOCOL = "OIP v1.0.30"
EVALUATION_DATASET = "Malawi Integrated Household Panel Survey (IHPS)"
DATASET_DOI = "10.48529/q5q1-2a34"

WAVES = [2010, 2013, 2016, 2019]

STATUS = "CANDIDATE_DATASET"

print(f"OIP version              : {OIP_VERSION}")
print(f"Governing protocol       : {GOVERNING_PROTOCOL}")
print(f"Dataset                  : {EVALUATION_DATASET}")
print(f"Dataset DOI              : {DATASET_DOI}")
print(f"Waves                    : {WAVES}")
print(f"Dataset status           : {STATUS}")

# ------------------------------------------------------------
# 2. IMPORTANT FAIL-CLOSED RULES
# ------------------------------------------------------------

RULES = {
    "no_invented_variables": True,
    "no_invented_mappings": True,
    "no_invented_outcomes": True,
    "no_random_fallback": True,
    "no_silent_imputation": True,
    "missing_is_not_zero": True,
    "missing_is_not_negative": True,
    "approval_requires_documentary_evidence": True,
    "candidate_is_not_approved": True,
    "discovery_is_not_approval": True,
    "ids_zero_liv_is_undefined": True,
    "no_epsilon_for_ids_zero": True,
    "no_posthoc_formula_changes": True,
    "no_posthoc_cohort_optimization": True,
    "no_synthetic_outcomes_for_empirical_claims": True,
    "fail_closed_on_insufficient_evidence": True
}

print("\nFail-closed rules loaded:")
for key, value in RULES.items():
    print(f"  {key}: {value}")

# ------------------------------------------------------------
# 3. KAGGLE ENVIRONMENT INVENTORY
# ------------------------------------------------------------

SEARCH_ROOTS = [
    Path("/kaggle/input"),
    Path("/kaggle/working")
]

files_found = []

for root in SEARCH_ROOTS:
    if root.exists():
        for path in root.rglob("*"):
            if path.is_file():
                files_found.append(path)

print("\nFiles discovered:")
print(f"  Total files: {len(files_found)}")

for path in files_found[:100]:
    print(f"  {path}")

if len(files_found) > 100:
    print(f"  ... {len(files_found) - 100} additional files not displayed")

# ------------------------------------------------------------
# 4. FILE HASH FUNCTION
# ------------------------------------------------------------

def sha256_file(path, chunk_size=1024 * 1024):
    h = hashlib.sha256()

    with open(path, "rb") as f:
        while True:
            chunk = f.read(chunk_size)
            if not chunk:
                break
            h.update(chunk)

    return h.hexdigest()

# ------------------------------------------------------------
# 5. PROVENANCE INVENTORY
# ------------------------------------------------------------

file_inventory = []

for path in files_found:
    try:
        stat = path.stat()

        record = {
            "path": str(path),
            "filename": path.name,
            "suffix": path.suffix.lower(),
            "size_bytes": stat.st_size,
            "sha256": sha256_file(path)
        }

        file_inventory.append(record)

    except Exception as e:
        file_inventory.append({
            "path": str(path),
            "filename": path.name,
            "error": str(e)
        })

print("\nSHA-256 provenance inventory created.")
print(f"  Records: {len(file_inventory)}")

# ------------------------------------------------------------
# 6. VERSION 31 GATE
# ------------------------------------------------------------

VERSION_LOCK = {
    "oip_version": OIP_VERSION,
    "governing_protocol": GOVERNING_PROTOCOL,
    "dataset": EVALUATION_DATASET,
    "dataset_doi": DATASET_DOI,
    "waves": WAVES,
    "dataset_status": STATUS,
    "created_utc": datetime.now(timezone.utc).isoformat(),
    "rules": RULES,
    "file_inventory": file_inventory
}

# ------------------------------------------------------------
# 7. WRITE IMMUTABLE-STYLE MANIFEST
# ------------------------------------------------------------

OUTPUT_DIR = Path("/kaggle/working/oip_v1_0_31_provenance")
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

manifest_path = OUTPUT_DIR / "01_PROVENANCE_LOCK.json"

with open(manifest_path, "w", encoding="utf-8") as f:
    json.dump(
        VERSION_LOCK,
        f,
        indent=2,
        ensure_ascii=False
    )

print("\nManifest written:")
print(f"  {manifest_path}")

# ------------------------------------------------------------
# 8. FINAL CELL STATUS
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("CELL 01 STATUS")
print("=" * 80)

print("PROVENANCE_LOCK = CREATED")
print("DATASET_STATUS   = CANDIDATE_DATASET")
print("CONSTRUCTS       = NOT YET APPROVED")
print("OUTCOME          = NOT YET APPROVED")
print("COHORT           = NOT YET CREATED")
print("SCORES            = NOT CALCULATED")
print("EMPIRICAL RESULT = NOT CLAIMED")

print("\nCell 01 completed without construct inference or score calculation.")
```

```python
# ============================================================
# OIP v1.0.31 — CELL 02
# DATASET + FILE + SCHEMA INVENTORY
# ============================================================
#
# PURPOSE:
#   Structural inventory of the Malawi IHPS package.
#
# THIS CELL DOES NOT:
#   - infer constructs
#   - approve variables
#   - create mappings
#   - approve outcomes
#   - create cohorts
#   - calculate scores
#   - modify source data
#
# FAIL-CLOSED PRINCIPLE:
#   Structural information is recorded exactly as observed.
#   Meaning is NOT inferred from filenames or variable names.
# ============================================================

import json
import sys
from pathlib import Path
from datetime import datetime, timezone

import pandas as pd

print("=" * 80)
print("OIP v1.0.31 — CELL 02: DATASET + FILE + SCHEMA INVENTORY")
print("=" * 80)

# ------------------------------------------------------------
# 1. PATHS
# ------------------------------------------------------------

INPUT_ROOT = Path(
    "/kaggle/input/datasets/sudharsandas27/oip-31"
)

OUTPUT_ROOT = Path(
    "/kaggle/working/oip_v1_0_31_provenance"
)

OUTPUT_ROOT.mkdir(
    parents=True,
    exist_ok=True
)

print()
print("Input root :", INPUT_ROOT)
print("Output root:", OUTPUT_ROOT)

if not INPUT_ROOT.exists():

    raise RuntimeError(
        f"FAIL-CLOSED: Input root not found: {INPUT_ROOT}"
    )

# ------------------------------------------------------------
# 2. DISCOVER ALL FILES
# ------------------------------------------------------------

all_files = sorted(
    [
        p for p in INPUT_ROOT.rglob("*")
        if p.is_file()
    ],
    key=lambda p: str(p).lower()
)

print()
print("Total files discovered:", len(all_files))

if len(all_files) == 0:

    raise RuntimeError(
        "FAIL-CLOSED: No files discovered."
    )

# ------------------------------------------------------------
# 3. FILE TYPE INVENTORY
# ------------------------------------------------------------

extension_counts = {}

for path in all_files:

    extension = path.suffix.lower()

    if extension == "":
        extension = "[NO_EXTENSION]"

    extension_counts[extension] = (
        extension_counts.get(extension, 0) + 1
    )

print()
print("File type inventory:")

for extension, count in sorted(
    extension_counts.items()
):

    print(
        f"  {extension:20s} : {count}"
    )

# ------------------------------------------------------------
# 4. FILE INVENTORY
# ------------------------------------------------------------

file_inventory = []

for path in all_files:

    try:
        size_bytes = path.stat().st_size
    except Exception:
        size_bytes = None

    file_inventory.append({

        "relative_path": str(
            path.relative_to(INPUT_ROOT)
        ),

        "extension": (
            path.suffix.lower()
            if path.suffix
            else "[NO_EXTENSION]"
        ),

        "size_bytes": size_bytes

    })

# ------------------------------------------------------------
# 5. IDENTIFY DATA AND DOCUMENT FILES
# ------------------------------------------------------------

data_extensions = {
    ".dta",
    ".csv",
    ".sav",
    ".sas7bdat",
    ".parquet",
    ".xlsx",
    ".xls"
}

documentation_extensions = {
    ".pdf",
    ".doc",
    ".docx",
    ".txt",
    ".html",
    ".htm"
}

data_files = [
    p for p in all_files
    if p.suffix.lower() in data_extensions
]

documentation_files = [
    p for p in all_files
    if p.suffix.lower() in documentation_extensions
]

print()
print(
    "Potential data files:",
    len(data_files)
)

print(
    "Potential documentation files:",
    len(documentation_files)
)

# ------------------------------------------------------------
# 6. SCHEMA INVENTORY
#
# IMPORTANT:
# Only structural metadata are recorded.
# Variable meaning is NOT interpreted.
# ------------------------------------------------------------

schema_inventory = []

for index, path in enumerate(
    data_files,
    start=1
):

    extension = path.suffix.lower()

    record = {

        "relative_path": str(
            path.relative_to(INPUT_ROOT)
        ),

        "extension": extension,

        "read_status": "NOT_ATTEMPTED",

        "rows": None,

        "columns": None,

        "column_names": [],

        "error_type": None,

        "error": None

    }

    try:

        # ----------------------------------------------------
        # STATA
        # ----------------------------------------------------

        if extension == ".dta":

            # Read only the dataset structure.
            # No semantic interpretation is performed.
            df = pd.read_stata(
                path,
                convert_categoricals=False
            )

            record["rows"] = int(
                df.shape[0]
            )

            record["columns"] = int(
                df.shape[1]
            )

            record["column_names"] = [
                str(c)
                for c in df.columns
            ]

            record["read_status"] = "SUCCESS"

            # Release memory immediately.
            del df

        # ----------------------------------------------------
        # CSV
        # ----------------------------------------------------

        elif extension == ".csv":

            df = pd.read_csv(
                path,
                nrows=0,
                low_memory=False
            )

            record["columns"] = int(
                len(df.columns)
            )

            record["column_names"] = [
                str(c)
                for c in df.columns
            ]

            record["read_status"] = "SUCCESS"

            del df

        # ----------------------------------------------------
        # PARQUET
        # ----------------------------------------------------

        elif extension == ".parquet":

            df = pd.read_parquet(path)

            record["rows"] = int(
                df.shape[0]
            )

            record["columns"] = int(
                df.shape[1]
            )

            record["column_names"] = [
                str(c)
                for c in df.columns
            ]

            record["read_status"] = "SUCCESS"

            del df

        # ----------------------------------------------------
        # EXCEL
        # ----------------------------------------------------

        elif extension in {
            ".xlsx",
            ".xls"
        }:

            workbook = pd.ExcelFile(path)

            record["columns"] = int(
                len(workbook.sheet_names)
            )

            record["column_names"] = [
                f"SHEET::{sheet}"
                for sheet in workbook.sheet_names
            ]

            record["read_status"] = "SUCCESS"

            del workbook

        # ----------------------------------------------------
        # OTHER FORMATS
        # ----------------------------------------------------

        else:

            record["read_status"] = (
                "UNSUPPORTED_FORMAT"
            )

    except Exception as e:

        record["read_status"] = "ERROR"

        record["error_type"] = (
            type(e).__name__
        )

        record["error"] = str(e)

    schema_inventory.append(record)

    # Progress indicator every 25 files
    if (
        index % 25 == 0
        or index == len(data_files)
    ):

        print(
            f"  Schema progress: "
            f"{index}/{len(data_files)}"
        )

# ------------------------------------------------------------
# 7. SCHEMA SUMMARY
# ------------------------------------------------------------

successful_schema = [
    r for r in schema_inventory
    if r["read_status"] == "SUCCESS"
]

failed_schema = [
    r for r in schema_inventory
    if r["read_status"] == "ERROR"
]

unsupported_schema = [
    r for r in schema_inventory
    if r["read_status"] == "UNSUPPORTED_FORMAT"
]

print()
print("=" * 80)
print("SCHEMA INSPECTION SUMMARY")
print("=" * 80)

print(
    "Successful:",
    len(successful_schema)
)

print(
    "Failed:",
    len(failed_schema)
)

print(
    "Unsupported:",
    len(unsupported_schema)
)

# ------------------------------------------------------------
# 8. DOCUMENTATION INVENTORY
#
# This records files only.
# It does NOT interpret their contents.
# ------------------------------------------------------------

documentation_inventory = []

for path in documentation_files:

    documentation_inventory.append({

        "relative_path": str(
            path.relative_to(INPUT_ROOT)
        ),

        "extension": (
            path.suffix.lower()
        ),

        "size_bytes": path.stat().st_size

    })

# ------------------------------------------------------------
# 9. DATASET STRUCTURAL SUMMARY
# ------------------------------------------------------------

dataset_summary = {

    "protocol_version": "OIP v1.0.31",

    "governing_protocol": "OIP v1.0.30",

    "inventory_timestamp_utc": (
        datetime.now(
            timezone.utc
        ).isoformat()
    ),

    "input_root": str(
        INPUT_ROOT
    ),

    "total_files": len(
        all_files
    ),

    "potential_data_files": len(
        data_files
    ),

    "potential_documentation_files": len(
        documentation_files
    ),

    "file_type_counts": (
        extension_counts
    ),

    "schema_success_count": len(
        successful_schema
    ),

    "schema_error_count": len(
        failed_schema
    ),

    "schema_unsupported_count": len(
        unsupported_schema
    ),

    "construct_status": (
        "NOT_APPROVED"
    ),

    "mapping_status": (
        "NOT_APPROVED"
    ),

    "outcome_status": (
        "NOT_APPROVED"
    ),

    "cohort_status": (
        "NOT_CREATED"
    ),

    "score_status": (
        "NOT_CALCULATED"
    ),

    "empirical_result_status": (
        "NOT_CLAIMED"
    )

}

# ------------------------------------------------------------
# 10. SAVE FILE INVENTORY
# ------------------------------------------------------------

file_inventory_path = (
    OUTPUT_ROOT /
    "02_FILE_INVENTORY.json"
)

with open(
    file_inventory_path,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        file_inventory,
        f,
        indent=2,
        ensure_ascii=False
    )

# ------------------------------------------------------------
# 11. SAVE SCHEMA INVENTORY
# ------------------------------------------------------------

schema_inventory_path = (
    OUTPUT_ROOT /
    "02_SCHEMA_INVENTORY.json"
)

with open(
    schema_inventory_path,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        schema_inventory,
        f,
        indent=2,
        ensure_ascii=False
    )

# ------------------------------------------------------------
# 12. SAVE DOCUMENTATION INVENTORY
# ------------------------------------------------------------

documentation_inventory_path = (
    OUTPUT_ROOT /
    "02_DOCUMENTATION_INVENTORY.json"
)

with open(
    documentation_inventory_path,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        documentation_inventory,
        f,
        indent=2,
        ensure_ascii=False
    )

# ------------------------------------------------------------
# 13. SAVE STRUCTURAL SUMMARY
# ------------------------------------------------------------

summary_path = (
    OUTPUT_ROOT /
    "02_DATASET_STRUCTURAL_SUMMARY.json"
)

with open(
    summary_path,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        dataset_summary,
        f,
        indent=2,
        ensure_ascii=False
    )

# ------------------------------------------------------------
# 14. FINAL FAIL-CLOSED STATUS
# ------------------------------------------------------------

print()
print("=" * 80)
print("CELL 02 STATUS")
print("=" * 80)

if len(successful_schema) > 0:

    print(
        "SCHEMA INVENTORY = CREATED"
    )

else:

    print(
        "SCHEMA INVENTORY = BLOCKED"
    )

print(
    "FILE INVENTORY    = CREATED"
)

print(
    "DOCUMENT INVENTORY = CREATED"
)

print(
    "STRUCTURAL SUMMARY = CREATED"
)

print()
print(
    "CONSTRUCTS        = NOT APPROVED"
)

print(
    "MAPPINGS          = NOT APPROVED"
)

print(
    "OUTCOME           = NOT APPROVED"
)

print(
    "COHORT            = NOT CREATED"
)

print(
    "SCORES            = NOT CALCULATED"
)

print(
    "EMPIRICAL RESULT  = NOT CLAIMED"
)

print()
print(
    "No construct inference was performed."
)

print(
    "No variable mapping was performed."
)

print(
    "No source data were modified."
)

print("=" * 80)
```

```python
# ============================================================
# OIP v1.0.31 — CELL 03
# PANEL ID + TEMPORAL STRUCTURE AUDIT
# FINAL CORRECTED VERSION
# ============================================================

import json
from pathlib import Path
import re

print("=" * 80)
print("OIP v1.0.31 — CELL 03: PANEL ID + TEMPORAL STRUCTURE AUDIT")
print("=" * 80)

BASE = Path("/kaggle/working/oip_v1_0_31_provenance")

schema_path = BASE / "02_SCHEMA_INVENTORY.json"

if not schema_path.exists():
    raise FileNotFoundError(
        f"Required Cell 02 artifact not found: {schema_path}"
    )

with open(schema_path, "r", encoding="utf-8") as f:
    schema = json.load(f)

# ------------------------------------------------------------
# 1. Recover schema records safely
# ------------------------------------------------------------

records = []

if isinstance(schema, list):
    records = schema

elif isinstance(schema, dict):
    for key in [
        "records",
        "schema_records",
        "schemas",
        "files",
        "data",
        "inventory"
    ]:
        value = schema.get(key)

        if isinstance(value, list):
            records = value
            break

        if isinstance(value, dict):
            records = list(value.values())
            break

print(f"Cell 02 schema inventory: FOUND")
print(f"Schema records recovered: {len(records)}")

# ------------------------------------------------------------
# 2. Extract file names and variable names recursively
# ------------------------------------------------------------

def recursive_strings(obj):
    """
    Collect strings from nested dictionaries/lists.
    Used only for structural inspection.
    """
    output = []

    if isinstance(obj, str):
        output.append(obj)

    elif isinstance(obj, dict):
        for value in obj.values():
            output.extend(recursive_strings(value))

    elif isinstance(obj, list):
        for item in obj:
            output.extend(recursive_strings(item))

    return output


all_strings = []

for record in records:
    all_strings.extend(recursive_strings(record))

# ------------------------------------------------------------
# 3. Identifier-name structural scan
# ------------------------------------------------------------

identifier_patterns = [
    r"^hhid$",
    r"^householdid$",
    r"^household_id$",
    r"^hh_id$",
    r"^pid$",
    r"^personid$",
    r"^person_id$",
    r"^individualid$",
    r"^individual_id$",
    r"^caseid$",
    r"^case_id$",
    r"^panelid$",
    r"^panel_id$",
    r"^respondentid$",
    r"^respondent_id$",
    r"^memberid$",
    r"^member_id$"
]

candidate_names = set()

for value in all_strings:
    name = value.strip().lower()

    for pattern in identifier_patterns:
        if re.fullmatch(pattern, name):
            candidate_names.add(name)
            break

print("\nIdentifier-name structural scan:")
print(f"  Unique candidate names: {len(candidate_names)}")

for name in sorted(candidate_names):
    print(f"  {name}")

# ------------------------------------------------------------
# 4. Filename / wave-token structural audit
# ------------------------------------------------------------

dta_files = sorted(
    p for p in Path("/kaggle/input").rglob("*")
    if p.is_file() and p.suffix.lower() == ".dta"
)

print("\nStata file count:")
print(f"  .dta: {len(dta_files)}")

wave_tokens = {
    "2010": 0,
    "2013": 0,
    "2016": 0,
    "2019": 0
}

for path in dta_files:
    filename = path.name.lower()

    for wave in wave_tokens:
        if re.search(rf"(?<!\d){wave}(?!\d)", filename):
            wave_tokens[wave] += 1

print("\nFilename-level wave token scan:")

for wave in sorted(wave_tokens):
    print(f"  {wave}: {wave_tokens[wave]} filename match(es)")

# ------------------------------------------------------------
# 5. Files containing candidate names
# ------------------------------------------------------------

files_with_candidates = set()

for record in records:

    record_strings = recursive_strings(record)

    record_lower = {
        s.strip().lower()
        for s in record_strings
        if isinstance(s, str)
    }

    matched = candidate_names.intersection(record_lower)

    if matched:
        # Try to recover a filename/path from the record
        possible_files = [
            s for s in record_strings
            if isinstance(s, str)
            and (
                s.lower().endswith(".dta")
                or ".dta/" in s.lower()
            )
        ]

        if possible_files:
            files_with_candidates.add(possible_files[0])

print("\nFiles containing identifier-name candidates:")
print(f"  {len(files_with_candidates)}")

for item in sorted(files_with_candidates):
    print(f"  {item}")

# ------------------------------------------------------------
# 6. FAIL-CLOSED STATUS
# ------------------------------------------------------------

panel_id_status = "NOT_APPROVED"
panel_linkage_status = "NOT_VERIFIED"

if candidate_names:
    structural_identifier_signal = "PRESENT"
else:
    structural_identifier_signal = "NOT_FOUND_IN_SCHEMA_ARTIFACT"

result = {
    "protocol_version": "OIP v1.0.31",
    "governing_protocol": "OIP v1.0.30",

    "stata_file_count": len(dta_files),

    "schema_inventory_found": True,
    "schema_records_recovered": len(records),

    "identifier_name_candidates": sorted(candidate_names),
    "identifier_name_candidate_count": len(candidate_names),

    "files_with_identifier_candidates": sorted(
        files_with_candidates
    ),

    "wave_filename_scan": wave_tokens,

    "structural_identifier_signal":
        structural_identifier_signal,

    "panel_id_status":
        panel_id_status,

    "panel_linkage_status":
        panel_linkage_status,

    "constructs":
        "NOT_APPROVED",

    "mappings":
        "NOT_APPROVED",

    "outcome":
        "NOT_APPROVED",

    "cohort":
        "NOT_CREATED",

    "scores":
        "NOT_CALCULATED",

    "empirical_result":
        "NOT_CLAIMED",

    "audit_notes": [
        "This cell performs structural inspection only.",
        "Identifier-name presence does not constitute documentary evidence.",
        "No variable was approved as a panel identifier.",
        "No longitudinal merge was performed.",
        "No analytical cohort was created.",
        "No source data were modified.",
        "Documentary evidence is still required before panel linkage approval."
    ]
}

output_path = BASE / "03_PANEL_STRUCTURAL_AUDIT.json"

with open(output_path, "w", encoding="utf-8") as f:
    json.dump(result, f, indent=2, ensure_ascii=False)

# ------------------------------------------------------------
# 7. FINAL STATUS
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("CELL 03 STATUS")
print("=" * 80)

print(f"STATA FILES DISCOVERED = {len(dta_files)}")
print(f"DTA FILE COUNT         = {len(dta_files)}")
print("FILE STRUCTURE         = RECORDED")
print("WAVE STRUCTURE         = STRUCTURALLY SCANNED")
print("ID NAME SCAN            = COMPLETED")

print(f"PANEL ID STATUS        = {panel_id_status}")
print(f"PANEL LINKAGE          = {panel_linkage_status}")

print("CONSTRUCTS             = NOT APPROVED")
print("MAPPINGS               = NOT APPROVED")
print("OUTCOME                = NOT APPROVED")
print("COHORT                 = NOT CREATED")
print("SCORES                 = NOT CALCULATED")
print("EMPIRICAL RESULT       = NOT CLAIMED")

print("\nNo construct inference was performed.")
print("No variable was approved as a panel identifier.")
print("No longitudinal merge was performed.")
print("No analytical cohort was created.")
print("No source data were modified.")

print(f"\nOutput: {output_path}")
print("=" * 80)
```

```python
# ============================================================
# OIP v1.0.31 — CELL 07
# IDENTIFIER VALUE-STRUCTURE AUDIT
# FAIL-CLOSED
# ============================================================

import json
from pathlib import Path
import pandas as pd

print("=" * 80)
print("OIP v1.0.31 — CELL 07: IDENTIFIER VALUE-STRUCTURE AUDIT")
print("=" * 80)

BASE = Path("/kaggle/working/oip_v1_0_31_provenance")

CELL03_PATH = BASE / "03_PANEL_STRUCTURAL_AUDIT.json"
CELL04_PATH = BASE / "04_PANEL_IDENTIFIER_STRUCTURAL_CROSSFILE_AUDIT.json"
CELL06_PATH = BASE / "06_DOCUMENTATION_ACCESSIBILITY_AUDIT.json"

for path in [
    CELL03_PATH,
    CELL04_PATH,
    CELL06_PATH
]:
    if not path.exists():
        raise FileNotFoundError(
            f"Required artifact not found: {path}"
        )

with open(CELL03_PATH, "r", encoding="utf-8") as f:
    cell03 = json.load(f)

with open(CELL04_PATH, "r", encoding="utf-8") as f:
    cell04 = json.load(f)

with open(CELL06_PATH, "r", encoding="utf-8") as f:
    cell06 = json.load(f)

# ------------------------------------------------------------
# 1. Candidate identifiers
# ------------------------------------------------------------

candidate_identifiers = [
    x.lower()
    for x in cell03.get(
        "identifier_name_candidates",
        []
    )
]

print("\nCandidate identifiers:")
for identifier in candidate_identifiers:
    print(f"  {identifier}")

# ------------------------------------------------------------
# 2. Discover relevant Stata files
# ------------------------------------------------------------

input_root = Path("/kaggle/input")

stata_files = sorted(
    p for p in input_root.rglob("*")
    if p.is_file()
    and p.suffix.lower() == ".dta"
)

print(
    f"\nTotal Stata files discovered: "
    f"{len(stata_files)}"
)

# ------------------------------------------------------------
# 3. Only inspect files whose schema contains a candidate ID
# ------------------------------------------------------------

cross_file_summary = cell04.get(
    "cross_file_identifier_summary",
    {}
)

candidate_files = set()

for identifier, info in cross_file_summary.items():

    for file_name in info.get("files", []):

        if file_name:
            candidate_files.add(file_name)

print(
    f"Candidate-containing schema records/files: "
    f"{len(candidate_files)}"
)

# ------------------------------------------------------------
# 4. Match schema paths back to actual input files
# ------------------------------------------------------------

matched_files = []

for path in stata_files:

    normalized_path = str(path).replace("\\", "/")

    relative_tail = normalized_path.split(
        "/kaggle/input/",
        1
    )[-1]

    filename = path.name

    if (
        filename in candidate_files
        or relative_tail in candidate_files
        or normalized_path in candidate_files
    ):
        matched_files.append(path)

# If schema stores only shortened paths, use filename matching
if not matched_files:

    candidate_basenames = {
        Path(x).name
        for x in candidate_files
        if isinstance(x, str)
    }

    matched_files = [
        p for p in stata_files
        if p.name in candidate_basenames
    ]

print(
    f"Matched physical Stata files: "
    f"{len(matched_files)}"
)

# ------------------------------------------------------------
# 5. Bounded value-structure inspection
# ------------------------------------------------------------

inspection = []

MAX_FILES = 25

files_to_inspect = matched_files[:MAX_FILES]

print(
    f"\nBounded inspection limit: "
    f"{MAX_FILES} files"
)

for path in files_to_inspect:

    try:

        df = pd.read_stata(
            path,
            convert_categoricals=False
        )

        present_ids = [
            identifier
            for identifier in candidate_identifiers
            if identifier in df.columns
        ]

        if not present_ids:
            continue

        record = {
            "file": str(path),
            "rows": int(len(df)),
            "columns": int(len(df.columns)),
            "identifier_columns": {},
        }

        for identifier in present_ids:

            series = df[identifier]

            nonmissing = series.notna()

            nonmissing_count = int(
                nonmissing.sum()
            )

            unique_nonmissing = int(
                series.loc[nonmissing].nunique(
                    dropna=True
                )
            )

            duplicate_nonmissing = (
                nonmissing_count -
                unique_nonmissing
            )

            record["identifier_columns"][identifier] = {
                "nonmissing_count":
                    nonmissing_count,

                "unique_nonmissing":
                    unique_nonmissing,

                "duplicate_nonmissing":
                    int(duplicate_nonmissing),

                "all_nonmissing_unique":
                    bool(
                        duplicate_nonmissing == 0
                    )
            }

        inspection.append(record)

        print(
            f"  inspected: "
            f"{path.name}"
        )

    except Exception as e:

        inspection.append({
            "file": str(path),
            "inspection_status": "ERROR",
            "error_type": type(e).__name__
        })

# ------------------------------------------------------------
# 6. No approval decision
# ------------------------------------------------------------

panel_id_status = "NOT_APPROVED"
panel_linkage_status = "NOT_VERIFIED"

# ------------------------------------------------------------
# 7. Save result
# ------------------------------------------------------------

result = {

    "protocol_version":
        "OIP v1.0.31",

    "governing_protocol":
        "OIP v1.0.30",

    "candidate_identifiers":
        candidate_identifiers,

    "total_stata_files":
        len(stata_files),

    "candidate_schema_files":
        len(candidate_files),

    "physical_files_matched":
        len(matched_files),

    "bounded_files_inspected":
        len(files_to_inspect),

    "identifier_value_structure":
        inspection,

    "documentation_accessibility":
        cell06.get(
            "accessibility_status"
        ),

    "panel_id_status":
        panel_id_status,

    "panel_linkage_status":
        panel_linkage_status,

    "constructs":
        "NOT_APPROVED",

    "mappings":
        "NOT_APPROVED",

    "outcome":
        "NOT_APPROVED",

    "cohort":
        "NOT_CREATED",

    "scores":
        "NOT_CALCULATED",

    "empirical_result":
        "NOT_CLAIMED",

    "audit_notes": [

        "Identifier value structure is descriptive only.",

        "Uniqueness does not establish identifier semantics.",

        "Duplicate patterns do not establish or reject panel linkage.",

        "No identifier was approved.",

        "No longitudinal merge was performed.",

        "No analytical cohort was created.",

        "No source data were modified.",

        "Documentary evidence remains required."
    ]
}

OUTPUT = (
    BASE /
    "07_IDENTIFIER_VALUE_STRUCTURE_AUDIT.json"
)

with open(OUTPUT, "w", encoding="utf-8") as f:
    json.dump(
        result,
        f,
        indent=2,
        ensure_ascii=False
    )

# ------------------------------------------------------------
# 8. Final status
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("CELL 07 STATUS")
print("=" * 80)

print(
    f"STATA FILES          = "
    f"{len(stata_files)}"
)

print(
    f"CANDIDATE ID FILES   = "
    f"{len(candidate_files)}"
)

print(
    f"FILES INSPECTED      = "
    f"{len(files_to_inspect)}"
)

print(
    f"PANEL ID STATUS      = "
    f"{panel_id_status}"
)

print(
    f"PANEL LINKAGE        = "
    f"{panel_linkage_status}"
)

print("CONSTRUCTS           = NOT APPROVED")
print("MAPPINGS             = NOT APPROVED")
print("OUTCOME              = NOT APPROVED")
print("COHORT               = NOT CREATED")
print("SCORES               = NOT CALCULATED")
print("EMPIRICAL RESULT     = NOT CLAIMED")

print("\nNo identifier was approved.")
print("No longitudinal merge was performed.")
print("No analytical cohort was created.")
print("No source data were modified.")

print(f"\nOutput: {OUTPUT}")
print("=" * 80)
```

```python
# ============================================================
# OIP v1.0.31 — CELL 10
# DOCUMENTATION SOURCE IDENTITY LOCK
# FAIL-CLOSED
# ============================================================

import json
import hashlib
from pathlib import Path

print("=" * 80)
print("OIP v1.0.31 — CELL 10: DOCUMENTATION SOURCE IDENTITY LOCK")
print("=" * 80)

BASE = Path("/kaggle/working/oip_v1_0_31_provenance")

CELL01_PATH = BASE / "01_PROVENANCE_LOCK.json"
CELL09_PATH = BASE / "09_DOCUMENTATION_METADATA_EVIDENCE_INVENTORY.json"

for path in [
    CELL01_PATH,
    CELL09_PATH
]:
    if not path.exists():
        raise FileNotFoundError(
            f"Required artifact not found: {path}"
        )

# ------------------------------------------------------------
# 1. Load previous artifacts
# ------------------------------------------------------------

with open(CELL01_PATH, "r", encoding="utf-8") as f:
    cell01 = json.load(f)

with open(CELL09_PATH, "r", encoding="utf-8") as f:
    cell09 = json.load(f)

# ------------------------------------------------------------
# 2. Recover documentation candidates
# ------------------------------------------------------------

candidates = cell09.get(
    "filesystem_documentation_candidates",
    []
)

print(
    f"\nDocumentation candidates recovered: "
    f"{len(candidates)}"
)

# ------------------------------------------------------------
# 3. Calculate SHA-256 identity
# ------------------------------------------------------------

source_records = []

for item in candidates:

    path = Path(
        item["path"]
    )

    if not path.exists():
        source_records.append({

            "path":
                str(path),

            "status":
                "FILE_NOT_FOUND"
        })

        continue

    sha256 = hashlib.sha256()

    with open(
        path,
        "rb"
    ) as f:

        for chunk in iter(
            lambda: f.read(1024 * 1024),
            b""
        ):

            sha256.update(chunk)

    source_records.append({

        "path":
            str(path),

        "filename":
            path.name,

        "extension":
            path.suffix.lower(),

        "size_bytes":
            int(path.stat().st_size),

        "sha256":
            sha256.hexdigest(),

        "status":
            "IDENTITY_RECORDED"
    })

    print(
        f"  recorded: {path.name}"
    )

# ------------------------------------------------------------
# 4. Evidence status remains fail-closed
# ------------------------------------------------------------

identity_recorded = (
    len(source_records) > 0
    and all(
        x.get("status") == "IDENTITY_RECORDED"
        for x in source_records
    )
)

documentary_evidence_status = (
    "SOURCE_IDENTITY_RECORDED"
    if identity_recorded
    else
    "SOURCE_IDENTITY_NOT_ESTABLISHED"
)

# IMPORTANT:
# Identity is NOT semantic evidence.

semantic_evidence_status = (
    "NOT_ESTABLISHED"
)

panel_id_status = "BLOCKED"
panel_linkage_status = "BLOCKED"

construct_status = "NOT_APPROVED"
mapping_status = "NOT_APPROVED"
outcome_status = "NOT_APPROVED"

# ------------------------------------------------------------
# 5. Save identity lock
# ------------------------------------------------------------

result = {

    "protocol_version":
        "OIP v1.0.31",

    "governing_protocol":
        "OIP v1.0.30",

    "cell":
        "10_DOCUMENTATION_SOURCE_IDENTITY_LOCK",

    "dataset_status":
        cell01.get(
            "dataset_status",
            "UNKNOWN"
        ),

    "documentation_source_count":
        len(source_records),

    "documentation_sources":
        source_records,

    "documentary_evidence_status":
        documentary_evidence_status,

    "semantic_evidence_status":
        semantic_evidence_status,

    "panel_id_status":
        panel_id_status,

    "panel_linkage_status":
        panel_linkage_status,

    "constructs":
        construct_status,

    "mappings":
        mapping_status,

    "outcome":
        outcome_status,

    "approval_decisions": {

        "identifier":
            False,

        "construct":
            False,

        "mapping":
            False,

        "outcome":
            False
    },

    "audit_notes": [

        "SHA-256 records identify the documentation source file.",

        "File identity does not establish documentary semantics.",

        "File identity does not establish panel identifier semantics.",

        "File identity does not establish longitudinal linkage.",

        "File identity does not establish construct mappings.",

        "File identity does not establish outcome definition.",

        "No semantic evidence was inferred.",

        "No identifier was approved.",

        "No longitudinal merge was performed.",

        "No analytical cohort was created.",

        "No source data were modified."
    ]
}

OUTPUT = (
    BASE /
    "10_DOCUMENTATION_SOURCE_IDENTITY_LOCK.json"
)

with open(
    OUTPUT,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        result,
        f,
        indent=2,
        ensure_ascii=False
    )

# ------------------------------------------------------------
# 6. Final status
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("CELL 10 STATUS")
print("=" * 80)

print(
    f"DOCUMENTATION SOURCES = "
    f"{len(source_records)}"
)

print(
    f"IDENTITY STATUS       = "
    f"{documentary_evidence_status}"
)

print(
    f"SEMANTIC EVIDENCE     = "
    f"{semantic_evidence_status}"
)

print(
    f"PANEL ID STATUS       = "
    f"{panel_id_status}"
)

print(
    f"PANEL LINKAGE         = "
    f"{panel_linkage_status}"
)

print("CONSTRUCTS            = NOT APPROVED")
print("MAPPINGS              = NOT APPROVED")
print("OUTCOME               = NOT APPROVED")

print("IDENTIFIER APPROVAL   = NONE")
print("CONSTRUCT APPROVAL    = NONE")
print("MAPPING APPROVAL      = NONE")
print("OUTCOME APPROVAL      = NONE")

print("\nNo semantic inference was performed.")
print("No identifier was approved.")
print("No longitudinal merge was performed.")
print("No analytical cohort was created.")
print("No score was calculated.")
print("No empirical result was claimed.")
print("No source data were modified.")

print(f"\nOutput: {OUTPUT}")
print("=" * 80)
```

```python
# ============================================================
# OIP v1.0.31 — CELL 11
# PDF ENGINE CAPABILITY RECOVERY AUDIT
# FAIL-CLOSED
# ============================================================

import json
import shutil
import subprocess
from pathlib import Path

print("=" * 80)
print("OIP v1.0.31 — CELL 11: PDF ENGINE CAPABILITY RECOVERY AUDIT")
print("=" * 80)

BASE = Path("/kaggle/working/oip_v1_0_31_provenance")

CELL10_PATH = (
    BASE /
    "10_DOCUMENTATION_SOURCE_IDENTITY_LOCK.json"
)

CELL06_PATH = (
    BASE /
    "06_DOCUMENTATION_ACCESSIBILITY_AUDIT.json"
)

if not CELL10_PATH.exists():
    raise FileNotFoundError(
        f"Required artifact not found: {CELL10_PATH}"
    )

if not CELL06_PATH.exists():
    raise FileNotFoundError(
        f"Required artifact not found: {CELL06_PATH}"
    )

with open(
    CELL10_PATH,
    "r",
    encoding="utf-8"
) as f:
    cell10 = json.load(f)

with open(
    CELL06_PATH,
    "r",
    encoding="utf-8"
) as f:
    cell06 = json.load(f)

# ------------------------------------------------------------
# 1. Recover documentation source
# ------------------------------------------------------------

sources = cell10.get(
    "documentation_sources",
    []
)

if not sources:
    raise RuntimeError(
        "No documentation source was recorded in Cell 10."
    )

source_path = Path(
    sources[0].get("path", "")
)

if not source_path.exists():
    raise FileNotFoundError(
        f"Documentation source not found: {source_path}"
    )

print(
    f"\nDocumentation source:"
    f"\n  {source_path}"
)

# ------------------------------------------------------------
# 2. Detect system PDF utilities
# ------------------------------------------------------------

available_tools = {}

for tool_name in [
    "pdfinfo",
    "pdftotext"
]:

    tool_path = shutil.which(tool_name)

    available_tools[tool_name] = (
        str(tool_path)
        if tool_path
        else None
    )

print("\nSystem PDF utilities:")

for tool_name, tool_path in available_tools.items():

    print(
        f"  {tool_name}: "
        f"{tool_path if tool_path else 'NOT_FOUND'}"
    )

# ------------------------------------------------------------
# 3. Bounded pdfinfo test
# ------------------------------------------------------------

pdfinfo_status = "NOT_RUN"
pdfinfo_output = ""
pdfinfo_error = ""

pdfinfo_path = available_tools["pdfinfo"]

if pdfinfo_path:

    try:

        completed = subprocess.run(
            [
                pdfinfo_path,
                str(source_path)
            ],
            capture_output=True,
            text=True,
            timeout=30
        )

        pdfinfo_output = (
            completed.stdout.strip()
        )

        pdfinfo_error = (
            completed.stderr.strip()
        )

        if completed.returncode == 0:

            pdfinfo_status = "PASS"

        else:

            pdfinfo_status = (
                "FAILED_RETURN_CODE"
            )

    except subprocess.TimeoutExpired:

        pdfinfo_status = "TIMEOUT"

    except Exception as e:

        pdfinfo_status = (
            f"ERROR_{type(e).__name__}"
        )

else:

    pdfinfo_status = "NOT_AVAILABLE"

# ------------------------------------------------------------
# 4. Bounded first-page text accessibility test
# ------------------------------------------------------------

pdftotext_status = "NOT_RUN"
first_page_text_length = 0
pdftotext_error = ""

pdftotext_path = available_tools["pdftotext"]

if pdftotext_path:

    try:

        completed = subprocess.run(
            [
                pdftotext_path,
                "-f", "1",
                "-l", "1",
                "-layout",
                str(source_path),
                "-"
            ],
            capture_output=True,
            text=True,
            timeout=30
        )

        first_page_text = (
            completed.stdout
            if completed.stdout is not None
            else ""
        )

        first_page_text_length = len(
            first_page_text.strip()
        )

        pdftotext_error = (
            completed.stderr.strip()
        )

        if (
            completed.returncode == 0
            and first_page_text_length > 0
        ):

            pdftotext_status = "PASS"

        elif (
            completed.returncode == 0
            and first_page_text_length == 0
        ):

            pdftotext_status = (
                "PASS_NO_TEXT_DETECTED"
            )

        else:

            pdftotext_status = (
                "FAILED_RETURN_CODE"
            )

    except subprocess.TimeoutExpired:

        pdftotext_status = "TIMEOUT"

    except Exception as e:

        pdftotext_status = (
            f"ERROR_{type(e).__name__}"
        )

else:

    pdftotext_status = "NOT_AVAILABLE"

# ------------------------------------------------------------
# 5. Fail-closed accessibility classification
# ------------------------------------------------------------

if (
    pdfinfo_status == "PASS"
    and pdftotext_status == "PASS"
):

    accessibility_status = (
        "PDF_SYSTEM_ACCESSIBILITY_AVAILABLE"
    )

elif (
    pdfinfo_status == "PASS"
    or pdftotext_status == "PASS"
):

    accessibility_status = (
        "PDF_PARTIAL_SYSTEM_ACCESSIBILITY"
    )

else:

    accessibility_status = (
        "DOCUMENT_ACCESS_BLOCKED"
    )

# IMPORTANT:
# Accessibility does NOT establish semantic evidence.

semantic_evidence_status = (
    "NOT_ESTABLISHED"
)

panel_id_status = "BLOCKED"
panel_linkage_status = "BLOCKED"

construct_status = "NOT_APPROVED"
mapping_status = "NOT_APPROVED"
outcome_status = "NOT_APPROVED"

# ------------------------------------------------------------
# 6. Save result
# ------------------------------------------------------------

result = {

    "protocol_version":
        "OIP v1.0.31",

    "governing_protocol":
        "OIP v1.0.30",

    "cell":
        "11_PDF_ENGINE_CAPABILITY_RECOVERY_AUDIT",

    "documentation_source":
        str(source_path),

    "previous_accessibility_status":
        cell06.get(
            "accessibility_status",
            "UNKNOWN"
        ),

    "system_pdf_tools":
        available_tools,

    "pdfinfo_status":
        pdfinfo_status,

    "pdfinfo_output":
        pdfinfo_output,

    "pdfinfo_error":
        pdfinfo_error,

    "pdftotext_status":
        pdftotext_status,

    "first_page_text_length":
        first_page_text_length,

    "pdftotext_error":
        pdftotext_error,

    "accessibility_status":
        accessibility_status,

    "semantic_evidence_status":
        semantic_evidence_status,

    "panel_id_status":
        panel_id_status,

    "panel_linkage_status":
        panel_linkage_status,

    "constructs":
        construct_status,

    "mappings":
        mapping_status,

    "outcome":
        outcome_status,

    "approval_decisions": {

        "identifier":
            False,

        "construct":
            False,

        "mapping":
            False,

        "outcome":
            False
    },

    "audit_notes": [

        "This cell tests PDF accessibility mechanisms only.",

        "No variable semantics were inferred.",

        "No documentation claim was approved from filename or tool availability.",

        "A successful PDF read does not itself establish authoritative semantic evidence.",

        "No panel identifier was approved.",

        "No longitudinal linkage was approved.",

        "No construct mapping was approved.",

        "No outcome was approved.",

        "No source data were modified."
    ]
}

OUTPUT = (
    BASE /
    "11_PDF_ENGINE_CAPABILITY_RECOVERY_AUDIT.json"
)

with open(
    OUTPUT,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        result,
        f,
        indent=2,
        ensure_ascii=False
    )

# ------------------------------------------------------------
# 7. Final status
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("CELL 11 STATUS")
print("=" * 80)

print(
    f"PDFINFO STATUS           = "
    f"{pdfinfo_status}"
)

print(
    f"PDFTOTEXT STATUS         = "
    f"{pdftotext_status}"
)

print(
    f"FIRST PAGE TEXT LENGTH   = "
    f"{first_page_text_length}"
)

print(
    f"ACCESSIBILITY STATUS     = "
    f"{accessibility_status}"
)

print(
    f"SEMANTIC EVIDENCE        = "
    f"{semantic_evidence_status}"
)

print(
    f"PANEL ID STATUS          = "
    f"{panel_id_status}"
)

print(
    f"PANEL LINKAGE            = "
    f"{panel_linkage_status}"
)

print("CONSTRUCTS               = NOT APPROVED")
print("MAPPINGS                 = NOT APPROVED")
print("OUTCOME                  = NOT APPROVED")

print("IDENTIFIER APPROVAL      = NONE")
print("CONSTRUCT APPROVAL       = NONE")
print("MAPPING APPROVAL         = NONE")
print("OUTCOME APPROVAL         = NONE")

print("\nNo semantic inference was performed.")
print("No identifier was approved.")
print("No longitudinal merge was performed.")
print("No construct mapping was approved.")
print("No outcome was approved.")
print("No source data were modified.")

print(f"\nOutput: {OUTPUT}")
print("=" * 80)
```

```python
# ============================================================
# OIP v1.0.31 — CELL 12
# STATA NATIVE METADATA INVENTORY
# FAIL-CLOSED
# ============================================================

import json
from pathlib import Path
import pandas as pd

print("=" * 80)
print("OIP v1.0.31 — CELL 12: STATA NATIVE METADATA INVENTORY")
print("=" * 80)

BASE = Path("/kaggle/working/oip_v1_0_31_provenance")

CELL03_PATH = BASE / "03_PANEL_STRUCTURAL_AUDIT.json"
CELL11_PATH = BASE / "11_PDF_ENGINE_CAPABILITY_RECOVERY_AUDIT.json"

if not CELL03_PATH.exists():
    raise FileNotFoundError(
        f"Required artifact not found: {CELL03_PATH}"
    )

if not CELL11_PATH.exists():
    raise FileNotFoundError(
        f"Required artifact not found: {CELL11_PATH}"
    )

with open(CELL03_PATH, "r", encoding="utf-8") as f:
    cell03 = json.load(f)

with open(CELL11_PATH, "r", encoding="utf-8") as f:
    cell11 = json.load(f)

# ------------------------------------------------------------
# 1. Candidate identifiers
# ------------------------------------------------------------

candidate_identifiers = [
    str(x).lower()
    for x in cell03.get(
        "identifier_name_candidates",
        []
    )
]

print("\nCandidate identifiers:")

for identifier in candidate_identifiers:
    print(f"  {identifier}")

# ------------------------------------------------------------
# 2. Discover Stata files
# ------------------------------------------------------------

input_root = Path("/kaggle/input")

stata_files = sorted(
    p
    for p in input_root.rglob("*")
    if p.is_file()
    and p.suffix.lower() == ".dta"
)

print(
    f"\nTotal Stata files discovered: "
    f"{len(stata_files)}"
)

# ------------------------------------------------------------
# 3. Bounded native metadata inspection
# ------------------------------------------------------------

MAX_FILES = 25

files_to_inspect = stata_files[:MAX_FILES]

print(
    f"Bounded metadata inspection limit: "
    f"{MAX_FILES} files"
)

metadata_records = []

for path in files_to_inspect:

    try:

        # Context manager avoids deprecated reader.close()
        with pd.io.stata.StataReader(str(path)) as reader:

            record = {
                "file":
                    str(path),

                "status":
                    "READ_SUCCESS",

                "data_label":
                    None,

                "variable_labels":
                    {},

                "value_label_names":
                    {},

                "metadata_methods_observed":
                    []
            }

            # ------------------------------------------------
            # Dataset label
            # ------------------------------------------------

            try:

                data_label = reader.data_label

                record["data_label"] = (
                    str(data_label)
                    if data_label is not None
                    else None
                )

                record[
                    "metadata_methods_observed"
                ].append(
                    "data_label"
                )

            except Exception as e:

                record[
                    "data_label_error"
                ] = type(e).__name__

            # ------------------------------------------------
            # Variable labels
            # ------------------------------------------------

            try:

                variable_labels = (
                    reader.variable_labels()
                )

                if isinstance(
                    variable_labels,
                    dict
                ):

                    record[
                        "variable_labels"
                    ] = {
                        str(k): str(v)
                        for k, v
                        in variable_labels.items()
                    }

                record[
                    "metadata_methods_observed"
                ].append(
                    "variable_labels"
                )

            except Exception as e:

                record[
                    "variable_labels_error"
                ] = type(e).__name__

            # ------------------------------------------------
            # Value-label names
            # ------------------------------------------------

            try:

                value_label_names = (
                    reader.value_labels()
                )

                if isinstance(
                    value_label_names,
                    dict
                ):

                    record[
                        "value_label_names"
                    ] = {
                        str(k): str(v)
                        for k, v
                        in value_label_names.items()
                    }

                record[
                    "metadata_methods_observed"
                ].append(
                    "value_labels"
                )

            except Exception as e:

                record[
                    "value_label_names_error"
                ] = type(e).__name__

            # ------------------------------------------------
            # Candidate identifier metadata
            # ------------------------------------------------

            record[
                "candidate_identifier_metadata"
            ] = {}

            for identifier in candidate_identifiers:

                if identifier in record[
                    "variable_labels"
                ]:

                    record[
                        "candidate_identifier_metadata"
                    ][identifier] = {

                        "variable_label":
                            record[
                                "variable_labels"
                            ][identifier]
                    }

            metadata_records.append(record)

            print(
                f"  inspected: {path.name}"
            )

    except Exception as e:

        metadata_records.append({

            "file":
                str(path),

            "status":
                "ERROR",

            "error_type":
                type(e).__name__,

            "error_message":
                str(e)
        })

# ------------------------------------------------------------
# 4. Metadata coverage summary
# ------------------------------------------------------------

successful_records = [
    x
    for x in metadata_records
    if x.get("status") == "READ_SUCCESS"
]

files_with_dataset_label = sum(
    1
    for x in successful_records
    if x.get("data_label") not in (
        None,
        ""
    )
)

files_with_variable_labels = sum(
    1
    for x in successful_records
    if isinstance(
        x.get("variable_labels"),
        dict
    )
    and len(
        x.get(
            "variable_labels",
            {}
        )
    ) > 0
)

files_with_value_labels = sum(
    1
    for x in successful_records
    if isinstance(
        x.get("value_label_names"),
        dict
    )
    and len(
        x.get(
            "value_label_names",
            {}
        )
    ) > 0
)

files_with_candidate_identifier_labels = sum(
    1
    for x in successful_records
    if len(
        x.get(
            "candidate_identifier_metadata",
            {}
        )
    ) > 0
)

if (
    files_with_dataset_label > 0
    or files_with_variable_labels > 0
    or files_with_value_labels > 0
):

    native_metadata_status = (
        "NATIVE_METADATA_EXPOSED"
    )

else:

    native_metadata_status = (
        "NO_NATIVE_METADATA_EXPOSED_BY_READER"
    )

# ------------------------------------------------------------
# 5. Fail-closed decision layer
# ------------------------------------------------------------

semantic_evidence_status = (
    "NOT_ESTABLISHED"
)

panel_id_status = "BLOCKED"
panel_linkage_status = "BLOCKED"

construct_status = "NOT_APPROVED"
mapping_status = "NOT_APPROVED"
outcome_status = "NOT_APPROVED"

# ------------------------------------------------------------
# 6. Build result artifact
# ------------------------------------------------------------

result = {

    "protocol_version":
        "OIP v1.0.31",

    "governing_protocol":
        "OIP v1.0.30",

    "cell":
        "12_STATA_NATIVE_METADATA_INVENTORY",

    "total_stata_files":
        len(stata_files),

    "bounded_files_inspected":
        len(files_to_inspect),

    "candidate_identifiers":
        candidate_identifiers,

    "successful_metadata_reads":
        len(successful_records),

    "metadata_read_errors":
        len(metadata_records)
        - len(successful_records),

    "files_with_dataset_label":
        files_with_dataset_label,

    "files_with_variable_labels":
        files_with_variable_labels,

    "files_with_value_labels":
        files_with_value_labels,

    "files_with_candidate_identifier_labels":
        files_with_candidate_identifier_labels,

    "native_metadata_status":
        native_metadata_status,

    "metadata_records":
        metadata_records,

    "pdf_accessibility_status":
        cell11.get(
            "accessibility_status",
            "UNKNOWN"
        ),

    "semantic_evidence_status":
        semantic_evidence_status,

    "panel_id_status":
        panel_id_status,

    "panel_linkage_status":
        panel_linkage_status,

    "constructs":
        construct_status,

    "mappings":
        mapping_status,

    "outcome":
        outcome_status,

    "approval_decisions": {

        "identifier":
            False,

        "construct":
            False,

        "mapping":
            False,

        "outcome":
            False
    },

    "audit_notes": [

        "This cell inspects metadata exposed by pandas StataReader.",

        "Inspection is bounded to the first 25 Stata files.",

        "Native metadata presence does not establish semantic authority.",

        "Variable labels are evidence for review, not automatic variable approval.",

        "Value-label metadata are evidence for review, not automatic construct mapping.",

        "Absence of metadata in the bounded sample does not prove absence from all files.",

        "No variable semantics were approved.",

        "No panel identifier was approved.",

        "No longitudinal linkage was approved.",

        "No construct mapping was approved.",

        "No outcome was approved.",

        "No source data were modified."
    ]
}

# ------------------------------------------------------------
# 7. Save artifact
# ------------------------------------------------------------

OUTPUT = (
    BASE /
    "12_STATA_NATIVE_METADATA_INVENTORY.json"
)

with open(
    OUTPUT,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        result,
        f,
        indent=2,
        ensure_ascii=False
    )

# ------------------------------------------------------------
# 8. Final status
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("CELL 12 STATUS")
print("=" * 80)

print(
    f"STATA FILES                    = "
    f"{len(stata_files)}"
)

print(
    f"FILES INSPECTED                = "
    f"{len(files_to_inspect)}"
)

print(
    f"SUCCESSFUL METADATA READS      = "
    f"{len(successful_records)}"
)

print(
    f"METADATA READ ERRORS           = "
    f"{len(metadata_records) - len(successful_records)}"
)

print(
    f"FILES WITH DATASET LABEL       = "
    f"{files_with_dataset_label}"
)

print(
    f"FILES WITH VARIABLE LABELS     = "
    f"{files_with_variable_labels}"
)

print(
    f"FILES WITH VALUE LABELS        = "
    f"{files_with_value_labels}"
)

print(
    f"FILES WITH CANDIDATE ID LABELS = "
    f"{files_with_candidate_identifier_labels}"
)

print(
    f"NATIVE METADATA STATUS         = "
    f"{native_metadata_status}"
)

print(
    f"PDF ACCESSIBILITY              = "
    f"{cell11.get('accessibility_status', 'UNKNOWN')}"
)

print(
    f"SEMANTIC EVIDENCE              = "
    f"{semantic_evidence_status}"
)

print(
    f"PANEL ID STATUS                = "
    f"{panel_id_status}"
)

print(
    f"PANEL LINKAGE                  = "
    f"{panel_linkage_status}"
)

print("CONSTRUCTS                     = NOT APPROVED")
print("MAPPINGS                       = NOT APPROVED")
print("OUTCOME                        = NOT APPROVED")

print("IDENTIFIER APPROVAL            = NONE")
print("CONSTRUCT APPROVAL             = NONE")
print("MAPPING APPROVAL               = NONE")
print("OUTCOME APPROVAL               = NONE")

print("\nNo semantic inference was performed.")
print("No identifier was approved.")
print("No longitudinal merge was performed.")
print("No construct mapping was approved.")
print("No outcome was approved.")
print("No source data were modified.")

print(f"\nOutput: {OUTPUT}")

print("\n" + "=" * 80)
print("CELL 12 COMPLETE")
print("=" * 80)
```

```python
# ============================================================
# OIP v1.0.31 — CELL 18
# IDENTIFIER WAVE-CONTEXT FORENSIC AUDIT
# FAIL-CLOSED
# ============================================================

import json
import re
from pathlib import Path
from collections import defaultdict

print("=" * 80)
print("OIP v1.0.31 — CELL 18: IDENTIFIER WAVE-CONTEXT FORENSIC AUDIT")
print("=" * 80)

BASE = Path("/kaggle/working/oip_v1_0_31_provenance")

INPUT = (
    BASE /
    "17_EXHAUSTIVE_IDENTIFIER_LABEL_INVENTORY.json"
)

if not INPUT.exists():
    raise FileNotFoundError(
        f"Required artifact not found: {INPUT}"
    )

with open(
    INPUT,
    "r",
    encoding="utf-8"
) as f:
    data = json.load(f)

candidate_identifiers = [
    str(x).lower()
    for x in data.get(
        "candidate_identifiers",
        []
    )
]

records = data.get(
    "identifier_records",
    []
)

if not isinstance(records, list):
    records = []

print("\nCandidate identifiers:")

for identifier in candidate_identifiers:
    print(f"  {identifier}")

print(
    f"\nIdentifier evidence records loaded: "
    f"{len(records)}"
)

# ------------------------------------------------------------
# 1. Conservative wave-token extraction
# ------------------------------------------------------------

def extract_wave_tokens(text):

    text = str(text)

    tokens = set()

    patterns = [

        r'(?<!\d)(2010|2013|2016|2019)(?!\d)',

        r'(?<!\d)(10|13|16|19)(?!\d)'
    ]

    for pattern in patterns:

        for match in re.findall(
            pattern,
            text,
            flags=re.IGNORECASE
        ):

            token = str(match)

            if token in {
                "10",
                "13",
                "16",
                "19"
            }:

                token = {
                    "10": "2010",
                    "13": "2013",
                    "16": "2016",
                    "19": "2019"
                }[token]

            tokens.add(token)

    return sorted(tokens)


# ------------------------------------------------------------
# 2. Build forensic records
# ------------------------------------------------------------

forensic_records = []

for record in records:

    file_path = str(
        record.get(
            "file",
            ""
        )
    )

    identifier = str(
        record.get(
            "identifier",
            ""
        )
    ).lower()

    label = str(
        record.get(
            "variable_label_exact",
            ""
        )
    )

    file_name = Path(
        file_path
    ).name

    file_context = (
        f"{file_path} {file_name}"
    )

    wave_tokens = extract_wave_tokens(
        file_context
    )

    forensic_records.append({

        "identifier":
            identifier,

        "file":
            file_path,

        "file_name":
            file_name,

        "variable_label_exact":
            label,

        "wave_tokens_from_filename":
            wave_tokens,

        "wave_context_status":
            (
                "WAVE_TOKEN_PRESENT"
                if wave_tokens
                else
                "NO_WAVE_TOKEN_IN_FILENAME"
            ),

        "approval_status":
            "NOT_APPROVED"
    })


# ------------------------------------------------------------
# 3. Identifier-level wave coverage
# ------------------------------------------------------------

coverage = {}

for identifier in candidate_identifiers:

    identifier_records = [
        x
        for x in forensic_records
        if x.get("identifier") == identifier
    ]

    wave_counts = defaultdict(int)

    no_wave_count = 0

    for record in identifier_records:

        waves = record.get(
            "wave_tokens_from_filename",
            []
        )

        if waves:

            for wave in waves:
                wave_counts[wave] += 1

        else:

            no_wave_count += 1

    coverage[identifier] = {

        "evidence_record_count":
            len(identifier_records),

        "wave_file_counts":
            dict(
                sorted(
                    wave_counts.items()
                )
            ),

        "records_without_filename_wave_token":
            no_wave_count,

        "approval_status":
            "NOT_APPROVED"
    }


# ------------------------------------------------------------
# 4. Label-by-wave distribution
# ------------------------------------------------------------

label_wave_distribution = {}

for identifier in candidate_identifiers:

    identifier_records = [
        x
        for x in forensic_records
        if x.get("identifier") == identifier
    ]

    distribution = defaultdict(
        lambda: defaultdict(int)
    )

    for record in identifier_records:

        label = record.get(
            "variable_label_exact",
            ""
        )

        waves = record.get(
            "wave_tokens_from_filename",
            []
        )

        if waves:

            for wave in waves:

                distribution[
                    label
                ][wave] += 1

        else:

            distribution[
                label
            ]["NO_WAVE_TOKEN"] += 1

    label_wave_distribution[
        identifier
    ] = {

        label: dict(
            sorted(
                wave_counts.items()
            )
        )

        for label, wave_counts
        in distribution.items()
    }


# ------------------------------------------------------------
# 5. Print forensic summary
# ------------------------------------------------------------

print("\n" + "-" * 80)
print("IDENTIFIER WAVE-CONTEXT SUMMARY")
print("-" * 80)

for identifier in candidate_identifiers:

    summary = coverage[
        identifier
    ]

    print(
        f"\n[{identifier}]"
    )

    print(
        f"Evidence records: "
        f"{summary['evidence_record_count']}"
    )

    print(
        "Filename wave-token counts:"
    )

    if summary[
        "wave_file_counts"
    ]:

        for wave, count in (
            summary[
                "wave_file_counts"
            ].items()
        ):

            print(
                f"  {wave}: {count}"
            )

    else:

        print(
            "  NONE"
        )

    print(
        "Records without filename wave token: "
        f"{summary['records_without_filename_wave_token']}"
    )


# ------------------------------------------------------------
# 6. Print label distribution
# ------------------------------------------------------------

print("\n" + "-" * 80)
print("IDENTIFIER LABEL × WAVE DISTRIBUTION")
print("-" * 80)

for identifier in candidate_identifiers:

    print(
        f"\n[{identifier}]"
    )

    distribution = (
        label_wave_distribution[
            identifier
        ]
    )

    if not distribution:

        print(
            "  No evidence records."
        )

        continue

    for label, waves in distribution.items():

        print(
            f"\n  Label: {label}"
        )

        for wave, count in waves.items():

            print(
                f"    {wave}: {count}"
            )


# ------------------------------------------------------------
# 7. Explicit anti-inference decision
# ------------------------------------------------------------

identifier_approval = {
    identifier: False
    for identifier in candidate_identifiers
}

panel_id_status = "BLOCKED"
panel_linkage_status = "BLOCKED"

# IMPORTANT:
# Filename wave tokens are structural context only.
# They do NOT prove longitudinal linkage.
#
# This cell deliberately does not:
# - merge files
# - test ID uniqueness
# - create a panel
# - approve an identifier
# - infer baseline/new-member semantics
# - map OIP constructs


# ------------------------------------------------------------
# 8. Build output artifact
# ------------------------------------------------------------

result = {

    "protocol_version":
        "OIP v1.0.31",

    "governing_protocol":
        "OIP v1.0.30",

    "cell":
        "18_IDENTIFIER_WAVE_CONTEXT_FORENSIC_AUDIT",

    "input_artifact":
        str(INPUT),

    "candidate_identifiers":
        candidate_identifiers,

    "evidence_record_count":
        len(records),

    "forensic_records":
        forensic_records,

    "identifier_wave_coverage":
        coverage,

    "identifier_label_wave_distribution":
        label_wave_distribution,

    "identifier_approval":
        identifier_approval,

    "panel_id_status":
        panel_id_status,

    "panel_linkage_status":
        panel_linkage_status,

    "semantic_evidence_status":
        "NOT_ESTABLISHED",

    "constructs":
        "NOT_APPROVED",

    "mappings":
        "NOT_APPROVED",

    "outcome":
        "NOT_APPROVED",

    "analytical_cohort":
        "NOT_CREATED",

    "scores":
        "NOT_CALCULATED",

    "empirical_result":
        "NOT_CLAIMED",

    "audit_notes": [

        "This cell analyzes wave context from filenames only.",

        "Filename tokens are structural metadata and are not treated as documentary proof of survey-wave semantics.",

        "Presence of an identifier across multiple files does not prove longitudinal panel linkage.",

        "Presence of a year token does not prove that the identifier is valid for longitudinal linkage.",

        "No identifier was approved.",

        "No longitudinal merge was performed.",

        "No uniqueness claim was made.",

        "No construct mapping was approved.",

        "No outcome was approved.",

        "No analytical cohort was created.",

        "No score was calculated.",

        "No empirical result was claimed.",

        "No source data were modified."
    ]
}

OUTPUT = (
    BASE /
    "18_IDENTIFIER_WAVE_CONTEXT_FORENSIC_AUDIT.json"
)

with open(
    OUTPUT,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        result,
        f,
        indent=2,
        ensure_ascii=False
    )

# ------------------------------------------------------------
# 9. Final status
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("CELL 18 STATUS")
print("=" * 80)

print(
    f"CANDIDATE IDENTIFIERS       = "
    f"{len(candidate_identifiers)}"
)

print(
    f"EVIDENCE RECORDS            = "
    f"{len(records)}"
)

for identifier in candidate_identifiers:

    summary = coverage[
        identifier
    ]

    print(
        f"{identifier} EVIDENCE RECORDS     = "
        f"{summary['evidence_record_count']}"
    )

    print(
        f"{identifier} WAVE CONTEXT         = "
        f"{summary['wave_file_counts']}"
    )

    print(
        f"{identifier} NO-WAVE FILENAMES    = "
        f"{summary['records_without_filename_wave_token']}"
    )

print(
    "IDENTIFIER APPROVAL          = NONE"
)

print(
    "PANEL ID STATUS              = BLOCKED"
)

print(
    "PANEL LINKAGE                = BLOCKED"
)

print(
    "SEMANTIC EVIDENCE            = NOT ESTABLISHED"
)

print(
    "CONSTRUCTS                   = NOT APPROVED"
)

print(
    "MAPPINGS                     = NOT APPROVED"
)

print(
    "OUTCOME                      = NOT APPROVED"
)

print(
    "ANALYTICAL COHORT            = NOT CREATED"
)

print(
    "SCORES                       = NOT CALCULATED"
)

print(
    "EMPIRICAL RESULT             = NOT CLAIMED"
)

print("\nWave-context evidence was inventoried.")
print("No longitudinal linkage was inferred.")
print("No identifier was approved.")
print("No source data were modified.")

print(f"\nOutput: {OUTPUT}")

print("\n" + "=" * 80)
print("CELL 18 COMPLETE")
print("=" * 80)
```

```python
# ============================================================
# OIP v1.0.31 — CELL 20
# CROSS-FILE IDENTIFIER OVERLAP FORENSIC AUDIT
# FAIL-CLOSED
# ============================================================

import json
from pathlib import Path
import pandas as pd

print("=" * 80)
print("OIP v1.0.31 — CELL 20: CROSS-FILE IDENTIFIER OVERLAP FORENSIC AUDIT")
print("=" * 80)

BASE = Path("/kaggle/working/oip_v1_0_31_provenance")

CELL17_PATH = (
    BASE /
    "17_EXHAUSTIVE_IDENTIFIER_LABEL_INVENTORY.json"
)

if not CELL17_PATH.exists():
    raise FileNotFoundError(
        f"Required artifact not found: {CELL17_PATH}"
    )

with open(
    CELL17_PATH,
    "r",
    encoding="utf-8"
) as f:
    cell17 = json.load(f)

candidate_identifiers = [
    str(x).lower()
    for x in cell17.get(
        "candidate_identifiers",
        []
    )
]

identifier_records = cell17.get(
    "identifier_records",
    []
)

if not isinstance(
    identifier_records,
    list
):
    identifier_records = []

candidate_files = sorted(
    set(
        str(
            record.get(
                "file",
                ""
            )
        )
        for record in identifier_records
        if record.get("file")
    )
)

print("\nCandidate identifiers:")

for identifier in candidate_identifiers:
    print(f"  {identifier}")

print(
    f"\nCandidate physical files: "
    f"{len(candidate_files)}"
)

# ------------------------------------------------------------
# 1. Read unique nonmissing identifier values per file
# ------------------------------------------------------------

file_value_sets = {}

read_errors = []

for file_path in candidate_files:

    path = Path(file_path)

    try:

        df = pd.read_stata(
            path,
            convert_categoricals=False
        )

        normalized_columns = {
            str(col).lower(): col
            for col in df.columns
        }

        file_value_sets[
            file_path
        ] = {}

        for identifier in candidate_identifiers:

            if identifier not in normalized_columns:
                continue

            original_column = (
                normalized_columns[
                    identifier
                ]
            )

            values = df[
                original_column
            ]

            values = values[
                values.notna()
            ]

            unique_values = set(
                values.tolist()
            )

            file_value_sets[
                file_path
            ][identifier] = unique_values

    except Exception as e:

        read_errors.append({

            "file":
                file_path,

            "error_type":
                type(e).__name__,

            "error_message":
                str(e)
        })

# ------------------------------------------------------------
# 2. Pairwise overlap analysis
# ------------------------------------------------------------

pairwise_records = []

for identifier in candidate_identifiers:

    relevant_files = [
        file_path
        for file_path in candidate_files
        if identifier
        in file_value_sets.get(
            file_path,
            {}
        )
    ]

    for i in range(
        len(relevant_files)
    ):

        file_a = relevant_files[i]

        values_a = file_value_sets[
            file_a
        ][
            identifier
        ]

        for j in range(
            i + 1,
            len(relevant_files)
        ):

            file_b = relevant_files[j]

            values_b = file_value_sets[
                file_b
            ][
                identifier
            ]

            intersection = (
                values_a
                &
                values_b
            )

            union = (
                values_a
                |
                values_b
            )

            overlap_count = len(
                intersection
            )

            min_set_size = min(
                len(values_a),
                len(values_b)
            )

            overlap_rate_min = (
                overlap_count /
                min_set_size
                if min_set_size > 0
                else None
            )

            jaccard = (
                overlap_count /
                len(union)
                if len(union) > 0
                else None
            )

            pairwise_records.append({

                "identifier":
                    identifier,

                "file_a":
                    file_a,

                "file_b":
                    file_b,

                "unique_values_a":
                    len(values_a),

                "unique_values_b":
                    len(values_b),

                "intersection_count":
                    overlap_count,

                "minimum_set_size":
                    min_set_size,

                "overlap_rate_against_smaller_set":
                    overlap_rate_min,

                "jaccard_overlap":
                    jaccard,

                "linkage_status":
                    "NOT_ESTABLISHED",

                "approval_status":
                    "NOT_APPROVED"
            })

# ------------------------------------------------------------
# 3. Summarize overlap evidence
# ------------------------------------------------------------

overlap_summary = {}

for identifier in candidate_identifiers:

    records = [
        x
        for x in pairwise_records
        if x.get("identifier") == identifier
    ]

    positive_overlap = [
        x
        for x in records
        if x[
            "intersection_count"
        ] > 0
    ]

    overlap_summary[identifier] = {

        "pairwise_comparisons":
            len(records),

        "pairs_with_nonzero_overlap":
            len(positive_overlap),

        "pairs_with_zero_overlap":
            (
                len(records)
                -
                len(positive_overlap)
            ),

        "maximum_intersection_count":
            (
                max(
                    x[
                        "intersection_count"
                    ]
                    for x in records
                )
                if records
                else 0
            ),

        "maximum_overlap_rate_against_smaller_set":
            (
                max(
                    x[
                        "overlap_rate_against_smaller_set"
                    ]
                    for x in records
                    if x[
                        "overlap_rate_against_smaller_set"
                    ] is not None
                )
                if any(
                    x[
                        "overlap_rate_against_smaller_set"
                    ] is not None
                    for x in records
                )
                else None
            ),

        "maximum_jaccard_overlap":
            (
                max(
                    x[
                        "jaccard_overlap"
                    ]
                    for x in records
                    if x[
                        "jaccard_overlap"
                    ] is not None
                )
                if any(
                    x[
                        "jaccard_overlap"
                    ] is not None
                    for x in records
                )
                else None
            ),

        "linkage_status":
            "NOT_ESTABLISHED",

        "approval_status":
            "NOT_APPROVED"
    }

# ------------------------------------------------------------
# 4. Print summary
# ------------------------------------------------------------

print("\n" + "-" * 80)
print("CROSS-FILE IDENTIFIER OVERLAP SUMMARY")
print("-" * 80)

for identifier in candidate_identifiers:

    summary = overlap_summary[
        identifier
    ]

    print(
        f"\n[{identifier}]"
    )

    print(
        f"Pairwise comparisons: "
        f"{summary['pairwise_comparisons']}"
    )

    print(
        f"Pairs with nonzero overlap: "
        f"{summary['pairs_with_nonzero_overlap']}"
    )

    print(
        f"Pairs with zero overlap: "
        f"{summary['pairs_with_zero_overlap']}"
    )

    print(
        f"Maximum intersection count: "
        f"{summary['maximum_intersection_count']}"
    )

    if (
        summary[
            "maximum_overlap_rate_against_smaller_set"
        ]
        is not None
    ):

        print(
            "Maximum overlap rate "
            "against smaller set: "
            f"{summary['maximum_overlap_rate_against_smaller_set']:.6f}"
        )

    if (
        summary[
            "maximum_jaccard_overlap"
        ]
        is not None
    ):

        print(
            "Maximum Jaccard overlap: "
            f"{summary['maximum_jaccard_overlap']:.6f}"
        )

# ------------------------------------------------------------
# 5. Explicit anti-inference rule
# ------------------------------------------------------------

identifier_approval = {
    identifier: False
    for identifier in candidate_identifiers
}

panel_id_status = "BLOCKED"
panel_linkage_status = "BLOCKED"

# IMPORTANT:
#
# Cross-file value overlap is descriptive evidence only.
#
# It does NOT prove:
# - the identifier's semantic authority
# - wave equivalence
# - longitudinal identity
# - household/person-level linkage validity
#
# Therefore no merge or approval occurs here.

# ------------------------------------------------------------
# 6. Save artifact
# ------------------------------------------------------------

result = {

    "protocol_version":
        "OIP v1.0.31",

    "governing_protocol":
        "OIP v1.0.30",

    "cell":
        "20_CROSS_FILE_IDENTIFIER_OVERLAP_FORENSIC_AUDIT",

    "input_artifact":
        str(CELL17_PATH),

    "candidate_identifiers":
        candidate_identifiers,

    "candidate_physical_files":
        len(candidate_files),

    "pairwise_comparison_count":
        len(pairwise_records),

    "read_errors":
        read_errors,

    "overlap_summary":
        overlap_summary,

    "pairwise_records":
        pairwise_records,

    "identifier_approval":
        identifier_approval,

    "panel_id_status":
        panel_id_status,

    "panel_linkage_status":
        panel_linkage_status,

    "semantic_evidence_status":
        "NOT_ESTABLISHED",

    "constructs":
        "NOT_APPROVED",

    "mappings":
        "NOT_APPROVED",

    "outcome":
        "NOT_APPROVED",

    "analytical_cohort":
        "NOT_CREATED",

    "scores":
        "NOT_CALCULATED",

    "empirical_result":
        "NOT_CLAIMED",

    "audit_notes": [

        "Cross-file identifier value overlap was measured without merging datasets.",

        "Nonzero value intersection is descriptive evidence only.",

        "Value overlap does not by itself prove longitudinal identity.",

        "Zero overlap does not by itself prove identifier invalidity.",

        "No wave equivalence was inferred from value overlap.",

        "No identifier was approved.",

        "No longitudinal merge was performed.",

        "No construct mapping was approved.",

        "No outcome was approved.",

        "No analytical cohort was created.",

        "No score was calculated.",

        "No empirical result was claimed.",

        "No source data were modified."
    ]
}

OUTPUT = (
    BASE /
    "20_CROSS_FILE_IDENTIFIER_OVERLAP_FORENSIC_AUDIT.json"
)

with open(
    OUTPUT,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        result,
        f,
        indent=2,
        ensure_ascii=False
    )

# ------------------------------------------------------------
# 7. Final status
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("CELL 20 STATUS")
print("=" * 80)

print(
    f"CANDIDATE IDENTIFIERS       = "
    f"{len(candidate_identifiers)}"
)

print(
    f"PHYSICAL ID FILES           = "
    f"{len(candidate_files)}"
)

print(
    f"PAIRWISE COMPARISONS        = "
    f"{len(pairwise_records)}"
)

print(
    f"VALUE READ ERRORS           = "
    f"{len(read_errors)}"
)

for identifier in candidate_identifiers:

    summary = overlap_summary[
        identifier
    ]

    print(
        f"{identifier} NONZERO OVERLAP PAIRS = "
        f"{summary['pairs_with_nonzero_overlap']}"
    )

print(
    "IDENTIFIER APPROVAL          = NONE"
)

print(
    "PANEL ID STATUS              = BLOCKED"
)

print(
    "PANEL LINKAGE                = BLOCKED"
)

print(
    "SEMANTIC EVIDENCE            = NOT ESTABLISHED"
)

print(
    "CONSTRUCTS                   = NOT APPROVED"
)

print(
    "MAPPINGS                     = NOT APPROVED"
)

print(
    "OUTCOME                      = NOT APPROVED"
)

print(
    "ANALYTICAL COHORT            = NOT CREATED"
)

print(
    "SCORES                       = NOT CALCULATED"
)

print(
    "EMPIRICAL RESULT             = NOT CLAIMED"
)

print("\nCross-file identifier overlap was measured.")
print("No longitudinal linkage was inferred.")
print("No identifier was approved.")
print("No source data were modified.")

print(f"\nOutput: {OUTPUT}")

print("\n" + "=" * 80)
print("CELL 20 COMPLETE")
print("=" * 80)
```

```python
# ============================================================
# OIP v1.0.31 — CELL 21
# IDENTIFIER LABEL + WAVE + OVERLAP CONSISTENCY AUDIT
# FAIL-CLOSED
# ============================================================

import json
from pathlib import Path
from collections import defaultdict

print("=" * 80)
print("OIP v1.0.31 — CELL 21: IDENTIFIER LABEL + WAVE + OVERLAP CONSISTENCY AUDIT")
print("=" * 80)

BASE = Path("/kaggle/working/oip_v1_0_31_provenance")

CELL18_PATH = (
    BASE /
    "18_IDENTIFIER_WAVE_CONTEXT_FORENSIC_AUDIT.json"
)

CELL20_PATH = (
    BASE /
    "20_CROSS_FILE_IDENTIFIER_OVERLAP_FORENSIC_AUDIT.json"
)

if not CELL18_PATH.exists():
    raise FileNotFoundError(
        f"Required artifact not found: {CELL18_PATH}"
    )

if not CELL20_PATH.exists():
    raise FileNotFoundError(
        f"Required artifact not found: {CELL20_PATH}"
    )

with open(
    CELL18_PATH,
    "r",
    encoding="utf-8"
) as f:
    cell18 = json.load(f)

with open(
    CELL20_PATH,
    "r",
    encoding="utf-8"
) as f:
    cell20 = json.load(f)

candidate_identifiers = [
    str(x).lower()
    for x in cell18.get(
        "candidate_identifiers",
        []
    )
]

forensic_records = cell18.get(
    "forensic_records",
    []
)

pairwise_records = cell20.get(
    "pairwise_records",
    []
)

if not isinstance(
    forensic_records,
    list
):
    forensic_records = []

if not isinstance(
    pairwise_records,
    list
):
    pairwise_records = []

# ------------------------------------------------------------
# 1. Build file-level evidence index
# ------------------------------------------------------------

file_evidence = defaultdict(
    dict
)

for record in forensic_records:

    identifier = str(
        record.get(
            "identifier",
            ""
        )
    ).lower()

    file_path = str(
        record.get(
            "file",
            ""
        )
    )

    if not identifier or not file_path:
        continue

    file_evidence[
        identifier
    ][
        file_path
    ] = {

        "file_name":
            record.get(
                "file_name",
                ""
            ),

        "variable_label_exact":
            record.get(
                "variable_label_exact",
                ""
            ),

        "wave_tokens":
            record.get(
                "wave_tokens_from_filename",
                []
            )
    }

# ------------------------------------------------------------
# 2. Build pairwise overlap index
# ------------------------------------------------------------

pairwise_index = defaultdict(
    list
)

for record in pairwise_records:

    identifier = str(
        record.get(
            "identifier",
            ""
        )
    ).lower()

    file_a = str(
        record.get(
            "file_a",
            ""
        )
    )

    file_b = str(
        record.get(
            "file_b",
            ""
        )
    )

    if (
        identifier
        and file_a
        and file_b
    ):

        pairwise_index[
            identifier
        ].append(
            record
        )

# ------------------------------------------------------------
# 3. Analyze wave-pair overlap
# ------------------------------------------------------------

wave_pair_summary = {}

for identifier in candidate_identifiers:

    summary = defaultdict(
        lambda: {
            "pair_count": 0,
            "nonzero_overlap_pairs": 0,
            "zero_overlap_pairs": 0,
            "maximum_intersection": 0
        }
    )

    records = pairwise_index[
        identifier
    ]

    for record in records:

        file_a = record[
            "file_a"
        ]

        file_b = record[
            "file_b"
        ]

        evidence_a = file_evidence[
            identifier
        ].get(
            file_a,
            {}
        )

        evidence_b = file_evidence[
            identifier
        ].get(
            file_b,
            {}
        )

        waves_a = evidence_a.get(
            "wave_tokens",
            []
        )

        waves_b = evidence_b.get(
            "wave_tokens",
            []
        )

        intersection_count = int(
            record.get(
                "intersection_count",
                0
            )
        )

        if not waves_a:
            waves_a = [
                "NO_WAVE_TOKEN"
            ]

        if not waves_b:
            waves_b = [
                "NO_WAVE_TOKEN"
            ]

        for wave_a in waves_a:

            for wave_b in waves_b:

                if wave_a == wave_b:

                    pair_type = (
                        f"{wave_a}↔{wave_b}"
                    )

                else:

                    pair_type = (
                        f"{wave_a}↔{wave_b}"
                    )

                summary[
                    pair_type
                ][
                    "pair_count"
                ] += 1

                if intersection_count > 0:

                    summary[
                        pair_type
                    ][
                        "nonzero_overlap_pairs"
                    ] += 1

                else:

                    summary[
                        pair_type
                    ][
                        "zero_overlap_pairs"
                    ] += 1

                summary[
                    pair_type
                ][
                    "maximum_intersection"
                ] = max(
                    summary[
                        pair_type
                    ][
                        "maximum_intersection"
                    ],
                    intersection_count
                )

    wave_pair_summary[
        identifier
    ] = {
        key: dict(value)
        for key, value
        in sorted(
            summary.items()
        )
    }

# ------------------------------------------------------------
# 4. Label-level evidence counts
# ------------------------------------------------------------

label_summary = {}

for identifier in candidate_identifiers:

    counts = defaultdict(int)

    files_by_label = defaultdict(
        set
    )

    for file_path, evidence in (
        file_evidence[
            identifier
        ].items()
    ):

        label = evidence.get(
            "variable_label_exact",
            ""
        )

        counts[label] += 1

        files_by_label[
            label
        ].add(
            file_path
        )

    label_summary[
        identifier
    ] = {

        label: {

            "file_count":
                count,

            "files":
                sorted(
                    files_by_label[
                        label
                    ]
                )
        }

        for label, count
        in sorted(
            counts.items()
        )
    }

# ------------------------------------------------------------
# 5. Print consistency summary
# ------------------------------------------------------------

print("\n" + "-" * 80)
print("IDENTIFIER LABEL + WAVE CONSISTENCY")
print("-" * 80)

for identifier in candidate_identifiers:

    print(
        f"\n[{identifier}]"
    )

    print(
        "Native label distribution:"
    )

    for label, info in (
        label_summary[
            identifier
        ].items()
    ):

        print(
            f"  {label}"
        )

        print(
            f"    files: "
            f"{info['file_count']}"
        )

    print(
        "\nWave-pair overlap:"
    )

    summary = wave_pair_summary[
        identifier
    ]

    for pair_type, stats in (
        summary.items()
    ):

        print(
            f"  {pair_type}: "
            f"{stats['pair_count']} pairs, "
            f"{stats['nonzero_overlap_pairs']} "
            f"nonzero, "
            f"{stats['zero_overlap_pairs']} "
            f"zero"
        )

# ------------------------------------------------------------
# 6. Fail-closed decision
# ------------------------------------------------------------

identifier_approval = {
    identifier: False
    for identifier in candidate_identifiers
}

panel_id_status = "BLOCKED"
panel_linkage_status = "BLOCKED"

# No semantic reconciliation is performed.
# Conflicting labels remain conflicting evidence.
# Cross-wave overlap remains descriptive evidence.
# No identifier is approved.

# ------------------------------------------------------------
# 7. Save artifact
# ------------------------------------------------------------

result = {

    "protocol_version":
        "OIP v1.0.31",

    "governing_protocol":
        "OIP v1.0.30",

    "cell":
        "21_IDENTIFIER_LABEL_WAVE_OVERLAP_CONSISTENCY_AUDIT",

    "inputs": {

        "cell18":
            str(CELL18_PATH),

        "cell20":
            str(CELL20_PATH)
    },

    "candidate_identifiers":
        candidate_identifiers,

    "label_summary":
        label_summary,

    "wave_pair_summary":
        wave_pair_summary,

    "identifier_approval":
        identifier_approval,

    "panel_id_status":
        panel_id_status,

    "panel_linkage_status":
        panel_linkage_status,

    "semantic_evidence_status":
        "NOT_ESTABLISHED",

    "constructs":
        "NOT_APPROVED",

    "mappings":
        "NOT_APPROVED",

    "outcome":
        "NOT_APPROVED",

    "analytical_cohort":
        "NOT_CREATED",

    "scores":
        "NOT_CALCULATED",

    "empirical_result":
        "NOT_CLAIMED",

    "audit_notes": [

        "This cell combines previously extracted native labels, filename wave context and measured cross-file overlap.",

        "The analysis is descriptive and forensic only.",

        "Different native labels are preserved as distinct evidence.",

        "Cross-wave overlap does not establish longitudinal identity by itself.",

        "Same-wave overlap does not establish identifier authority by itself.",

        "No label reconciliation was performed.",

        "No identifier was approved.",

        "No longitudinal merge was performed.",

        "No construct mapping was approved.",

        "No outcome was approved.",

        "No analytical cohort was created.",

        "No score was calculated.",

        "No empirical result was claimed.",

        "No source data were modified."
    ]
}

OUTPUT = (
    BASE /
    "21_IDENTIFIER_LABEL_WAVE_OVERLAP_CONSISTENCY_AUDIT.json"
)

with open(
    OUTPUT,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        result,
        f,
        indent=2,
        ensure_ascii=False
    )

# ------------------------------------------------------------
# 8. Final status
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("CELL 21 STATUS")
print("=" * 80)

print(
    f"CANDIDATE IDENTIFIERS       = "
    f"{len(candidate_identifiers)}"
)

for identifier in candidate_identifiers:

    print(
        f"{identifier} LABEL GROUPS         = "
        f"{len(label_summary[identifier])}"
    )

    print(
        f"{identifier} WAVE-PAIR GROUPS      = "
        f"{len(wave_pair_summary[identifier])}"
    )

print(
    "IDENTIFIER APPROVAL          = NONE"
)

print(
    "PANEL ID STATUS              = BLOCKED"
)

print(
    "PANEL LINKAGE                = BLOCKED"
)

print(
    "SEMANTIC EVIDENCE            = NOT ESTABLISHED"
)

print(
    "CONSTRUCTS                   = NOT APPROVED"
)

print(
    "MAPPINGS                     = NOT APPROVED"
)

print(
    "OUTCOME                      = NOT APPROVED"
)

print(
    "ANALYTICAL COHORT            = NOT CREATED"
)

print(
    "SCORES                       = NOT CALCULATED"
)

print(
    "EMPIRICAL RESULT             = NOT CLAIMED"
)

print("\nIdentifier label, wave and overlap consistency was audited.")
print("No semantic inference was performed.")
print("No identifier was approved.")
print("No longitudinal merge was performed.")
print("No source data were modified.")

print(f"\nOutput: {OUTPUT}")

print("\n" + "=" * 80)
print("CELL 21 COMPLETE")
print("=" * 80)
```

```python
# ============================================================
# OIP v1.0.31 — CELL 24
# LIGHTWEIGHT PDF TEXT STREAM RECOVERY
# FAIL-CLOSED
# ============================================================

import json
import re
import zlib
import hashlib
from pathlib import Path

print("=" * 80)
print("OIP v1.0.31 — CELL 24: LIGHTWEIGHT PDF TEXT STREAM RECOVERY")
print("=" * 80)

BASE = Path("/kaggle/working/oip_v1_0_31_provenance")

INPUT_ROOT = Path(
    "/kaggle/input/datasets/sudharsandas27/oip-31"
)

PDF_PATH = (
    INPUT_ROOT /
    "ddi-documentation-english_microdata-3819.pdf"
)

if not PDF_PATH.exists():
    raise FileNotFoundError(
        f"Documentation PDF not found: {PDF_PATH}"
    )

# ------------------------------------------------------------
# 1. Read complete PDF as bytes
# ------------------------------------------------------------

with open(
    PDF_PATH,
    "rb"
) as f:
    pdf_bytes = f.read()

file_size = len(pdf_bytes)

sha256 = hashlib.sha256(
    pdf_bytes
).hexdigest()

# ------------------------------------------------------------
# 2. Locate PDF stream boundaries
# ------------------------------------------------------------

stream_matches = list(
    re.finditer(
        rb"stream\r?\n",
        pdf_bytes
    )
)

print(
    f"Stream boundaries discovered: "
    f"{len(stream_matches)}"
)

# ------------------------------------------------------------
# 3. Recover compressed streams
# ------------------------------------------------------------

recovered_streams = []

flate_count = 0
decompressed_count = 0
failed_decompression_count = 0

MAX_STREAM_BYTES = 5 * 1024 * 1024

for idx, match in enumerate(
    stream_matches
):

    start = match.end()

    end_match = re.search(
        rb"\r?\nendstream",
        pdf_bytes[start:]
    )

    if end_match is None:
        continue

    end = (
        start
        +
        end_match.start()
    )

    raw_stream = pdf_bytes[
        start:end
    ]

    if len(raw_stream) > MAX_STREAM_BYTES:
        continue

    # Check whether the immediately preceding
    # object area contains FlateDecode.

    object_start = pdf_bytes.rfind(
        b"obj",
        0,
        match.start()
    )

    object_header = (
        pdf_bytes[
            max(0, object_start - 1000):
            match.start()
        ]
        if object_start != -1
        else b""
    )

    is_flate = (
        b"/FlateDecode"
        in object_header
    )

    if not is_flate:
        continue

    flate_count += 1

    decompressed = None

    # First attempt: standard zlib
    try:

        decompressed = zlib.decompress(
            raw_stream
        )

    except Exception:

        # Second attempt: raw DEFLATE
        try:

            decompressed = zlib.decompress(
                raw_stream,
                -15
            )

        except Exception:

            failed_decompression_count += 1
            continue

    if decompressed is None:
        continue

    decompressed_count += 1

    # Keep bounded evidence only
    sample = decompressed[
        :2_000_000
    ]

    # --------------------------------------------------------
    # 4. Look for PDF text operators
    # --------------------------------------------------------

    tj_count = len(
        re.findall(
            rb"\bTJ\b",
            sample
        )
    )

    tj_single_count = len(
        re.findall(
            rb"\bTj\b",
            sample
        )
    )

    quote_count = len(
        re.findall(
            rb"(?:^|[\s])['\"]",
            sample
        )
    )

    # --------------------------------------------------------
    # 5. Extract printable text-like fragments
    # --------------------------------------------------------

    ascii_fragments = re.findall(
        rb"[\x20-\x7e]{4,}",
        sample
    )

    decoded_fragments = []

    for fragment in ascii_fragments:

        try:

            text = fragment.decode(
                "latin-1",
                errors="ignore"
            )

        except Exception:

            continue

        text = re.sub(
            r"\s+",
            " ",
            text
        ).strip()

        if len(text) < 4:
            continue

        decoded_fragments.append(
            text
        )

    # deterministic de-duplication
    decoded_fragments = list(
        dict.fromkeys(
            decoded_fragments
        )
    )

    # --------------------------------------------------------
    # 6. Search for exact documentation signals
    # --------------------------------------------------------

    signal_terms = [

        "questionnaire",
        "codebook",
        "variable",
        "household",
        "individual",
        "IHPS",
        "IHS3",
        "case_id",
        "hhid",
        "pid",
        "Unique HH Identifier",
        "Person ID",
        "identifier",
        "serial number",
        "assigned unique"
    ]

    signal_hits = []

    combined_text = "\n".join(
        decoded_fragments
    ).lower()

    for term in signal_terms:

        if term.lower() in combined_text:

            signal_hits.append(
                term
            )

    # --------------------------------------------------------
    # 7. Save bounded stream evidence
    # --------------------------------------------------------

    recovered_streams.append({

        "stream_index":
            idx,

        "raw_stream_bytes":
            len(raw_stream),

        "decompressed_bytes":
            len(decompressed),

        "text_operator_TJ":
            tj_count,

        "text_operator_Tj":
            tj_single_count,

        "text_fragment_count":
            len(decoded_fragments),

        "signal_hits":
            sorted(
                set(signal_hits)
            ),

        "text_fragments":
            decoded_fragments[
                :200
            ]
    })

# ------------------------------------------------------------
# 8. Aggregate recovery result
# ------------------------------------------------------------

streams_with_text = sum(
    1
    for record in recovered_streams
    if record["text_fragment_count"] > 0
)

streams_with_signal_hits = sum(
    1
    for record in recovered_streams
    if record["signal_hits"]
)

all_fragments = []

for record in recovered_streams:

    all_fragments.extend(
        record["text_fragments"]
    )

all_fragments = list(
    dict.fromkeys(
        all_fragments
    )
)

# ------------------------------------------------------------
# 9. Fail-closed semantic state
# ------------------------------------------------------------

semantic_evidence_status = (
    "RECOVERY_ATTEMPTED_NOT_APPROVED"
)

identifier_approval = {

    "case_id": False,
    "hhid": False,
    "pid": False
}

panel_id_status = "BLOCKED"
panel_linkage_status = "BLOCKED"

# ------------------------------------------------------------
# 10. Save artifact
# ------------------------------------------------------------

result = {

    "protocol_version":
        "OIP v1.0.31",

    "governing_protocol":
        "OIP v1.0.30",

    "cell":
        "24_LIGHTWEIGHT_PDF_TEXT_STREAM_RECOVERY",

    "pdf_path":
        str(PDF_PATH),

    "file_size_bytes":
        file_size,

    "sha256":
        sha256,

    "stream_boundaries":
        len(stream_matches),

    "flate_streams_detected":
        flate_count,

    "streams_decompressed":
        decompressed_count,

    "failed_decompression":
        failed_decompression_count,

    "streams_with_text_fragments":
        streams_with_text,

    "streams_with_documentation_signals":
        streams_with_signal_hits,

    "recovered_stream_records":
        recovered_streams,

    "unique_recovered_fragments":
        all_fragments[
            :5000
        ],

    "semantic_evidence_status":
        semantic_evidence_status,

    "identifier_approval":
        identifier_approval,

    "panel_id_status":
        panel_id_status,

    "panel_linkage_status":
        panel_linkage_status,

    "constructs":
        "NOT_APPROVED",

    "mappings":
        "NOT_APPROVED",

    "outcome":
        "NOT_APPROVED",

    "analytical_cohort":
        "NOT_CREATED",

    "scores":
        "NOT_CALCULATED",

    "empirical_result":
        "NOT_CLAIMED",

    "audit_notes": [

        "This cell attempts lightweight recovery of Flate-compressed PDF streams.",

        "Recovered fragments are treated as forensic evidence only.",

        "Text fragments may be incomplete, encoded, compressed, or structurally reordered.",

        "Presence of keywords does not establish semantic authority.",

        "No identifier was approved.",

        "No construct mapping was approved.",

        "No outcome was approved.",

        "No longitudinal merge was performed.",

        "No analytical cohort was created.",

        "No score was calculated.",

        "No empirical result was claimed.",

        "No source data were modified."
    ]
}

OUTPUT = (
    BASE /
    "24_LIGHTWEIGHT_PDF_TEXT_STREAM_RECOVERY.json"
)

with open(
    OUTPUT,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        result,
        f,
        indent=2,
        ensure_ascii=False
    )

# ------------------------------------------------------------
# 11. Final status
# ------------------------------------------------------------

print("\n" + "-" * 80)
print("PDF TEXT RECOVERY SUMMARY")
print("-" * 80)

print(
    f"PDF size                         = "
    f"{file_size:,} bytes"
)

print(
    f"SHA-256                          = "
    f"{sha256}"
)

print(
    f"Stream boundaries                = "
    f"{len(stream_matches)}"
)

print(
    f"Flate streams detected           = "
    f"{flate_count}"
)

print(
    f"Streams decompressed             = "
    f"{decompressed_count}"
)

print(
    f"Failed decompression             = "
    f"{failed_decompression_count}"
)

print(
    f"Streams with text fragments      = "
    f"{streams_with_text}"
)

print(
    f"Streams with documentation hits  = "
    f"{streams_with_signal_hits}"
)

print(
    f"Unique recovered fragments       = "
    f"{len(all_fragments)}"
)

print("\n" + "=" * 80)
print("CELL 24 STATUS")
print("=" * 80)

print(
    "SEMANTIC EVIDENCE            = "
    "RECOVERY ATTEMPTED NOT APPROVED"
)

print(
    "IDENTIFIER APPROVAL          = NONE"
)

print(
    "PANEL ID STATUS              = BLOCKED"
)

print(
    "PANEL LINKAGE                = BLOCKED"
)

print(
    "CONSTRUCTS                   = NOT APPROVED"
)

print(
    "MAPPINGS                     = NOT APPROVED"
)

print(
    "OUTCOME                      = NOT APPROVED"
)

print(
    "ANALYTICAL COHORT            = NOT CREATED"
)

print(
    "SCORES                       = NOT CALCULATED"
)

print(
    "EMPIRICAL RESULT             = NOT CLAIMED"
)

print("\nLightweight PDF text-stream recovery completed.")
print("No semantic inference was performed.")
print("No identifier was approved.")
print("No source data were modified.")

print(f"\nOutput: {OUTPUT}")

print("\n" + "=" * 80)
print("CELL 24 COMPLETE")
print("=" * 80)
```

```python
# ============================================================
# OIP v1.0.31 — CELL 28
# TOUNICODE-AWARE PDF TEXT RECOVERY
# HARDENED UNICODE SERIALIZATION
# FAIL-CLOSED
# ============================================================

import json
import re
import zlib
from pathlib import Path

print("=" * 80)
print("OIP v1.0.31 — CELL 28: TOUNICODE-AWARE PDF TEXT RECOVERY")
print("=" * 80)

BASE = Path(
    "/kaggle/working/oip_v1_0_31_provenance"
)

INPUT_ROOT = Path(
    "/kaggle/input/datasets/sudharsandas27/oip-31"
)

PDF_PATH = (
    INPUT_ROOT /
    "ddi-documentation-english_microdata-3819.pdf"
)

if not PDF_PATH.exists():
    raise FileNotFoundError(
        f"Documentation PDF not found: {PDF_PATH}"
    )

# ------------------------------------------------------------
# 1. Safe Unicode sanitizer
# ------------------------------------------------------------

def sanitize_text(value):

    if not isinstance(value, str):
        return value

    # Remove invalid UTF-16 surrogate code points
    return "".join(
        ch
        for ch in value
        if not (
            0xD800
            <= ord(ch)
            <= 0xDFFF
        )
    )

# ------------------------------------------------------------
# 2. Recursive JSON sanitizer
# ------------------------------------------------------------

def sanitize_for_json(value):

    if isinstance(value, str):

        return sanitize_text(
            value
        )

    if isinstance(value, dict):

        return {
            sanitize_text(str(k)):
                sanitize_for_json(v)

            for k, v in value.items()
        }

    if isinstance(value, list):

        return [
            sanitize_for_json(v)
            for v in value
        ]

    if isinstance(value, tuple):

        return [
            sanitize_for_json(v)
            for v in value
        ]

    return value

# ------------------------------------------------------------
# 3. Read PDF
# ------------------------------------------------------------

with open(
    PDF_PATH,
    "rb"
) as f:

    pdf_bytes = f.read()

print(
    f"\nPDF size: "
    f"{len(pdf_bytes):,} bytes"
)

# ------------------------------------------------------------
# 4. Locate PDF objects
# ------------------------------------------------------------

object_pattern = re.compile(
    rb"(\d+)\s+(\d+)\s+obj\b(.*?)endobj",
    re.DOTALL
)

objects = {}

for match in object_pattern.finditer(
    pdf_bytes
):

    obj_num = int(
        match.group(1)
    )

    generation = int(
        match.group(2)
    )

    objects[obj_num] = {

        "generation":
            generation,

        "content":
            match.group(3)
    }

print(
    f"Objects recovered: "
    f"{len(objects)}"
)

# ------------------------------------------------------------
# 5. Stream extraction
# ------------------------------------------------------------

def get_stream(
    content
):

    match = re.search(
        rb"stream\r?\n(.*?)\r?\nendstream",
        content,
        re.DOTALL
    )

    if not match:
        return None

    return match.group(1)

# ------------------------------------------------------------
# 6. Stream decompression
# ------------------------------------------------------------

def decompress_stream(
    content
):

    stream = get_stream(
        content
    )

    if stream is None:
        return None

    if b"/FlateDecode" not in content:

        return stream

    try:

        return zlib.decompress(
            stream
        )

    except Exception:

        try:

            return zlib.decompress(
                stream,
                -15
            )

        except Exception:

            return None

# ------------------------------------------------------------
# 7. ToUnicode CMap parser
# ------------------------------------------------------------

def decode_utf16be_safe(
    raw_bytes
):

    try:

        text = raw_bytes.decode(
            "utf-16-be",
            errors="surrogatepass"
        )

    except Exception:

        return ""

    return sanitize_text(
        text
    )

def parse_cmap(
    cmap_bytes
):

    if cmap_bytes is None:

        return {}

    try:

        text = cmap_bytes.decode(
            "latin-1",
            errors="ignore"
        )

    except Exception:

        return {}

    mapping = {}

    # --------------------------------------------------------
    # bfchar
    # --------------------------------------------------------

    bfchar_blocks = re.findall(
        r"beginbfchar(.*?)endbfchar",
        text,
        re.DOTALL
    )

    for block in bfchar_blocks:

        pairs = re.findall(
            r"<([0-9A-Fa-f]+)>\s+<([0-9A-Fa-f]+)>",
            block
        )

        for src_hex, dst_hex in pairs:

            try:

                src = int(
                    src_hex,
                    16
                )

                dst = decode_utf16be_safe(
                    bytes.fromhex(
                        dst_hex
                    )
                )

                if dst:

                    mapping[src] = dst

            except Exception:

                continue

    # --------------------------------------------------------
    # bfrange
    # --------------------------------------------------------

    bfrange_blocks = re.findall(
        r"beginbfrange(.*?)endbfrange",
        text,
        re.DOTALL
    )

    for block in bfrange_blocks:

        lines = block.splitlines()

        for line in lines:

            line = line.strip()

            if not line:
                continue

            # <start> <end> <destination>

            m = re.match(
                r"<([0-9A-Fa-f]+)>\s+"
                r"<([0-9A-Fa-f]+)>\s+"
                r"<([0-9A-Fa-f]+)>",
                line
            )

            if m:

                start = int(
                    m.group(1),
                    16
                )

                end = int(
                    m.group(2),
                    16
                )

                dst_start = int(
                    m.group(3),
                    16
                )

                for offset, src in enumerate(
                    range(
                        start,
                        end + 1
                    )
                ):

                    dst_code = (
                        dst_start
                        +
                        offset
                    )

                    try:

                        mapping[src] = (
                            sanitize_text(
                                chr(
                                    dst_code
                                )
                            )
                        )

                    except Exception:

                        continue

                continue

            # <start> <end> [<dst1> <dst2> ...]

            m = re.match(
                r"<([0-9A-Fa-f]+)>\s+"
                r"<([0-9A-Fa-f]+)>\s+"
                r"\[(.*?)\]",
                line
            )

            if m:

                start = int(
                    m.group(1),
                    16
                )

                end = int(
                    m.group(2),
                    16
                )

                destinations = re.findall(
                    r"<([0-9A-Fa-f]+)>",
                    m.group(3)
                )

                for offset, dst_hex in enumerate(
                    destinations
                ):

                    src = (
                        start
                        +
                        offset
                    )

                    if src > end:
                        break

                    try:

                        dst = decode_utf16be_safe(
                            bytes.fromhex(
                                dst_hex
                            )
                        )

                        if dst:

                            mapping[src] = dst

                    except Exception:

                        continue

    return mapping

# ------------------------------------------------------------
# 8. Locate ToUnicode objects
# ------------------------------------------------------------

tounicode_refs = []

for obj_num, record in objects.items():

    content = record[
        "content"
    ]

    refs = re.findall(
        rb"/ToUnicode\s+(\d+)\s+\d+\s+R",
        content
    )

    for ref in refs:

        tounicode_refs.append(
            int(ref)
        )

tounicode_refs = sorted(
    set(
        tounicode_refs
    )
)

print(
    f"ToUnicode object references: "
    f"{tounicode_refs}"
)

# ------------------------------------------------------------
# 9. Recover CMaps
# ------------------------------------------------------------

cmap_records = []

for obj_num in tounicode_refs:

    if obj_num not in objects:
        continue

    raw = decompress_stream(
        objects[
            obj_num
        ][
            "content"
        ]
    )

    mapping = parse_cmap(
        raw
    )

    cmap_records.append({

        "object":
            obj_num,

        "raw_bytes":
            len(raw)
            if raw is not None
            else 0,

        "mapping_entries":
            len(mapping),

        "mapping":
            {
                str(k):
                    sanitize_text(v)

                for k, v in mapping.items()
            }
    })

print("\n" + "-" * 80)
print("TOUNICODE MAP SUMMARY")
print("-" * 80)

for record in cmap_records:

    print(
        f"Object {record['object']}: "
        f"{record['mapping_entries']} "
        f"mapping entries"
    )

# ------------------------------------------------------------
# 10. Identify text content streams
# ------------------------------------------------------------

content_objects = []

for obj_num, record in objects.items():

    content = record[
        "content"
    ]

    if (
        b"/Length" in content
        and
        b"stream" in content
    ):

        decoded = decompress_stream(
            content
        )

        if decoded is None:
            continue

        if (
            b"BT" in decoded
            or
            b"Tj" in decoded
            or
            b"TJ" in decoded
        ):

            content_objects.append({

                "object":
                    obj_num,

                "decoded":
                    decoded
            })

print(
    f"\nText-content objects identified: "
    f"{len(content_objects)}"
)

# ------------------------------------------------------------
# 11. Combined mapping
# ------------------------------------------------------------

combined_mapping = {}

for record in cmap_records:

    for key, value in record[
        "mapping"
    ].items():

        combined_mapping[
            int(key)
        ] = sanitize_text(
            value
        )

print(
    f"Combined Unicode mapping entries: "
    f"{len(combined_mapping)}"
)

# ------------------------------------------------------------
# 12. PDF text string patterns
# ------------------------------------------------------------

hex_string_pattern = re.compile(
    rb"<([0-9A-Fa-f\s]+)>"
)

literal_string_pattern = re.compile(
    rb"\((?:\\.|[^\\)])*\)"
)

decoded_text_records = []

invalid_surrogate_count = 0

# ------------------------------------------------------------
# 13. Decode content streams
# ------------------------------------------------------------

for record in content_objects:

    obj_num = record[
        "object"
    ]

    content = record[
        "decoded"
    ]

    # --------------------------------------------------------
    # Hex strings
    # --------------------------------------------------------

    for match in hex_string_pattern.finditer(
        content
    ):

        raw_hex = re.sub(
            rb"\s+",
            b"",
            match.group(1)
        )

        if not raw_hex:
            continue

        try:

            raw_bytes = bytes.fromhex(
                raw_hex.decode(
                    "ascii"
                )
            )

        except Exception:

            continue

        chars = []

        # Identity-H normally uses 2-byte CIDs
        for i in range(
            0,
            len(raw_bytes) - 1,
            2
        ):

            cid = int.from_bytes(
                raw_bytes[
                    i:i + 2
                ],
                "big"
            )

            if cid in combined_mapping:

                value = combined_mapping[
                    cid
                ]

                if any(
                    0xD800
                    <= ord(ch)
                    <= 0xDFFF
                    for ch in value
                ):

                    invalid_surrogate_count += 1

                chars.append(
                    sanitize_text(
                        value
                    )
                )

        text = sanitize_text(
            "".join(chars)
        ).strip()

        if text:

            decoded_text_records.append({

                "object":
                    obj_num,

                "encoding":
                    "hex",

                "text":
                    text
            })

    # --------------------------------------------------------
    # Literal strings
    # --------------------------------------------------------

    for match in literal_string_pattern.finditer(
        content
    ):

        raw = match.group(0)

        inner = raw[
            1:-1
        ]

        inner = re.sub(
            rb"\\([\\()])",
            rb"\1",
            inner
        )

        chars = []

        for byte in inner:

            if byte in combined_mapping:

                value = combined_mapping[
                    byte
                ]

                if any(
                    0xD800
                    <= ord(ch)
                    <= 0xDFFF
                    for ch in value
                ):

                    invalid_surrogate_count += 1

                chars.append(
                    sanitize_text(
                        value
                    )
                )

            elif (
                32
                <= byte
                <= 126
            ):

                chars.append(
                    chr(byte)
                )

        text = sanitize_text(
            "".join(chars)
        ).strip()

        if text:

            decoded_text_records.append({

                "object":
                    obj_num,

                "encoding":
                    "literal",

                "text":
                    text
            })

# ------------------------------------------------------------
# 14. Deterministic de-duplication
# ------------------------------------------------------------

seen = set()
unique_decoded = []

for record in decoded_text_records:

    text = sanitize_text(
        record["text"]
    )

    key = (
        record["object"],
        record["encoding"],
        text
    )

    if key in seen:
        continue

    seen.add(key)

    unique_decoded.append({

        "object":
            record["object"],

        "encoding":
            record["encoding"],

        "text":
            text
    })

decoded_text_records = (
    unique_decoded
)

print(
    f"\nDecoded text records: "
    f"{len(decoded_text_records)}"
)

print(
    f"Invalid surrogate observations removed: "
    f"{invalid_surrogate_count}"
)

# ------------------------------------------------------------
# 15. Search recovered Unicode text
# ------------------------------------------------------------

search_terms = [

    "identifier",
    "household",
    "hhid",
    "household id",
    "household identifier",
    "unique household",
    "unique hh",
    "person id",
    "personid",
    "pid",
    "individual id",
    "individual",
    "ihps",
    "ihs3",
    "baseline",
    "serial number",
    "case id",
    "case_id",
    "questionnaire",
    "codebook",
    "variable",
    "data dictionary",
    "survey",
    "panel",
    "longitudinal",
    "tracking",
    "2010",
    "2013",
    "2016",
    "2019"
]

evidence_hits = []

for record in decoded_text_records:

    text = sanitize_text(
        record["text"]
    )

    lower = text.lower()

    matched = [

        term

        for term in search_terms

        if term.lower()
        in lower
    ]

    if matched:

        evidence_hits.append({

            "object":
                record["object"],

            "encoding":
                record["encoding"],

            "text":
                text,

            "matched_terms":
                sorted(
                    set(matched)
                )
        })

# ------------------------------------------------------------
# 16. Print recovered evidence
# ------------------------------------------------------------

print("\n" + "-" * 80)
print("DECODED DOCUMENTATION EVIDENCE")
print("-" * 80)

print(
    f"Evidence hits: "
    f"{len(evidence_hits)}"
)

for item in evidence_hits[
    :100
]:

    print(
        f"\nObject {item['object']} "
        f"({item['encoding']}):"
    )

    print(
        f"  {item['text']}"
    )

    print(
        f"  matched: "
        f"{item['matched_terms']}"
    )

# ------------------------------------------------------------
# 17. Fail-closed state
# ------------------------------------------------------------

semantic_evidence_status = (
    "TOUNICODE_TEXT_RECOVERED_NOT_APPROVED"
)

identifier_approval = {

    "case_id": False,
    "hhid": False,
    "pid": False
}

panel_id_status = "BLOCKED"
panel_linkage_status = "BLOCKED"

# ------------------------------------------------------------
# 18. Build result
# ------------------------------------------------------------

result = {

    "protocol_version":
        "OIP v1.0.31",

    "governing_protocol":
        "OIP v1.0.30",

    "cell":
        "28_TOUNICODE_AWARE_PDF_TEXT_RECOVERY",

    "pdf_path":
        str(PDF_PATH),

    "file_size_bytes":
        len(pdf_bytes),

    "object_count":
        len(objects),

    "tounicode_references":
        tounicode_refs,

    "cmap_records":
        cmap_records,

    "content_object_count":
        len(content_objects),

    "combined_mapping_entries":
        len(combined_mapping),

    "decoded_text_record_count":
        len(decoded_text_records),

    "decoded_text_records":
        decoded_text_records[
            :10000
        ],

    "invalid_surrogate_observations_removed":
        invalid_surrogate_count,

    "evidence_hit_count":
        len(evidence_hits),

    "evidence_hits":
        evidence_hits[
            :5000
        ],

    "semantic_evidence_status":
        semantic_evidence_status,

    "identifier_approval":
        identifier_approval,

    "panel_id_status":
        panel_id_status,

    "panel_linkage_status":
        panel_linkage_status,

    "constructs":
        "NOT_APPROVED",

    "mappings":
        "NOT_APPROVED",

    "outcome":
        "NOT_APPROVED",

    "analytical_cohort":
        "NOT_CREATED",

    "scores":
        "NOT_CALCULATED",

    "empirical_result":
        "NOT_CLAIMED",

    "audit_notes": [

        "This cell attempts ToUnicode-aware recovery of PDF text.",

        "Invalid Unicode surrogate characters are removed only for safe serialization.",

        "Recovered Unicode text is forensic evidence only.",

        "Recovered text is not automatically treated as authoritative documentation.",

        "Keyword hits are discovery signals and require contextual verification.",

        "No identifier was approved.",

        "No construct mapping was approved.",

        "No outcome was approved.",

        "No longitudinal merge was performed.",

        "No analytical cohort was created.",

        "No score was calculated.",

        "No empirical result was claimed.",

        "No source data were modified."
    ]
}

result = sanitize_for_json(
    result
)

# ------------------------------------------------------------
# 19. Save JSON safely
# ------------------------------------------------------------

OUTPUT = (
    BASE /
    "28_TOUNICODE_AWARE_PDF_TEXT_RECOVERY.json"
)

with open(
    OUTPUT,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        result,
        f,
        indent=2,
        ensure_ascii=True,
        allow_nan=False
    )

# ------------------------------------------------------------
# 20. Final status
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("CELL 28 STATUS")
print("=" * 80)

print(
    f"OBJECTS                      = "
    f"{len(objects)}"
)

print(
    f"TOUNICODE REFERENCES         = "
    f"{len(tounicode_refs)}"
)

print(
    f"CMAP RECORDS                 = "
    f"{len(cmap_records)}"
)

print(
    f"UNICODE MAPPING ENTRIES      = "
    f"{len(combined_mapping)}"
)

print(
    f"TEXT CONTENT OBJECTS         = "
    f"{len(content_objects)}"
)

print(
    f"DECODED TEXT RECORDS         = "
    f"{len(decoded_text_records)}"
)

print(
    f"INVALID SURROGATES REMOVED   = "
    f"{invalid_surrogate_count}"
)

print(
    f"DOCUMENTATION EVIDENCE HITS  = "
    f"{len(evidence_hits)}"
)

print(
    "\nSEMANTIC EVIDENCE            = "
    "TOUNICODE TEXT RECOVERED NOT APPROVED"
)

print(
    "IDENTIFIER APPROVAL          = NONE"
)

print(
    "PANEL ID STATUS              = BLOCKED"
)

print(
    "PANEL LINKAGE                = BLOCKED"
)

print(
    "CONSTRUCTS                   = NOT APPROVED"
)

print(
    "MAPPINGS                     = NOT APPROVED"
)

print(
    "OUTCOME                      = NOT APPROVED"
)

print(
    "ANALYTICAL COHORT            = NOT CREATED"
)

print(
    "SCORES                       = NOT CALCULATED"
)

print(
    "EMPIRICAL RESULT             = NOT CLAIMED"
)

print(
    "\nToUnicode-aware text recovery completed."
)

print(
    "Unicode serialization was hardened."
)

print(
    "No semantic inference was performed."
)

print(
    "No identifier was approved."
)

print(
    "No source data were modified."
)

print(
    f"\nOutput: {OUTPUT}"
)

print("\n" + "=" * 80)
print("CELL 28 COMPLETE")
print("=" * 80)
```

```python
# ============================================================
# OIP v1.0.31 — CELL 32A
# CELL 31 JSON SCHEMA INSPECTION
# READ-ONLY / FAIL-CLOSED
# ============================================================

import json
from pathlib import Path

print("=" * 80)
print("OIP v1.0.31 — CELL 32A: CELL 31 JSON SCHEMA INSPECTION")
print("=" * 80)

BASE = Path(
    "/kaggle/working/oip_v1_0_31_provenance"
)

INPUT = (
    BASE /
    "31_DOCUMENTARY_CONSTRUCT_EVIDENCE_GATE.json"
)

if not INPUT.exists():
    raise FileNotFoundError(
        f"Cell 31 output not found: {INPUT}"
    )

with open(
    INPUT,
    "r",
    encoding="utf-8"
) as f:
    data = json.load(f)

print("\nCell 31 JSON loaded successfully.")

print("\n" + "-" * 80)
print("TOP-LEVEL KEYS")
print("-" * 80)

for key in data.keys():
    print(
        f"{key!r} : {type(data[key]).__name__}"
    )

# ------------------------------------------------------------
# Recursive schema inspection
# ------------------------------------------------------------

def inspect_schema(
    obj,
    path="ROOT",
    depth=0,
    max_depth=5
):

    if depth > max_depth:
        return

    if isinstance(obj, dict):

        for key, value in obj.items():

            current_path = (
                f"{path}.{key}"
            )

            print(
                f"{current_path} "
                f"-> {type(value).__name__}"
            )

            inspect_schema(
                value,
                current_path,
                depth + 1,
                max_depth
            )

    elif isinstance(obj, list):

        print(
            f"{path} "
            f"-> LIST(length={len(obj)})"
        )

        if len(obj) > 0:

            inspect_schema(
                obj[0],
                f"{path}[0]",
                depth + 1,
                max_depth
            )

print("\n" + "-" * 80)
print("RECURSIVE JSON SCHEMA")
print("-" * 80)

inspect_schema(
    data
)

# ------------------------------------------------------------
# Search for construct-related keys
# ------------------------------------------------------------

print("\n" + "-" * 80)
print("CONSTRUCT-RELATED KEYS")
print("-" * 80)

matches = []

def find_construct_keys(
    obj,
    path="ROOT"
):

    if isinstance(obj, dict):

        for key, value in obj.items():

            key_text = str(key).lower()

            if any(
                token in key_text
                for token in [
                    "gfl",
                    "ids",
                    "aml",
                    "approval",
                    "status",
                    "decision"
                ]
            ):

                matches.append(
                    (
                        f"{path}.{key}",
                        value
                    )
                )

            find_construct_keys(
                value,
                f"{path}.{key}"
            )

    elif isinstance(obj, list):

        for index, value in enumerate(obj):

            find_construct_keys(
                value,
                f"{path}[{index}]"
            )

find_construct_keys(
    data
)

for path, value in matches:

    print(
        f"{path} = {value!r}"
    )

# ------------------------------------------------------------
# Final status
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("CELL 32A STATUS")
print("=" * 80)

print(
    "PURPOSE                  = "
    "READ-ONLY JSON SCHEMA INSPECTION"
)

print(
    "EVIDENCE INTERPRETATION  = NONE"
)

print(
    "CONSTRUCT APPROVAL       = NONE"
)

print(
    "IDENTIFIER APPROVAL      = NONE"
)

print(
    "PANEL LINKAGE            = NONE"
)

print(
    "COHORT CREATION          = NONE"
)

print(
    "SCORE CALCULATION        = NONE"
)

print(
    "EMPIRICAL RESULT         = NOT CLAIMED"
)

print(
    "\nCell 31 JSON schema inspection completed."
)

print(
    "No source data were modified."
)

print("=" * 80)
print("CELL 32A COMPLETE")
print("=" * 80)
```

```python
# ============================================================
# OIP v1.0.31 — CELL 32
# FINAL EVIDENCE AVAILABILITY + FAIL-CLOSED GATE
# ============================================================

import json
from pathlib import Path

print("=" * 80)
print("OIP v1.0.31 — CELL 32: FINAL EVIDENCE AVAILABILITY GATE")
print("=" * 80)

BASE = Path(
    "/kaggle/working/oip_v1_0_31_provenance"
)

INPUT = (
    BASE /
    "31_DOCUMENTARY_CONSTRUCT_EVIDENCE_GATE.json"
)

OUTPUT = (
    BASE /
    "32_FINAL_EVIDENCE_AVAILABILITY_GATE.json"
)

# ------------------------------------------------------------
# 1. Load Cell 31
# ------------------------------------------------------------

if not INPUT.exists():
    raise FileNotFoundError(
        f"Required Cell 31 output not found: {INPUT}"
    )

with open(
    INPUT,
    "r",
    encoding="utf-8"
) as f:

    gate = json.load(f)

print("\nCell 31 evidence gate loaded successfully.")

# ------------------------------------------------------------
# 2. Read exact Cell 31 construct approval schema
# ------------------------------------------------------------

construct_approval = gate.get(
    "construct_approval"
)

if not isinstance(
    construct_approval,
    dict
):

    raise ValueError(
        "Cell 31 construct_approval structure "
        "is missing or invalid."
    )

required_constructs = [
    "GFL",
    "IDS",
    "AML"
]

construct_states = {}

for construct in required_constructs:

    record = construct_approval.get(
        construct
    )

    if not isinstance(
        record,
        dict
    ):

        raise ValueError(
            f"Missing construct record: {construct}"
        )

    if "approved" not in record:
        raise ValueError(
            f"Missing approval field: {construct}"
        )

    if "status" not in record:
        raise ValueError(
            f"Missing status field: {construct}"
        )

    approved = record["approved"]
    status = record["status"]

    # --------------------------------------------------------
    # Fail-closed type enforcement
    # --------------------------------------------------------

    if not isinstance(
        approved,
        bool
    ):

        raise ValueError(
            f"Approval field for {construct} "
            f"is not boolean."
        )

    construct_states[
        construct
    ] = {

        "approved":
            approved,

        "status":
            str(status)
    }

# ------------------------------------------------------------
# 3. Extract explicit states
# ------------------------------------------------------------

gfl_approved = (
    construct_states["GFL"]["approved"]
)

gfl_status = (
    construct_states["GFL"]["status"]
)

ids_approved = (
    construct_states["IDS"]["approved"]
)

ids_status = (
    construct_states["IDS"]["status"]
)

aml_approved = (
    construct_states["AML"]["approved"]
)

aml_status = (
    construct_states["AML"]["status"]
)

overall_status = str(
    gate.get(
        "overall_status",
        "UNKNOWN"
    )
)

# ------------------------------------------------------------
# 4. Fail-closed construct gate
# ------------------------------------------------------------

all_constructs_approved = (
    gfl_approved
    and
    ids_approved
    and
    aml_approved
)

source_gate_blocked = (
    overall_status
    in {
        "CONSTRUCT_EVIDENCE_INSUFFICIENT",
        "BLOCKED",
        "PENDING_EVIDENCE",
        "UNKNOWN"
    }
)

if (
    all_constructs_approved
    and
    not source_gate_blocked
):

    final_decision = (
        "EVIDENCE_AVAILABLE_FOR_NEXT_GATE"
    )

    execution_authorization = (
        "CONSTRUCT_GATE_CLEARED_ONLY"
    )

else:

    final_decision = "BLOCKED"

    execution_authorization = (
        "NOT_AUTHORIZED"
    )

# ------------------------------------------------------------
# 5. Downstream execution lock
# ------------------------------------------------------------

if final_decision == "BLOCKED":

    identifier_approval = "NONE"

    panel_id_status = "BLOCKED"

    panel_linkage_status = "BLOCKED"

    outcome_approval = "NONE"

    analytical_cohort = "NOT_CREATED"

    scores = "NOT_PERFORMED"

    empirical_result = "NOT_CLAIMED"

else:

    #
    # Construct gate clearance does not authorize
    # identifiers, linkage, outcome, cohort, or scores.
    #

    identifier_approval = "NONE"

    panel_id_status = "REQUIRES_SEPARATE_GATE"

    panel_linkage_status = "REQUIRES_SEPARATE_GATE"

    outcome_approval = "NONE"

    analytical_cohort = "NOT_CREATED"

    scores = "NOT_PERFORMED"

    empirical_result = "NOT_CLAIMED"

# ------------------------------------------------------------
# 6. Final immutable audit record
# ------------------------------------------------------------

result = {

    "protocol_version":
        "OIP v1.0.31",

    "governing_protocol":
        "OIP v1.0.30",

    "cell":
        "32_FINAL_EVIDENCE_AVAILABILITY_GATE",

    "source_cell":
        "31_DOCUMENTARY_CONSTRUCT_EVIDENCE_GATE",

    "source_overall_status":
        overall_status,

    "construct_approval":
        construct_states,

    "all_constructs_approved":
        all_constructs_approved,

    "final_decision":
        final_decision,

    "execution_authorization":
        execution_authorization,

    "identifier_approval":
        identifier_approval,

    "panel_id_status":
        panel_id_status,

    "panel_linkage_status":
        panel_linkage_status,

    "outcome_approval":
        outcome_approval,

    "analytical_cohort":
        analytical_cohort,

    "scores":
        scores,

    "empirical_result":
        empirical_result,

    "claim_discipline": {

        "empirical_validation":
            "NOT_ESTABLISHED",

        "independent_replication":
            "NOT_ESTABLISHED",

        "hardware_validation":
            "NOT_ESTABLISHED",

        "production_validation":
            "NOT_ESTABLISHED",

        "economic_validation":
            "NOT_ESTABLISHED"
    },

    "fail_closed_rules": [

        "Construct approval requires explicit boolean approval.",

        "Pending evidence cannot be treated as approval.",

        "Unknown status cannot authorize progression.",

        "Insufficient construct evidence blocks progression.",

        "No identifier approval was performed.",

        "No panel linkage was performed.",

        "No outcome was approved.",

        "No analytical cohort was created.",

        "No score was calculated.",

        "No empirical result was claimed.",

        "No source data were modified.",

        "No post-hoc construct approval was performed."
    ]
}

# ------------------------------------------------------------
# 7. Safe serialization
# ------------------------------------------------------------

with open(
    OUTPUT,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        result,
        f,
        indent=2,
        ensure_ascii=True,
        allow_nan=False
    )

# ------------------------------------------------------------
# 8. Final status
# ------------------------------------------------------------

print("\n" + "-" * 80)
print("EXPLICIT CONSTRUCT EVIDENCE")
print("-" * 80)

print(
    f"GFL APPROVED = {gfl_approved}"
)

print(
    f"GFL STATUS   = {gfl_status}"
)

print(
    f"IDS APPROVED = {ids_approved}"
)

print(
    f"IDS STATUS   = {ids_status}"
)

print(
    f"AML APPROVED = {aml_approved}"
)

print(
    f"AML STATUS   = {aml_status}"
)

print("\n" + "=" * 80)
print("CELL 32 STATUS")
print("=" * 80)

print(
    f"SOURCE GATE STATUS       = "
    f"{overall_status}"
)

print(
    f"ALL CONSTRUCTS APPROVED  = "
    f"{all_constructs_approved}"
)

print(
    f"FINAL DECISION           = "
    f"{final_decision}"
)

print(
    f"EXECUTION AUTHORIZATION  = "
    f"{execution_authorization}"
)

print(
    f"IDENTIFIER APPROVAL      = "
    f"{identifier_approval}"
)

print(
    f"PANEL ID STATUS          = "
    f"{panel_id_status}"
)

print(
    f"PANEL LINKAGE            = "
    f"{panel_linkage_status}"
)

print(
    f"OUTCOME                  = "
    f"{outcome_approval}"
)

print(
    f"ANALYTICAL COHORT        = "
    f"{analytical_cohort}"
)

print(
    f"SCORES                   = "
    f"{scores}"
)

print(
    f"EMPIRICAL RESULT         = "
    f"{empirical_result}"
)

print(
    "\nEMPIRICAL VALIDATION     = "
    "NOT ESTABLISHED"
)

print(
    "INDEPENDENT REPLICATION  = "
    "NOT ESTABLISHED"
)

print(
    "HARDWARE VALIDATION      = "
    "NOT ESTABLISHED"
)

print(
    "\nNo construct mapping was created."
)

print(
    "No identifier was approved."
)

print(
    "No source data were modified."
)

print(
    f"\nOutput: {OUTPUT}"
)

print("\n" + "=" * 80)
print("CELL 32 COMPLETE")
print("=" * 80)
```
