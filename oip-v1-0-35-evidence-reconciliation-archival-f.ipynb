```python
# ============================================================
# OIP v1.0.35 — C00
# SOURCE SCOPE & ENVIRONMENT LOCK
# ============================================================
#
# PURPOSE
# -------
# Establish the execution environment and identify the source
# scope actually accessible to this notebook.
#
# C00 IS RECONCILIATION-ONLY.
#
# C00 MUST NOT:
#   - modify source files
#   - invent or infer variables
#   - construct OIP variables
#   - calculate an OIP score
#   - construct an outcome
#   - perform empirical evaluation
#   - resolve source conflicts
#   - assign substantive source authority
#
# Governing sequence:
# Evidence → Measurement → Score → Evaluation → Claim
#
# C00 establishes SCOPE ONLY.
# Complete artifact inventory and SHA-256 verification are
# performed in C01.
# ============================================================

from pathlib import Path
from datetime import datetime, timezone
import json
import platform
import sys


# ------------------------------------------------------------
# 1. FIXED PATHS
# ------------------------------------------------------------

KAGGLE_INPUT = Path("/kaggle/input")
KAGGLE_WORKING = Path("/kaggle/working")

RUN_ROOT = KAGGLE_WORKING / "oip_v1_0_35"
C00_DIR = RUN_ROOT / "c00_source_scope"

C00_JSON = C00_DIR / "c00_source_scope.json"
C00_TXT = C00_DIR / "c00_source_scope.txt"

PROTOCOL_VERSION = "OIP v1.0.35"
CELL_ID = "C00"
CELL_PURPOSE = "SOURCE_SCOPE_AND_ENVIRONMENT_LOCK"


# ------------------------------------------------------------
# 2. CREATE WORKING OUTPUT DIRECTORY
# ------------------------------------------------------------

C00_DIR.mkdir(parents=True, exist_ok=True)


# ------------------------------------------------------------
# 3. EXECUTION ENVIRONMENT
# ------------------------------------------------------------

timestamp_utc = datetime.now(timezone.utc).replace(
    microsecond=0
).isoformat()

environment = {
    "python_version": sys.version,
    "python_executable": sys.executable,
    "platform": platform.platform(),
    "system": platform.system(),
    "release": platform.release(),
    "machine": platform.machine(),
    "processor": platform.processor(),
    "working_directory": str(Path.cwd()),
    "kaggle_input_exists": KAGGLE_INPUT.exists(),
    "kaggle_working_exists": KAGGLE_WORKING.exists(),
}


# ------------------------------------------------------------
# 4. OBSERVE ATTACHED INPUT ROOTS
# ------------------------------------------------------------
#
# No expected dataset name is hard-coded.
# No expected file count is hard-coded.
#
# We observe what is actually accessible.
# ------------------------------------------------------------

input_entries = []

if KAGGLE_INPUT.exists():

    for p in sorted(
        KAGGLE_INPUT.iterdir(),
        key=lambda x: x.name.lower()
    ):

        input_entries.append({
            "name": p.name,
            "path": str(p),
            "type": (
                "directory"
                if p.is_dir()
                else "file"
            ),
        })


source_roots = [
    item
    for item in input_entries
    if item["type"] == "directory"
]


# ------------------------------------------------------------
# 5. NON-RECURSIVE ROOT OBSERVATION
# ------------------------------------------------------------
#
# C00 does NOT perform the complete artifact inventory.
#
# C01 will recursively enumerate every accessible artifact
# and calculate SHA-256 hashes.
# ------------------------------------------------------------

root_summaries = []

for root_info in source_roots:

    root = Path(root_info["path"])

    try:

        immediate_children = sorted(
            list(root.iterdir()),
            key=lambda x: x.name.lower()
        )

        root_summaries.append({
            "name": root.name,
            "path": str(root),
            "immediate_child_count": len(
                immediate_children
            ),
            "immediate_child_names": [
                x.name
                for x in immediate_children
            ],
        })

    except Exception as exc:

        root_summaries.append({
            "name": root.name,
            "path": str(root),
            "immediate_child_count": None,
            "immediate_child_names": [],
            "read_error": repr(exc),
        })


# ------------------------------------------------------------
# 6. C00 SCOPE STATUS
# ------------------------------------------------------------
#
# PASS means ONLY:
#   - /kaggle/input is accessible
#   - at least one source root is visible
#
# PASS does NOT mean:
#   - provenance verified
#   - completeness verified
#   - authority established
#   - construct authorized
#   - score authorized
#   - outcome estimable
#   - empirical validity established
# ------------------------------------------------------------

if not KAGGLE_INPUT.exists():

    scope_status = "FAIL"

    scope_reason = (
        "KAGGLE_INPUT directory is not accessible."
    )

elif len(source_roots) == 0:

    scope_status = "FAIL"

    scope_reason = (
        "No attached source root was detected "
        "under /kaggle/input."
    )

else:

    scope_status = "PASS"

    scope_reason = (
        "At least one attached source root is accessible. "
        "This establishes execution scope only. It does not "
        "establish artifact identity, provenance, completeness, "
        "authority, construct authorization, outcome estimability, "
        "or empirical validity."
    )


# ------------------------------------------------------------
# 7. EXPLICIT CLAIM BOUNDARY
# ------------------------------------------------------------

claim_boundary = {

    "artifact_identity_verified": False,

    "artifact_provenance_verified": False,

    "artifact_completeness_verified": False,

    "authority_assigned": False,

    "construct_authorization_performed": False,

    "score_computation_performed": False,

    "outcome_construction_performed": False,

    "empirical_evaluation_performed": False,

    "empirical_validity_claim_authorized": False,
}


# ------------------------------------------------------------
# 8. C00 RECORD
# ------------------------------------------------------------

c00_record = {

    "protocol": PROTOCOL_VERSION,

    "cell": CELL_ID,

    "cell_purpose": CELL_PURPOSE,

    "timestamp_utc": timestamp_utc,

    "scope_status": scope_status,

    "scope_reason": scope_reason,

    "environment": environment,

    "kaggle_input_entries": input_entries,

    "attached_source_roots": root_summaries,

    "claim_boundary": claim_boundary,

    "next_cell": (
        "C01 — ARTIFACT INVENTORY & SHA-256"
    ),

    "governing_statement": (
        "C00 establishes source scope and execution "
        "environment only. No source authority, construct "
        "authorization, score, outcome, empirical evaluation, "
        "or empirical validity is established by C00."
    ),
}


# ------------------------------------------------------------
# 9. WRITE MACHINE-READABLE RECORD
# ------------------------------------------------------------

with open(
    C00_JSON,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        c00_record,
        f,
        indent=2,
        ensure_ascii=False
    )


# ------------------------------------------------------------
# 10. WRITE HUMAN-READABLE RECORD
# ------------------------------------------------------------

with open(
    C00_TXT,
    "w",
    encoding="utf-8"
) as f:

    f.write("=" * 72 + "\n")

    f.write(
        "OIP v1.0.35 — "
        "C00 SOURCE SCOPE & ENVIRONMENT LOCK\n"
    )

    f.write("=" * 72 + "\n\n")

    f.write(
        f"Protocol      : {PROTOCOL_VERSION}\n"
    )

    f.write(
        f"Cell          : {CELL_ID}\n"
    )

    f.write(
        f"Timestamp UTC : {timestamp_utc}\n"
    )

    f.write(
        f"Status        : {scope_status}\n\n"
    )

    f.write("SCOPE REASON\n")
    f.write("-" * 72 + "\n")
    f.write(scope_reason + "\n\n")

    f.write("ATTACHED SOURCE ROOTS\n")
    f.write("-" * 72 + "\n")

    if root_summaries:

        for root in root_summaries:

            f.write(
                f"Name     : {root['name']}\n"
            )

            f.write(
                f"Path     : {root['path']}\n"
            )

            f.write(
                f"Children : "
                f"{root.get('immediate_child_count')}\n"
            )

            f.write("\n")

    else:

        f.write("NONE\n\n")

    f.write("CLAIM BOUNDARY\n")
    f.write("-" * 72 + "\n")

    for key, value in claim_boundary.items():

        f.write(
            f"{key}: {value}\n"
        )

    f.write("\n")

    f.write("GOVERNING STATEMENT\n")
    f.write("-" * 72 + "\n")

    f.write(
        c00_record["governing_statement"]
        + "\n"
    )


# ------------------------------------------------------------
# 11. CONSOLE OUTPUT
# ------------------------------------------------------------

print("=" * 72)

print(
    "OIP v1.0.35 — "
    "C00: SOURCE SCOPE & ENVIRONMENT LOCK"
)

print("=" * 72)

print(
    f"\nProtocol : {PROTOCOL_VERSION}"
)

print(
    f"Cell     : {CELL_ID}"
)

print(
    f"Status   : {scope_status}"
)

print("\nAttached source roots:")

if root_summaries:

    for root in root_summaries:

        print(
            f"  - {root['name']} "
            f"({root.get('immediate_child_count')} "
            f"immediate entries)"
        )

else:

    print("  NONE")

print("\nImportant:")

print(
    "  C00 establishes SCOPE ONLY."
)

print(
    "  No construct, score, outcome, evaluation,"
)

print(
    "  or empirical validity is established."
)

print("\nArtifacts written:")

print(
    f"  {C00_JSON}"
)

print(
    f"  {C00_TXT}"
)

print("\nNext:")

print(
    "  C01 — ARTIFACT INVENTORY & SHA-256"
)

print("=" * 72)
```

```python
# ============================================================
# OIP v1.0.35 — C01
# ARTIFACT INVENTORY & SHA-256
# ============================================================
#
# PURPOSE
# -------
# Recursively inventory all accessible source artifacts under
# the source scope established by C00 and calculate SHA-256
# hashes for regular files.
#
# C01 IS INVENTORY-ONLY.
#
# C01 MUST NOT:
#   - modify source files
#   - open or interpret substantive dataset contents
#   - infer variables
#   - authorize constructs
#   - calculate OIP scores
#   - construct outcomes
#   - perform empirical evaluation
#   - resolve source conflicts
#   - assign substantive evidentiary authority
#
# C01 establishes:
#   SOURCE INVENTORY + FILE METADATA + SHA-256
#
# Governing sequence:
# Evidence → Measurement → Score → Evaluation → Claim
#
# IMPORTANT:
# The source tree is observed as it actually exists in the
# current Kaggle execution environment. Previous report counts
# are NOT used as inventory truth.
# ============================================================

from pathlib import Path
from datetime import datetime, timezone
import hashlib
import json
import os
import platform
import sys


# ------------------------------------------------------------
# 1. FIXED PATHS
# ------------------------------------------------------------

KAGGLE_INPUT = Path("/kaggle/input")
KAGGLE_WORKING = Path("/kaggle/working")

RUN_ROOT = KAGGLE_WORKING / "oip_v1_0_35"
C01_DIR = RUN_ROOT / "c01_artifact_inventory"

C01_JSON = C01_DIR / "c01_artifact_inventory.json"
C01_TXT = C01_DIR / "c01_artifact_inventory.txt"
C01_MANIFEST = C01_DIR / "c01_sha256_manifest.txt"

PROTOCOL_VERSION = "OIP v1.0.35"
CELL_ID = "C01"
CELL_PURPOSE = "ARTIFACT_INVENTORY_AND_SHA256"


# ------------------------------------------------------------
# 2. CREATE WORKING OUTPUT DIRECTORY
# ------------------------------------------------------------

C01_DIR.mkdir(parents=True, exist_ok=True)


# ------------------------------------------------------------
# 3. TIMESTAMP
# ------------------------------------------------------------

timestamp_utc = datetime.now(
    timezone.utc
).replace(
    microsecond=0
).isoformat()


# ------------------------------------------------------------
# 4. IDENTIFY SOURCE ROOT
# ------------------------------------------------------------
#
# C00 established /kaggle/input as accessible and observed
# a source root named "datasets".
#
# C01 does not assume that this is the only root.
# It enumerates all directories directly under /kaggle/input.
# ------------------------------------------------------------

if not KAGGLE_INPUT.exists():

    raise RuntimeError(
        "C01 STOPPED: /kaggle/input is not accessible."
    )


source_roots = sorted(
    [
        p
        for p in KAGGLE_INPUT.iterdir()
        if p.is_dir()
    ],
    key=lambda p: p.name.lower()
)


if len(source_roots) == 0:

    raise RuntimeError(
        "C01 STOPPED: no source root is available "
        "under /kaggle/input."
    )


# ------------------------------------------------------------
# 5. SHA-256 FUNCTION
# ------------------------------------------------------------

def sha256_file(path, chunk_size=1024 * 1024):
    """
    Calculate SHA-256 using streaming reads.

    The source file is opened read-only.
    No source modification occurs.
    """

    digest = hashlib.sha256()

    with open(path, "rb") as f:

        while True:

            chunk = f.read(chunk_size)

            if not chunk:
                break

            digest.update(chunk)

    return digest.hexdigest()


# ------------------------------------------------------------
# 6. INVENTORY CONTAINER
# ------------------------------------------------------------

artifacts = []
directories = []
errors = []


# ------------------------------------------------------------
# 7. RECURSIVE SOURCE INVENTORY
# ------------------------------------------------------------

for root in source_roots:

    root = root.resolve()

    for current_root, dirnames, filenames in os.walk(
        root,
        topdown=True,
        followlinks=False
    ):

        current_path = Path(current_root)

        # Deterministic traversal order
        dirnames.sort(key=str.lower)
        filenames.sort(key=str.lower)

        # ----------------------------------------------------
        # Record directories
        # ----------------------------------------------------

        for dirname in dirnames:

            directory_path = current_path / dirname

            try:

                rel_path = directory_path.relative_to(
                    Path("/kaggle/input")
                ).as_posix()

                directories.append({
                    "path": rel_path,
                    "name": dirname,
                    "type": "directory",
                })

            except Exception as exc:

                errors.append({
                    "path": str(directory_path),
                    "operation": "directory_inventory",
                    "error": repr(exc),
                })

        # ----------------------------------------------------
        # Record files
        # ----------------------------------------------------

        for filename in filenames:

            file_path = current_path / filename

            try:

                # Never follow symbolic links as files.
                if file_path.is_symlink():

                    target = os.readlink(file_path)

                    rel_path = file_path.relative_to(
                        Path("/kaggle/input")
                    ).as_posix()

                    artifacts.append({
                        "path": rel_path,
                        "name": filename,
                        "type": "symlink",
                        "size_bytes": None,
                        "sha256": None,
                        "extension": file_path.suffix.lower(),
                        "symlink_target": target,
                        "hash_status": "NOT_HASHED",
                    })

                    continue


                # Regular file check
                if not file_path.is_file():

                    rel_path = file_path.relative_to(
                        Path("/kaggle/input")
                    ).as_posix()

                    artifacts.append({
                        "path": rel_path,
                        "name": filename,
                        "type": "non_regular_file",
                        "size_bytes": None,
                        "sha256": None,
                        "extension": file_path.suffix.lower(),
                        "hash_status": "NOT_HASHED",
                    })

                    continue


                # File metadata
                stat_info = file_path.stat()

                size_bytes = stat_info.st_size

                rel_path = file_path.relative_to(
                    Path("/kaggle/input")
                ).as_posix()


                # SHA-256
                digest = sha256_file(file_path)


                artifacts.append({
                    "path": rel_path,
                    "name": filename,
                    "type": "file",
                    "size_bytes": size_bytes,
                    "sha256": digest,
                    "extension": file_path.suffix.lower(),
                    "hash_status": "HASHED",
                })


            except Exception as exc:

                try:
                    rel_path = file_path.relative_to(
                        Path("/kaggle/input")
                    ).as_posix()
                except Exception:
                    rel_path = str(file_path)

                errors.append({
                    "path": rel_path,
                    "operation": "file_inventory_or_hash",
                    "error": repr(exc),
                })


# ------------------------------------------------------------
# 8. DETERMINISTIC SORTING
# ------------------------------------------------------------

artifacts.sort(
    key=lambda x: x["path"].lower()
)

directories.sort(
    key=lambda x: x["path"].lower()
)

errors.sort(
    key=lambda x: (
        x.get("path", "").lower(),
        x.get("operation", "")
    )
)


# ------------------------------------------------------------
# 9. COUNTS
# ------------------------------------------------------------

regular_files = [
    x
    for x in artifacts
    if x["type"] == "file"
]

hashed_files = [
    x
    for x in artifacts
    if x.get("hash_status") == "HASHED"
]

symlinks = [
    x
    for x in artifacts
    if x["type"] == "symlink"
]

non_regular_files = [
    x
    for x in artifacts
    if x["type"] == "non_regular_file"
]


inventory_complete = (
    len(errors) == 0
    and len(regular_files) == len(hashed_files)
)


# ------------------------------------------------------------
# 10. CLAIM BOUNDARY
# ------------------------------------------------------------

claim_boundary = {

    "source_inventory_performed": True,

    "sha256_inventory_performed": True,

    "artifact_identity_verified": False,

    "artifact_provenance_verified": False,

    "artifact_authority_assigned": False,

    "artifact_completeness_verified": False,

    "construct_authorization_performed": False,

    "score_computation_performed": False,

    "outcome_construction_performed": False,

    "empirical_evaluation_performed": False,

    "empirical_validity_claim_authorized": False,
}


# ------------------------------------------------------------
# 11. MACHINE-READABLE RECORD
# ------------------------------------------------------------

c01_record = {

    "protocol": PROTOCOL_VERSION,

    "cell": CELL_ID,

    "cell_purpose": CELL_PURPOSE,

    "timestamp_utc": timestamp_utc,

    "source_roots": [
        str(root)
        for root in source_roots
    ],

    "inventory_summary": {

        "directory_count": len(directories),

        "artifact_entry_count": len(artifacts),

        "regular_file_count": len(
            regular_files
        ),

        "hashed_file_count": len(
            hashed_files
        ),

        "symlink_count": len(
            symlinks
        ),

        "non_regular_file_count": len(
            non_regular_files
        ),

        "error_count": len(
            errors
        ),

        "inventory_complete": inventory_complete,
    },

    "directories": directories,

    "artifacts": artifacts,

    "errors": errors,

    "claim_boundary": claim_boundary,

    "governing_statement": (
        "C01 establishes the observed source inventory and "
        "SHA-256 hashes only. It does not establish artifact "
        "authority, provenance, completeness, construct "
        "authorization, score authorization, outcome "
        "estimability, empirical evaluation, or empirical "
        "validity."
    ),

    "next_cell": (
        "C02 — ARTIFACT IDENTITY VERIFICATION"
    ),
}


with open(
    C01_JSON,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        c01_record,
        f,
        indent=2,
        ensure_ascii=False
    )


# ------------------------------------------------------------
# 12. SHA-256 MANIFEST
# ------------------------------------------------------------
#
# One deterministic line per successfully hashed file:
#
# SHA256  relative/path
#
# This is a RECORD of the observed source state.
# ------------------------------------------------------------

with open(
    C01_MANIFEST,
    "w",
    encoding="utf-8"
) as f:

    for item in hashed_files:

        f.write(
            f"{item['sha256']}  {item['path']}\n"
        )


# ------------------------------------------------------------
# 13. HUMAN-READABLE REPORT
# ------------------------------------------------------------

with open(
    C01_TXT,
    "w",
    encoding="utf-8"
) as f:

    f.write("=" * 72 + "\n")

    f.write(
        "OIP v1.0.35 — "
        "C01 ARTIFACT INVENTORY & SHA-256\n"
    )

    f.write("=" * 72 + "\n\n")

    f.write(
        f"Protocol : {PROTOCOL_VERSION}\n"
    )

    f.write(
        f"Cell     : {CELL_ID}\n"
    )

    f.write(
        f"Timestamp UTC : {timestamp_utc}\n\n"
    )

    f.write("SOURCE ROOTS\n")
    f.write("-" * 72 + "\n")

    for root in source_roots:

        f.write(
            f"{root}\n"
        )

    f.write("\n")

    f.write("INVENTORY SUMMARY\n")
    f.write("-" * 72 + "\n")

    f.write(
        f"Directories          : "
        f"{len(directories)}\n"
    )

    f.write(
        f"Artifact entries     : "
        f"{len(artifacts)}\n"
    )

    f.write(
        f"Regular files        : "
        f"{len(regular_files)}\n"
    )

    f.write(
        f"SHA-256 hashed files : "
        f"{len(hashed_files)}\n"
    )

    f.write(
        f"Symlinks             : "
        f"{len(symlinks)}\n"
    )

    f.write(
        f"Non-regular files    : "
        f"{len(non_regular_files)}\n"
    )

    f.write(
        f"Errors               : "
        f"{len(errors)}\n"
    )

    f.write(
        f"Inventory complete   : "
        f"{inventory_complete}\n\n"
    )


    f.write("ARTIFACTS\n")
    f.write("-" * 72 + "\n")

    for item in artifacts:

        f.write(
            f"{item['path']}\n"
        )

        f.write(
            f"  type      : "
            f"{item['type']}\n"
        )

        f.write(
            f"  size      : "
            f"{item.get('size_bytes')}\n"
        )

        f.write(
            f"  sha256    : "
            f"{item.get('sha256')}\n"
        )

        f.write(
            f"  hash      : "
            f"{item.get('hash_status')}\n"
        )

        f.write("\n")


    if errors:

        f.write("ERRORS\n")
        f.write("-" * 72 + "\n")

        for error in errors:

            f.write(
                f"{error}\n"
            )

        f.write("\n")


    f.write("CLAIM BOUNDARY\n")
    f.write("-" * 72 + "\n")

    for key, value in claim_boundary.items():

        f.write(
            f"{key}: {value}\n"
        )

    f.write("\n")

    f.write("GOVERNING STATEMENT\n")
    f.write("-" * 72 + "\n")

    f.write(
        c01_record["governing_statement"]
        + "\n"
    )


# ------------------------------------------------------------
# 14. CONSOLE OUTPUT
# ------------------------------------------------------------

print("=" * 72)

print(
    "OIP v1.0.35 — "
    "C01: ARTIFACT INVENTORY & SHA-256"
)

print("=" * 72)

print(
    f"\nSource roots       : "
    f"{len(source_roots)}"
)

print(
    f"Directories        : "
    f"{len(directories)}"
)

print(
    f"Artifact entries   : "
    f"{len(artifacts)}"
)

print(
    f"Regular files      : "
    f"{len(regular_files)}"
)

print(
    f"SHA-256 hashed     : "
    f"{len(hashed_files)}"
)

print(
    f"Symlinks            : "
    f"{len(symlinks)}"
)

print(
    f"Non-regular files   : "
    f"{len(non_regular_files)}"
)

print(
    f"Errors              : "
    f"{len(errors)}"
)

print(
    f"Inventory complete  : "
    f"{inventory_complete}"
)

print("\nImportant:")

print(
    "  C01 establishes observed inventory and SHA-256 only."
)

print(
    "  It does NOT establish artifact authority or provenance."
)

print(
    "  It does NOT authorize constructs, scores, outcomes,"
)

print(
    "  empirical evaluation, or empirical validity."
)

print("\nArtifacts written:")

print(
    f"  {C01_JSON}"
)

print(
    f"  {C01_TXT}"
)

print(
    f"  {C01_MANIFEST}"
)

print("\nNext:")

print(
    "  C02 — ARTIFACT IDENTITY VERIFICATION"
)

print("=" * 72)
```

```python
# ============================================================
# OIP v1.0.35 — C02
# ARTIFACT IDENTITY VERIFICATION
# ============================================================
#
# PURPOSE
# -------
# Verify that every artifact recorded by C01:
#   1. still exists
#   2. remains a regular file
#   3. retains the recorded byte size
#   4. retains the recorded SHA-256
#
# C02 verifies artifact identity/integrity ONLY.
#
# C02 MUST NOT:
#   - interpret document contents
#   - infer variables
#   - assign substantive authority
#   - authorize constructs
#   - calculate scores
#   - construct outcomes
#   - perform empirical evaluation
#   - resolve source conflicts
#   - establish empirical validity
#
# Duplicate SHA-256 values are reported, not automatically
# treated as errors.
#
# Governing sequence:
# Evidence → Measurement → Score → Evaluation → Claim
# ============================================================

from pathlib import Path
from datetime import datetime, timezone
from collections import defaultdict
import hashlib
import json


# ------------------------------------------------------------
# 1. FIXED PATHS
# ------------------------------------------------------------

KAGGLE_INPUT = Path("/kaggle/input")
KAGGLE_WORKING = Path("/kaggle/working")

RUN_ROOT = KAGGLE_WORKING / "oip_v1_0_35"

C01_DIR = RUN_ROOT / "c01_artifact_inventory"
C01_JSON = C01_DIR / "c01_artifact_inventory.json"

C02_DIR = RUN_ROOT / "c02_artifact_identity"
C02_JSON = C02_DIR / "c02_artifact_identity.json"
C02_TXT = C02_DIR / "c02_artifact_identity.txt"

PROTOCOL_VERSION = "OIP v1.0.35"
CELL_ID = "C02"
CELL_PURPOSE = "ARTIFACT_IDENTITY_VERIFICATION"


# ------------------------------------------------------------
# 2. CREATE OUTPUT DIRECTORY
# ------------------------------------------------------------

C02_DIR.mkdir(
    parents=True,
    exist_ok=True
)


# ------------------------------------------------------------
# 3. TIMESTAMP
# ------------------------------------------------------------

timestamp_utc = datetime.now(
    timezone.utc
).replace(
    microsecond=0
).isoformat()


# ------------------------------------------------------------
# 4. LOAD C01 INVENTORY
# ------------------------------------------------------------

if not C01_JSON.exists():
    raise RuntimeError(
        "C02 STOPPED: C01 inventory record was not found."
    )

with open(
    C01_JSON,
    "r",
    encoding="utf-8"
) as f:
    c01_record = json.load(f)


# ------------------------------------------------------------
# 5. VERIFY C01 STRUCTURE
# ------------------------------------------------------------

required_c01_keys = [
    "protocol",
    "cell",
    "inventory_summary",
    "artifacts",
]

missing_c01_keys = [
    key
    for key in required_c01_keys
    if key not in c01_record
]

if missing_c01_keys:
    raise RuntimeError(
        "C02 STOPPED: required C01 fields are missing: "
        + ", ".join(missing_c01_keys)
    )

if c01_record["protocol"] != PROTOCOL_VERSION:
    raise RuntimeError(
        "C02 STOPPED: C01 protocol version mismatch."
    )

if c01_record["cell"] != "C01":
    raise RuntimeError(
        "C02 STOPPED: supplied inventory is not marked as C01."
    )

c01_artifacts = c01_record["artifacts"]


# ------------------------------------------------------------
# 6. SHA-256 FUNCTION
# ------------------------------------------------------------

def sha256_file(path, chunk_size=1024 * 1024):
    digest = hashlib.sha256()

    with open(path, "rb") as f:
        while True:
            chunk = f.read(chunk_size)

            if not chunk:
                break

            digest.update(chunk)

    return digest.hexdigest()


# ------------------------------------------------------------
# 7. VERIFY EACH C01 ARTIFACT
# ------------------------------------------------------------

verification_records = []

missing_files = []
non_regular_files = []
size_mismatches = []
hash_mismatches = []
verification_errors = []

for item in c01_artifacts:

    relative_path = item.get("path")
    recorded_type = item.get("type")
    recorded_size = item.get("size_bytes")
    recorded_hash = item.get("sha256")

    if not relative_path:
        verification_errors.append({
            "path": None,
            "error": "C01 artifact has no recorded path."
        })
        continue

    current_path = (
        KAGGLE_INPUT / Path(relative_path)
    )

    record = {
        "path": relative_path,
        "recorded_type": recorded_type,
        "recorded_size_bytes": recorded_size,
        "recorded_sha256": recorded_hash,
        "current_exists": False,
        "current_regular_file": False,
        "current_size_bytes": None,
        "current_sha256": None,
        "identity_status": "NOT_VERIFIED",
        "error": None,
    }

    # --------------------------------------------------------
    # EXISTENCE
    # --------------------------------------------------------

    if not current_path.exists():

        record["identity_status"] = "MISSING"

        missing_files.append(relative_path)

        verification_records.append(record)

        continue

    record["current_exists"] = True


    # --------------------------------------------------------
    # REGULAR FILE CHECK
    # --------------------------------------------------------

    if not current_path.is_file():

        record["identity_status"] = "NOT_REGULAR_FILE"

        non_regular_files.append(relative_path)

        verification_records.append(record)

        continue

    record["current_regular_file"] = True


    # --------------------------------------------------------
    # SIZE + SHA-256
    # --------------------------------------------------------

    try:

        current_size = current_path.stat().st_size

        record["current_size_bytes"] = current_size

        if recorded_size != current_size:

            size_mismatches.append({
                "path": relative_path,
                "recorded_size_bytes": recorded_size,
                "current_size_bytes": current_size,
            })


        current_hash = sha256_file(current_path)

        record["current_sha256"] = current_hash

        if recorded_hash != current_hash:

            record["identity_status"] = "HASH_MISMATCH"

            hash_mismatches.append({
                "path": relative_path,
                "recorded_sha256": recorded_hash,
                "current_sha256": current_hash,
            })

        elif recorded_size != current_size:

            record["identity_status"] = (
                "HASH_MATCH_SIZE_MISMATCH"
            )

        else:

            record["identity_status"] = "VERIFIED"


    except Exception as exc:

        record["identity_status"] = "ERROR"

        record["error"] = repr(exc)

        verification_errors.append({
            "path": relative_path,
            "error": repr(exc),
        })


    verification_records.append(record)


# ------------------------------------------------------------
# 8. DUPLICATE SHA-256 GROUPS
# ------------------------------------------------------------

hash_to_paths = defaultdict(list)

for record in verification_records:

    if (
        record["identity_status"] == "VERIFIED"
        and record["current_sha256"]
    ):

        hash_to_paths[
            record["current_sha256"]
        ].append(
            record["path"]
        )


duplicate_hash_groups = []

for digest, paths in sorted(
    hash_to_paths.items(),
    key=lambda x: x[0]
):

    if len(paths) > 1:

        duplicate_hash_groups.append({
            "sha256": digest,
            "path_count": len(paths),
            "paths": sorted(paths),
        })


# ------------------------------------------------------------
# 9. STATUS COUNTS
# ------------------------------------------------------------

verified_count = sum(
    1
    for r in verification_records
    if r["identity_status"] == "VERIFIED"
)

hash_match_size_mismatch_count = sum(
    1
    for r in verification_records
    if r["identity_status"]
    == "HASH_MATCH_SIZE_MISMATCH"
)

hash_mismatch_count = sum(
    1
    for r in verification_records
    if r["identity_status"] == "HASH_MISMATCH"
)

missing_count = sum(
    1
    for r in verification_records
    if r["identity_status"] == "MISSING"
)

non_regular_count = sum(
    1
    for r in verification_records
    if r["identity_status"] == "NOT_REGULAR_FILE"
)

error_count = sum(
    1
    for r in verification_records
    if r["identity_status"] == "ERROR"
)


# ------------------------------------------------------------
# 10. FINAL IDENTITY STATUS
# ------------------------------------------------------------

identity_status = "PASS"

identity_reason = (
    "All C01-recorded regular files remain present as "
    "regular files with matching size and SHA-256."
)

if (
    missing_count > 0
    or non_regular_count > 0
    or hash_mismatch_count > 0
    or hash_match_size_mismatch_count > 0
    or error_count > 0
):

    identity_status = "FAIL"

    identity_reason = (
        "One or more C01-recorded artifacts failed identity "
        "or integrity verification. See detailed records."
    )


# ------------------------------------------------------------
# 11. CLAIM BOUNDARY
# ------------------------------------------------------------

claim_boundary = {

    "artifact_paths_verified": (
        identity_status == "PASS"
    ),

    "artifact_size_integrity_verified": (
        identity_status == "PASS"
    ),

    "artifact_sha256_integrity_verified": (
        identity_status == "PASS"
    ),

    "artifact_identity_verified": (
        identity_status == "PASS"
    ),

    "artifact_provenance_verified": False,

    "artifact_authority_assigned": False,

    "artifact_completeness_verified": False,

    "construct_authorization_performed": False,

    "score_computation_performed": False,

    "outcome_construction_performed": False,

    "empirical_evaluation_performed": False,

    "empirical_validity_claim_authorized": False,
}


# ------------------------------------------------------------
# 12. MACHINE-READABLE RECORD
# ------------------------------------------------------------

c02_record = {

    "protocol": PROTOCOL_VERSION,

    "cell": CELL_ID,

    "cell_purpose": CELL_PURPOSE,

    "timestamp_utc": timestamp_utc,

    "c01_inventory_source": str(C01_JSON),

    "identity_status": identity_status,

    "identity_reason": identity_reason,

    "verification_summary": {

        "c01_artifact_count": len(c01_artifacts),

        "verified_count": verified_count,

        "hash_match_size_mismatch_count":
            hash_match_size_mismatch_count,

        "hash_mismatch_count":
            hash_mismatch_count,

        "missing_count":
            missing_count,

        "non_regular_count":
            non_regular_count,

        "error_count":
            error_count,

        "duplicate_hash_group_count":
            len(duplicate_hash_groups),
    },

    "missing_files": missing_files,

    "non_regular_files": non_regular_files,

    "size_mismatches": size_mismatches,

    "hash_mismatches": hash_mismatches,

    "verification_errors": verification_errors,

    "duplicate_hash_groups":
        duplicate_hash_groups,

    "verification_records":
        verification_records,

    "claim_boundary":
        claim_boundary,

    "governing_statement": (
        "C02 verifies artifact identity and byte-level "
        "integrity through path, regular-file status, "
        "size, and SHA-256 comparison. It does not "
        "establish substantive provenance, authority, "
        "semantic validity, construct authorization, "
        "score authorization, outcome estimability, "
        "empirical evaluation, or empirical validity."
    ),

    "next_cell":
        "C03 — AUTHORITY ASSIGNMENT",
}


# ------------------------------------------------------------
# 13. WRITE JSON
# ------------------------------------------------------------

with open(
    C02_JSON,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        c02_record,
        f,
        indent=2,
        ensure_ascii=False
    )


# ------------------------------------------------------------
# 14. WRITE HUMAN-READABLE REPORT
# ------------------------------------------------------------

with open(
    C02_TXT,
    "w",
    encoding="utf-8"
) as f:

    f.write("=" * 72 + "\n")

    f.write(
        "OIP v1.0.35 — "
        "C02 ARTIFACT IDENTITY VERIFICATION\n"
    )

    f.write("=" * 72 + "\n\n")

    f.write(
        f"Protocol : {PROTOCOL_VERSION}\n"
    )

    f.write(
        f"Cell     : {CELL_ID}\n"
    )

    f.write(
        f"Timestamp UTC : {timestamp_utc}\n\n"
    )

    f.write(
        f"Identity status : {identity_status}\n"
    )

    f.write(
        f"Reason : {identity_reason}\n\n"
    )

    f.write("VERIFICATION SUMMARY\n")
    f.write("-" * 72 + "\n")

    f.write(
        f"C01 artifacts              : "
        f"{len(c01_artifacts)}\n"
    )

    f.write(
        f"Verified                   : "
        f"{verified_count}\n"
    )

    f.write(
        f"Hash/size anomaly          : "
        f"{hash_match_size_mismatch_count}\n"
    )

    f.write(
        f"Hash mismatches            : "
        f"{hash_mismatch_count}\n"
    )

    f.write(
        f"Missing                    : "
        f"{missing_count}\n"
    )

    f.write(
        f"Non-regular                : "
        f"{non_regular_count}\n"
    )

    f.write(
        f"Errors                     : "
        f"{error_count}\n"
    )

    f.write(
        f"Duplicate SHA-256 groups   : "
        f"{len(duplicate_hash_groups)}\n\n"
    )


    if missing_files:

        f.write("MISSING FILES\n")
        f.write("-" * 72 + "\n")

        for path in missing_files:
            f.write(f"{path}\n")

        f.write("\n")


    if non_regular_files:

        f.write("NON-REGULAR FILES\n")
        f.write("-" * 72 + "\n")

        for path in non_regular_files:
            f.write(f"{path}\n")

        f.write("\n")


    if size_mismatches:

        f.write("SIZE MISMATCHES\n")
        f.write("-" * 72 + "\n")

        for item in size_mismatches:
            f.write(f"{item}\n")

        f.write("\n")


    if hash_mismatches:

        f.write("SHA-256 MISMATCHES\n")
        f.write("-" * 72 + "\n")

        for item in hash_mismatches:
            f.write(f"{item}\n")

        f.write("\n")


    if verification_errors:

        f.write("VERIFICATION ERRORS\n")
        f.write("-" * 72 + "\n")

        for item in verification_errors:
            f.write(f"{item}\n")

        f.write("\n")


    if duplicate_hash_groups:

        f.write("DUPLICATE SHA-256 GROUPS\n")
        f.write("-" * 72 + "\n")

        for group in duplicate_hash_groups:

            f.write(
                f"SHA-256: {group['sha256']}\n"
            )

            for path in group["paths"]:
                f.write(f"  - {path}\n")

            f.write("\n")


    f.write("CLAIM BOUNDARY\n")
    f.write("-" * 72 + "\n")

    for key, value in claim_boundary.items():

        f.write(
            f"{key}: {value}\n"
        )

    f.write("\n")

    f.write("GOVERNING STATEMENT\n")
    f.write("-" * 72 + "\n")

    f.write(
        c02_record["governing_statement"]
        + "\n"
    )


# ------------------------------------------------------------
# 15. CONSOLE OUTPUT
# ------------------------------------------------------------

print("=" * 72)

print(
    "OIP v1.0.35 — "
    "C02: ARTIFACT IDENTITY VERIFICATION"
)

print("=" * 72)

print(
    f"\nC01 artifacts          : "
    f"{len(c01_artifacts)}"
)

print(
    f"Verified               : "
    f"{verified_count}"
)

print(
    f"Hash/size anomaly      : "
    f"{hash_match_size_mismatch_count}"
)

print(
    f"Hash mismatches        : "
    f"{hash_mismatch_count}"
)

print(
    f"Missing                : "
    f"{missing_count}"
)

print(
    f"Non-regular            : "
    f"{non_regular_count}"
)

print(
    f"Errors                 : "
    f"{error_count}"
)

print(
    f"Duplicate hash groups  : "
    f"{len(duplicate_hash_groups)}"
)

print(
    f"\nIdentity status        : "
    f"{identity_status}"
)

print("\nImportant:")

print(
    "  C02 verifies artifact identity and byte-level integrity."
)

print(
    "  C02 does NOT establish provenance or substantive authority."
)

print(
    "  C02 does NOT authorize constructs, scores, outcomes,"
)

print(
    "  empirical evaluation, or empirical validity."
)

print("\nArtifacts written:")

print(
    f"  {C02_JSON}"
)

print(
    f"  {C02_TXT}"
)

print("\nNext:")

print(
    "  C03 — AUTHORITY ASSIGNMENT"
)

print("=" * 72)
```

```python
# ============================================================
# OIP v1.0.35 — C03
# AUTHORITY ASSIGNMENT
# ============================================================
#
# PURPOSE
# -------
# Establish an explicit, fail-closed authority register for
# the artifacts verified by C02.
#
# CORE RULE
# ---------
# Artifact existence, filename similarity, extension, or path
# does NOT by itself establish substantive authority.
#
# Executable artifacts receive PRIMARY EXECUTION-RECORD status
# only after identity, provenance, completeness, execution state,
# and canonical authority have been established.
#
# Therefore C03 distinguishes:
#
#   1. SOURCE FORM
#   2. AUTHORITY STATUS
#   3. AUTHORITY BASIS
#   4. VERIFICATION STATE
#
# C03 MUST NOT:
#   - infer scientific validity
#   - approve OIP constructs
#   - calculate scores
#   - construct outcomes
#   - perform empirical evaluation
#   - resolve unresolved source conflicts
#   - treat filename similarity as proof
#   - treat AI interpretation as authoritative evidence
#
# Governing sequence:
# Evidence → Measurement → Score → Evaluation → Claim
# ============================================================

from pathlib import Path
from datetime import datetime, timezone
import json


# ------------------------------------------------------------
# 1. FIXED PATHS
# ------------------------------------------------------------

KAGGLE_INPUT = Path("/kaggle/input")
KAGGLE_WORKING = Path("/kaggle/working")

RUN_ROOT = KAGGLE_WORKING / "oip_v1_0_35"

C02_DIR = RUN_ROOT / "c02_artifact_identity"
C02_JSON = C02_DIR / "c02_artifact_identity.json"

C03_DIR = RUN_ROOT / "c03_authority_assignment"
C03_JSON = C03_DIR / "c03_authority_assignment.json"
C03_TXT = C03_DIR / "c03_authority_assignment.txt"

PROTOCOL_VERSION = "OIP v1.0.35"
CELL_ID = "C03"
CELL_PURPOSE = "AUTHORITY_ASSIGNMENT"


# ------------------------------------------------------------
# 2. CREATE OUTPUT DIRECTORY
# ------------------------------------------------------------

C03_DIR.mkdir(
    parents=True,
    exist_ok=True
)


# ------------------------------------------------------------
# 3. TIMESTAMP
# ------------------------------------------------------------

timestamp_utc = datetime.now(
    timezone.utc
).replace(
    microsecond=0
).isoformat()


# ------------------------------------------------------------
# 4. LOAD C02 RECORD
# ------------------------------------------------------------

if not C02_JSON.exists():
    raise RuntimeError(
        "C03 STOPPED: C02 identity record was not found."
    )

with open(
    C02_JSON,
    "r",
    encoding="utf-8"
) as f:
    c02_record = json.load(f)


# ------------------------------------------------------------
# 5. VERIFY C02 PREREQUISITES
# ------------------------------------------------------------

required_c02_keys = [
    "protocol",
    "cell",
    "identity_status",
    "verification_summary",
    "verification_records",
]

missing_c02_keys = [
    key
    for key in required_c02_keys
    if key not in c02_record
]

if missing_c02_keys:
    raise RuntimeError(
        "C03 STOPPED: required C02 fields are missing: "
        + ", ".join(missing_c02_keys)
    )

if c02_record["protocol"] != PROTOCOL_VERSION:
    raise RuntimeError(
        "C03 STOPPED: C02 protocol version mismatch."
    )

if c02_record["cell"] != "C02":
    raise RuntimeError(
        "C03 STOPPED: supplied record is not marked as C02."
    )

if c02_record["identity_status"] != "PASS":
    raise RuntimeError(
        "C03 STOPPED: C02 identity status is not PASS."
    )

verification_records = c02_record[
    "verification_records"
]


# ------------------------------------------------------------
# 6. SOURCE-FORM CLASSIFICATION
# ------------------------------------------------------------
#
# These rules classify observable artifact FORM only.
#
# They do NOT establish substantive authority.
#
# Classification is an organizational aid for later review.
# ------------------------------------------------------------

def classify_source_form(path_string):

    path = Path(path_string)

    name = path.name.lower()
    suffix = path.suffix.lower()
    normalized = path_string.lower()


    # --------------------------------------------------------
    # OIP v1.0.30 normative protocol
    # --------------------------------------------------------

    if (
        "v1_0_30" in normalized
        and (
            "normative" in normalized
            or "protocol" in normalized
        )
    ):
        return "NORMATIVE_PROTOCOL_CANDIDATE"


    # --------------------------------------------------------
    # OIP v1.0.27 addendum
    # --------------------------------------------------------

    if (
        "v1_0_27" in normalized
        and "addendum" in normalized
    ):
        return "SPECIFICATION_ADDENDUM_CANDIDATE"


    # --------------------------------------------------------
    # OIP v1.0.28 historical evaluation report
    # --------------------------------------------------------

    if (
        "v1_0_28" in normalized
        and (
            "empirical" in normalized
            or "report" in normalized
        )
    ):
        return "HISTORICAL_EVALUATION_REPORT"


    # --------------------------------------------------------
    # OIP v1.0.34 synthesis / evidence boundary
    # --------------------------------------------------------

    if (
        "v1_0_34" in normalized
        and (
            "evidence" in normalized
            or "boundary" in normalized
            or "four_dataset" in normalized
        )
    ):
        return "SYNTHESIS_REPORT_CANDIDATE"


    # --------------------------------------------------------
    # OIP v1.0.26 historical specification candidate
    # --------------------------------------------------------

    if (
        "v1_0_26" in normalized
        or "organic intelligence protocol" in normalized
    ):
        return "HISTORICAL_SPECIFICATION_CANDIDATE"


    # --------------------------------------------------------
    # Executable / audit artifacts
    # --------------------------------------------------------

    if (
        suffix in [".ipynb", ".py"]
        or "notebook" in name
        or "audit" in name
    ):
        return "EXECUTABLE_OR_AUDIT_ARTIFACT_CANDIDATE"


    # --------------------------------------------------------
    # Data artifacts
    # --------------------------------------------------------

    if suffix in [
        ".dta",
        ".csv",
        ".sav",
        ".sas7bdat"
    ]:
        return "DATA_ARTIFACT"


    # --------------------------------------------------------
    # Documentary artifacts
    # --------------------------------------------------------

    if suffix == ".pdf":
        return "DOCUMENTARY_ARTIFACT"


    # --------------------------------------------------------
    # Text / metadata
    # --------------------------------------------------------

    if suffix in [
        ".md",
        ".txt",
        ".json"
    ]:
        return "TEXTUAL_OR_METADATA_ARTIFACT"


    # --------------------------------------------------------
    # Archives
    # --------------------------------------------------------

    if suffix in [
        ".zip",
        ".tar",
        ".gz",
        ".7z"
    ]:
        return "ARCHIVE_ARTIFACT"


    return "UNKNOWN_ARTIFACT_FORM"


# ------------------------------------------------------------
# 7. FAIL-CLOSED AUTHORITY STATUS
# ------------------------------------------------------------
#
# C03 does NOT grant substantive authority.
#
# Identity was verified by C02.
#
# However:
#   provenance               = not established here
#   completeness             = not established here
#   execution state          = not established globally
#   canonical authority      = not established here
#
# Therefore authority remains pending.
# ------------------------------------------------------------

def determine_authority_status(source_form):

    if source_form == "NORMATIVE_PROTOCOL_CANDIDATE":
        return "AUTHORITY_PENDING_DOCUMENTARY_VERIFICATION"

    if source_form == "SPECIFICATION_ADDENDUM_CANDIDATE":
        return "AUTHORITY_PENDING_DOCUMENTARY_VERIFICATION"

    if source_form == "HISTORICAL_EVALUATION_REPORT":
        return "HISTORICAL_RECORD_PENDING_PROVENANCE_RECONCILIATION"

    if source_form == "SYNTHESIS_REPORT_CANDIDATE":
        return "SECONDARY_SYNTHESIS_RECORD"

    if source_form == "HISTORICAL_SPECIFICATION_CANDIDATE":
        return "HISTORICAL_RECORD_PENDING_PROVENANCE_RECONCILIATION"

    if source_form == "EXECUTABLE_OR_AUDIT_ARTIFACT_CANDIDATE":
        return "EXECUTION_AUTHORITY_PENDING"

    if source_form == "DATA_ARTIFACT":
        return "DATA_SOURCE_AUTHORITY_PENDING"

    if source_form == "DOCUMENTARY_ARTIFACT":
        return "DOCUMENTARY_AUTHORITY_PENDING"

    if source_form == "TEXTUAL_OR_METADATA_ARTIFACT":
        return "TEXTUAL_AUTHORITY_PENDING"

    if source_form == "ARCHIVE_ARTIFACT":
        return "ARCHIVE_AUTHORITY_PENDING"

    return "AUTHORITY_UNCLASSIFIED"


# ------------------------------------------------------------
# 8. BUILD AUTHORITY REGISTER
# ------------------------------------------------------------

authority_records = []

for record in verification_records:

    path = record["path"]

    source_form = classify_source_form(path)

    authority_status = determine_authority_status(
        source_form
    )

    authority_records.append({

        "path": path,

        "sha256": record["current_sha256"],

        "size_bytes": record["current_size_bytes"],

        "identity_status": record["identity_status"],

        "source_form": source_form,

        "authority_status": authority_status,

        "authority_basis": [
            "C02 identity verification only"
        ],

        "provenance_verified": False,

        "completeness_verified": False,

        "execution_state_verified": False,

        "canonical_authority_verified": False,

        "substantive_authority_assigned": False,

        "note": (
            "Filename/path classification is not treated "
            "as substantive authority."
        ),
    })


# ------------------------------------------------------------
# 9. SUMMARY COUNTS
# ------------------------------------------------------------

source_form_counts = {}

authority_status_counts = {}

for record in authority_records:

    source_form = record["source_form"]
    authority_status = record["authority_status"]

    source_form_counts[source_form] = (
        source_form_counts.get(
            source_form,
            0
        ) + 1
    )

    authority_status_counts[authority_status] = (
        authority_status_counts.get(
            authority_status,
            0
        ) + 1
    )


# ------------------------------------------------------------
# 10. SAFETY CHECK
# ------------------------------------------------------------

substantive_authority_count = sum(
    1
    for record in authority_records
    if record["substantive_authority_assigned"]
)

if substantive_authority_count != 0:
    raise RuntimeError(
        "C03 SAFETY STOP: substantive authority was assigned "
        "without the required verification chain."
    )


# ------------------------------------------------------------
# 11. CLAIM BOUNDARY
# ------------------------------------------------------------

claim_boundary = {

    "c01_inventory_verified": True,

    "c02_identity_verified": True,

    "source_form_classification_performed": True,

    "authority_register_created": True,

    "substantive_authority_assigned": False,

    "provenance_verified": False,

    "completeness_verified": False,

    "execution_state_verified": False,

    "canonical_authority_verified": False,

    "construct_authorization_performed": False,

    "score_computation_performed": False,

    "outcome_construction_performed": False,

    "empirical_evaluation_performed": False,

    "empirical_validity_claim_authorized": False,
}


# ------------------------------------------------------------
# 12. MACHINE-READABLE RECORD
# ------------------------------------------------------------

c03_record = {

    "protocol": PROTOCOL_VERSION,

    "cell": CELL_ID,

    "cell_purpose": CELL_PURPOSE,

    "timestamp_utc": timestamp_utc,

    "prerequisite": {

        "c02_identity_status":
            c02_record["identity_status"],

        "c02_verified_artifacts":
            c02_record[
                "verification_summary"
            ][
                "verified_count"
            ],
    },

    "authority_rule": (
        "Artifact existence, filename similarity, extension, "
        "or hash identity does not by itself establish "
        "substantive authority."
    ),

    "source_form_counts":
        source_form_counts,

    "authority_status_counts":
        authority_status_counts,

    "substantive_authority_count":
        substantive_authority_count,

    "authority_records":
        authority_records,

    "claim_boundary":
        claim_boundary,

    "governing_statement": (
        "C03 establishes an explicit authority register and "
        "source-form classification while preserving a "
        "fail-closed boundary. No substantive authority is "
        "assigned solely from filename, extension, path, or "
        "artifact existence."
    ),

    "next_cell":
        "C04 — CONFLICT REGISTER",
}


# ------------------------------------------------------------
# 13. WRITE JSON
# ------------------------------------------------------------

with open(
    C03_JSON,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        c03_record,
        f,
        indent=2,
        ensure_ascii=False
    )


# ------------------------------------------------------------
# 14. WRITE HUMAN-READABLE REPORT
# ------------------------------------------------------------

with open(
    C03_TXT,
    "w",
    encoding="utf-8"
) as f:

    f.write("=" * 72 + "\n")

    f.write(
        "OIP v1.0.35 — "
        "C03 AUTHORITY ASSIGNMENT\n"
    )

    f.write("=" * 72 + "\n\n")

    f.write(
        f"Protocol : {PROTOCOL_VERSION}\n"
    )

    f.write(
        f"Cell     : {CELL_ID}\n"
    )

    f.write(
        f"Timestamp UTC : {timestamp_utc}\n\n"
    )

    f.write(
        "C02 prerequisite status\n"
    )

    f.write("-" * 72 + "\n")

    f.write(
        f"Identity status : "
        f"{c02_record['identity_status']}\n"
    )

    f.write(
        f"Verified artifacts : "
        f"{c02_record['verification_summary']['verified_count']}\n\n"
    )

    f.write(
        "SOURCE FORM COUNTS\n"
    )

    f.write("-" * 72 + "\n")

    for key, value in sorted(
        source_form_counts.items()
    ):

        f.write(
            f"{key}: {value}\n"
        )

    f.write("\n")

    f.write(
        "AUTHORITY STATUS COUNTS\n"
    )

    f.write("-" * 72 + "\n")

    for key, value in sorted(
        authority_status_counts.items()
    ):

        f.write(
            f"{key}: {value}\n"
        )

    f.write("\n")

    f.write(
        "SUBSTANTIVE AUTHORITY ASSIGNED\n"
    )

    f.write("-" * 72 + "\n")

    f.write(
        f"{substantive_authority_count}\n\n"
    )

    f.write(
        "AUTHORITY RULE\n"
    )

    f.write("-" * 72 + "\n")

    f.write(
        c03_record["authority_rule"]
        + "\n\n"
    )

    f.write(
        "CLAIM BOUNDARY\n"
    )

    f.write("-" * 72 + "\n")

    for key, value in claim_boundary.items():

        f.write(
            f"{key}: {value}\n"
        )

    f.write("\n")

    f.write(
        "GOVERNING STATEMENT\n"
    )

    f.write("-" * 72 + "\n")

    f.write(
        c03_record["governing_statement"]
        + "\n"
    )


# ------------------------------------------------------------
# 15. CONSOLE OUTPUT
# ------------------------------------------------------------

print("=" * 72)

print(
    "OIP v1.0.35 — "
    "C03: AUTHORITY ASSIGNMENT"
)

print("=" * 72)

print(
    f"\nC02 identity status : "
    f"{c02_record['identity_status']}"
)

print(
    f"Verified artifacts  : "
    f"{c02_record['verification_summary']['verified_count']}"
)

print("\nSource-form counts:")

for key, value in sorted(
    source_form_counts.items()
):

    print(
        f"  {key} : {value}"
    )

print("\nAuthority-status counts:")

for key, value in sorted(
    authority_status_counts.items()
):

    print(
        f"  {key} : {value}"
    )

print(
    "\nSubstantive authority assigned : "
    f"{substantive_authority_count}"
)

print("\nImportant:")

print(
    "  Filename/path similarity does NOT establish authority."
)

print(
    "  Artifact existence does NOT establish authority."
)

print(
    "  C03 preserves unresolved authority as PENDING."
)

print(
    "  No construct, score, outcome, or empirical evaluation"
)

print(
    "  is authorized by C03."
)

print("\nArtifacts written:")

print(
    f"  {C03_JSON}"
)

print(
    f"  {C03_TXT}"
)

print("\nNext:")

print(
    "  C04 — CONFLICT REGISTER"
)

print("=" * 72)
```

```python
# ============================================================
# OIP v1.0.35 — CELL 05
# PRIMARY ARTIFACT VERIFICATION (MALAWI)
# ============================================================
#
# PURPOSE
# -------
# Directly locate and inspect Malawi-related primary artifacts
# available in the current Kaggle execution environment.
#
# C05 verifies what the available artifacts actually contain.
#
# C05 MUST NOT:
#   - resolve C04 conflicts
#   - select a canonical source
#   - declare substantive authority
#   - approve/reject an identifier
#   - decide whether 395 or 396 is correct
#   - decide whether hh_b06_1 is approved
#   - authorize constructs
#   - compute a score
#   - evaluate an outcome
#   - establish empirical validity
#
# ============================================================

from pathlib import Path
from datetime import datetime, timezone
import json
import re
import zipfile
import io
import hashlib


# ------------------------------------------------------------
# 0. CONSTANTS
# ------------------------------------------------------------

PROTOCOL = "OIP v1.0.35"
CELL_ID = "C05"
CELL_NAME = "PRIMARY ARTIFACT VERIFICATION (MALAWI)"

KAGGLE_INPUT = Path("/kaggle/input")
RUN_ROOT = Path("/kaggle/working/oip_v1_0_35")

C04_JSON = (
    RUN_ROOT
    / "c04_conflict_register"
    / "c04_conflict_register.json"
)

C05_DIR = (
    RUN_ROOT
    / "c05_malawi_primary_verification"
)

C05_DIR.mkdir(parents=True, exist_ok=True)

C05_JSON = C05_DIR / "c05_malawi_primary_verification.json"
C05_TXT = C05_DIR / "c05_malawi_primary_verification.txt"


# ------------------------------------------------------------
# 1. HEADER
# ------------------------------------------------------------

print("=" * 72)
print(f"{PROTOCOL} — {CELL_ID}: {CELL_NAME}")
print("=" * 72)


# ------------------------------------------------------------
# 2. LOAD C04
# ------------------------------------------------------------

if not C04_JSON.exists():
    raise FileNotFoundError(
        "C05 FAIL-CLOSED: required C04 register not found:\n"
        f"{C04_JSON}"
    )

with open(C04_JSON, "r", encoding="utf-8") as f:
    c04 = json.load(f)


# C04 schema created by final C04 cell uses register_status.
c04_status = c04.get("register_status")

if c04_status != "PASS":
    raise RuntimeError(
        "C05 FAIL-CLOSED: C04 register status is not PASS.\n"
        f"Observed: {c04_status!r}"
    )

if c04.get("protocol") != PROTOCOL:
    raise RuntimeError(
        "C05 FAIL-CLOSED: protocol mismatch in C04."
    )

if c04.get("cell_id") != "C04":
    raise RuntimeError(
        "C05 FAIL-CLOSED: unexpected upstream cell identifier."
    )


# ------------------------------------------------------------
# 3. SEARCH PARAMETERS
# ------------------------------------------------------------

MALAWI_TERMS = (
    "malawi",
    "mwi_",
    "ihps",
    "ihps2010",
    "ihps_",
    "v1_0_31",
    "v1-0-31",
)

NOTEBOOK_EXTENSIONS = {".ipynb"}
PDF_EXTENSIONS = {".pdf"}
ARCHIVE_EXTENSIONS = {".zip"}


# ------------------------------------------------------------
# 4. HELPERS
# ------------------------------------------------------------

def sha256_file(path, chunk_size=1024 * 1024):
    h = hashlib.sha256()

    with open(path, "rb") as f:
        while True:
            chunk = f.read(chunk_size)
            if not chunk:
                break
            h.update(chunk)

    return h.hexdigest()


def path_is_malawi_related(path_text):
    p = str(path_text).lower()
    return any(term in p for term in MALAWI_TERMS)


def normalize_text(value):
    if value is None:
        return ""

    if isinstance(value, list):
        return "".join(str(x) for x in value)

    return str(value)


def context_snippets(text, pattern, window=180, max_snippets=8):
    """
    Return exact surrounding text for matching evidence.
    Presence alone is not treated as semantic verification.
    """

    if not text:
        return []

    snippets = []

    try:
        matches = list(
            re.finditer(
                pattern,
                text,
                flags=re.IGNORECASE | re.DOTALL
            )
        )
    except re.error:
        matches = []

    for m in matches[:max_snippets]:

        start = max(0, m.start() - window)
        end = min(len(text), m.end() + window)

        snippet = text[start:end]

        snippets.append(
            snippet.replace("\x00", " ").strip()
        )

    return snippets


def inspect_notebook(path):
    """
    Inspect notebook code, markdown and execution outputs separately.
    """

    record = {
        "artifact_type": "NOTEBOOK",
        "path": str(path),
        "sha256": None,
        "read_status": None,

        "cells": 0,
        "code_cells": 0,
        "markdown_cells": 0,
        "executed_cells": 0,

        "code_matches": [],
        "output_matches": [],
        "markdown_matches": [],

        "execution_outputs_present": False
    }

    try:
        raw = path.read_bytes()
        record["sha256"] = hashlib.sha256(raw).hexdigest()

        nb = json.loads(raw.decode("utf-8"))

    except Exception as e:

        record["read_status"] = (
            f"READ_ERROR:{type(e).__name__}"
        )

        return record

    cells = nb.get("cells", [])

    record["cells"] = len(cells)

    for idx, cell in enumerate(cells):

        cell_type = str(
            cell.get("cell_type", "")
        ).lower()

        source = normalize_text(
            cell.get("source", [])
        )

        outputs_text = []

        outputs = cell.get("outputs", [])

        if outputs:
            record["execution_outputs_present"] = True

        for out in outputs:

            text_value = out.get("text", [])

            if text_value:
                outputs_text.append(
                    normalize_text(text_value)
                )

            data = out.get("data", {})

            if isinstance(data, dict):

                for value in data.values():

                    if isinstance(value, str):
                        outputs_text.append(value)

                    elif isinstance(value, list):
                        outputs_text.append(
                            normalize_text(value)
                        )

        output_text = "\n".join(outputs_text)

        if cell_type == "code":

            record["code_cells"] += 1

            if outputs:
                record["executed_cells"] += 1

            matches = []

            for name, pattern in {

                "395": r"\b395\b",
                "396": r"\b396\b",

                "hh_b06_1":
                    r"\bhh_b06_1\b",

                "approved_identifiers_none":
                    r"APPROVED\s+IDENTIFIERS\s*[:=]\s*NONE",

                "hh_b06_1_approved":
                    r"\bhh_b06_1\b.{0,160}\bapproved\b",

                "processed_395":
                    r"\b395\b.{0,160}\bprocess",

                "inventoried_396":
                    r"\b396\b.{0,160}\binventor"
            }.items():

                snippets = context_snippets(
                    source,
                    pattern
                )

                if snippets:

                    matches.append({
                        "cell_index": idx,
                        "claim_signal": name,
                        "snippets": snippets
                    })

            if matches:
                record["code_matches"].extend(matches)

            # ---------------------------------------------
            # Execution output inspection
            # ---------------------------------------------

            if output_text:

                output_matches = []

                for name, pattern in {

                    "395":
                        r"\b395\b",

                    "396":
                        r"\b396\b",

                    "hh_b06_1":
                        r"\bhh_b06_1\b",

                    "approved_identifiers_none":
                        r"APPROVED\s+IDENTIFIERS\s*[:=]\s*NONE",

                    "hh_b06_1_approved":
                        r"\bhh_b06_1\b.{0,160}\bapproved\b",

                    "processed_395":
                        r"\b395\b.{0,160}\bprocess",

                    "inventoried_396":
                        r"\b396\b.{0,160}\binventor"
                }.items():

                    snippets = context_snippets(
                        output_text,
                        pattern
                    )

                    if snippets:

                        output_matches.append({
                            "cell_index": idx,
                            "claim_signal": name,
                            "snippets": snippets
                        })

                if output_matches:
                    record["output_matches"].extend(
                        output_matches
                    )

        elif cell_type == "markdown":

            record["markdown_cells"] += 1

            markdown_matches = []

            for name, pattern in {

                "395":
                    r"\b395\b",

                "396":
                    r"\b396\b",

                "hh_b06_1":
                    r"\bhh_b06_1\b",

                "approved_identifiers_none":
                    r"APPROVED\s+IDENTIFIERS\s*[:=]\s*NONE",

                "hh_b06_1_approved":
                    r"\bhh_b06_1\b.{0,160}\bapproved\b"
            }.items():

                snippets = context_snippets(
                    source,
                    pattern
                )

                if snippets:

                    markdown_matches.append({
                        "cell_index": idx,
                        "claim_signal": name,
                        "snippets": snippets
                    })

            if markdown_matches:
                record["markdown_matches"].extend(
                    markdown_matches
                )

    record["read_status"] = "READ_SUCCESS"

    return record


def inspect_pdf(path):
    """
    Extract PDF text and record exact evidence snippets.
    """

    record = {
        "artifact_type": "PDF",
        "path": str(path),
        "sha256": None,
        "read_status": None,

        "page_count": None,

        "matches": []
    }

    try:
        raw = path.read_bytes()

        record["sha256"] = (
            hashlib.sha256(raw).hexdigest()
        )

    except Exception as e:

        record["read_status"] = (
            f"READ_ERROR:{type(e).__name__}"
        )

        return record

    try:

        from pypdf import PdfReader

        reader = PdfReader(
            io.BytesIO(raw),
            strict=False
        )

        record["page_count"] = len(
            reader.pages
        )

        pages = []

        for page_no, page in enumerate(
            reader.pages,
            start=1
        ):

            try:
                text = page.extract_text() or ""
            except Exception:
                text = ""

            pages.append(
                (page_no, text)
            )

        patterns = {

            "395":
                r"\b395\b",

            "396":
                r"\b396\b",

            "hh_b06_1":
                r"\bhh_b06_1\b",

            "approved_identifiers":
                r"\bhh_b06_1\b.{0,160}\bapproved\b",

            "processed":
                r"\b395\b.{0,160}\bprocess"
        }

        for page_no, text in pages:

            for name, pattern in patterns.items():

                snippets = context_snippets(
                    text,
                    pattern
                )

                for snippet in snippets:

                    record["matches"].append({

                        "page":
                            page_no,

                        "claim_signal":
                            name,

                        "snippet":
                            snippet
                    })

        record["read_status"] = "READ_SUCCESS"

    except ImportError:

        record["read_status"] = (
            "PDF_LIBRARY_NOT_AVAILABLE"
        )

    except Exception as e:

        record["read_status"] = (
            f"PDF_EXTRACTION_ERROR:{type(e).__name__}"
        )

    return record


def inspect_archive(path):
    """
    Inspect ZIP members without assuming that all primary
    artifacts are extracted into the filesystem.
    """

    record = {
        "artifact_type": "ZIP_ARCHIVE",
        "path": str(path),
        "sha256": None,
        "read_status": None,
        "members": []
    }

    try:

        raw = path.read_bytes()

        record["sha256"] = (
            hashlib.sha256(raw).hexdigest()
        )

        with zipfile.ZipFile(
            io.BytesIO(raw),
            "r"
        ) as z:

            for name in z.namelist():

                lower = name.lower()

                # Only inspect likely Malawi artifacts.
                relevant = (
                    path_is_malawi_related(name)
                    or lower.endswith(".ipynb")
                    or lower.endswith(".pdf")
                )

                if not relevant:
                    continue

                record["members"].append({
                    "member":
                        name,
                    "extension":
                        Path(name).suffix.lower()
                })

        record["read_status"] = "READ_SUCCESS"

    except zipfile.BadZipFile:

        record["read_status"] = "INVALID_ZIP"

    except Exception as e:

        record["read_status"] = (
            f"ARCHIVE_READ_ERROR:{type(e).__name__}"
        )

    return record


# ------------------------------------------------------------
# 5. DIRECT SOURCE DISCOVERY
# ------------------------------------------------------------

all_files = []

for path in KAGGLE_INPUT.rglob("*"):

    try:

        if path.is_file():

            all_files.append(path)

    except OSError:
        continue


all_files = sorted(
    all_files,
    key=lambda p: str(p).lower()
)


candidate_paths = []

for path in all_files:

    lower_path = str(path).lower()

    suffix = path.suffix.lower()

    if path_is_malawi_related(lower_path):

        candidate_paths.append(path)

        continue

    # Include notebooks/PDFs that may be the Malawi
    # audit artifacts even if their parent path lacks
    # the word Malawi.
    if suffix in (
        NOTEBOOK_EXTENSIONS
        | PDF_EXTENSIONS
    ):

        name = path.name.lower()

        if (
            "audit" in name
            or "fail_closed" in name
            or "ihps" in name
            or "v1_0_31" in name
            or "v1-0-31" in name
        ):

            candidate_paths.append(path)


candidate_paths = sorted(
    set(candidate_paths),
    key=lambda p: str(p).lower()
)


# ------------------------------------------------------------
# 6. PRIMARY ARTIFACT INSPECTION
# ------------------------------------------------------------

verification_records = []

for path in candidate_paths:

    suffix = path.suffix.lower()

    if suffix in NOTEBOOK_EXTENSIONS:

        verification_records.append(
            inspect_notebook(path)
        )

    elif suffix in PDF_EXTENSIONS:

        verification_records.append(
            inspect_pdf(path)
        )

    elif suffix in ARCHIVE_EXTENSIONS:

        verification_records.append(
            inspect_archive(path)
        )


# ------------------------------------------------------------
# 7. CLASSIFY OBSERVED EVIDENCE
# ------------------------------------------------------------
#
# This section summarizes what was OBSERVED.
# It does NOT resolve C04.
# ------------------------------------------------------------

observed_395 = []
observed_396 = []

observed_hh = []
observed_none = []
observed_approved = []

for record in verification_records:

    # Notebook code
    for item in record.get("code_matches", []):

        signal = item.get("claim_signal")

        if signal == "395":
            observed_395.append(
                (record["path"], "NOTEBOOK_CODE")
            )

        if signal == "396":
            observed_396.append(
                (record["path"], "NOTEBOOK_CODE")
            )

        if signal == "hh_b06_1":
            observed_hh.append(
                (record["path"], "NOTEBOOK_CODE")
            )

        if signal == "approved_identifiers_none":
            observed_none.append(
                (record["path"], "NOTEBOOK_CODE")
            )

        if signal == "hh_b06_1_approved":
            observed_approved.append(
                (record["path"], "NOTEBOOK_CODE")
            )

    # Notebook execution output
    for item in record.get("output_matches", []):

        signal = item.get("claim_signal")

        if signal == "395":
            observed_395.append(
                (record["path"], "NOTEBOOK_OUTPUT")
            )

        if signal == "396":
            observed_396.append(
                (record["path"], "NOTEBOOK_OUTPUT")
            )

        if signal == "hh_b06_1":
            observed_hh.append(
                (record["path"], "NOTEBOOK_OUTPUT")
            )

        if signal == "approved_identifiers_none":
            observed_none.append(
                (record["path"], "NOTEBOOK_OUTPUT")
            )

        if signal == "hh_b06_1_approved":
            observed_approved.append(
                (record["path"], "NOTEBOOK_OUTPUT")
            )

    # Notebook markdown
    for item in record.get("markdown_matches", []):

        signal = item.get("claim_signal")

        if signal == "395":
            observed_395.append(
                (record["path"], "NOTEBOOK_MARKDOWN")
            )

        if signal == "396":
            observed_396.append(
                (record["path"], "NOTEBOOK_MARKDOWN")
            )

        if signal == "hh_b06_1":
            observed_hh.append(
                (record["path"], "NOTEBOOK_MARKDOWN")
            )

        if signal == "approved_identifiers_none":
            observed_none.append(
                (record["path"], "NOTEBOOK_MARKDOWN")
            )

        if signal == "hh_b06_1_approved":
            observed_approved.append(
                (record["path"], "NOTEBOOK_MARKDOWN")
            )

    # PDF
    for item in record.get("matches", []):

        signal = item.get("claim_signal")

        if signal == "395":
            observed_395.append(
                (record["path"], "PDF")
            )

        if signal == "396":
            observed_396.append(
                (record["path"], "PDF")
            )

        if signal == "hh_b06_1":
            observed_hh.append(
                (record["path"], "PDF")
            )

        if signal == "approved_identifiers":
            observed_approved.append(
                (record["path"], "PDF")
            )


# Remove duplicates while preserving deterministic order.

def unique_pairs(items):
    return sorted(
        set(items),
        key=lambda x: (x[0], x[1])
    )


observed_395 = unique_pairs(observed_395)
observed_396 = unique_pairs(observed_396)
observed_hh = unique_pairs(observed_hh)
observed_none = unique_pairs(observed_none)
observed_approved = unique_pairs(observed_approved)


# ------------------------------------------------------------
# 8. EVIDENCE AVAILABILITY STATES
# ------------------------------------------------------------

pdf_records = [
    r for r in verification_records
    if r["artifact_type"] == "PDF"
]

notebook_records = [
    r for r in verification_records
    if r["artifact_type"] == "NOTEBOOK"
]

archive_records = [
    r for r in verification_records
    if r["artifact_type"] == "ZIP_ARCHIVE"
]


def readable_count(records):
    return sum(
        1
        for r in records
        if r.get("read_status") == "READ_SUCCESS"
    )


pdf_primary_state = (
    "PRIMARY_ARTIFACT_VERIFIED_FOR_TEXT_CONTENT"
    if readable_count(pdf_records) > 0
    else "PRIMARY_ARTIFACT_NOT_AVAILABLE_OR_NOT_READABLE"
)

notebook_primary_state = (
    "PRIMARY_ARTIFACT_VERIFIED_FOR_SOURCE_AND_OUTPUT_CONTENT"
    if readable_count(notebook_records) > 0
    else "PRIMARY_ARTIFACT_NOT_AVAILABLE_OR_NOT_READABLE"
)


# ------------------------------------------------------------
# 9. NO-RESOLUTION ASSERTION
# ------------------------------------------------------------

conflict_resolution_performed = False
canonical_source_declared = False

claim_boundary = {

    "inventory_scope_conflict_resolved":
        False,

    "identifier_conflict_resolved":
        False,

    "canonical_source_declared":
        False,

    "construct_authorized":
        False,

    "score_authorized":
        False,

    "outcome_authorized":
        False,

    "empirical_evaluation_authorized":
        False,

    "empirical_validity_established":
        False
}


# ------------------------------------------------------------
# 10. OUTPUT OBJECT
# ------------------------------------------------------------

timestamp = datetime.now(
    timezone.utc
).replace(
    microsecond=0
).isoformat()


c05_record = {

    "protocol":
        PROTOCOL,

    "cell_id":
        CELL_ID,

    "cell_name":
        CELL_NAME,

    "timestamp_utc":
        timestamp,

    "upstream":

        {
            "c04_status":
                c04_status,

            "c04_conflict_count":
                c04.get("conflict_count"),

            "c04_open_item_count":
                c04.get("open_item_count")
        },

    "source_environment":

        {
            "root":
                str(KAGGLE_INPUT),

            "candidate_file_count":
                len(candidate_paths),

            "candidate_files":
                [
                    str(p)
                    for p in candidate_paths
                ]
        },

    "artifact_summary":

        {
            "notebook_count":
                len(notebook_records),

            "notebook_readable":
                readable_count(notebook_records),

            "pdf_count":
                len(pdf_records),

            "pdf_readable":
                readable_count(pdf_records),

            "archive_count":
                len(archive_records),

            "archive_readable":
                readable_count(archive_records)
        },

    "verification_records":
        verification_records,

    "observed_evidence":

        {
            "395_mentions":
                observed_395,

            "396_mentions":
                observed_396,

            "hh_b06_1_mentions":
                observed_hh,

            "approved_identifiers_none":
                observed_none,

            "hh_b06_1_approved_signals":
                observed_approved
        },

    "evidence_state":

        {
            "malawi_notebook":
                notebook_primary_state,

            "malawi_pdf":
                pdf_primary_state,

            "396_signal_observed":
                bool(observed_396),

            "395_signal_observed":
                bool(observed_395),

            "hh_b06_1_signal_observed":
                bool(observed_hh),

            "approved_identifiers_none_signal_observed":
                bool(observed_none),

            "hh_b06_1_approved_signal_observed":
                bool(observed_approved)
        },

    "conflict_resolution_performed":
        conflict_resolution_performed,

    "canonical_source_declared":
        canonical_source_declared,

    "claim_boundary":
        claim_boundary,

    "governing_statement":
        (
            "C05 directly inspects Malawi-related artifacts available "
            "in the current execution environment and records observable "
            "source/code/output evidence. Presence of a number or phrase "
            "is recorded as an observed signal and is not by itself "
            "treated as semantic resolution. C05 does not resolve C04 "
            "conflicts, declare canonical authority, authorize constructs, "
            "scores, outcomes, or empirical evaluation."
        ),

    "next_cell":
        "C06 — PRIMARY ARTIFACT VERIFICATION (ETHIOPIA C21R2)"
}


# ------------------------------------------------------------
# 11. WRITE JSON
# ------------------------------------------------------------

with open(
    C05_JSON,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        c05_record,
        f,
        indent=2,
        ensure_ascii=False
    )


# ------------------------------------------------------------
# 12. WRITE HUMAN-READABLE REPORT
# ------------------------------------------------------------

lines = []

lines.append("=" * 72)
lines.append(
    "OIP v1.0.35 — C05: PRIMARY ARTIFACT VERIFICATION (MALAWI)"
)
lines.append("=" * 72)
lines.append("")

lines.append("UPSTREAM")
lines.append("--------")
lines.append(
    f"C04 register status          : {c04_status}"
)
lines.append(
    f"C04 conflict count           : "
    f"{c04.get('conflict_count')}"
)
lines.append(
    f"C04 open item count          : "
    f"{c04.get('open_item_count')}"
)
lines.append("")

lines.append("ARTIFACT DISCOVERY")
lines.append("------------------")
lines.append(
    f"Candidate files              : "
    f"{len(candidate_paths)}"
)
lines.append(
    f"Notebook artifacts           : "
    f"{len(notebook_records)}"
)
lines.append(
    f"Readable notebooks           : "
    f"{readable_count(notebook_records)}"
)
lines.append(
    f"PDF artifacts                : "
    f"{len(pdf_records)}"
)
lines.append(
    f"Readable PDFs                : "
    f"{readable_count(pdf_records)}"
)
lines.append(
    f"ZIP archives                 : "
    f"{len(archive_records)}"
)
lines.append("")

if candidate_paths:

    lines.append("CANDIDATE PATHS")
    lines.append("---------------")

    for p in candidate_paths:
        lines.append(str(p))

    lines.append("")


lines.append("OBSERVED EVIDENCE SIGNALS")
lines.append("-------------------------")
lines.append(
    f"395 observed                : "
    f"{bool(observed_395)}"
)
lines.append(
    f"396 observed                : "
    f"{bool(observed_396)}"
)
lines.append(
    f"hh_b06_1 observed           : "
    f"{bool(observed_hh)}"
)
lines.append(
    f"APPROVED IDENTIFIERS=NONE  : "
    f"{bool(observed_none)}"
)
lines.append(
    f"hh_b06_1 approved signal    : "
    f"{bool(observed_approved)}"
)
lines.append("")


lines.append("PRIMARY ARTIFACT STATES")
lines.append("-----------------------")
lines.append(
    f"Malawi notebook             : "
    f"{notebook_primary_state}"
)
lines.append(
    f"Malawi PDF                  : "
    f"{pdf_primary_state}"
)
lines.append("")


for record in verification_records:

    lines.append("-" * 72)
    lines.append(
        f"ARTIFACT: {record['path']}"
    )
    lines.append(
        f"TYPE    : {record['artifact_type']}"
    )
    lines.append(
        f"READ    : {record.get('read_status')}"
    )

    if record.get("sha256"):
        lines.append(
            f"SHA-256 : {record['sha256']}"
        )

    if record["artifact_type"] == "NOTEBOOK":

        lines.append(
            f"Cells   : {record.get('cells')}"
        )
        lines.append(
            f"Code    : {record.get('code_cells')}"
        )
        lines.append(
            f"Markdown: {record.get('markdown_cells')}"
        )
        lines.append(
            f"Executed: {record.get('executed_cells')}"
        )
        lines.append(
            f"Outputs present: "
            f"{record.get('execution_outputs_present')}"
        )

        if record.get("code_matches"):
            lines.append("")

            lines.append("CODE MATCHES:")

            for item in record["code_matches"]:

                lines.append(
                    f"  Cell {item['cell_index']} | "
                    f"{item['claim_signal']}"
                )

                for snippet in item["snippets"]:
                    lines.append(
                        "    " +
                        " ".join(
                            snippet.split()
                        )
                    )

        if record.get("output_matches"):
            lines.append("")

            lines.append("EXECUTION OUTPUT MATCHES:")

            for item in record["output_matches"]:

                lines.append(
                    f"  Cell {item['cell_index']} | "
                    f"{item['claim_signal']}"
                )

                for snippet in item["snippets"]:
                    lines.append(
                        "    " +
                        " ".join(
                            snippet.split()
                        )
                    )

        if record.get("markdown_matches"):
            lines.append("")

            lines.append("MARKDOWN MATCHES:")

            for item in record["markdown_matches"]:

                lines.append(
                    f"  Cell {item['cell_index']} | "
                    f"{item['claim_signal']}"
                )

                for snippet in item["snippets"]:
                    lines.append(
                        "    " +
                        " ".join(
                            snippet.split()
                        )
                    )

    elif record["artifact_type"] == "PDF":

        lines.append(
            f"Pages   : {record.get('page_count')}"
        )

        if record.get("matches"):

            lines.append("")
            lines.append("PDF MATCHES:")

            for item in record["matches"]:

                lines.append(
                    f"  Page {item['page']} | "
                    f"{item['claim_signal']}"
                )

                lines.append(
                    "    " +
                    " ".join(
                        item["snippet"].split()
                    )
                )

    elif record["artifact_type"] == "ZIP_ARCHIVE":

        if record.get("members"):

            lines.append("")
            lines.append("RELEVANT ARCHIVE MEMBERS:")

            for member in record["members"]:

                lines.append(
                    f"  {member['member']}"
                )


lines.append("")
lines.append("=" * 72)
lines.append("C05 GOVERNANCE BOUNDARY")
lines.append("=" * 72)
lines.append("")
lines.append(
    "Conflict resolution performed : NO"
)
lines.append(
    "Canonical source declared     : NO"
)
lines.append(
    "Construct authorization       : NO"
)
lines.append(
    "Score authorization           : NO"
)
lines.append(
    "Outcome authorization         : NO"
)
lines.append(
    "Empirical evaluation          : NO"
)
lines.append(
    "Empirical validity            : NO"
)
lines.append("")
lines.append(
    "C05 STATUS                    : PASS"
)
lines.append("")
lines.append(
    "PASS means the verification cell completed its controlled "
    "inspection procedure. It does NOT mean that either C04 "
    "conflict has been resolved."
)
lines.append("")
lines.append("Artifacts written:")
lines.append(
    f"  {C05_JSON}"
)
lines.append(
    f"  {C05_TXT}"
)
lines.append("")
lines.append(
    "Next:"
)
lines.append(
    "  C06 — PRIMARY ARTIFACT VERIFICATION (ETHIOPIA C21R2)"
)
lines.append("=" * 72)


with open(
    C05_TXT,
    "w",
    encoding="utf-8"
) as f:

    f.write(
        "\n".join(lines)
    )


# ------------------------------------------------------------
# 13. FINAL CONSOLE OUTPUT
# ------------------------------------------------------------

print("")
print(
    f"C04 register status          : {c04_status}"
)
print(
    f"Candidate Malawi files       : {len(candidate_paths)}"
)
print(
    f"Notebook artifacts           : {len(notebook_records)}"
)
print(
    f"Readable notebooks           : "
    f"{readable_count(notebook_records)}"
)
print(
    f"PDF artifacts                : {len(pdf_records)}"
)
print(
    f"Readable PDFs                : "
    f"{readable_count(pdf_records)}"
)
print("")
print(
    f"395 signal observed          : "
    f"{bool(observed_395)}"
)
print(
    f"396 signal observed          : "
    f"{bool(observed_396)}"
)
print(
    f"hh_b06_1 observed            : "
    f"{bool(observed_hh)}"
)
print(
    f"APPROVED IDENTIFIERS = NONE  : "
    f"{bool(observed_none)}"
)
print(
    f"hh_b06_1 approved signal     : "
    f"{bool(observed_approved)}"
)
print("")
print(
    "Conflict resolution performed : NO"
)
print(
    "Canonical source declared     : NO"
)
print("")
print(
    "C05 STATUS                   : PASS"
)
print("")
print("Artifacts written:")
print(f"  {C05_JSON}")
print(f"  {C05_TXT}")
print("")
print(
    "Next:"
)
print(
    "  C06 — PRIMARY ARTIFACT VERIFICATION (ETHIOPIA C21R2)"
)
print("=" * 72)
```

```python
# ============================================================
# OIP v1.0.35 — CELL 06
# PRIMARY ARTIFACT VERIFICATION (ETHIOPIA C21R2)
# ============================================================
#
# PURPOSE
# -------
# Locate and inspect Ethiopia-related primary artifacts available
# in the current Kaggle environment.
#
# C06 records observable artifact evidence only.
#
# C06 MUST NOT:
#   - resolve any C04 conflict
#   - declare a canonical source
#   - declare substantive authority
#   - convert signal presence into semantic verification
#   - resolve C11C/C12/C13R execution state
#   - infer C19 status from a generic "BLOCKED" string
#   - authorize constructs, scores, outcomes, or evaluation
#
# ============================================================

from pathlib import Path
from datetime import datetime, timezone
import json
import re
import hashlib
import io


# ------------------------------------------------------------
# 0. CONSTANTS
# ------------------------------------------------------------

PROTOCOL = "OIP v1.0.35"
CELL_ID = "C06"
CELL_NAME = "PRIMARY ARTIFACT VERIFICATION (ETHIOPIA C21R2)"

KAGGLE_INPUT = Path("/kaggle/input")
RUN_ROOT = Path("/kaggle/working/oip_v1_0_35")

C05_JSON = (
    RUN_ROOT
    / "c05_malawi_primary_verification"
    / "c05_malawi_primary_verification.json"
)

C06_DIR = (
    RUN_ROOT
    / "c06_ethiopia_primary_verification"
)

C06_DIR.mkdir(parents=True, exist_ok=True)

C06_JSON = (
    C06_DIR
    / "c06_ethiopia_primary_verification.json"
)

C06_TXT = (
    C06_DIR
    / "c06_ethiopia_primary_verification.txt"
)


# ------------------------------------------------------------
# 1. HEADER
# ------------------------------------------------------------

print("=" * 72)
print(f"{PROTOCOL} — {CELL_ID}: {CELL_NAME}")
print("=" * 72)


# ------------------------------------------------------------
# 2. LOAD C05
# ------------------------------------------------------------

if not C05_JSON.exists():
    raise FileNotFoundError(
        "C06 FAIL-CLOSED: required C05 artifact not found:\n"
        f"{C05_JSON}"
    )

with open(C05_JSON, "r", encoding="utf-8") as f:
    c05 = json.load(f)


if c05.get("protocol") != PROTOCOL:
    raise RuntimeError(
        "C06 FAIL-CLOSED: C05 protocol mismatch."
    )

if c05.get("cell_id") != "C05":
    raise RuntimeError(
        "C06 FAIL-CLOSED: unexpected C05 cell identifier."
    )


# ------------------------------------------------------------
# 3. SOURCE DISCOVERY TERMS
# ------------------------------------------------------------

ETHIOPIA_TERMS = (
    "ethiopia",
    "eth_",
    "ess4",
    "ess_",
    "v1_0_33",
    "v1-0-33",
    "c21r2",
    "c21",
    "c19",
    "c11c",
    "c12",
    "c13r",
    "audit-2"
)


# ------------------------------------------------------------
# 4. HELPERS
# ------------------------------------------------------------

def sha256_bytes(raw):
    return hashlib.sha256(raw).hexdigest()


def normalize_text(value):
    if value is None:
        return ""

    if isinstance(value, list):
        return "".join(str(x) for x in value)

    return str(value)


def context_snippets(
    text,
    pattern,
    window=220,
    max_snippets=10
):
    """
    Return surrounding context for an exact/controlled pattern.
    Presence is recorded as a signal only.
    """

    if not text:
        return []

    try:
        matches = list(
            re.finditer(
                pattern,
                text,
                flags=re.IGNORECASE | re.DOTALL
            )
        )
    except re.error:
        return []

    snippets = []

    for match in matches[:max_snippets]:

        start = max(
            0,
            match.start() - window
        )

        end = min(
            len(text),
            match.end() + window
        )

        snippets.append(
            text[start:end]
            .replace("\x00", " ")
            .strip()
        )

    return snippets


def inspect_notebook(path):
    """
    Inspect notebook source, markdown and execution outputs
    separately.
    """

    record = {

        "artifact_type":
            "NOTEBOOK",

        "path":
            str(path),

        "sha256":
            None,

        "read_status":
            None,

        "cell_count":
            0,

        "code_cell_count":
            0,

        "markdown_cell_count":
            0,

        "executed_code_cell_count":
            0,

        "execution_outputs_present":
            False,

        "code_signals":
            [],

        "markdown_signals":
            [],

        "output_signals":
            []
    }

    try:

        raw = path.read_bytes()

        record["sha256"] = sha256_bytes(raw)

        notebook = json.loads(
            raw.decode("utf-8")
        )

    except Exception as e:

        record["read_status"] = (
            f"READ_ERROR:{type(e).__name__}"
        )

        return record


    cells = notebook.get(
        "cells",
        []
    )

    record["cell_count"] = len(cells)


    # Controlled signals.
    #
    # Note that generic "23" is NOT searched globally.
    # Only contextual forms associated with the relevant
    # Ethiopia audit records are considered.

    patterns = {

        "C11C":
            r"\bC11C\b",

        "C12":
            r"\bC12\b",

        "C13R":
            r"\bC13R\b",

        "C21R2":
            r"\bC21R2\b",

        "C19":
            r"\bC19\b",

        "C19_BLOCKED":
            r"\bC19\b.{0,220}\bBLOCKED\b",

        "C19_BLOCKED_REVERSE":
            r"\bBLOCKED\b.{0,220}\bC19\b",

        "NOT_RUN":
            r"\bNOT[_\s-]?RUN\b",

        "RUN_STATUS_NOT_RUN":
            r"(?:RUN[_\s-]?STATUS|STATUS).{0,100}\bNOT[_\s-]?RUN\b",

        "1044":
            r"\b1044\b",

        "1021":
            r"\b1021\b",

        "1446":
            r"\b1446\b",

        "23_IDENTIFIER_CONTEXT":
            r"(?:structurally\s+unique|unique).{0,120}\b23\b",

        "23_CONTEXT_REVERSE":
            r"\b23\b.{0,120}(?:structurally\s+unique|unique)"
    }


    for idx, cell in enumerate(cells):

        cell_type = str(
            cell.get(
                "cell_type",
                ""
            )
        ).lower()

        source = normalize_text(
            cell.get(
                "source",
                []
            )
        )

        outputs = cell.get(
            "outputs",
            []
        )

        output_parts = []

        for output in outputs:

            text_value = output.get(
                "text",
                []
            )

            if text_value:
                output_parts.append(
                    normalize_text(
                        text_value
                    )
                )

            data = output.get(
                "data",
                {}
            )

            if isinstance(data, dict):

                for value in data.values():

                    if isinstance(value, str):
                        output_parts.append(
                            value
                        )

                    elif isinstance(value, list):
                        output_parts.append(
                            normalize_text(
                                value
                            )
                        )

        output_text = "\n".join(
            output_parts
        )


        if cell_type == "code":

            record["code_cell_count"] += 1

            if outputs:
                record[
                    "executed_code_cell_count"
                ] += 1

                record[
                    "execution_outputs_present"
                ] = True


            for signal_name, pattern in patterns.items():

                snippets = context_snippets(
                    source,
                    pattern
                )

                if snippets:

                    record[
                        "code_signals"
                    ].append({

                        "cell_index":
                            idx,

                        "signal":
                            signal_name,

                        "snippets":
                            snippets
                    })


            # Execution output is kept separate from source code.

            if output_text:

                for signal_name, pattern in patterns.items():

                    snippets = context_snippets(
                        output_text,
                        pattern
                    )

                    if snippets:

                        record[
                            "output_signals"
                        ].append({

                            "cell_index":
                                idx,

                            "signal":
                                signal_name,

                            "snippets":
                                snippets
                        })


        elif cell_type == "markdown":

            record[
                "markdown_cell_count"
            ] += 1


            for signal_name, pattern in patterns.items():

                snippets = context_snippets(
                    source,
                    pattern
                )

                if snippets:

                    record[
                        "markdown_signals"
                    ].append({

                        "cell_index":
                            idx,

                        "signal":
                            signal_name,

                        "snippets":
                            snippets
                    })


    record["read_status"] = "READ_SUCCESS"

    return record


def inspect_pdf(path):
    """
    Inspect PDF text when pypdf is available.
    """

    record = {

        "artifact_type":
            "PDF",

        "path":
            str(path),

        "sha256":
            None,

        "read_status":
            None,

        "page_count":
            None,

        "signals":
            []
    }

    try:

        raw = path.read_bytes()

        record["sha256"] = sha256_bytes(raw)

    except Exception as e:

        record["read_status"] = (
            f"READ_ERROR:{type(e).__name__}"
        )

        return record


    try:

        from pypdf import PdfReader

        reader = PdfReader(
            io.BytesIO(raw),
            strict=False
        )

        record["page_count"] = len(
            reader.pages
        )


        patterns = {

            "C11C":
                r"\bC11C\b",

            "C12":
                r"\bC12\b",

            "C13R":
                r"\bC13R\b",

            "C21R2":
                r"\bC21R2\b",

            "C19":
                r"\bC19\b",

            "C19_BLOCKED":
                r"\bC19\b.{0,220}\bBLOCKED\b",

            "NOT_RUN":
                r"\bNOT[_\s-]?RUN\b",

            "1044":
                r"\b1044\b",

            "1021":
                r"\b1021\b",

            "1446":
                r"\b1446\b"
        }


        for page_no, page in enumerate(
            reader.pages,
            start=1
        ):

            try:
                text = page.extract_text() or ""
            except Exception:
                text = ""


            for signal_name, pattern in patterns.items():

                snippets = context_snippets(
                    text,
                    pattern
                )

                for snippet in snippets:

                    record[
                        "signals"
                    ].append({

                        "page":
                            page_no,

                        "signal":
                            signal_name,

                        "snippet":
                            snippet
                    })


        record["read_status"] = (
            "READ_SUCCESS"
        )

    except ImportError:

        record["read_status"] = (
            "PDF_LIBRARY_NOT_AVAILABLE"
        )

    except Exception as e:

        record["read_status"] = (
            f"PDF_EXTRACTION_ERROR:{type(e).__name__}"
        )


    return record


# ------------------------------------------------------------
# 5. DISCOVER CURRENT SOURCE FILES
# ------------------------------------------------------------

all_files = []

for path in KAGGLE_INPUT.rglob("*"):

    try:

        if path.is_file():
            all_files.append(path)

    except OSError:
        continue


all_files = sorted(
    all_files,
    key=lambda p: str(p).lower()
)


candidate_paths = []

for path in all_files:

    lower_path = str(
        path
    ).lower()

    suffix = path.suffix.lower()

    # Direct Ethiopia-related path signal.
    if any(
        term in lower_path
        for term in ETHIOPIA_TERMS
    ):

        candidate_paths.append(
            path
        )

        continue


    # Secondary filename discovery for notebooks/PDFs.
    if suffix in {
        ".ipynb",
        ".pdf"
    }:

        name = path.name.lower()

        if (
            "audit" in name
            or "ess4" in name
            or "ethiopia" in name
            or "c21" in name
            or "c19" in name
            or "v1_0_33" in name
            or "v1-0-33" in name
        ):

            candidate_paths.append(
                path
            )


candidate_paths = sorted(
    set(candidate_paths),
    key=lambda p: str(p).lower()
)


# ------------------------------------------------------------
# 6. INSPECT CANDIDATES
# ------------------------------------------------------------

verification_records = []

for path in candidate_paths:

    suffix = path.suffix.lower()

    if suffix == ".ipynb":

        verification_records.append(
            inspect_notebook(path)
        )

    elif suffix == ".pdf":

        verification_records.append(
            inspect_pdf(path)
        )


# ------------------------------------------------------------
# 7. SUMMARIZE OBSERVED SIGNALS
# ------------------------------------------------------------

def signal_present(
    record,
    signal_name
):

    for group in (
        "code_signals",
        "markdown_signals",
        "output_signals",
        "signals"
    ):

        for item in record.get(
            group,
            []
        ):

            if item.get(
                "signal"
            ) == signal_name:

                return True

    return False


summary = {

    "C11C":
        False,

    "C12":
        False,

    "C13R":
        False,

    "C21R2":
        False,

    "C19":
        False,

    "C19_BLOCKED":
        False,

    "NOT_RUN":
        False,

    "1044":
        False,

    "1021":
        False,

    "1446":
        False,

    "23_identifier_context":
        False
}


for record in verification_records:

    for signal in summary.keys():

        if signal_present(
            record,
            signal
        ):

            summary[signal] = True


    # 23 has two controlled contextual forms.

    if (
        signal_present(
            record,
            "23_IDENTIFIER_CONTEXT"
        )
        or
        signal_present(
            record,
            "23_CONTEXT_REVERSE"
        )
    ):

        summary[
            "23_identifier_context"
        ] = True


# ------------------------------------------------------------
# 8. ARTIFACT STATE
# ------------------------------------------------------------

notebook_records = [
    r for r in verification_records
    if r["artifact_type"] == "NOTEBOOK"
]

pdf_records = [
    r for r in verification_records
    if r["artifact_type"] == "PDF"
]


readable_notebooks = sum(
    1
    for r in notebook_records
    if r.get("read_status")
    == "READ_SUCCESS"
)


readable_pdfs = sum(
    1
    for r in pdf_records
    if r.get("read_status")
    == "READ_SUCCESS"
)


if readable_notebooks > 0:

    notebook_state = (
        "PRIMARY_ARTIFACT_READABLE"
    )

else:

    notebook_state = (
        "PRIMARY_ARTIFACT_NOT_AVAILABLE_OR_NOT_READABLE"
    )


if readable_pdfs > 0:

    pdf_state = (
        "PRIMARY_ARTIFACT_READABLE"
    )

else:

    pdf_state = (
        "PRIMARY_ARTIFACT_NOT_AVAILABLE_OR_NOT_READABLE"
    )


# ------------------------------------------------------------
# 9. GOVERNANCE BOUNDARY
# ------------------------------------------------------------

claim_boundary = {

    "C11C_execution_state_resolved":
        False,

    "C12_execution_state_resolved":
        False,

    "C13R_execution_state_resolved":
        False,

    "C19_BLOCKED_state_resolved":
        False,

    "canonical_source_declared":
        False,

    "construct_authorized":
        False,

    "score_authorized":
        False,

    "outcome_authorized":
        False,

    "empirical_evaluation_authorized":
        False,

    "empirical_validity_established":
        False
}


# ------------------------------------------------------------
# 10. MACHINE-READABLE RECORD
# ------------------------------------------------------------

timestamp = (
    datetime.now(
        timezone.utc
    )
    .replace(
        microsecond=0
    )
    .isoformat()
)


c06_record = {

    "protocol":
        PROTOCOL,

    "cell_id":
        CELL_ID,

    "cell_name":
        CELL_NAME,

    "timestamp_utc":
        timestamp,

    "upstream":
        {
            "required_cell":
                "C05",

            "c05_protocol":
                c05.get("protocol"),

            "c05_cell_id":
                c05.get("cell_id")
        },

    "source_environment":
        {
            "root":
                str(KAGGLE_INPUT),

            "candidate_artifact_count":
                len(candidate_paths),

            "candidate_paths":
                [
                    str(p)
                    for p in candidate_paths
                ]
        },

    "artifact_state":
        {
            "notebook_count":
                len(notebook_records),

            "readable_notebook_count":
                readable_notebooks,

            "pdf_count":
                len(pdf_records),

            "readable_pdf_count":
                readable_pdfs,

            "notebook_state":
                notebook_state,

            "pdf_state":
                pdf_state
        },

    "verification_records":
        verification_records,

    "observed_signals":
        summary,

    "conflict_resolution_performed":
        False,

    "canonical_source_declared":
        False,

    "claim_boundary":
        claim_boundary,

    "governing_statement":
        (
            "C06 directly inspects Ethiopia-related artifacts "
            "available in the current execution environment. "
            "Signals are recorded with artifact and context "
            "information where available. Signal presence does "
            "not itself resolve execution state or semantic "
            "meaning. C06 does not resolve C04 items, declare "
            "canonical authority, authorize constructs, scores, "
            "outcomes, or empirical evaluation."
        ),

    "next_cell":
        "C07 — v1.0.28 PROVENANCE RECONCILIATION"
}


# ------------------------------------------------------------
# 11. WRITE JSON
# ------------------------------------------------------------

with open(
    C06_JSON,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        c06_record,
        f,
        indent=2,
        ensure_ascii=False
    )


# ------------------------------------------------------------
# 12. WRITE TXT
# ------------------------------------------------------------

lines = []

lines.append("=" * 72)
lines.append(
    "OIP v1.0.35 — C06: PRIMARY ARTIFACT VERIFICATION (ETHIOPIA C21R2)"
)
lines.append("=" * 72)
lines.append("")

lines.append("UPSTREAM")
lines.append("--------")
lines.append(
    f"C05 protocol                : "
    f"{c05.get('protocol')}"
)
lines.append(
    f"C05 cell                    : "
    f"{c05.get('cell_id')}"
)
lines.append("")

lines.append("ARTIFACT DISCOVERY")
lines.append("------------------")
lines.append(
    f"Candidate artifacts         : "
    f"{len(candidate_paths)}"
)
lines.append(
    f"Notebook artifacts          : "
    f"{len(notebook_records)}"
)
lines.append(
    f"Readable notebooks          : "
    f"{readable_notebooks}"
)
lines.append(
    f"PDF artifacts               : "
    f"{len(pdf_records)}"
)
lines.append(
    f"Readable PDFs               : "
    f"{readable_pdfs}"
)
lines.append("")

if candidate_paths:

    lines.append("CANDIDATE PATHS")
    lines.append("---------------")

    for p in candidate_paths:
        lines.append(
            str(p)
        )

    lines.append("")


lines.append("OBSERVED SIGNALS")
lines.append("----------------")

for key, value in summary.items():

    lines.append(
        f"{key:<32}: {value}"
    )

lines.append("")

lines.append("ARTIFACT STATES")
lines.append("---------------")
lines.append(
    f"Notebook state              : "
    f"{notebook_state}"
)
lines.append(
    f"PDF state                   : "
    f"{pdf_state}"
)
lines.append("")


# Detailed artifact evidence.

for record in verification_records:

    lines.append("-" * 72)

    lines.append(
        f"ARTIFACT: {record['path']}"
    )

    lines.append(
        f"TYPE    : {record['artifact_type']}"
    )

    lines.append(
        f"READ    : {record.get('read_status')}"
    )

    if record.get("sha256"):

        lines.append(
            f"SHA-256 : {record['sha256']}"
        )


    if record["artifact_type"] == "NOTEBOOK":

        lines.append(
            f"Cells   : "
            f"{record.get('cell_count')}"
        )

        lines.append(
            f"Code    : "
            f"{record.get('code_cell_count')}"
        )

        lines.append(
            f"Markdown: "
            f"{record.get('markdown_cell_count')}"
        )

        lines.append(
            f"Executed: "
            f"{record.get('executed_code_cell_count')}"
        )

        lines.append(
            f"Outputs : "
            f"{record.get('execution_outputs_present')}"
        )


        if record.get("code_signals"):

            lines.append("")
            lines.append(
                "CODE SIGNALS:"
            )

            for item in record[
                "code_signals"
            ]:

                lines.append(
                    f"  Cell {item['cell_index']} | "
                    f"{item['signal']}"
                )

                for snippet in item[
                    "snippets"
                ]:

                    lines.append(
                        "    "
                        + " ".join(
                            snippet.split()
                        )
                    )


        if record.get("output_signals"):

            lines.append("")
            lines.append(
                "EXECUTION OUTPUT SIGNALS:"
            )

            for item in record[
                "output_signals"
            ]:

                lines.append(
                    f"  Cell {item['cell_index']} | "
                    f"{item['signal']}"
                )

                for snippet in item[
                    "snippets"
                ]:

                    lines.append(
                        "    "
                        + " ".join(
                            snippet.split()
                        )
                    )


        if record.get("markdown_signals"):

            lines.append("")
            lines.append(
                "MARKDOWN SIGNALS:"
            )

            for item in record[
                "markdown_signals"
            ]:

                lines.append(
                    f"  Cell {item['cell_index']} | "
                    f"{item['signal']}"
                )

                for snippet in item[
                    "snippets"
                ]:

                    lines.append(
                        "    "
                        + " ".join(
                            snippet.split()
                        )
                    )


    elif record["artifact_type"] == "PDF":

        if record.get("page_count") is not None:

            lines.append(
                f"Pages   : "
                f"{record['page_count']}"
            )


        if record.get("signals"):

            lines.append("")
            lines.append(
                "PDF SIGNALS:"
            )

            for item in record[
                "signals"
            ]:

                lines.append(
                    f"  Page {item['page']} | "
                    f"{item['signal']}"
                )

                lines.append(
                    "    "
                    + " ".join(
                        item["snippet"].split()
                    )
                )


lines.append("")
lines.append("=" * 72)
lines.append(
    "C06 GOVERNANCE BOUNDARY"
)
lines.append("=" * 72)
lines.append("")
lines.append(
    "Conflict resolution performed : NO"
)
lines.append(
    "Canonical source declared     : NO"
)
lines.append(
    "Construct authorization       : NO"
)
lines.append(
    "Score authorization           : NO"
)
lines.append(
    "Outcome authorization         : NO"
)
lines.append(
    "Empirical evaluation          : NO"
)
lines.append(
    "Empirical validity            : NO"
)
lines.append("")

lines.append(
    "C06 STATUS                    : PASS"
)

lines.append("")
lines.append(
    "PASS means the controlled primary-artifact inspection "
    "procedure completed. It does not mean that any Ethiopia "
    "execution-state issue has been resolved."
)

lines.append("")
lines.append(
    "Artifacts written:"
)

lines.append(
    f"  {C06_JSON}"
)

lines.append(
    f"  {C06_TXT}"
)

lines.append("")
lines.append(
    "Next:"
)

lines.append(
    "  C07 — v1.0.28 PROVENANCE RECONCILIATION"
)

lines.append(
    "=" * 72
)


with open(
    C06_TXT,
    "w",
    encoding="utf-8"
) as f:

    f.write(
        "\n".join(lines)
    )


# ------------------------------------------------------------
# 13. FINAL CONSOLE OUTPUT
# ------------------------------------------------------------

print("")

print(
    f"C05 prerequisite             : "
    f"{c05.get('cell_id')}"
)

print(
    f"Candidate Ethiopia artifacts : "
    f"{len(candidate_paths)}"
)

print(
    f"Notebook artifacts           : "
    f"{len(notebook_records)}"
)

print(
    f"Readable notebooks           : "
    f"{readable_notebooks}"
)

print(
    f"PDF artifacts                : "
    f"{len(pdf_records)}"
)

print(
    f"Readable PDFs                : "
    f"{readable_pdfs}"
)

print("")

for key, value in summary.items():

    print(
        f"{key:<32}: {value}"
    )

print("")

print(
    "Conflict resolution performed : NO"
)

print(
    "Canonical source declared     : NO"
)

print("")

print(
    "C06 STATUS                   : PASS"
)

print("")

print("Artifacts written:")

print(
    f"  {C06_JSON}"
)

print(
    f"  {C06_TXT}"
)

print("")

print(
    "Next:"
)

print(
    "  C07 — v1.0.28 PROVENANCE RECONCILIATION"
)

print("=" * 72)
```

```python
# ============================================================
# OIP v1.0.35 — C07
# v1.0.28 PROVENANCE RECONCILIATION
#
# PURPOSE:
#   Record observable provenance/classification signals only.
#
# NON-NEGOTIABLE:
#   - No conflict resolution
#   - No canonical-source declaration
#   - No retraction declaration
#   - No synthetic-generator claim
#   - No empirical-validity authorization
#   - No score/outcome authorization
# ============================================================

from pathlib import Path
import json
import re
import hashlib
import sys

print("=" * 72)
print("OIP v1.0.35 — C07: v1.0.28 PROVENANCE RECONCILIATION")
print("=" * 72)

ROOT = Path("/kaggle/working/oip_v1_0_35")
C06_JSON = ROOT / "c06_ethiopia_primary_verification" / \
    "c06_ethiopia_primary_verification.json"

C01_JSON = ROOT / "c01_artifact_inventory" / \
    "c01_artifact_inventory.json"

OUT_DIR = ROOT / "c07_v1_0_28_provenance"
OUT_DIR.mkdir(parents=True, exist_ok=True)

C07_JSON = OUT_DIR / "c07_v1_0_28_provenance.json"
C07_TXT = OUT_DIR / "c07_v1_0_28_provenance.txt"


# ------------------------------------------------------------
# 1. REQUIRED UPSTREAM CHECKS
# ------------------------------------------------------------

if not C06_JSON.exists():
    raise FileNotFoundError(
        f"Required upstream artifact missing: {C06_JSON}"
    )

if not C01_JSON.exists():
    raise FileNotFoundError(
        f"Required upstream artifact missing: {C01_JSON}"
    )

with open(C06_JSON, "r", encoding="utf-8") as f:
    c06 = json.load(f)

with open(C01_JSON, "r", encoding="utf-8") as f:
    c01 = json.load(f)

print("C06 prerequisite : PRESENT")
print("C01 prerequisite : PRESENT")


# ------------------------------------------------------------
# 2. DISCOVER v1.0.28-RELATED ARTIFACTS
# ------------------------------------------------------------

records = c01.get("artifacts", [])

if not isinstance(records, list):
    raise RuntimeError(
        "C01 artifact inventory schema is not compatible: "
        "'artifacts' is not a list."
    )

keywords = (
    "v1_0_28",
    "v1-0-28",
    "empirical_report",
    "preliminary_empirical",
)

candidate_records = []

for rec in records:
    if not isinstance(rec, dict):
        continue

    path_value = str(
        rec.get("relative_path")
        or rec.get("path")
        or rec.get("artifact_path")
        or ""
    )

    lower = path_value.lower()

    if any(k in lower for k in keywords):
        candidate_records.append(rec)


print(f"v1.0.28 candidate artifacts : {len(candidate_records)}")


# ------------------------------------------------------------
# 3. SAFE TEXT EXTRACTION
# ------------------------------------------------------------

def extract_text(path):
    """
    Best-effort extraction.

    This function records observable text only.
    It does not infer provenance.
    """

    p = Path(path)

    if not p.exists():
        return ""

    suffix = p.suffix.lower()

    if suffix in {".txt", ".md", ".json", ".csv", ".py", ".ipynb"}:
        try:
            return p.read_text(
                encoding="utf-8",
                errors="ignore"
            )
        except Exception:
            return ""

    if suffix == ".pdf":
        try:
            import subprocess

            result = subprocess.run(
                ["pdftotext", "-layout", str(p), "-"],
                capture_output=True,
                text=True,
                timeout=60
            )

            if result.returncode == 0:
                return result.stdout

        except Exception:
            pass

    return ""


# ------------------------------------------------------------
# 4. COLLECT OBSERVABLE EVIDENCE
# ------------------------------------------------------------

evidence = []

for rec in candidate_records:

    path_value = str(
        rec.get("absolute_path")
        or rec.get("path")
        or rec.get("relative_path")
        or ""
    )

    # If C01 only stores relative paths, resolve against /kaggle/input.
    path_obj = Path(path_value)

    if not path_obj.is_absolute():
        possible = Path("/kaggle/input") / path_obj
        if possible.exists():
            path_obj = possible

    text = extract_text(path_obj)

    evidence.append({
        "path": str(path_obj),
        "suffix": path_obj.suffix.lower(),
        "exists": path_obj.exists(),
        "text_length": len(text),
        "contains_synthetic": bool(
            re.search(r"\bsynthetic\b", text, re.I)
        ),
        "contains_open_source_hr_benchmark": bool(
            re.search(
                r"Open[- ]Source\s+HR\s+Benchmark",
                text,
                re.I
            )
        ),
        "contains_n_54808": bool(
            re.search(r"\b54[, ]?808\b", text)
        ),
        "contains_np_random": bool(
            re.search(r"\bnp\s*\.\s*random\b", text, re.I)
        ),
        "contains_preliminary": bool(
            re.search(r"\bpreliminary\b", text, re.I)
        ),
        "contains_historical": bool(
            re.search(r"\bhistorical\b", text, re.I)
        ),
        "contains_retraction_word": bool(
            re.search(r"\bretract(?:ed|ion)?\b", text, re.I)
        ),
        "contains_classified": bool(
            re.search(
                r"\bclassif(?:ied|ication|y|ies)\b",
                text,
                re.I
            )
        ),
        "contains_7_96pp": bool(
            re.search(r"7\.96\s*pp", text, re.I)
        ),
        "contains_pvalue_3_59e_29": bool(
            re.search(
                r"3\.59\s*[x×*]?\s*10\s*(?:\^|[-−])?\s*29",
                text,
                re.I
            )
        ),
    })


# ------------------------------------------------------------
# 5. LINEAGE EVIDENCE
# ------------------------------------------------------------

lineage_candidates = []

for rec in records:

    path_value = str(
        rec.get("relative_path")
        or rec.get("path")
        or rec.get("artifact_path")
        or ""
    )

    lower = path_value.lower()

    if (
        "v1_0_34" in lower
        or "v1-0-34" in lower
        or "four_dataset" in lower
        or "evidence_boundary" in lower
    ):
        lineage_candidates.append(rec)


lineage_evidence = []

for rec in lineage_candidates:

    path_value = str(
        rec.get("absolute_path")
        or rec.get("path")
        or rec.get("relative_path")
        or ""
    )

    path_obj = Path(path_value)

    if not path_obj.is_absolute():
        possible = Path("/kaggle/input") / path_obj
        if possible.exists():
            path_obj = possible

    text = extract_text(path_obj)

    if not text:
        continue

    lineage_evidence.append({
        "path": str(path_obj),
        "contains_synthetic": bool(
            re.search(r"\bsynthetic\b", text, re.I)
        ),
        "contains_classified": bool(
            re.search(
                r"\bclassif(?:ied|ication|y|ies)\b",
                text,
                re.I
            )
        ),
        "contains_retraction_word": bool(
            re.search(r"\bretract(?:ed|ion)?\b", text, re.I)
        ),
        "contains_no_formal_retraction_language": bool(
            re.search(
                r"no\s+formal\s+retraction|formal\s+retraction.*not",
                text,
                re.I
            )
        ),
    })


# ------------------------------------------------------------
# 6. GOVERNANCE BOUNDARY
# ------------------------------------------------------------

result = {
    "protocol": "OIP v1.0.35",
    "cell_id": "C07",
    "cell": "v1.0.28 PROVENANCE RECONCILIATION",

    "upstream": {
        "c06_present": True,
        "c01_present": True,
    },

    "candidate_artifact_count": len(candidate_records),

    "v1_0_28_evidence": evidence,

    "lineage_evidence": lineage_evidence,

    "governance": {
        "conflict_resolution_performed": False,
        "canonical_source_declared": False,
        "formal_retraction_declared": False,
        "synthetic_generator_verified": False,
        "provenance_reconciled": False,
        "construct_authorization": False,
        "score_authorization": False,
        "outcome_authorization": False,
        "empirical_evaluation_authorization": False,
        "empirical_validity_claim": False,
    },

    "claim_boundary": {
        "original_report_provenance": "OBSERVATION_ONLY",
        "later_lineage_classification": "OBSERVATION_ONLY",
        "generator_artifact": "NOT_ESTABLISHED",
        "formal_retraction": "NOT_ESTABLISHED",
        "current_independent_validation_use": "NOT_AUTHORIZED",
    },

    "status": "PASS",
    "next_cell": "C09 — FINAL CONFLICT CONSISTENCY CHECK",
}


# ------------------------------------------------------------
# 7. WRITE JSON
# ------------------------------------------------------------

with open(C07_JSON, "w", encoding="utf-8") as f:
    json.dump(
        result,
        f,
        indent=2,
        ensure_ascii=False
    )


# ------------------------------------------------------------
# 8. WRITE HUMAN-READABLE TXT
# ------------------------------------------------------------

lines = [
    "=" * 72,
    "OIP v1.0.35 — C07: v1.0.28 PROVENANCE RECONCILIATION",
    "=" * 72,
    "",
    "C06 prerequisite              : PRESENT",
    "C01 prerequisite              : PRESENT",
    f"Candidate artifacts           : {len(candidate_records)}",
    "",
    "Conflict resolution           : NO",
    "Canonical source declared     : NO",
    "Formal retraction declared    : NO",
    "Synthetic generator verified  : NO",
    "Provenance reconciled         : NO",
    "Construct authorization       : NO",
    "Score authorization           : NO",
    "Outcome authorization         : NO",
    "Empirical evaluation          : NO",
    "Empirical validity claim      : NO",
    "",
    "Current provenance status     : OBSERVATION ONLY",
    "Formal retraction             : NOT ESTABLISHED",
    "Generator artifact            : NOT ESTABLISHED",
    "Independent validation use    : NOT AUTHORIZED",
    "",
    "C07 STATUS                    : PASS",
    "",
    f"JSON : {C07_JSON}",
    f"TXT  : {C07_TXT}",
    "",
    "Next: C09 — FINAL CONFLICT CONSISTENCY CHECK",
    "=" * 72,
]

with open(C07_TXT, "w", encoding="utf-8") as f:
    f.write("\n".join(lines))


# ------------------------------------------------------------
# 9. FINAL EXISTENCE CHECK
# ------------------------------------------------------------

if not C07_JSON.exists():
    raise RuntimeError(
        "C07 failed to create its required JSON artifact."
    )

if not C07_TXT.exists():
    raise RuntimeError(
        "C07 failed to create its required TXT artifact."
    )


print("\n" + "\n".join(lines))
```

```python
# ============================================================
# OIP v1.0.35 — C09
# FINAL CONFLICT CONSISTENCY CHECK
# ============================================================
#
# PURPOSE
# -------
# Final consistency gate before C10.
#
# This cell:
#   - loads the actual C04-C08 JSON artifacts
#   - checks prerequisite status
#   - checks unresolved governance boundaries
#   - checks that no conflict was resolved
#   - checks that no canonical source/formula was declared
#   - checks that no construct/score/outcome/evaluation
#     authorization was introduced
#   - checks the v1.0.28 and v1.0.26 claim boundaries
#
# FAIL-CLOSED:
#   Any forbidden TRUE authorization/resolution state => FAIL
#
# This cell DOES NOT resolve anything.
# ============================================================

import json
from pathlib import Path

print("=" * 72)
print("OIP v1.0.35 — C09: FINAL CONFLICT CONSISTENCY CHECK")
print("=" * 72)

# ------------------------------------------------------------
# 1. BASE + ACTUAL VERIFIED PATHS
# ------------------------------------------------------------

BASE = Path("/kaggle/working/oip_v1_0_35")

JSON_PATHS = {
    "C04": BASE / "c04_conflict_register"
                     / "c04_conflict_register.json",

    "C05": BASE / "c05_malawi_primary_verification"
                     / "c05_malawi_primary_verification.json",

    "C06": BASE / "c06_ethiopia_primary_verification"
                     / "c06_ethiopia_primary_verification.json",

    "C07": BASE / "c07_v1_0_28_provenance"
                     / "c07_v1_0_28_provenance.json",

    "C08": BASE / "c08_v1_0_26_provenance"
                     / "c08_v1_0_26_provenance.json",
}

# ------------------------------------------------------------
# 2. LOAD JSON
# ------------------------------------------------------------

def load_json(path, label):

    if not path.exists():
        raise FileNotFoundError(
            f"{label} JSON not found: {path}"
        )

    with open(path, "r", encoding="utf-8") as f:
        obj = json.load(f)

    if not isinstance(obj, dict):
        raise TypeError(
            f"{label} JSON root must be a dictionary."
        )

    return obj


docs = {}

for cell, path in JSON_PATHS.items():

    docs[cell] = load_json(
        path,
        cell
    )

# ------------------------------------------------------------
# 3. HELPERS
# ------------------------------------------------------------

def find_nested_key(obj, target_key):

    found = []

    if isinstance(obj, dict):

        for key, value in obj.items():

            if key == target_key:
                found.append(value)

            found.extend(
                find_nested_key(
                    value,
                    target_key
                )
            )

    elif isinstance(obj, list):

        for item in obj:

            found.extend(
                find_nested_key(
                    item,
                    target_key
                )
            )

    return found


def first_nested_value(obj, keys):

    for key in keys:

        values = find_nested_key(
            obj,
            key
        )

        for value in values:

            if value is not None:
                return value

    return None


def scalar_status(obj):

    return first_nested_value(
        obj,
        [
            "status",
            "cell_status",
            "overall_status",
            "final_status",
        ]
    )


def contains_forbidden_true(
    obj,
    forbidden_keys
):

    hits = []

    def walk(value, path="root"):

        if isinstance(value, dict):

            for key, item in value.items():

                current_path = (
                    f"{path}.{key}"
                )

                if (
                    key in forbidden_keys
                    and item is True
                ):

                    hits.append(
                        current_path
                    )

                walk(
                    item,
                    current_path
                )

        elif isinstance(value, list):

            for i, item in enumerate(value):

                walk(
                    item,
                    f"{path}[{i}]"
                )

    walk(obj)

    return hits


# ------------------------------------------------------------
# 4. CHECK ENGINE
# ------------------------------------------------------------

checks = []


def check(
    name,
    condition,
    detail=""
):

    checks.append(
        {
            "name": name,
            "passed": bool(condition),
            "detail": detail,
        }
    )


# ------------------------------------------------------------
# 5. INPUT ARTIFACT CHECK
# ------------------------------------------------------------

for cell, path in JSON_PATHS.items():

    check(
        f"{cell} JSON exists",
        path.exists(),
        str(path),
    )


# ------------------------------------------------------------
# 6. C07 / C08 STATUS
# ------------------------------------------------------------

c07_status = scalar_status(
    docs["C07"]
)

c08_status = scalar_status(
    docs["C08"]
)

check(
    "C07 status is PASS",
    c07_status == "PASS",
    f"observed={c07_status!r}",
)

check(
    "C08 status is PASS",
    c08_status == "PASS",
    f"observed={c08_status!r}",
)


# ------------------------------------------------------------
# 7. C07 GOVERNANCE BOUNDARY
# ------------------------------------------------------------

C07_FORBIDDEN_TRUE = [

    "conflict_resolution_performed",

    "canonical_source_declared",

    "formal_retraction_declared",

    "synthetic_generator_verified",

    "provenance_reconciled",

    "construct_authorization",

    "score_authorization",

    "outcome_authorization",

    "empirical_evaluation_authorization",

    "empirical_validity_claim",
]


for key in C07_FORBIDDEN_TRUE:

    values = find_nested_key(
        docs["C07"],
        key
    )

    if values:

        check(
            f"C07 {key} remains FALSE",
            all(
                value is False
                for value in values
            ),
            f"observed={values}",
        )


# ------------------------------------------------------------
# 8. C07 CLAIM BOUNDARY
# ------------------------------------------------------------

C07_EXPECTED = {

    "generator_artifact":
        "NOT_ESTABLISHED",

    "formal_retraction":
        "NOT_ESTABLISHED",

    "current_independent_validation_use":
        "NOT_AUTHORIZED",
}


for key, expected in C07_EXPECTED.items():

    values = find_nested_key(
        docs["C07"],
        key
    )

    if values:

        check(
            f"C07 claim boundary: {key}",
            expected in values,
            (
                f"expected={expected!r}; "
                f"observed={values}"
            ),
        )


# ------------------------------------------------------------
# 9. C08 GOVERNANCE BOUNDARY
# ------------------------------------------------------------

C08_FORBIDDEN_TRUE = [

    "conflict_resolution_performed",

    "canonical_source_declared",

    "legal_nda_adjudication",

    "authorship_adjudication",

    "formula_version_lock",

    "formula_conflict_resolution",

    "substantive_authority_assignment",

    "construct_authorization",

    "score_authorization",

    "outcome_authorization",

    "empirical_evaluation_authorization",

    "empirical_validity_claim",
]


for key in C08_FORBIDDEN_TRUE:

    values = find_nested_key(
        docs["C08"],
        key
    )

    if values:

        check(
            f"C08 {key} remains FALSE",
            all(
                value is False
                for value in values
            ),
            f"observed={values}",
        )


# ------------------------------------------------------------
# 10. C08 CLAIM BOUNDARY
# ------------------------------------------------------------

C08_EXPECTED = {

    "legal_nda_existence":
        "NOT_ESTABLISHED_BY_SUPPLIED_SOURCE_SET",

    "nda_with_openai_existence":
        "NOT_ESTABLISHED_BY_SUPPLIED_SOURCE_SET",

    "assistant_text_authorship":
        "NOT_ESTABLISHED",

    "formula_conflict_state":
        "UNRESOLVED",

    "canonical_formula":
        "NOT_ESTABLISHED",

    "substantive_authority":
        "NOT_ESTABLISHED",

    "empirical_validity":
        "NOT_AUTHORIZED",
}


for key, expected in C08_EXPECTED.items():

    values = find_nested_key(
        docs["C08"],
        key
    )

    if values:

        check(
            f"C08 claim boundary: {key}",
            expected in values,
            (
                f"expected={expected!r}; "
                f"observed={values}"
            ),
        )


# ------------------------------------------------------------
# 11. C04 — NO RESOLUTION / AUTHORIZATION
# ------------------------------------------------------------

C04_FORBIDDEN_TRUE = [

    "conflict_resolution",

    "conflict_resolution_performed",

    "resolution_performed",

    "canonical_source_declared",

    "construct_authorization",

    "score_authorization",

    "outcome_authorization",

    "empirical_evaluation",

    "empirical_validity",
]


for key in C04_FORBIDDEN_TRUE:

    values = find_nested_key(
        docs["C04"],
        key
    )

    if values:

        true_values = [
            value
            for value in values
            if value is True
        ]

        check(
            f"C04 {key} has no TRUE resolution state",
            len(true_values) == 0,
            f"observed={values}",
        )


# ------------------------------------------------------------
# 12. C05 / C06 OPTIONAL GOVERNANCE CHECK
# ------------------------------------------------------------

OPTIONAL_FORBIDDEN_TRUE = [

    "conflict_resolution_performed",

    "canonical_source_declared",

    "construct_authorization",

    "score_authorization",

    "outcome_authorization",

    "empirical_evaluation_authorization",

    "empirical_validity_claim",
]


for cell in ["C05", "C06"]:

    for key in OPTIONAL_FORBIDDEN_TRUE:

        values = find_nested_key(
            docs[cell],
            key
        )

        if values:

            true_values = [
                value
                for value in values
                if value is True
            ]

            check(
                f"{cell} {key} has no forbidden TRUE state",
                len(true_values) == 0,
                f"observed={values}",
            )


# ------------------------------------------------------------
# 13. GLOBAL FORBIDDEN TRUE SCAN
# ------------------------------------------------------------

GLOBAL_FORBIDDEN_TRUE = set(
    C07_FORBIDDEN_TRUE
    + C08_FORBIDDEN_TRUE
    + C04_FORBIDDEN_TRUE
)


for cell, doc in docs.items():

    hits = contains_forbidden_true(
        doc,
        GLOBAL_FORBIDDEN_TRUE
    )

    check(
        f"{cell} global forbidden-TRUE scan",
        len(hits) == 0,
        f"hits={hits}",
    )


# ------------------------------------------------------------
# 14. FORMULA CONFLICT MUST REMAIN UNRESOLVED
# ------------------------------------------------------------

formula_values = find_nested_key(
    docs["C08"],
    "formula_conflict_state"
)

if formula_values:

    check(
        "Formula conflict remains UNRESOLVED",
        "UNRESOLVED" in formula_values,
        f"observed={formula_values}",
    )


# ------------------------------------------------------------
# 15. CANONICAL FORMULA MUST NOT BE ESTABLISHED
# ------------------------------------------------------------

canonical_values = find_nested_key(
    docs["C08"],
    "canonical_formula"
)

if canonical_values:

    check(
        "Canonical formula remains NOT_ESTABLISHED",
        "NOT_ESTABLISHED" in canonical_values,
        f"observed={canonical_values}",
    )


# ------------------------------------------------------------
# 16. v1.0.28 PROVENANCE MUST REMAIN UNRECONCILED
# ------------------------------------------------------------

provenance_values = find_nested_key(
    docs["C07"],
    "provenance_reconciled"
)

if provenance_values:

    check(
        "v1.0.28 provenance remains unreconciled",
        all(
            value is False
            for value in provenance_values
        ),
        f"observed={provenance_values}",
    )


# ------------------------------------------------------------
# 17. INDEPENDENT VALIDATION USE
# ------------------------------------------------------------

validation_values = find_nested_key(
    docs["C07"],
    "current_independent_validation_use"
)

if validation_values:

    check(
        "v1.0.28 independent validation remains unauthorized",
        "NOT_AUTHORIZED" in validation_values,
        f"observed={validation_values}",
    )


# ------------------------------------------------------------
# 18. LEGAL NDA ADJUDICATION
# ------------------------------------------------------------

nda_values = find_nested_key(
    docs["C08"],
    "legal_nda_adjudication"
)

if nda_values:

    check(
        "Legal NDA adjudication remains FALSE",
        all(
            value is False
            for value in nda_values
        ),
        f"observed={nda_values}",
    )


# ------------------------------------------------------------
# 19. FINAL RESULT
# ------------------------------------------------------------

failed_checks = [
    item
    for item in checks
    if not item["passed"]
]

passed_checks = [
    item
    for item in checks
    if item["passed"]
]

overall_pass = (
    len(failed_checks) == 0
)


# ------------------------------------------------------------
# 20. OUTPUT DIRECTORY
# ------------------------------------------------------------

OUT_DIR = (
    BASE /
    "c09_final_conflict_consistency"
)

OUT_DIR.mkdir(
    parents=True,
    exist_ok=True
)

JSON_OUT = (
    OUT_DIR /
    "c09_final_conflict_consistency.json"
)

TXT_OUT = (
    OUT_DIR /
    "c09_final_conflict_consistency.txt"
)


# ------------------------------------------------------------
# 21. OUTPUT JSON
# ------------------------------------------------------------

output = {

    "protocol":
        "OIP v1.0.35",

    "cell_id":
        "C09",

    "cell_name":
        "FINAL CONFLICT CONSISTENCY CHECK",

    "input_artifacts": {
        cell: str(path)
        for cell, path in JSON_PATHS.items()
    },

    "checks_performed":
        len(checks),

    "checks_passed":
        len(passed_checks),

    "checks_failed":
        len(failed_checks),

    "failed_checks":
        failed_checks,

    "governance_boundary": {

        "conflict_resolution_performed":
            False,

        "canonical_source_declared":
            False,

        "construct_authorization":
            False,

        "score_authorization":
            False,

        "outcome_authorization":
            False,

        "empirical_evaluation_authorization":
            False,

        "empirical_validity_claim":
            False,
    },

    "status":
        "PASS"
        if overall_pass
        else
        "FAIL",

    "next_step":
        (
            "C10 — ARCHIVAL FREEZE PRECONDITION CHECK"
            if overall_pass
            else
            "STOP — DO NOT PROCEED TO C10"
        ),
}


with open(
    JSON_OUT,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        output,
        f,
        indent=2,
        ensure_ascii=False
    )


# ------------------------------------------------------------
# 22. OUTPUT TXT
# ------------------------------------------------------------

lines = []

lines.append("=" * 72)

lines.append(
    "OIP v1.0.35 — C09: FINAL CONFLICT CONSISTENCY CHECK"
)

lines.append("=" * 72)

lines.append("")

lines.append("Input JSON artifacts:")

for cell, path in JSON_PATHS.items():

    lines.append(
        f"  {cell} : {path}"
    )

lines.append("")

lines.append(
    f"Checks performed : {len(checks)}"
)

lines.append(
    f"Checks passed    : {len(passed_checks)}"
)

lines.append(
    f"Checks failed    : {len(failed_checks)}"
)

lines.append("")

lines.append(
    "Governance boundary:"
)

lines.append(
    "  Conflict resolution       : NO"
)

lines.append(
    "  Canonical source declared : NO"
)

lines.append(
    "  Construct authorization    : NO"
)

lines.append(
    "  Score authorization       : NO"
)

lines.append(
    "  Outcome authorization     : NO"
)

lines.append(
    "  Empirical evaluation      : NO"
)

lines.append(
    "  Empirical validity claim  : NO"
)

lines.append("")


if failed_checks:

    lines.append(
        "FAILED CHECKS"
    )

    lines.append(
        "-" * 72
    )

    for item in failed_checks:

        lines.append(
            f"[FAIL] {item['name']}"
        )

        if item["detail"]:

            lines.append(
                f"       {item['detail']}"
            )

    lines.append("")

    lines.append(
        "C09 STATUS : FAIL"
    )

    lines.append(
        "STOP — DO NOT PROCEED TO C10."
    )

else:

    lines.append(
        "FAILED CHECKS : NONE"
    )

    lines.append("")

    lines.append(
        "C09 STATUS : PASS"
    )

    lines.append("")

    lines.append(
        "Next: C10 — ARCHIVAL FREEZE PRECONDITION CHECK"
    )


with open(
    TXT_OUT,
    "w",
    encoding="utf-8"
) as f:

    f.write(
        "\n".join(lines)
    )


# ------------------------------------------------------------
# 23. CONSOLE OUTPUT
# ------------------------------------------------------------

print("")

print(
    f"Checks performed                  : {len(checks)}"
)

print(
    f"Checks passed                     : {len(passed_checks)}"
)

print(
    f"Checks failed                     : {len(failed_checks)}"
)

print("")

print(
    "Governance boundary:"
)

print(
    "  Conflict resolution             : NO"
)

print(
    "  Canonical source declared      : NO"
)

print(
    "  Construct authorization         : NO"
)

print(
    "  Score authorization             : NO"
)

print(
    "  Outcome authorization           : NO"
)

print(
    "  Empirical evaluation            : NO"
)

print(
    "  Empirical validity claim        : NO"
)

print("")


if failed_checks:

    print(
        "FAILED CHECKS"
    )

    print(
        "-" * 72
    )

    for item in failed_checks:

        print(
            f"[FAIL] {item['name']}"
        )

        if item["detail"]:

            print(
                f"       {item['detail']}"
            )

    print("")

    print(
        "C09 STATUS : FAIL"
    )

    print(
        "STOP — DO NOT PROCEED TO C10."
    )

    raise RuntimeError(
        "C09 FAILED. "
        "Do not proceed to C10."
    )


else:

    print(
        "FAILED CHECKS                    : NONE"
    )

    print("")

    print(
        "C09 STATUS                      : PASS"
    )

    print("")

    print(
        "Next: C10 — ARCHIVAL FREEZE PRECONDITION CHECK"
    )

    print("")

    print(
        "Artifacts:"
    )

    print(
        f"  {JSON_OUT}"
    )

    print(
        f"  {TXT_OUT}"
    )

print("=" * 72)
```

```python
# ============================================================
# OIP v1.0.35 — C10
# HARDENED ARCHIVAL FREEZE PRECONDITION CHECK
# ============================================================
#
# PURPOSE
# -------
# Verify that C00-C09 form a complete, internally consistent,
# fail-closed reconciliation chain before archival freeze.
#
# C10 DOES NOT:
#   - resolve conflicts
#   - select canonical sources
#   - select canonical formulas
#   - authorize constructs
#   - authorize scores
#   - authorize outcomes
#   - perform empirical evaluation
#   - establish empirical validity
#   - create the final archival freeze manifest
#
# FAIL-CLOSED RULE
# ----------------
# Missing / unknown / ambiguous required state => FAIL
# No default PASS is permitted.
# ============================================================

from pathlib import Path
from datetime import datetime, timezone
import json
import hashlib


print("=" * 72)
print("OIP v1.0.35 — C10: HARDENED ARCHIVAL FREEZE PRECONDITION CHECK")
print("=" * 72)


# ------------------------------------------------------------
# 1. ROOT
# ------------------------------------------------------------

ROOT = Path(
    "/kaggle/working/oip_v1_0_35"
)


# ------------------------------------------------------------
# 2. VERIFIED UPSTREAM ARTIFACT PATHS
# ------------------------------------------------------------

UPSTREAM = {

    "C00":
        ROOT /
        "c00_source_scope" /
        "c00_source_scope.json",

    "C01":
        ROOT /
        "c01_artifact_inventory" /
        "c01_artifact_inventory.json",

    "C02":
        ROOT /
        "c02_artifact_identity" /
        "c02_artifact_identity.json",

    "C03":
        ROOT /
        "c03_authority_assignment" /
        "c03_authority_assignment.json",

    "C04":
        ROOT /
        "c04_conflict_register" /
        "c04_conflict_register.json",

    "C05":
        ROOT /
        "c05_malawi_primary_verification" /
        "c05_malawi_primary_verification.json",

    "C06":
        ROOT /
        "c06_ethiopia_primary_verification" /
        "c06_ethiopia_primary_verification.json",

    "C07":
        ROOT /
        "c07_v1_0_28_provenance" /
        "c07_v1_0_28_provenance.json",

    "C08":
        ROOT /
        "c08_v1_0_26_provenance" /
        "c08_v1_0_26_provenance.json",

    "C09":
        ROOT /
        "c09_final_conflict_consistency" /
        "c09_final_conflict_consistency.json",
}


# ------------------------------------------------------------
# 3. OUTPUT PATHS
# ------------------------------------------------------------

OUT_DIR = (
    ROOT /
    "c10_archival_freeze_precondition"
)

OUT_DIR.mkdir(
    parents=True,
    exist_ok=True
)

OUT_JSON = (
    OUT_DIR /
    "c10_archival_freeze_precondition.json"
)

OUT_TXT = (
    OUT_DIR /
    "c10_archival_freeze_precondition.txt"
)


# ------------------------------------------------------------
# 4. CHECK ENGINE
# ------------------------------------------------------------

checks = []


def add_check(
    check_id,
    expected,
    observed,
    detail=""
):

    passed = (
        observed == expected
    )

    checks.append(
        {
            "id": check_id,
            "expected": expected,
            "observed": observed,
            "passed": passed,
            "detail": detail,
        }
    )


# ------------------------------------------------------------
# 5. LOAD UPSTREAM ARTIFACTS
# ------------------------------------------------------------

data = {}

missing = []

load_errors = []


for cell, path in UPSTREAM.items():

    if not path.exists():

        missing.append(
            {
                "cell": cell,
                "path": str(path),
            }
        )

        continue


    try:

        with open(
            path,
            "r",
            encoding="utf-8"
        ) as f:

            obj = json.load(f)


        if not isinstance(
            obj,
            dict
        ):

            raise TypeError(
                "JSON root is not a dictionary."
            )


        data[cell] = obj


    except Exception as exc:

        load_errors.append(
            {
                "cell": cell,
                "path": str(path),
                "error": repr(exc),
            }
        )


# ------------------------------------------------------------
# 6. BASIC INPUT CHECKS
# ------------------------------------------------------------

add_check(
    "upstream_missing_count",
    0,
    len(missing),
    f"missing={missing}",
)

add_check(
    "upstream_load_error_count",
    0,
    len(load_errors),
    f"errors={load_errors}",
)


# ------------------------------------------------------------
# 7. ALL C00-C09 MUST BE PRESENT
# ------------------------------------------------------------

expected_cells = list(
    UPSTREAM.keys()
)

observed_cells = sorted(
    data.keys()
)

add_check(
    "all_C00_C09_loaded",
    sorted(expected_cells),
    observed_cells,
)


# ------------------------------------------------------------
# 8. STATUS EXTRACTION
# ------------------------------------------------------------

def extract_explicit_status(
    cell,
    record
):

    # No generic fallback to PASS.
    #
    # A status must be explicitly present
    # in the artifact.

    if cell == "C02":

        candidates = [
            "identity_status",
            "status",
        ]

    else:

        candidates = [
            "status",
            "cell_status",
            "overall_status",
            "final_status",
        ]


    for key in candidates:

        if key in record:

            value = record[key]

            if isinstance(
                value,
                str
            ):

                return value


    return None


# ------------------------------------------------------------
# 9. REQUIRED STATUS CHECKS
# ------------------------------------------------------------

status_records = {}


for cell in expected_cells:

    if cell not in data:

        continue


    status = extract_explicit_status(
        cell,
        data[cell]
    )


    status_records[cell] = status


    add_check(
        f"{cell}_explicit_status_PASS",
        "PASS",
        status,
        (
            "Explicit status required. "
            "Missing/unknown status fails closed."
        ),
    )


# ------------------------------------------------------------
# 10. C09 SPECIFIC INTEGRITY
# ------------------------------------------------------------

if "C09" in data:

    c09 = data["C09"]


    failed_checks = c09.get(
        "failed_checks",
        None
    )


    if isinstance(
        failed_checks,
        list
    ):

        add_check(
            "C09_failed_checks_zero",
            0,
            len(failed_checks),
        )

    else:

        add_check(
            "C09_failed_checks_zero",
            0,
            None,
            (
                "C09 failed_checks field "
                "is missing or not a list."
            ),
        )


    c09_checks_failed = c09.get(
        "checks_failed",
        None
    )


    add_check(
        "C09_checks_failed_zero",
        0,
        c09_checks_failed,
        (
            "C09 checks_failed must explicitly "
            "equal zero."
        ),
    )


    c09_status = c09.get(
        "status",
        None
    )


    add_check(
        "C09_recorded_status_PASS",
        "PASS",
        c09_status,
    )


# ------------------------------------------------------------
# 11. RECURSIVE FORBIDDEN TRUE CHECK
# ------------------------------------------------------------

FORBIDDEN_TRUE_KEYS = {

    "conflict_resolution_performed",

    "canonical_source_declared",

    "construct_authorization",

    "score_authorization",

    "outcome_authorization",

    "empirical_evaluation_authorization",

    "empirical_validity_claim",

    "empirical_validity_claim_authorized",

    "substantive_authority_assigned",

    "substantive_authority_assignment",

    "legal_nda_adjudication",

    "authorship_adjudication",

    "formula_version_lock",

    "formula_conflict_resolution",

    "provenance_reconciled",

    "synthetic_generator_verified",

    "formal_retraction_declared",

    "reclassification_adjudicated",
}


def find_forbidden_true(
    obj,
    path="root"
):

    findings = []


    if isinstance(
        obj,
        dict
    ):

        for key, value in obj.items():

            current_path = (
                f"{path}.{key}"
            )


            if (
                key in FORBIDDEN_TRUE_KEYS
                and value is True
            ):

                findings.append(
                    current_path
                )


            findings.extend(
                find_forbidden_true(
                    value,
                    current_path
                )
            )


    elif isinstance(
        obj,
        list
    ):

        for index, value in enumerate(
            obj
        ):

            findings.extend(
                find_forbidden_true(
                    value,
                    f"{path}[{index}]"
                )
            )


    return findings


all_forbidden = []


for cell, record in data.items():

    hits = find_forbidden_true(
        record,
        cell
    )

    all_forbidden.extend(
        hits
    )


add_check(
    "no_forbidden_authorization_true",
    0,
    len(all_forbidden),
    f"findings={all_forbidden}",
)


# ------------------------------------------------------------
# 12. C04 CONFLICT REGISTER INTEGRITY
# ------------------------------------------------------------

def find_conflict_list(
    obj
):

    recognized_keys = {
        "conflicts",
        "registered_conflicts",
        "conflict_records",
        "conflict_register",
    }


    if isinstance(
        obj,
        dict
    ):

        for key, value in obj.items():

            if (
                key in recognized_keys
                and isinstance(
                    value,
                    list
                )
            ):

                return value


        for value in obj.values():

            found = find_conflict_list(
                value
            )

            if found is not None:

                return found


    elif isinstance(
        obj,
        list
    ):

        for item in obj:

            found = find_conflict_list(
                item
            )

            if found is not None:

                return found


    return None


if "C04" in data:

    conflicts = find_conflict_list(
        data["C04"]
    )


    if isinstance(
        conflicts,
        list
    ):

        add_check(
            "C04_conflict_register_nonempty",
            True,
            len(conflicts) > 0,
            (
                f"conflict_count={len(conflicts)}"
            ),
        )

    else:

        add_check(
            "C04_conflict_register_nonempty",
            True,
            None,
            (
                "No recognized conflict "
                "register list found."
            ),
        )


# ------------------------------------------------------------
# 13. C07 PROVENANCE BOUNDARY
# ------------------------------------------------------------

if "C07" in data:

    c07 = data["C07"]

    claim_boundary = c07.get(
        "claim_boundary",
        None
    )


    if not isinstance(
        claim_boundary,
        dict
    ):

        add_check(
            "C07_claim_boundary_present",
            True,
            False,
            (
                "claim_boundary missing "
                "or not a dictionary."
            ),
        )

    else:

        add_check(
            "C07_generator_artifact_NOT_ESTABLISHED",
            "NOT_ESTABLISHED",
            claim_boundary.get(
                "generator_artifact"
            ),
        )


        add_check(
            "C07_formal_retraction_NOT_ESTABLISHED",
            "NOT_ESTABLISHED",
            claim_boundary.get(
                "formal_retraction"
            ),
        )


        add_check(
            "C07_independent_validation_NOT_AUTHORIZED",
            "NOT_AUTHORIZED",
            claim_boundary.get(
                "current_independent_validation_use"
            ),
        )


# ------------------------------------------------------------
# 14. C08 PROVENANCE / FORMULA BOUNDARY
# ------------------------------------------------------------

if "C08" in data:

    c08 = data["C08"]

    claim_boundary = c08.get(
        "claim_boundary",
        None
    )


    if not isinstance(
        claim_boundary,
        dict
    ):

        add_check(
            "C08_claim_boundary_present",
            True,
            False,
            (
                "claim_boundary missing "
                "or not a dictionary."
            ),
        )

    else:

        C08_EXPECTED = {

            "legal_nda_existence":
                "NOT_ESTABLISHED_BY_SUPPLIED_SOURCE_SET",

            "nda_with_openai_existence":
                "NOT_ESTABLISHED_BY_SUPPLIED_SOURCE_SET",

            "assistant_text_authorship":
                "NOT_ESTABLISHED",

            "formula_conflict_state":
                "UNRESOLVED",

            "canonical_formula":
                "NOT_ESTABLISHED",

            "empirical_validity":
                "NOT_AUTHORIZED",
        }


        for key, expected in (
            C08_EXPECTED.items()
        ):

            add_check(
                f"C08_{key}",
                expected,
                claim_boundary.get(
                    key
                ),
            )


# ------------------------------------------------------------
# 15. C07/C08 GOVERNANCE STATE EXPLICITLY FALSE
# ------------------------------------------------------------

C07_FALSE_KEYS = [

    "conflict_resolution_performed",
    "canonical_source_declared",
    "formal_retraction_declared",
    "synthetic_generator_verified",
    "provenance_reconciled",
    "construct_authorization",
    "score_authorization",
    "outcome_authorization",
    "empirical_evaluation_authorization",
    "empirical_validity_claim",
]


C08_FALSE_KEYS = [

    "conflict_resolution_performed",
    "canonical_source_declared",
    "legal_nda_adjudication",
    "authorship_adjudication",
    "formula_version_lock",
    "formula_conflict_resolution",
    "substantive_authority_assignment",
    "construct_authorization",
    "score_authorization",
    "outcome_authorization",
    "empirical_evaluation_authorization",
    "empirical_validity_claim",
]


def check_explicit_false_keys(
    cell,
    keys
):

    if cell not in data:
        return


    record = data[cell]


    for key in keys:

        if key in record:

            value = record[key]

            add_check(
                f"{cell}_{key}_FALSE",
                False,
                value,
            )


        else:

            # Search nested only if the top-level
            # key is absent.
            #
            # If found, verify all occurrences.
            #
            # If not found, do not invent a state.

            def recursive_find(
                obj,
                target
            ):

                found = []

                if isinstance(
                    obj,
                    dict
                ):

                    for k, v in obj.items():

                        if k == target:
                            found.append(v)

                        found.extend(
                            recursive_find(
                                v,
                                target
                            )
                        )

                elif isinstance(
                    obj,
                    list
                ):

                    for item in obj:

                        found.extend(
                            recursive_find(
                                item,
                                target
                            )
                        )

                return found


            values = recursive_find(
                record,
                key
            )


            if values:

                add_check(
                    f"{cell}_{key}_FALSE",
                    True,
                    all(
                        value is False
                        for value in values
                    ),
                    f"observed={values}",
                )


check_explicit_false_keys(
    "C07",
    C07_FALSE_KEYS
)

check_explicit_false_keys(
    "C08",
    C08_FALSE_KEYS
)


# ------------------------------------------------------------
# 16. BEMP EXECUTION STATE BOUNDARY
# ------------------------------------------------------------
#
# C04/C05 may not contain this field.
# If present, it must remain the known
# fail-closed state.
#
# Missing state is NOT converted to PASS.
# ------------------------------------------------------------

BEMP_STATE_EXPECTED = (
    "EXECUTION_STATE_NOT_RECONSTRUCTIBLE"
)


def recursive_find_values(
    obj,
    target
):

    found = []

    if isinstance(
        obj,
        dict
    ):

        for key, value in obj.items():

            if key == target:

                found.append(
                    value
                )

            found.extend(
                recursive_find_values(
                    value,
                    target
                )
            )

    elif isinstance(
        obj,
        list
    ):

        for item in obj:

            found.extend(
                recursive_find_values(
                    item,
                    target
                )
            )

    return found


bemp_state_found = []


for cell, record in data.items():

    values = recursive_find_values(
        record,
        "execution_state"
    )

    bemp_state_found.extend(
        values
    )


# If the field exists, it must not contradict
# the known boundary.
if bemp_state_found:

    add_check(
        "BEMP_execution_state_preserved",
        BEMP_STATE_EXPECTED,
        (
            BEMP_STATE_EXPECTED
            if BEMP_STATE_EXPECTED
            in bemp_state_found
            else None
        ),
        f"observed={bemp_state_found}",
    )


# ------------------------------------------------------------
# 17. HASH UPSTREAM ARTIFACTS
# ------------------------------------------------------------

def sha256_file(
    path,
    chunk_size=1024 * 1024
):

    digest = hashlib.sha256()


    with open(
        path,
        "rb"
    ) as f:

        while True:

            chunk = f.read(
                chunk_size
            )

            if not chunk:
                break

            digest.update(
                chunk
            )


    return digest.hexdigest()


upstream_hashes = {}


for cell, path in UPSTREAM.items():

    if path.exists():

        upstream_hashes[cell] = {

            "filename":
                path.name,

            "size_bytes":
                path.stat().st_size,

            "sha256":
                sha256_file(path),
        }


# ------------------------------------------------------------
# 18. FINAL EVALUATION
# ------------------------------------------------------------

failed_checks = [
    item
    for item in checks
    if not item["passed"]
]


passed_checks = [
    item
    for item in checks
    if item["passed"]
]


overall_pass = (
    len(failed_checks) == 0
)


# ------------------------------------------------------------
# 19. C10 GOVERNANCE BOUNDARY
# ------------------------------------------------------------

governance = {

    "conflict_resolution_performed":
        False,

    "canonical_source_declared":
        False,

    "construct_authorization":
        False,

    "score_authorization":
        False,

    "outcome_authorization":
        False,

    "empirical_evaluation_authorization":
        False,

    "empirical_validity_claim":
        False,

    "final_manifest_created":
        False,
}


# ------------------------------------------------------------
# 20. TIMESTAMP
# ------------------------------------------------------------

timestamp = (
    datetime.now(
        timezone.utc
    )
    .replace(
        microsecond=0
    )
    .isoformat()
)


# ------------------------------------------------------------
# 21. RESULT OBJECT
# ------------------------------------------------------------

result = {

    "protocol":
        "OIP v1.0.35",

    "cell":
        "C10",

    "cell_purpose":
        "ARCHIVAL_FREEZE_PRECONDITION_CHECK",

    "timestamp_utc":
        timestamp,

    "upstream_present":
        sorted(data.keys()),

    "upstream_missing":
        missing,

    "upstream_load_errors":
        load_errors,

    "status_records":
        status_records,

    "checks_performed":
        len(checks),

    "checks_passed":
        len(passed_checks),

    "checks_failed":
        len(failed_checks),

    "failed_checks":
        failed_checks,

    "forbidden_authorization_findings":
        all_forbidden,

    "upstream_hashes":
        upstream_hashes,

    "governance":
        governance,

    "status":
        (
            "PASS"
            if overall_pass
            else
            "FAIL"
        ),

    "interpretation":
        (
            "Archival-freeze preconditions are satisfied "
            "for continuation to the next documented "
            "archival step. No conflict was resolved, "
            "no canonical source or formula was declared, "
            "and no construct, score, outcome, evaluation, "
            "or empirical-validity authorization was granted."
            if overall_pass
            else
            "One or more archival-freeze preconditions "
            "failed. The failed state must be reconciled "
            "before proceeding."
        ),

    "next_step":
        (
            "C11 — CORRECTED PUBLIC RECORD"
            if overall_pass
            else
            "STOP — RECONCILE FAILED PRECONDITIONS"
        ),
}


# ------------------------------------------------------------
# 22. WRITE JSON
# ------------------------------------------------------------

with open(
    OUT_JSON,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        result,
        f,
        indent=2,
        ensure_ascii=False
    )


# ------------------------------------------------------------
# 23. WRITE HUMAN-READABLE TXT
# ------------------------------------------------------------

lines = [

    "=" * 72,

    "OIP v1.0.35 — C10: HARDENED ARCHIVAL FREEZE PRECONDITION CHECK",

    "=" * 72,

    "",

    f"Timestamp UTC : {timestamp}",

    "",

    "Upstream artifacts:",

]


for cell, path in UPSTREAM.items():

    state = (
        "PRESENT"
        if path.exists()
        else
        "MISSING"
    )

    lines.append(
        f"  {cell} : {state} : {path}"
    )


lines.extend([

    "",

    f"Upstream missing      : {len(missing)}",

    f"Upstream load errors  : {len(load_errors)}",

    "",

    f"Checks performed      : {len(checks)}",

    f"Checks passed         : {len(passed_checks)}",

    f"Checks failed         : {len(failed_checks)}",

    "",
])


if failed_checks:

    lines.append(
        "FAILED CHECKS:"
    )

    lines.append(
        "-" * 72
    )


    for item in failed_checks:

        lines.append(
            f"[FAIL] {item['id']}"
        )

        lines.append(
            f"       expected={item['expected']!r}"
        )

        lines.append(
            f"       observed={item['observed']!r}"
        )

        if item["detail"]:

            lines.append(
                f"       detail={item['detail']}"
            )


    lines.append("")


lines.extend([

    "Governance boundary (C10 itself):",

    "  Conflict resolution              : NO",

    "  Canonical source declared        : NO",

    "  Construct authorization          : NO",

    "  Score authorization              : NO",

    "  Outcome authorization            : NO",

    "  Empirical evaluation             : NO",

    "  Empirical validity claim         : NO",

    "  Final manifest created           : NO",

    "",

    (
        "C10 STATUS                       : PASS"
        if overall_pass
        else
        "C10 STATUS                       : FAIL"
    ),

    "",

    result["interpretation"],

    "",

    f"Next: {result['next_step']}",

    "",

    f"JSON : {OUT_JSON}",

    f"TXT  : {OUT_TXT}",

    "=" * 72,
])


with open(
    OUT_TXT,
    "w",
    encoding="utf-8"
) as f:

    f.write(
        "\n".join(lines)
    )


# ------------------------------------------------------------
# 24. CONSOLE OUTPUT
# ------------------------------------------------------------

print("")

print(
    f"Upstream missing      : {len(missing)}"
)

print(
    f"Upstream load errors  : {len(load_errors)}"
)

print("")

print(
    f"Checks performed      : {len(checks)}"
)

print(
    f"Checks passed         : {len(passed_checks)}"
)

print(
    f"Checks failed         : {len(failed_checks)}"
)

print("")


if failed_checks:

    print(
        "FAILED CHECKS"
    )

    print(
        "-" * 72
    )


    for item in failed_checks:

        print(
            f"[FAIL] {item['id']}"
        )

        print(
            f"       expected={item['expected']!r}"
        )

        print(
            f"       observed={item['observed']!r}"
        )

        if item["detail"]:

            print(
                f"       detail={item['detail']}"
            )


    print("")

    print(
        "C10 STATUS                       : FAIL"
    )

    print(
        "STOP — DO NOT PROCEED."
    )

    raise RuntimeError(
        "C10 FAILED. "
        "Do not proceed to the next archival step."
    )


else:

    print(
        "FAILED CHECKS                    : NONE"
    )

    print("")

    print(
        "Governance boundary:"
    )

    print(
        "  Conflict resolution             : NO"
    )

    print(
        "  Canonical source declared      : NO"
    )

    print(
        "  Construct authorization         : NO"
    )

    print(
        "  Score authorization             : NO"
    )

    print(
        "  Outcome authorization           : NO"
    )

    print(
        "  Empirical evaluation            : NO"
    )

    print(
        "  Empirical validity claim        : NO"
    )

    print(
        "  Final manifest created          : NO"
    )

    print("")

    print(
        "C10 STATUS                       : PASS"
    )

    print("")

    print(
        "Next: C11 — CORRECTED PUBLIC RECORD"
    )

    print("")

    print(
        f"JSON : {OUT_JSON}"
    )

    print(
        f"TXT  : {OUT_TXT}"
    )

print("=" * 72)
```

```python
# ============================================================
# OIP v1.0.35 — C10
# ARCHIVAL FREEZE PRECONDITION CHECK
# SCHEMA-ALIGNED / FAIL-CLOSED
# ============================================================
#
# PURPOSE
# -------
# Verify that C00-C09 satisfy their own recorded integrity
# conditions before proceeding to C11/C12.
#
# C10 DOES NOT:
#   - resolve conflicts
#   - declare canonical sources
#   - declare canonical formulas
#   - authorize constructs
#   - authorize scores
#   - authorize outcomes
#   - perform empirical evaluation
#   - establish empirical validity
#   - create the final archival manifest
#
# IMPORTANT
# ---------
# This cell uses the actual schemas established by C00-C09.
# It does NOT assume every cell contains a generic "status" key.
#
# FAIL-CLOSED:
#   Missing required field
#   unexpected state
#   forbidden TRUE authorization
#   unresolved schema ambiguity
#   => FAIL
# ============================================================

from pathlib import Path
from datetime import datetime, timezone
import json
import hashlib


print("=" * 72)
print("OIP v1.0.35 — C10: ARCHIVAL FREEZE PRECONDITION CHECK")
print("=" * 72)


# ------------------------------------------------------------
# 1. ROOT
# ------------------------------------------------------------

ROOT = Path(
    "/kaggle/working/oip_v1_0_35"
)


# ------------------------------------------------------------
# 2. ACTUAL VERIFIED ARTIFACT PATHS
# ------------------------------------------------------------

UPSTREAM = {

    "C00":
        ROOT /
        "c00_source_scope" /
        "c00_source_scope.json",

    "C01":
        ROOT /
        "c01_artifact_inventory" /
        "c01_artifact_inventory.json",

    "C02":
        ROOT /
        "c02_artifact_identity" /
        "c02_artifact_identity.json",

    "C03":
        ROOT /
        "c03_authority_assignment" /
        "c03_authority_assignment.json",

    "C04":
        ROOT /
        "c04_conflict_register" /
        "c04_conflict_register.json",

    "C05":
        ROOT /
        "c05_malawi_primary_verification" /
        "c05_malawi_primary_verification.json",

    "C06":
        ROOT /
        "c06_ethiopia_primary_verification" /
        "c06_ethiopia_primary_verification.json",

    "C07":
        ROOT /
        "c07_v1_0_28_provenance" /
        "c07_v1_0_28_provenance.json",

    "C08":
        ROOT /
        "c08_v1_0_26_provenance" /
        "c08_v1_0_26_provenance.json",

    "C09":
        ROOT /
        "c09_final_conflict_consistency" /
        "c09_final_conflict_consistency.json",
}


# ------------------------------------------------------------
# 3. OUTPUT
# ------------------------------------------------------------

OUT_DIR = (
    ROOT /
    "c10_archival_freeze_precondition"
)

OUT_DIR.mkdir(
    parents=True,
    exist_ok=True
)

OUT_JSON = (
    OUT_DIR /
    "c10_archival_freeze_precondition.json"
)

OUT_TXT = (
    OUT_DIR /
    "c10_archival_freeze_precondition.txt"
)


# ------------------------------------------------------------
# 4. LOAD ALL UPSTREAM ARTIFACTS
# ------------------------------------------------------------

data = {}

missing = []

load_errors = []


for cell, path in UPSTREAM.items():

    if not path.exists():

        missing.append(
            {
                "cell": cell,
                "path": str(path)
            }
        )

        continue


    try:

        with open(
            path,
            "r",
            encoding="utf-8"
        ) as f:

            obj = json.load(f)


        if not isinstance(
            obj,
            dict
        ):

            raise TypeError(
                "JSON root is not a dictionary."
            )


        data[cell] = obj


    except Exception as exc:

        load_errors.append(
            {
                "cell": cell,
                "path": str(path),
                "error": repr(exc)
            }
        )


# ------------------------------------------------------------
# 5. CHECK ENGINE
# ------------------------------------------------------------

checks = []


def add_check(
    check_id,
    expected,
    observed,
    detail=""
):

    checks.append(
        {
            "id": check_id,
            "expected": expected,
            "observed": observed,
            "passed": observed == expected,
            "detail": detail,
        }
    )


# ------------------------------------------------------------
# 6. BASIC INPUT INTEGRITY
# ------------------------------------------------------------

add_check(
    "upstream_missing_count",
    0,
    len(missing),
    f"missing={missing}",
)


add_check(
    "upstream_load_error_count",
    0,
    len(load_errors),
    f"errors={load_errors}",
)


add_check(
    "all_C00_C09_loaded",
    sorted(UPSTREAM.keys()),
    sorted(data.keys()),
)


# ------------------------------------------------------------
# 7. C00
# ------------------------------------------------------------
#
# Actual schema:
#   scope_status
# ------------------------------------------------------------

if "C00" in data:

    c00 = data["C00"]

    add_check(
        "C00_scope_status_PASS",
        "PASS",
        c00.get("scope_status"),
    )

else:

    add_check(
        "C00_loaded",
        True,
        False,
    )


# ------------------------------------------------------------
# 8. C01
# ------------------------------------------------------------
#
# C01 does not have a generic status field.
#
# Actual execution state is represented through:
#   inventory_summary
#   errors
#   artifacts
#
# We require:
#   inventory complete = True
#   errors = 0
#   artifact count > 0
# ------------------------------------------------------------

if "C01" in data:

    c01 = data["C01"]

    inventory_summary = c01.get(
        "inventory_summary"
    )

    errors = c01.get(
        "errors"
    )

    artifacts = c01.get(
        "artifacts"
    )


    if not isinstance(
        inventory_summary,
        dict
    ):

        add_check(
            "C01_inventory_summary_present",
            True,
            False,
        )

    else:

        inventory_complete = (
            inventory_summary.get(
                "inventory_complete"
            )
        )

        add_check(
            "C01_inventory_complete",
            True,
            inventory_complete,
        )


    if isinstance(
        errors,
        list
    ):

        add_check(
            "C01_errors_zero",
            0,
            len(errors),
        )

    else:

        add_check(
            "C01_errors_field_is_list",
            True,
            False,
        )


    if isinstance(
        artifacts,
        list
    ):

        add_check(
            "C01_artifact_inventory_nonempty",
            True,
            len(artifacts) > 0,
            f"artifact_count={len(artifacts)}",
        )

    else:

        add_check(
            "C01_artifacts_field_is_list",
            True,
            False,
        )


# ------------------------------------------------------------
# 9. C02
# ------------------------------------------------------------
#
# Actual schema:
#   identity_status
# ------------------------------------------------------------

if "C02" in data:

    c02 = data["C02"]

    add_check(
        "C02_identity_status_PASS",
        "PASS",
        c02.get("identity_status"),
    )


# ------------------------------------------------------------
# 10. C03
# ------------------------------------------------------------
#
# C03 does not have generic status.
#
# Actual boundary:
#   substantive_authority_count == 0
#
# This is critical because C03 is explicitly designed
# not to assign substantive authority.
# ------------------------------------------------------------

if "C03" in data:

    c03 = data["C03"]

    add_check(
        "C03_substantive_authority_count_zero",
        0,
        c03.get(
            "substantive_authority_count"
        ),
    )


    authority_records = c03.get(
        "authority_records"
    )

    add_check(
        "C03_authority_records_present",
        True,
        isinstance(
            authority_records,
            list
        ),
    )


# ------------------------------------------------------------
# 11. C04
# ------------------------------------------------------------
#
# Actual schema:
#   register_status
#   conflict_count
#   open_item_count
# ------------------------------------------------------------

if "C04" in data:

    c04 = data["C04"]

    add_check(
        "C04_register_status_PASS",
        "PASS",
        c04.get("register_status"),
    )


    conflicts = c04.get(
        "conflicts"
    )

    open_items = c04.get(
        "open_governance_items"
    )


    add_check(
        "C04_conflicts_field_present",
        True,
        isinstance(
            conflicts,
            list
        ),
    )


    add_check(
        "C04_conflict_count_matches",
        c04.get(
            "conflict_count"
        ),
        (
            len(conflicts)
            if isinstance(
                conflicts,
                list
            )
            else None
        ),
    )


    add_check(
        "C04_open_governance_items_present",
        True,
        isinstance(
            open_items,
            list
        ),
    )


    add_check(
        "C04_open_item_count_matches",
        c04.get(
            "open_item_count"
        ),
        (
            len(open_items)
            if isinstance(
                open_items,
                list
            )
            else None
        ),
    )


# ------------------------------------------------------------
# 12. C05
# ------------------------------------------------------------
#
# Actual schema:
#   evidence_state
#   conflict_resolution_performed
#   canonical_source_declared
#
# C05 does NOT expose generic status.
#
# Therefore we verify its actual governance boundary.
# ------------------------------------------------------------

if "C05" in data:

    c05 = data["C05"]

    evidence_state = c05.get(
        "evidence_state"
    )

    add_check(
        "C05_evidence_state_present",
        True,
        isinstance(
            evidence_state,
            dict
        ),
    )


    add_check(
        "C05_conflict_resolution_FALSE",
        False,
        c05.get(
            "conflict_resolution_performed"
        ),
    )


    add_check(
        "C05_canonical_source_FALSE",
        False,
        c05.get(
            "canonical_source_declared"
        ),
    )


# ------------------------------------------------------------
# 13. C06
# ------------------------------------------------------------
#
# Actual schema:
#   artifact_state
#   conflict_resolution_performed
#   canonical_source_declared
# ------------------------------------------------------------

if "C06" in data:

    c06 = data["C06"]

    artifact_state = c06.get(
        "artifact_state"
    )

    add_check(
        "C06_artifact_state_present",
        True,
        isinstance(
            artifact_state,
            dict
        ),
    )


    add_check(
        "C06_conflict_resolution_FALSE",
        False,
        c06.get(
            "conflict_resolution_performed"
        ),
    )


    add_check(
        "C06_canonical_source_FALSE",
        False,
        c06.get(
            "canonical_source_declared"
        ),
    )


# ------------------------------------------------------------
# 14. C07
# ------------------------------------------------------------

if "C07" in data:

    c07 = data["C07"]

    add_check(
        "C07_status_PASS",
        "PASS",
        c07.get("status"),
    )


    governance = c07.get(
        "governance"
    )

    claim_boundary = c07.get(
        "claim_boundary"
    )


    add_check(
        "C07_governance_present",
        True,
        isinstance(
            governance,
            dict
        ),
    )


    add_check(
        "C07_claim_boundary_present",
        True,
        isinstance(
            claim_boundary,
            dict
        ),
    )


    if isinstance(
        governance,
        dict
    ):

        for key in [

            "conflict_resolution_performed",
            "canonical_source_declared",
            "formal_retraction_declared",
            "synthetic_generator_verified",
            "provenance_reconciled",
            "construct_authorization",
            "score_authorization",
            "outcome_authorization",
            "empirical_evaluation_authorization",
            "empirical_validity_claim",

        ]:

            if key in governance:

                add_check(
                    f"C07_{key}_FALSE",
                    False,
                    governance.get(key),
                )


    if isinstance(
        claim_boundary,
        dict
    ):

        add_check(
            "C07_generator_artifact_NOT_ESTABLISHED",
            "NOT_ESTABLISHED",
            claim_boundary.get(
                "generator_artifact"
            ),
        )


        add_check(
            "C07_formal_retraction_NOT_ESTABLISHED",
            "NOT_ESTABLISHED",
            claim_boundary.get(
                "formal_retraction"
            ),
        )


        add_check(
            "C07_independent_validation_NOT_AUTHORIZED",
            "NOT_AUTHORIZED",
            claim_boundary.get(
                "current_independent_validation_use"
            ),
        )


# ------------------------------------------------------------
# 15. C08
# ------------------------------------------------------------

if "C08" in data:

    c08 = data["C08"]

    add_check(
        "C08_status_PASS",
        "PASS",
        c08.get("status"),
    )


    governance = c08.get(
        "governance"
    )

    claim_boundary = c08.get(
        "claim_boundary"
    )


    add_check(
        "C08_governance_present",
        True,
        isinstance(
            governance,
            dict
        ),
    )


    add_check(
        "C08_claim_boundary_present",
        True,
        isinstance(
            claim_boundary,
            dict
        ),
    )


    if isinstance(
        governance,
        dict
    ):

        for key in [

            "conflict_resolution_performed",
            "canonical_source_declared",
            "legal_nda_adjudication",
            "authorship_adjudication",
            "formula_version_lock",
            "formula_conflict_resolution",
            "substantive_authority_assignment",
            "construct_authorization",
            "score_authorization",
            "outcome_authorization",
            "empirical_evaluation_authorization",
            "empirical_validity_claim",

        ]:

            if key in governance:

                add_check(
                    f"C08_{key}_FALSE",
                    False,
                    governance.get(key),
                )


    if isinstance(
        claim_boundary,
        dict
    ):

        expected_c08 = {

            "legal_nda_existence":
                "NOT_ESTABLISHED_BY_SUPPLIED_SOURCE_SET",

            "nda_with_openai_existence":
                "NOT_ESTABLISHED_BY_SUPPLIED_SOURCE_SET",

            "assistant_text_authorship":
                "NOT_ESTABLISHED",

            "formula_conflict_state":
                "UNRESOLVED",

            "canonical_formula":
                "NOT_ESTABLISHED",

            "empirical_validity":
                "NOT_AUTHORIZED",
        }


        for key, expected in (
            expected_c08.items()
        ):

            add_check(
                f"C08_{key}",
                expected,
                claim_boundary.get(
                    key
                ),
            )


# ------------------------------------------------------------
# 16. C09
# ------------------------------------------------------------

if "C09" in data:

    c09 = data["C09"]


    add_check(
        "C09_status_PASS",
        "PASS",
        c09.get("status"),
    )


    add_check(
        "C09_checks_failed_zero",
        0,
        c09.get(
            "checks_failed"
        ),
    )


    failed_checks = c09.get(
        "failed_checks"
    )


    add_check(
        "C09_failed_checks_empty",
        0,
        (
            len(failed_checks)
            if isinstance(
                failed_checks,
                list
            )
            else None
        ),
    )


    add_check(
        "C09_governance_boundary_present",
        True,
        isinstance(
            c09.get(
                "governance_boundary"
            ),
            dict
        ),
    )


# ------------------------------------------------------------
# 17. GLOBAL FORBIDDEN TRUE SCAN
# ------------------------------------------------------------

FORBIDDEN_TRUE_KEYS = {

    "conflict_resolution_performed",
    "canonical_source_declared",
    "construct_authorization",
    "score_authorization",
    "outcome_authorization",
    "empirical_evaluation_authorization",
    "empirical_validity_claim",
    "empirical_validity_claim_authorized",
    "substantive_authority_assigned",
    "substantive_authority_assignment",
    "legal_nda_adjudication",
    "authorship_adjudication",
    "formula_version_lock",
    "formula_conflict_resolution",
    "provenance_reconciled",
    "synthetic_generator_verified",
    "formal_retraction_declared",
    "reclassification_adjudicated",
}


def find_forbidden_true(
    obj,
    path="root"
):

    hits = []


    if isinstance(
        obj,
        dict
    ):

        for key, value in obj.items():

            current = (
                f"{path}.{key}"
            )


            if (
                key in FORBIDDEN_TRUE_KEYS
                and value is True
            ):

                hits.append(
                    current
                )


            hits.extend(
                find_forbidden_true(
                    value,
                    current
                )
            )


    elif isinstance(
        obj,
        list
    ):

        for i, value in enumerate(
            obj
        ):

            hits.extend(
                find_forbidden_true(
                    value,
                    f"{path}[{i}]"
                )
            )


    return hits


all_forbidden = []


for cell, record in data.items():

    all_forbidden.extend(
        find_forbidden_true(
            record,
            cell
        )
    )


add_check(
    "global_forbidden_authorization_true_zero",
    0,
    len(all_forbidden),
    f"findings={all_forbidden}",
)


# ------------------------------------------------------------
# 18. UPSTREAM HASHES
# ------------------------------------------------------------

def sha256_file(
    path,
    chunk_size=1024 * 1024
):

    h = hashlib.sha256()


    with open(
        path,
        "rb"
    ) as f:

        while True:

            chunk = f.read(
                chunk_size
            )

            if not chunk:
                break

            h.update(
                chunk
            )


    return h.hexdigest()


upstream_hashes = {}


for cell, path in UPSTREAM.items():

    if path.exists():

        upstream_hashes[cell] = {

            "filename":
                path.name,

            "size_bytes":
                path.stat().st_size,

            "sha256":
                sha256_file(path),
        }


# ------------------------------------------------------------
# 19. FINAL RESULT
# ------------------------------------------------------------

failed_checks = [
    item
    for item in checks
    if not item["passed"]
]


passed_checks = [
    item
    for item in checks
    if item["passed"]
]


overall_pass = (
    len(failed_checks) == 0
)


# ------------------------------------------------------------
# 20. C10 GOVERNANCE BOUNDARY
# ------------------------------------------------------------

governance = {

    "conflict_resolution_performed":
        False,

    "canonical_source_declared":
        False,

    "construct_authorization":
        False,

    "score_authorization":
        False,

    "outcome_authorization":
        False,

    "empirical_evaluation_authorization":
        False,

    "empirical_validity_claim":
        False,

    "final_manifest_created":
        False,
}


# ------------------------------------------------------------
# 21. RESULT
# ------------------------------------------------------------

timestamp = (
    datetime.now(
        timezone.utc
    )
    .replace(
        microsecond=0
    )
    .isoformat()
)


result = {

    "protocol":
        "OIP v1.0.35",

    "cell":
        "C10",

    "cell_purpose":
        "ARCHIVAL_FREEZE_PRECONDITION_CHECK",

    "timestamp_utc":
        timestamp,

    "upstream_present":
        sorted(data.keys()),

    "upstream_missing":
        missing,

    "upstream_load_errors":
        load_errors,

    "checks_performed":
        len(checks),

    "checks_passed":
        len(passed_checks),

    "checks_failed":
        len(failed_checks),

    "failed_checks":
        failed_checks,

    "forbidden_authorization_findings":
        all_forbidden,

    "upstream_hashes":
        upstream_hashes,

    "governance":
        governance,

    "status":
        (
            "PASS"
            if overall_pass
            else
            "FAIL"
        ),

    "interpretation":
        (
            "Archival-freeze preconditions are satisfied "
            "for continuation to the next documented "
            "archival step. No conflict was resolved, "
            "no canonical source or formula was declared, "
            "and no construct, score, outcome, evaluation, "
            "or empirical-validity authorization was granted."
            if overall_pass
            else
            "One or more archival-freeze preconditions "
            "failed. The failed state must be reconciled "
            "before proceeding."
        ),

    "next_step":
        (
            "C11 — CORRECTED PUBLIC RECORD"
            if overall_pass
            else
            "STOP — RECONCILE FAILED PRECONDITIONS"
        ),
}


# ------------------------------------------------------------
# 22. WRITE JSON
# ------------------------------------------------------------

with open(
    OUT_JSON,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        result,
        f,
        indent=2,
        ensure_ascii=False
    )


# ------------------------------------------------------------
# 23. WRITE TXT
# ------------------------------------------------------------

lines = [

    "=" * 72,

    "OIP v1.0.35 — C10: ARCHIVAL FREEZE PRECONDITION CHECK",

    "=" * 72,

    "",

    f"Timestamp UTC : {timestamp}",

    "",

    f"Upstream missing      : {len(missing)}",

    f"Upstream load errors  : {len(load_errors)}",

    "",

    f"Checks performed      : {len(checks)}",

    f"Checks passed         : {len(passed_checks)}",

    f"Checks failed         : {len(failed_checks)}",

    "",
]


if failed_checks:

    lines.append(
        "FAILED CHECKS:"
    )

    lines.append(
        "-" * 72
    )


    for item in failed_checks:

        lines.append(
            f"[FAIL] {item['id']}"
        )

        lines.append(
            f"       expected={item['expected']!r}"
        )

        lines.append(
            f"       observed={item['observed']!r}"
        )

        if item["detail"]:

            lines.append(
                f"       detail={item['detail']}"
            )


    lines.append("")


lines.extend([

    "Governance boundary (C10 itself):",

    "  Conflict resolution              : NO",

    "  Canonical source declared        : NO",

    "  Construct authorization          : NO",

    "  Score authorization              : NO",

    "  Outcome authorization            : NO",

    "  Empirical evaluation             : NO",

    "  Empirical validity claim         : NO",

    "  Final manifest created           : NO",

    "",

    (
        "C10 STATUS                       : PASS"
        if overall_pass
        else
        "C10 STATUS                       : FAIL"
    ),

    "",

    result["interpretation"],

    "",

    f"Next: {result['next_step']}",

    "",

    f"JSON : {OUT_JSON}",

    f"TXT  : {OUT_TXT}",

    "=" * 72,
])


with open(
    OUT_TXT,
    "w",
    encoding="utf-8"
) as f:

    f.write(
        "\n".join(lines)
    )


# ------------------------------------------------------------
# 24. CONSOLE OUTPUT
# ------------------------------------------------------------

print("")

print(
    f"Upstream missing      : {len(missing)}"
)

print(
    f"Upstream load errors  : {len(load_errors)}"
)

print("")

print(
    f"Checks performed      : {len(checks)}"
)

print(
    f"Checks passed         : {len(passed_checks)}"
)

print(
    f"Checks failed         : {len(failed_checks)}"
)

print("")


if failed_checks:

    print(
        "FAILED CHECKS"
    )

    print(
        "-" * 72
    )


    for item in failed_checks:

        print(
            f"[FAIL] {item['id']}"
        )

        print(
            f"       expected={item['expected']!r}"
        )

        print(
            f"       observed={item['observed']!r}"
        )

        if item["detail"]:

            print(
                f"       detail={item['detail']}"
            )


    print("")

    print(
        "C10 STATUS : FAIL"
    )

    print(
        "STOP — DO NOT PROCEED TO C11/C12."
    )

    raise RuntimeError(
        "C10 FAILED. "
        "Do not proceed to C11/C12."
    )


else:

    print(
        "FAILED CHECKS                    : NONE"
    )

    print("")

    print(
        "Governance boundary:"
    )

    print(
        "  Conflict resolution             : NO"
    )

    print(
        "  Canonical source declared      : NO"
    )

    print(
        "  Construct authorization         : NO"
    )

    print(
        "  Score authorization             : NO"
    )

    print(
        "  Outcome authorization           : NO"
    )

    print(
        "  Empirical evaluation            : NO"
    )

    print(
        "  Empirical validity              : NO"
    )

    print(
        "  Final manifest created          : NO"
    )

    print("")

    print(
        "C10 STATUS                      : PASS"
    )

    print("")

    print(
        "Next: C11 — CORRECTED PUBLIC RECORD"
    )

    print("")

    print(
        f"JSON : {OUT_JSON}"
    )

    print(
        f"TXT  : {OUT_TXT}"
    )

print("=" * 72)
```

```python
# ============================================================
# OIP v1.0.35 — CELL 11
# CORRECTED PUBLIC RECORD
# FINAL SCHEMA-ALIGNED / FAIL-CLOSED VERSION
# ============================================================

import json
from pathlib import Path
from datetime import datetime, timezone

print("=" * 72)
print("OIP v1.0.35 — C11: CORRECTED PUBLIC RECORD")
print("=" * 72)

# ------------------------------------------------------------
# 1. PATHS
# ------------------------------------------------------------

BASE = Path("/kaggle/working/oip_v1_0_35")

UPSTREAM = {
    "C04": BASE / "c04_conflict_register"
          / "c04_conflict_register.json",

    "C05": BASE / "c05_malawi_primary_verification"
          / "c05_malawi_primary_verification.json",

    "C06": BASE / "c06_ethiopia_primary_verification"
          / "c06_ethiopia_primary_verification.json",

    "C07": BASE / "c07_v1_0_28_provenance"
          / "c07_v1_0_28_provenance.json",

    "C08": BASE / "c08_v1_0_26_provenance"
          / "c08_v1_0_26_provenance.json",

    "C09": BASE / "c09_final_conflict_consistency"
          / "c09_final_conflict_consistency.json",

    "C10": BASE / "c10_archival_freeze_precondition"
          / "c10_archival_freeze_precondition.json",
}

OUTDIR = BASE / "c11_corrected_public_record"
OUTDIR.mkdir(parents=True, exist_ok=True)

OUT_JSON = OUTDIR / "c11_corrected_public_record.json"
OUT_TXT = OUTDIR / "c11_corrected_public_record.txt"


# ------------------------------------------------------------
# 2. LOAD REQUIRED UPSTREAM ARTIFACTS
# ------------------------------------------------------------

def load_json(path):
    if not path.exists():
        raise FileNotFoundError(
            f"Required upstream artifact missing: {path}"
        )

    with path.open("r", encoding="utf-8") as f:
        return json.load(f)


artifacts = {
    name: load_json(path)
    for name, path in UPSTREAM.items()
}


# ------------------------------------------------------------
# 3. EXACT TOP-LEVEL / KNOWN-SCHEMA HELPERS
# ------------------------------------------------------------

def exact_key(obj, key, default=None):
    if isinstance(obj, dict) and key in obj:
        return obj[key]
    return default


def require_exact_value(
    cell,
    key,
    expected,
    failures,
):
    value = exact_key(artifacts[cell], key, None)

    if value != expected:
        failures.append(
            f"{cell}: {key}={value!r}; expected {expected!r}"
        )

    return value


def require_bool_false(
    cell,
    key,
    failures,
):
    value = exact_key(artifacts[cell], key, None)

    if value is not False:
        failures.append(
            f"{cell}: {key}={value!r}; expected False"
        )

    return value


failures = []
warnings = []


# ------------------------------------------------------------
# 4. EXACT PREREQUISITE VALIDATION
#
# Known schemas are used here rather than broad recursive
# matching. This prevents a nested unrelated "status" field
# from being treated as the cell status.
# ------------------------------------------------------------

# C04
require_exact_value(
    "C04",
    "register_status",
    "PASS",
    failures,
)

# C05
if not isinstance(
    exact_key(artifacts["C05"], "evidence_state", None),
    dict
):
    failures.append(
        "C05: evidence_state missing or not a dictionary"
    )

# C06
if not isinstance(
    exact_key(artifacts["C06"], "artifact_state", None),
    dict
):
    failures.append(
        "C06: artifact_state missing or not a dictionary"
    )

# C07
require_exact_value(
    "C07",
    "status",
    "PASS",
    failures,
)

# C08
require_exact_value(
    "C08",
    "status",
    "PASS",
    failures,
)

# C09
require_exact_value(
    "C09",
    "status",
    "PASS",
    failures,
)

c09_failed = exact_key(
    artifacts["C09"],
    "checks_failed",
    None,
)

if c09_failed != 0:
    failures.append(
        f"C09: checks_failed={c09_failed!r}; expected 0"
    )

c09_failed_checks = exact_key(
    artifacts["C09"],
    "failed_checks",
    None,
)

if not isinstance(c09_failed_checks, list):
    failures.append(
        "C09: failed_checks missing or not a list"
    )
elif len(c09_failed_checks) != 0:
    failures.append(
        "C09: failed_checks is not empty"
    )

# C10
require_exact_value(
    "C10",
    "status",
    "PASS",
    failures,
)


# ------------------------------------------------------------
# 5. GOVERNANCE BOUNDARY VALIDATION
# ------------------------------------------------------------

# C05
require_bool_false(
    "C05",
    "conflict_resolution_performed",
    failures,
)

require_bool_false(
    "C05",
    "canonical_source_declared",
    failures,
)


# C06
require_bool_false(
    "C06",
    "conflict_resolution_performed",
    failures,
)

require_bool_false(
    "C06",
    "canonical_source_declared",
    failures,
)


# ------------------------------------------------------------
# 6. C03/C04/C07/C08/C10 GOVERNANCE SAFETY
#
# These are read only from their known locations.
# ------------------------------------------------------------

# C07
c07_governance = exact_key(
    artifacts["C07"],
    "governance",
    {},
)

if not isinstance(c07_governance, dict):
    failures.append(
        "C07: governance is missing or not a dictionary"
    )
else:
    for key, value in c07_governance.items():
        if isinstance(value, bool) and value is True:
            failures.append(
                f"C07 governance violation: {key}=True"
            )


# C08
c08_governance = exact_key(
    artifacts["C08"],
    "governance",
    {},
)

if not isinstance(c08_governance, dict):
    failures.append(
        "C08: governance is missing or not a dictionary"
    )
else:
    for key, value in c08_governance.items():
        if isinstance(value, bool) and value is True:
            failures.append(
                f"C08 governance violation: {key}=True"
            )


# C10
c10_governance = exact_key(
    artifacts["C10"],
    "governance_boundary",
    {},
)

if not isinstance(c10_governance, dict):
    failures.append(
        "C10: governance_boundary is missing or not a dictionary"
    )
else:
    for key, value in c10_governance.items():
        if isinstance(value, bool) and value is True:
            failures.append(
                f"C10 governance violation: {key}=True"
            )


# ------------------------------------------------------------
# 7. C07 CLAIM BOUNDARY
# ------------------------------------------------------------

c07_claim = exact_key(
    artifacts["C07"],
    "claim_boundary",
    {},
)

if not isinstance(c07_claim, dict):
    failures.append(
        "C07: claim_boundary missing or not a dictionary"
    )
else:

    expected_c07_claims = {
        "generator_artifact":
            "NOT_ESTABLISHED",

        "formal_retraction":
            "NOT_ESTABLISHED",

        "current_independent_validation_use":
            "NOT_AUTHORIZED",
    }

    for key, expected in expected_c07_claims.items():

        actual = c07_claim.get(key, None)

        if actual != expected:
            failures.append(
                f"C07 claim boundary mismatch: "
                f"{key}={actual!r}; expected {expected!r}"
            )


# ------------------------------------------------------------
# 8. C08 CLAIM BOUNDARY
# ------------------------------------------------------------

c08_claim = exact_key(
    artifacts["C08"],
    "claim_boundary",
    {},
)

if not isinstance(c08_claim, dict):
    failures.append(
        "C08: claim_boundary missing or not a dictionary"
    )
else:

    expected_c08_claims = {

        "legal_nda_existence":
            "NOT_ESTABLISHED_BY_SUPPLIED_SOURCE_SET",

        "nda_with_openai_existence":
            "NOT_ESTABLISHED_BY_SUPPLIED_SOURCE_SET",

        "assistant_text_authorship":
            "NOT_ESTABLISHED",

        "formula_conflict_state":
            "UNRESOLVED",

        "canonical_formula":
            "NOT_ESTABLISHED",

        "empirical_validity":
            "NOT_AUTHORIZED",
    }

    for key, expected in expected_c08_claims.items():

        actual = c08_claim.get(key, None)

        if actual != expected:
            failures.append(
                f"C08 claim boundary mismatch: "
                f"{key}={actual!r}; expected {expected!r}"
            )


# ------------------------------------------------------------
# 9. C04 CONFLICT REGISTER
# ------------------------------------------------------------

c04 = artifacts["C04"]

conflicts = exact_key(
    c04,
    "conflicts",
    None,
)

open_items = exact_key(
    c04,
    "open_governance_items",
    None,
)

conflict_count = exact_key(
    c04,
    "conflict_count",
    None,
)

open_item_count = exact_key(
    c04,
    "open_item_count",
    None,
)

if not isinstance(conflicts, list):
    failures.append(
        "C04: conflicts missing or not a list"
    )
    conflicts = []

if not isinstance(open_items, list):
    failures.append(
        "C04: open_governance_items missing or not a list"
    )
    open_items = []

if conflict_count != len(conflicts):
    failures.append(
        f"C04: conflict_count={conflict_count!r} "
        f"but conflicts length={len(conflicts)}"
    )

if open_item_count != len(open_items):
    failures.append(
        f"C04: open_item_count={open_item_count!r} "
        f"but open_governance_items length={len(open_items)}"
    )


# ------------------------------------------------------------
# 10. OBSERVED PRIMARY-ARTIFACT STATES
# ------------------------------------------------------------

c05_evidence = exact_key(
    artifacts["C05"],
    "evidence_state",
    {},
)

c06_signals = exact_key(
    artifacts["C06"],
    "observed_signals",
    {},
)

if not isinstance(c05_evidence, dict):
    c05_evidence = {}

if not isinstance(c06_signals, dict):
    c06_signals = {}


# ------------------------------------------------------------
# 11. EXPLICIT GLOBAL AUTHORIZATION SAFETY
#
# Only exact governance fields are inspected.
# No recursive inference is performed.
# ------------------------------------------------------------

explicit_false_fields = [
    ("C05", "conflict_resolution_performed"),
    ("C05", "canonical_source_declared"),
    ("C06", "conflict_resolution_performed"),
    ("C06", "canonical_source_declared"),
]

for cell, key in explicit_false_fields:

    value = exact_key(
        artifacts[cell],
        key,
        None,
    )

    if value is not False:
        failures.append(
            f"{cell}: forbidden governance state "
            f"{key}={value!r}"
        )


# ------------------------------------------------------------
# 12. BUILD CORRECTED PUBLIC RECORD
# ------------------------------------------------------------

timestamp = (
    datetime.now(timezone.utc)
    .replace(microsecond=0)
    .isoformat()
)

public_record = {

    "protocol":
        "OIP v1.0.35",

    "cell_id":
        "C11",

    "cell_name":
        "CORRECTED PUBLIC RECORD",

    "created_utc":
        timestamp,

    "purpose":
        (
            "Create a source-bounded public record of the "
            "current evidence, unresolved conflicts, and "
            "claim boundaries without resolving conflicts "
            "or creating substantive scientific authorization."
        ),

    "upstream":
        {
            name: str(path)
            for name, path in UPSTREAM.items()
        },

    "prerequisite_state":
        {
            "C04": "PASS",
            "C05": "PRIMARY_ARTIFACT_STATE_RECORDED",
            "C06": "PRIMARY_ARTIFACT_STATE_RECORDED",
            "C07": "PASS",
            "C08": "PASS",
            "C09": "PASS",
            "C10": "PASS",
        },

    "registered_conflicts":
        {
            "count": len(conflicts),
            "records": conflicts,
        },

    "open_governance_items":
        {
            "count": len(open_items),
            "records": open_items,
        },

    "malawi_observed_evidence":
        c05_evidence,

    "ethiopia_observed_signals":
        c06_signals,

    "current_evidence_boundary":
        {
            "conflict_resolution":
                "NOT_PERFORMED",

            "canonical_source":
                "NOT_DECLARED",

            "construct_authorization":
                "NOT_AUTHORIZED",

            "score_authorization":
                "NOT_AUTHORIZED",

            "outcome_authorization":
                "NOT_AUTHORIZED",

            "empirical_evaluation":
                "NOT_AUTHORIZED",

            "empirical_validity":
                "NOT_AUTHORIZED",

            "v1_0_28_generator_artifact":
                c07_claim.get(
                    "generator_artifact",
                    "NOT_ESTABLISHED"
                ),

            "v1_0_28_formal_retraction":
                c07_claim.get(
                    "formal_retraction",
                    "NOT_ESTABLISHED"
                ),

            "v1_0_28_independent_validation_use":
                c07_claim.get(
                    "current_independent_validation_use",
                    "NOT_AUTHORIZED"
                ),

            "v1_0_26_legal_NDA_existence":
                c08_claim.get(
                    "legal_nda_existence",
                    "NOT_ESTABLISHED_BY_SUPPLIED_SOURCE_SET"
                ),

            "v1_0_26_NDA_with_OpenAI":
                c08_claim.get(
                    "nda_with_openai_existence",
                    "NOT_ESTABLISHED_BY_SUPPLIED_SOURCE_SET"
                ),

            "v1_0_26_assistant_text_authorship":
                c08_claim.get(
                    "assistant_text_authorship",
                    "NOT_ESTABLISHED"
                ),

            "v1_0_26_formula_conflict":
                c08_claim.get(
                    "formula_conflict_state",
                    "UNRESOLVED"
                ),

            "v1_0_26_canonical_formula":
                c08_claim.get(
                    "canonical_formula",
                    "NOT_ESTABLISHED"
                ),
        },

    "v1_0_28_public_record":
        {
            "status":
                "HISTORICAL_RECORD_PENDING_PROVENANCE_RECONCILIATION",

            "formal_retraction":
                c07_claim.get(
                    "formal_retraction",
                    "NOT_ESTABLISHED"
                ),

            "generator_artifact":
                c07_claim.get(
                    "generator_artifact",
                    "NOT_ESTABLISHED"
                ),

            "current_independent_validation_use":
                c07_claim.get(
                    "current_independent_validation_use",
                    "NOT_AUTHORIZED"
                ),
        },

    "v1_0_26_public_record":
        {
            "legal_NDA_existence":
                c08_claim.get(
                    "legal_nda_existence",
                    "NOT_ESTABLISHED_BY_SUPPLIED_SOURCE_SET"
                ),

            "NDA_with_OpenAI_existence":
                c08_claim.get(
                    "nda_with_openai_existence",
                    "NOT_ESTABLISHED_BY_SUPPLIED_SOURCE_SET"
                ),

            "assistant_text_authorship":
                c08_claim.get(
                    "assistant_text_authorship",
                    "NOT_ESTABLISHED"
                ),

            "formula_conflict":
                c08_claim.get(
                    "formula_conflict_state",
                    "UNRESOLVED"
                ),

            "canonical_formula":
                c08_claim.get(
                    "canonical_formula",
                    "NOT_ESTABLISHED"
                ),

            "formula_version_lock":
                "REQUIRED_BEFORE_CONFIRMATORY_IMPLEMENTATION",
        },

    "bemp_boundary_statement":
        (
            "The supplied BEMP notebook preserves the fail-closed "
            "audit logic, but its executed final runtime state is "
            "not independently reconstructible from the supplied "
            "source set."
        ),

    "malawi_boundary_statement":
        (
            "The Malawi record preserves an inventory-execution "
            "count discrepancy and an identifier-state conflict "
            "for further primary-artifact verification; this "
            "cell does not resolve either conflict."
        ),

    "ethiopia_boundary_statement":
        (
            "The Ethiopia primary artifacts expose the observed "
            "signals recorded by C06. Signal presence alone is "
            "not treated as semantic or execution-state resolution."
        ),

    "public_claim_statement":
        (
            "The OIP v1.0.35 reconciliation record documents "
            "source-level observations, unresolved provenance "
            "and formula conflicts, and explicit evidence "
            "boundaries. The current record does not establish "
            "empirical validity, predictive validity, causal "
            "validity, universal validity, superiority, or "
            "authorized downstream empirical evaluation of OIP."
        ),

    "governance":
        {
            "conflict_resolution_performed":
                False,

            "canonical_source_declared":
                False,

            "construct_authorization":
                False,

            "score_authorization":
                False,

            "outcome_authorization":
                False,

            "empirical_evaluation":
                False,

            "empirical_validity":
                False,

            "formal_retraction":
                False,

            "formula_version_lock":
                False,

            "final_manifest_created":
                False,
        },

    "fail_closed_statement":
        (
            "Unresolved evidence remains unresolved. No missing "
            "evidence is replaced by inference, assumption, proxy "
            "substitution, synthetic outcome, silent imputation, "
            "random fallback, or post-hoc selection."
        ),

    "warnings":
        warnings,

    "status":
        "PASS" if not failures else "FAIL",

    "failed_checks":
        failures,

    "next_cell":
        (
            "C12 — FINAL ARCHIVAL FREEZE MANIFEST"
            if not failures
            else
            "STOP — RECONCILE FAILED CHECKS"
        ),
}


# ------------------------------------------------------------
# 13. FAIL-CLOSED STOP
# ------------------------------------------------------------

if failures:

    print()
    print("C11 STATUS : FAIL")
    print("-" * 72)

    for failure in failures:
        print("FAIL:", failure)

    print()
    print("C12 MUST NOT BE EXECUTED.")

    raise RuntimeError(
        "C11 failed closed. Review failed checks before C12."
    )


# ------------------------------------------------------------
# 14. WRITE JSON
# ------------------------------------------------------------

with OUT_JSON.open(
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        public_record,
        f,
        indent=2,
        ensure_ascii=False,
    )


# ------------------------------------------------------------
# 15. WRITE TXT
# ------------------------------------------------------------

txt = []

txt.append("=" * 72)
txt.append(
    "OIP v1.0.35 — C11: CORRECTED PUBLIC RECORD"
)
txt.append("=" * 72)
txt.append("")

txt.append(
    "C04 register status          : PASS"
)

txt.append(
    "C05 primary artifact state  : RECORDED"
)

txt.append(
    "C06 primary artifact state  : RECORDED"
)

txt.append(
    "C07 status                  : PASS"
)

txt.append(
    "C08 status                  : PASS"
)

txt.append(
    "C09 status                  : PASS"
)

txt.append(
    "C10 status                  : PASS"
)

txt.append("")

txt.append(
    f"Registered conflicts        : {len(conflicts)}"
)

txt.append(
    f"Open governance items      : {len(open_items)}"
)

txt.append("")

txt.append("Governance boundary:")

txt.append(
    "  Conflict resolution       : NO"
)

txt.append(
    "  Canonical source declared : NO"
)

txt.append(
    "  Construct authorization   : NO"
)

txt.append(
    "  Score authorization       : NO"
)

txt.append(
    "  Outcome authorization     : NO"
)

txt.append(
    "  Empirical evaluation      : NO"
)

txt.append(
    "  Empirical validity        : NO"
)

txt.append(
    "  Formal retraction         : NO"
)

txt.append(
    "  Formula version lock      : NO"
)

txt.append(
    "  Final manifest created    : NO"
)

txt.append("")

txt.append("Claim boundary:")

txt.append(
    "  v1.0.28 generator artifact : "
    + str(
        c07_claim.get(
            "generator_artifact",
            "NOT_ESTABLISHED"
        )
    )
)

txt.append(
    "  v1.0.28 formal retraction  : "
    + str(
        c07_claim.get(
            "formal_retraction",
            "NOT_ESTABLISHED"
        )
    )
)

txt.append(
    "  v1.0.28 independent use    : "
    + str(
        c07_claim.get(
            "current_independent_validation_use",
            "NOT_AUTHORIZED"
        )
    )
)

txt.append(
    "  v1.0.26 legal NDA existence: "
    + str(
        c08_claim.get(
            "legal_nda_existence",
            "NOT_ESTABLISHED_BY_SUPPLIED_SOURCE_SET"
        )
    )
)

txt.append(
    "  v1.0.26 NDA with OpenAI    : "
    + str(
        c08_claim.get(
            "nda_with_openai_existence",
            "NOT_ESTABLISHED_BY_SUPPLIED_SOURCE_SET"
        )
    )
)

txt.append(
    "  v1.0.26 assistant authorship: "
    + str(
        c08_claim.get(
            "assistant_text_authorship",
            "NOT_ESTABLISHED"
        )
    )
)

txt.append(
    "  v1.0.26 formula conflict   : "
    + str(
        c08_claim.get(
            "formula_conflict_state",
            "UNRESOLVED"
        )
    )
)

txt.append(
    "  v1.0.26 canonical formula  : "
    + str(
        c08_claim.get(
            "canonical_formula",
            "NOT_ESTABLISHED"
        )
    )
)

txt.append("")

txt.append(
    "Public claim boundary:"
)

txt.append(
    "  No empirical validity claim is authorized."
)

txt.append(
    "  No causal validity claim is authorized."
)

txt.append(
    "  No universal validity claim is authorized."
)

txt.append(
    "  No superiority claim is authorized."
)

txt.append(
    "  No downstream empirical evaluation is authorized."
)

txt.append("")

if warnings:

    txt.append("Warnings:")

    for warning in warnings:
        txt.append(
            "  - " + warning
        )

    txt.append("")

txt.append(
    "C11 STATUS                  : PASS"
)

txt.append("")

txt.append(
    "Next: C12 — FINAL ARCHIVAL FREEZE MANIFEST"
)

txt.append("")

txt.append(
    "JSON : " + str(OUT_JSON)
)

txt.append(
    "TXT  : " + str(OUT_TXT)
)

txt.append("=" * 72)

with OUT_TXT.open(
    "w",
    encoding="utf-8"
) as f:

    f.write(
        "\n".join(txt)
    )


# ------------------------------------------------------------
# 16. FINAL CONSOLE OUTPUT
# ------------------------------------------------------------

print()

for line in txt:
    print(line)
```

```python
# ============================================================
# OIP v1.0.35 — CELL 12
# FINAL ARCHIVAL FREEZE MANIFEST
# CORRECTED FINAL VERSION
# ============================================================

import json
import hashlib
import csv
from pathlib import Path
from datetime import datetime, timezone

print("=" * 72)
print("OIP v1.0.35 — C12: FINAL ARCHIVAL FREEZE MANIFEST")
print("=" * 72)

# ------------------------------------------------------------
# 1. PATHS
# ------------------------------------------------------------

BASE = Path("/kaggle/working/oip_v1_0_35")

UPSTREAM = {
    "C00": BASE / "c00_source_scope" /
           "c00_source_scope.json",

    "C01": BASE / "c01_artifact_inventory" /
           "c01_artifact_inventory.json",

    "C02": BASE / "c02_artifact_identity" /
           "c02_artifact_identity.json",

    "C03": BASE / "c03_authority_assignment" /
           "c03_authority_assignment.json",

    "C04": BASE / "c04_conflict_register" /
           "c04_conflict_register.json",

    "C05": BASE / "c05_malawi_primary_verification" /
           "c05_malawi_primary_verification.json",

    "C06": BASE / "c06_ethiopia_primary_verification" /
           "c06_ethiopia_primary_verification.json",

    "C07": BASE / "c07_v1_0_28_provenance" /
           "c07_v1_0_28_provenance.json",

    "C08": BASE / "c08_v1_0_26_provenance" /
           "c08_v1_0_26_provenance.json",

    "C09": BASE / "c09_final_conflict_consistency" /
           "c09_final_conflict_consistency.json",

    "C10": BASE / "c10_archival_freeze_precondition" /
           "c10_archival_freeze_precondition.json",

    "C11": BASE / "c11_corrected_public_record" /
           "c11_corrected_public_record.json",
}

OUTDIR = BASE / "c12_final_archival_freeze_manifest"
OUTDIR.mkdir(parents=True, exist_ok=True)

OUT_JSON = OUTDIR / "c12_final_archival_freeze_manifest.json"
OUT_TXT = OUTDIR / "c12_final_archival_freeze_manifest.txt"
OUT_CSV = OUTDIR / "c12_artifact_sha256.csv"
OUT_HASH = OUTDIR / "c12_manifest_sha256.txt"


# ------------------------------------------------------------
# 2. HELPERS
# ------------------------------------------------------------

def load_json(path):
    if not path.exists():
        raise FileNotFoundError(
            f"Required artifact missing: {path}"
        )

    with path.open("r", encoding="utf-8") as f:
        return json.load(f)


def sha256_file(path, chunk_size=1024 * 1024):
    h = hashlib.sha256()

    with path.open("rb") as f:
        while True:
            chunk = f.read(chunk_size)

            if not chunk:
                break

            h.update(chunk)

    return h.hexdigest()


def require(condition, message, failures):
    if not condition:
        failures.append(message)


# ------------------------------------------------------------
# 3. LOAD C00-C11
# ------------------------------------------------------------

failures = []
artifacts = {}

for cell, path in UPSTREAM.items():

    if not path.exists():

        failures.append(
            f"{cell}: required artifact missing: {path}"
        )

        continue

    try:
        artifacts[cell] = load_json(path)

    except Exception as exc:

        failures.append(
            f"{cell}: JSON load failure: {exc}"
        )


if failures:

    print()
    print("C12 STATUS : FAIL")
    print("-" * 72)

    for failure in failures:
        print("FAIL:", failure)

    print()
    print("FINAL MANIFEST NOT CREATED.")

    raise RuntimeError(
        "C12 failed closed during upstream loading."
    )


# ------------------------------------------------------------
# 4. C00
# ------------------------------------------------------------

require(
    artifacts["C00"].get("scope_status") == "PASS",
    "C00 scope_status is not PASS",
    failures,
)


# ------------------------------------------------------------
# 5. C01
# ------------------------------------------------------------

c01 = artifacts["C01"]
c01_summary = c01.get("inventory_summary")

require(
    isinstance(c01_summary, dict),
    "C01 inventory_summary missing or invalid",
    failures,
)

if isinstance(c01_summary, dict):

    require(
        c01_summary.get("inventory_complete") is True,
        "C01 inventory_complete is not True",
        failures,
    )

require(
    isinstance(c01.get("errors"), list),
    "C01 errors is not a list",
    failures,
)

if isinstance(c01.get("errors"), list):

    require(
        len(c01["errors"]) == 0,
        "C01 contains inventory errors",
        failures,
    )

require(
    isinstance(c01.get("artifacts"), list)
    and len(c01.get("artifacts")) > 0,
    "C01 artifacts list is missing or empty",
    failures,
)


# ------------------------------------------------------------
# 6. C02
# ------------------------------------------------------------

require(
    artifacts["C02"].get("identity_status") == "PASS",
    "C02 identity_status is not PASS",
    failures,
)


# ------------------------------------------------------------
# 7. C03
# ------------------------------------------------------------

c03 = artifacts["C03"]

require(
    c03.get("substantive_authority_count") == 0,
    "C03 substantive authority count is not zero",
    failures,
)

require(
    isinstance(c03.get("authority_records"), list),
    "C03 authority_records is not a list",
    failures,
)


# ------------------------------------------------------------
# 8. C04
# ------------------------------------------------------------

c04 = artifacts["C04"]

require(
    c04.get("register_status") == "PASS",
    "C04 register_status is not PASS",
    failures,
)

c04_conflicts = c04.get("conflicts")
c04_open_items = c04.get("open_governance_items")

require(
    isinstance(c04_conflicts, list),
    "C04 conflicts is not a list",
    failures,
)

require(
    isinstance(c04_open_items, list),
    "C04 open_governance_items is not a list",
    failures,
)

if isinstance(c04_conflicts, list):

    require(
        c04.get("conflict_count") == len(c04_conflicts),
        "C04 conflict_count does not match conflicts",
        failures,
    )

if isinstance(c04_open_items, list):

    require(
        c04.get("open_item_count") == len(c04_open_items),
        "C04 open_item_count does not match open items",
        failures,
    )


# ------------------------------------------------------------
# 9. C05
# ------------------------------------------------------------

c05 = artifacts["C05"]

require(
    isinstance(c05.get("evidence_state"), dict),
    "C05 evidence_state missing or invalid",
    failures,
)

require(
    c05.get("conflict_resolution_performed") is False,
    "C05 conflict resolution was not FALSE",
    failures,
)

require(
    c05.get("canonical_source_declared") is False,
    "C05 canonical source was not FALSE",
    failures,
)


# ------------------------------------------------------------
# 10. C06
# ------------------------------------------------------------

c06 = artifacts["C06"]

require(
    isinstance(c06.get("artifact_state"), dict),
    "C06 artifact_state missing or invalid",
    failures,
)

require(
    c06.get("conflict_resolution_performed") is False,
    "C06 conflict resolution was not FALSE",
    failures,
)

require(
    c06.get("canonical_source_declared") is False,
    "C06 canonical source was not FALSE",
    failures,
)


# ------------------------------------------------------------
# 11. C07
# ------------------------------------------------------------

c07 = artifacts["C07"]

require(
    c07.get("status") == "PASS",
    "C07 status is not PASS",
    failures,
)

c07_claim = c07.get("claim_boundary")

require(
    isinstance(c07_claim, dict),
    "C07 claim_boundary missing or invalid",
    failures,
)

if isinstance(c07_claim, dict):

    require(
        c07_claim.get("generator_artifact")
        == "NOT_ESTABLISHED",
        "C07 generator_artifact boundary changed",
        failures,
    )

    require(
        c07_claim.get("formal_retraction")
        == "NOT_ESTABLISHED",
        "C07 formal_retraction boundary changed",
        failures,
    )

    require(
        c07_claim.get("current_independent_validation_use")
        == "NOT_AUTHORIZED",
        "C07 independent validation-use boundary changed",
        failures,
    )


# ------------------------------------------------------------
# 12. C08
# ------------------------------------------------------------

c08 = artifacts["C08"]

require(
    c08.get("status") == "PASS",
    "C08 status is not PASS",
    failures,
)

c08_claim = c08.get("claim_boundary")

require(
    isinstance(c08_claim, dict),
    "C08 claim_boundary missing or invalid",
    failures,
)

if isinstance(c08_claim, dict):

    expected_c08 = {
        "legal_nda_existence":
            "NOT_ESTABLISHED_BY_SUPPLIED_SOURCE_SET",

        "nda_with_openai_existence":
            "NOT_ESTABLISHED_BY_SUPPLIED_SOURCE_SET",

        "assistant_text_authorship":
            "NOT_ESTABLISHED",

        "formula_conflict_state":
            "UNRESOLVED",

        "canonical_formula":
            "NOT_ESTABLISHED",

        "empirical_validity":
            "NOT_AUTHORIZED",
    }

    for key, expected in expected_c08.items():

        require(
            c08_claim.get(key) == expected,
            f"C08 claim boundary changed: {key}",
            failures,
        )


# ------------------------------------------------------------
# 13. C09
# ------------------------------------------------------------

c09 = artifacts["C09"]

require(
    c09.get("status") == "PASS",
    "C09 status is not PASS",
    failures,
)

require(
    c09.get("checks_failed") == 0,
    "C09 checks_failed is not zero",
    failures,
)

require(
    c09.get("failed_checks") == [],
    "C09 failed_checks is not empty",
    failures,
)


# ------------------------------------------------------------
# 14. C10
#
# IMPORTANT:
# C10 does NOT contain governance_boundary.
# Only its actual PASS state is validated here.
# ------------------------------------------------------------

c10 = artifacts["C10"]

require(
    c10.get("status") == "PASS",
    "C10 status is not PASS",
    failures,
)


# ------------------------------------------------------------
# 15. C11
# ------------------------------------------------------------

c11 = artifacts["C11"]

require(
    c11.get("status") == "PASS",
    "C11 status is not PASS",
    failures,
)

require(
    c11.get("failed_checks") == [],
    "C11 failed_checks is not empty",
    failures,
)

require(
    c11.get("next_cell")
    == "C12 — FINAL ARCHIVAL FREEZE MANIFEST",
    "C11 next_cell is not C12",
    failures,
)


# ------------------------------------------------------------
# 16. C11 GOVERNANCE SAFETY
# ------------------------------------------------------------

c11_governance = c11.get("governance")

require(
    isinstance(c11_governance, dict),
    "C11 governance is missing or invalid",
    failures,
)

if isinstance(c11_governance, dict):

    for key, value in c11_governance.items():

        if isinstance(value, bool) and value is True:

            failures.append(
                f"C11 forbidden governance TRUE: {key}=True"
            )


# ------------------------------------------------------------
# 17. FINAL CLAIM BOUNDARY
# ------------------------------------------------------------

c11_boundary = c11.get("current_evidence_boundary")

require(
    isinstance(c11_boundary, dict),
    "C11 current_evidence_boundary missing or invalid",
    failures,
)


# ------------------------------------------------------------
# 18. FINAL FAIL-CLOSED GATE
# ------------------------------------------------------------

if failures:

    print()
    print("C12 STATUS : FAIL")
    print("-" * 72)

    for failure in failures:
        print("FAIL:", failure)

    print()
    print("FINAL MANIFEST NOT CREATED.")

    raise RuntimeError(
        "C12 failed closed. Review failed checks."
    )


# ------------------------------------------------------------
# 19. SHA-256 C00-C11
# ------------------------------------------------------------

artifact_records = []

for cell, path in UPSTREAM.items():

    digest = sha256_file(path)
    size = path.stat().st_size

    artifact_records.append(
        {
            "cell": cell,
            "artifact": path.name,
            "path": str(path),
            "size_bytes": size,
            "sha256": digest,
        }
    )


# ------------------------------------------------------------
# 20. FINAL GOVERNANCE BOUNDARY
# ------------------------------------------------------------

final_governance = {

    "conflict_resolution_performed": False,
    "canonical_source_declared": False,

    "construct_authorization": False,
    "score_authorization": False,
    "outcome_authorization": False,

    "empirical_evaluation": False,
    "empirical_validity": False,

    "formal_retraction": False,
    "formula_version_lock": False,

    "new_dataset_introduced": False,
    "new_construct_introduced": False,
    "new_outcome_introduced": False,
    "new_score_introduced": False,

    "post_hoc_tuning": False,
    "silent_reconciliation": False,
}


# ------------------------------------------------------------
# 21. FINAL DATASET STATE
# ------------------------------------------------------------

final_dataset_state = {

    "BEMP_v1_0_29":
        "FAIL_CLOSED_LOGIC_PRESERVED_EXECUTION_NOT_RECONSTRUCTIBLE",

    "Malawi_v1_0_31":
        "DOCUMENTED_FAIL_CLOSED_RECORD_WITH_UNRESOLVED_CONFLICTS",

    "Uganda_v1_0_32":
        "DOCUMENTED_FAIL_CLOSED_RECORD",

    "Ethiopia_v1_0_33":
        "DOCUMENTED_FAIL_CLOSED_RECORD_WITH_OBSERVED_PRIMARY_ARTIFACT_SIGNALS",
}


# ------------------------------------------------------------
# 22. PUBLIC CLAIM
# ------------------------------------------------------------

final_public_claim = (
    "The OIP v1.0.35 reconciliation record documents an "
    "evidence-first, fail-closed audit framework and preserves "
    "the current evidence boundary across the audited records. "
    "The current record does not establish empirical validity, "
    "predictive validity, causal validity, universal validity, "
    "or superiority of OIP. Several artifact-level and "
    "provenance-level items remain explicitly unresolved and "
    "are preserved as unresolved in this archival freeze."
)

final_fail_closed_statement = (
    "Unresolved evidence remains unresolved. No missing evidence "
    "is replaced by inference, assumption, proxy substitution, "
    "synthetic outcome, silent imputation, random fallback, "
    "or post-hoc selection."
)


# ------------------------------------------------------------
# 23. BUILD MANIFEST
# ------------------------------------------------------------

timestamp = (
    datetime.now(timezone.utc)
    .replace(microsecond=0)
    .isoformat()
)

manifest = {

    "protocol":
        "OIP v1.0.35",

    "cell":
        "C12",

    "cell_name":
        "FINAL ARCHIVAL FREEZE MANIFEST",

    "created_utc":
        timestamp,

    "release_type":
        "RECONCILIATION_AND_ARCHIVAL_FREEZE",

    "release_scope":
        {
            "introduces_new_dataset": False,
            "introduces_new_construct": False,
            "introduces_new_outcome": False,
            "introduces_new_score": False,
            "introduces_new_empirical_evaluation": False,
            "introduces_new_validity_claim": False,
        },

    "upstream_cells":
        list(UPSTREAM.keys()),

    "upstream_artifact_count":
        len(artifact_records),

    "artifact_records":
        artifact_records,

    "registered_conflicts_count":
        len(c04_conflicts),

    "open_governance_items_count":
        len(c04_open_items),

    "final_governance_boundary":
        final_governance,

    "final_claim_boundary":
        c11_boundary,

    "final_dataset_state":
        final_dataset_state,

    "public_claim_statement":
        final_public_claim,

    "fail_closed_statement":
        final_fail_closed_statement,

    "manifest_status":
        "FROZEN",
}


# ------------------------------------------------------------
# 24. WRITE FINAL JSON
# ------------------------------------------------------------

with OUT_JSON.open(
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        manifest,
        f,
        indent=2,
        ensure_ascii=False,
    )


# ------------------------------------------------------------
# 25. WRITE SHA-256 CSV
#
# IMPORTANT:
# CSV fieldnames now include ALL fields in artifact_records,
# including 'path'.
# ------------------------------------------------------------

with OUT_CSV.open(
    "w",
    newline="",
    encoding="utf-8"
) as f:

    writer = csv.DictWriter(
        f,
        fieldnames=[
            "cell",
            "artifact",
            "path",
            "size_bytes",
            "sha256",
        ],
    )

    writer.writeheader()

    for record in artifact_records:

        writer.writerow(record)


# ------------------------------------------------------------
# 26. HASH FINAL MANIFEST JSON
# ------------------------------------------------------------

manifest_digest = sha256_file(
    OUT_JSON
)

with OUT_HASH.open(
    "w",
    encoding="utf-8"
) as f:

    f.write(
        manifest_digest + "\n"
    )


# ------------------------------------------------------------
# 27. HUMAN-READABLE RECORD
# ------------------------------------------------------------

lines = []

lines.append("=" * 72)

lines.append(
    "OIP v1.0.35 — C12: FINAL ARCHIVAL FREEZE MANIFEST"
)

lines.append("=" * 72)

lines.append("")

lines.append(
    f"Created UTC : {timestamp}"
)

lines.append(
    "Release type : RECONCILIATION_AND_ARCHIVAL_FREEZE"
)

lines.append("")

lines.append(
    f"Upstream cells frozen : "
    f"{len(artifact_records)}/12"
)

lines.append("")

lines.append(
    "Artifact SHA-256:"
)

for record in artifact_records:

    lines.append(
        f"  {record['cell']} : "
        f"{record['sha256']}"
    )

lines.append("")

lines.append(
    f"Registered conflicts   : "
    f"{len(c04_conflicts)}"
)

lines.append(
    f"Open governance items  : "
    f"{len(c04_open_items)}"
)

lines.append("")

lines.append(
    "Final governance boundary:"
)

for key, value in final_governance.items():

    lines.append(
        f"  {key:38s} : "
        f"{'YES' if value else 'NO'}"
    )

lines.append("")

lines.append(
    "Final dataset state:"
)

for key, value in final_dataset_state.items():

    lines.append(
        f"  {key:22s} : {value}"
    )

lines.append("")

lines.append(
    "Public claim:"
)

lines.append(
    f"  {final_public_claim}"
)

lines.append("")

lines.append(
    "Fail-closed statement:"
)

lines.append(
    f"  {final_fail_closed_statement}"
)

lines.append("")

lines.append(
    "MANIFEST STATUS : FROZEN"
)

lines.append(
    f"MANIFEST SHA-256 : {manifest_digest}"
)

lines.append("")

lines.append(
    f"JSON : {OUT_JSON}"
)

lines.append(
    f"TXT  : {OUT_TXT}"
)

lines.append(
    f"CSV  : {OUT_CSV}"
)

lines.append(
    f"SHA  : {OUT_HASH}"
)

lines.append("")

lines.append(
    "OIP v1.0.35 ARCHIVAL FREEZE COMPLETE"
)

lines.append("=" * 72)


with OUT_TXT.open(
    "w",
    encoding="utf-8"
) as f:

    f.write(
        "\n".join(lines)
    )


# ------------------------------------------------------------
# 28. FINAL OUTPUT
# ------------------------------------------------------------

print()
print("\n".join(lines))
```

```python
# ============================================================
# OIP v1.0.35 — ARCHIVAL CORRECTION NOTE
# POST-FREEZE SCHEMA DIAGNOSTIC OBSERVATION
#
# IMPORTANT:
# This cell does NOT modify C12.
# This cell does NOT re-run C12.
# This cell does NOT change the frozen manifest SHA-256.
# ============================================================

import json
import hashlib
from pathlib import Path
from datetime import datetime, timezone

print("=" * 72)
print("OIP v1.0.35 — ARCHIVAL CORRECTION NOTE")
print("=" * 72)

# ------------------------------------------------------------
# 1. PATHS
# ------------------------------------------------------------

BASE = Path("/kaggle/working/oip_v1_0_35")

C10_PATH = (
    BASE
    / "c10_archival_freeze_precondition"
    / "c10_archival_freeze_precondition.json"
)

C12_DIR = (
    BASE
    / "c12_final_archival_freeze_manifest"
)

C12_PATH = (
    C12_DIR
    / "c12_final_archival_freeze_manifest.json"
)

C12_HASH_PATH = (
    C12_DIR
    / "c12_manifest_sha256.txt"
)

CORRECTION_PATH = (
    C12_DIR
    / "c12_correction_note.json"
)


# ------------------------------------------------------------
# 2. LOAD C10 AND C12
# ------------------------------------------------------------

if not C10_PATH.exists():
    raise FileNotFoundError(
        f"C10 artifact missing: {C10_PATH}"
    )

if not C12_PATH.exists():
    raise FileNotFoundError(
        f"Frozen C12 manifest missing: {C12_PATH}"
    )

if not C12_HASH_PATH.exists():
    raise FileNotFoundError(
        f"C12 SHA-256 file missing: {C12_HASH_PATH}"
    )


with C10_PATH.open(
    "r",
    encoding="utf-8"
) as f:
    c10 = json.load(f)


with C12_PATH.open(
    "r",
    encoding="utf-8"
) as f:
    c12 = json.load(f)


# ------------------------------------------------------------
# 3. VERIFY C10 ACTUAL SCHEMA
# ------------------------------------------------------------

c10_governance = c10.get("governance")

if not isinstance(c10_governance, dict):
    raise RuntimeError(
        "C10 governance is not a dictionary. "
        "Correction note not created."
    )


if "governance_boundary" in c10:
    raise RuntimeError(
        "Unexpected C10 governance_boundary key detected. "
        "Correction note requires manual review."
    )


if c10.get("status") != "PASS":
    raise RuntimeError(
        "C10 status is not PASS. "
        "Correction note not created."
    )


# ------------------------------------------------------------
# 4. VERIFY C10 GOVERNANCE VALUES
# ------------------------------------------------------------

expected_governance = {
    "conflict_resolution_performed": False,
    "canonical_source_declared": False,
    "construct_authorization": False,
    "score_authorization": False,
    "outcome_authorization": False,
    "empirical_evaluation_authorization": False,
    "empirical_validity_claim": False,
    "final_manifest_created": False,
}


governance_mismatches = {}

for key, expected in expected_governance.items():

    actual = c10_governance.get(key)

    if actual != expected:

        governance_mismatches[key] = {
            "expected": expected,
            "actual": actual,
        }


if governance_mismatches:

    print()
    print("C10 GOVERNANCE MISMATCH DETECTED")
    print(json.dumps(
        governance_mismatches,
        indent=2,
        ensure_ascii=False
    ))

    raise RuntimeError(
        "C10 governance values do not match the verified state. "
        "Correction note not created."
    )


# ------------------------------------------------------------
# 5. VERIFY FROZEN C12
# ------------------------------------------------------------

if c12.get("manifest_status") != "FROZEN":

    raise RuntimeError(
        "C12 manifest is not FROZEN. "
        "Correction note not created."
    )


if c12.get("upstream_artifact_count") != 12:

    raise RuntimeError(
        "C12 upstream artifact count is not 12. "
        "Correction note not created."
    )


# ------------------------------------------------------------
# 6. READ EXISTING C12 SHA
# ------------------------------------------------------------

recorded_sha = (
    C12_HASH_PATH
    .read_text(encoding="utf-8")
    .strip()
)


# ------------------------------------------------------------
# 7. INDEPENDENTLY COMPUTE C12 SHA
# ------------------------------------------------------------

def sha256_file(path):

    h = hashlib.sha256()

    with path.open("rb") as f:

        while True:

            chunk = f.read(1024 * 1024)

            if not chunk:
                break

            h.update(chunk)

    return h.hexdigest()


actual_sha = sha256_file(C12_PATH)


if recorded_sha != actual_sha:

    raise RuntimeError(
        "C12 recorded SHA-256 does not match the actual "
        "frozen manifest. Correction note not created."
    )


# ------------------------------------------------------------
# 8. CREATE CORRECTION NOTE
# ------------------------------------------------------------

created_utc = (
    datetime.now(timezone.utc)
    .replace(microsecond=0)
    .isoformat()
)


correction_note = {

    "protocol":
        "OIP v1.0.35",

    "note_type":
        "POST_FREEZE_SCHEMA_DIAGNOSTIC_OBSERVATION",

    "created_utc":
        created_utc,

    "status":
        "DOCUMENTED_OBSERVATION_NOT_RESOLUTION",

    "observation":
        (
            "An auxiliary read-only diagnostic reported one failed "
            "check: C10 governance_boundary expected a dictionary "
            "but returned NoneType."
        ),

    "verified_root_cause":
        (
            "The diagnostic referenced a C10 JSON key named "
            "'governance_boundary'. Direct read-only inspection "
            "of the frozen C10 artifact established that the "
            "actual C10 schema contains 'governance' and does not "
            "contain 'governance_boundary'."
        ),

    "c10_verification":
        {
            "status":
                c10.get("status"),

            "governance_key_present":
                "governance" in c10,

            "governance_boundary_key_present":
                "governance_boundary" in c10,

            "governance_values_verified":
                True,
        },

    "diagnostic_interpretation":
        (
            "The reported diagnostic failure was caused by a "
            "schema-key mismatch in the auxiliary diagnostic "
            "check. It does not establish a failure of the C10 "
            "archival-freeze precondition artifact."
        ),

    "c12_frozen_manifest":
        {
            "manifest_status":
                c12.get("manifest_status"),

            "upstream_artifact_count":
                c12.get("upstream_artifact_count"),

            "sha256":
                actual_sha,

            "recorded_sha256":
                recorded_sha,

            "sha256_verified":
                True,
        },

    "governance":
        {
            "conflict_resolution_performed": False,
            "canonical_source_declared": False,
            "construct_authorization": False,
            "score_authorization": False,
            "outcome_authorization": False,
            "empirical_evaluation": False,
            "empirical_validity": False,
            "manifest_modified": False,
            "new_analysis_introduced": False,
            "c12_rerun_performed": False,
            "frozen_manifest_modified": False,
        },

    "preservation_rules":
        [
            "The C12 frozen manifest is not modified.",
            "The C12 frozen manifest is not re-run.",
            "The C12 SHA-256 is not changed.",
            "No conflict is resolved by this note.",
            "No construct is authorized by this note.",
            "No score is authorized by this note.",
            "No outcome is authorized by this note.",
            "No empirical evaluation is authorized by this note.",
            "No empirical validity claim is introduced by this note.",
        ],

    "evidence_boundary":
        (
            "This correction note records a verified schema "
            "observation only. It does not establish empirical "
            "validity, predictive validity, causal validity, "
            "universal validity, superiority, or any new "
            "downstream empirical result."
        ),
}


# ------------------------------------------------------------
# 9. WRITE SEPARATE CORRECTION NOTE
# ------------------------------------------------------------

with CORRECTION_PATH.open(
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        correction_note,
        f,
        indent=2,
        ensure_ascii=False,
    )


# ------------------------------------------------------------
# 10. VERIFY CORRECTION NOTE
# ------------------------------------------------------------

with CORRECTION_PATH.open(
    "r",
    encoding="utf-8"
) as f:

    correction_check = json.load(f)


if correction_check.get(
    "status"
) != "DOCUMENTED_OBSERVATION_NOT_RESOLUTION":

    raise RuntimeError(
        "Correction note verification failed."
    )


# ------------------------------------------------------------
# 11. FINAL OUTPUT
# ------------------------------------------------------------

print()
print("C10 schema verification        : PASS")
print("C10 status                     : PASS")
print("C10 'governance'               : PRESENT")
print("C10 'governance_boundary'      : ABSENT")
print("C10 governance values          : VERIFIED")
print()
print("C12 manifest status             : FROZEN")
print("C12 upstream artifacts          : 12")
print("C12 SHA-256 independently verified : YES")
print("C12 frozen manifest modified    : NO")
print("C12 re-run performed            : NO")
print()
print("Correction note status          : CREATED")
print()
print("Correction note:")
print(f"  {CORRECTION_PATH}")
print()
print("Frozen C12 SHA-256:")
print(f"  {actual_sha}")
print()
print("ARCHIVAL CORRECTION NOTE COMPLETE")
print("OIP v1.0.35 FROZEN MANIFEST PRESERVED")
print("=" * 72)
```
